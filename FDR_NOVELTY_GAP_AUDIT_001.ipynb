```python
# ================================================================================================================
# CELL 1 v001 — LITERATURE GAP AUDIT SEARCH PROTOCOL
# ================================================================================================================

print("""
================================================================================================================
CELL 1 v001 — LITERATURE GAP AUDIT SEARCH PROTOCOL
================================================================================================================
""")

# ================================================================================================================
# STEP 0 — SCIENTIFIC VALUE GATE
# ================================================================================================================

print("""
STEP 0 — SCIENTIFIC VALUE GATE
----------------------------------------------------------------------------------------------------------------
Mandatory question:

ما فائدتها للرسالة العلمية أو لبناء الـdataset؟

Answer:
To freeze the literature-search protocol before reviewing results,
so that the novelty/gap assessment is systematic, reproducible,
bounded, and not driven by post-hoc selection of favorable papers.

✓ DIRECT VALUE TO THE THESIS
✓ SUPPORTS A DEFENSIBLE NOVELTY CLAIM
✓ NO DATASET MODIFICATION
✓ NO MODEL INFERENCE
✓ NO TRAINING
✓ NO ARCHITECTURE IMPLEMENTATION
""")


# ================================================================================================================
# STEP 1 — EXPERIMENT IDENTITY
# ================================================================================================================

EXPERIMENT_ID = "FDR-NOVELTY-GAP-001"
CELL_ID = "CELL_1"
CELL_VERSION = "V001"

print("STEP 1 — EXPERIMENT IDENTITY")
print("-" * 112)
print(f"Experiment ID : {EXPERIMENT_ID}")
print(f"Cell          : {CELL_ID}")
print(f"Version       : {CELL_VERSION}")


# ================================================================================================================
# STEP 2 — SEARCH OBJECTIVE
# ================================================================================================================

SEARCH_OBJECTIVE = """
Identify prior work that is substantially relevant to the proposed
F-D+R contribution, with particular emphasis on post-retrieval
evidence verification and evidence suitability before generation.

The search must determine whether the proposed contribution is:

A) clearly distinct,
B) partially overlapping but still defensible, or
C) substantially covered by prior work.
""".strip()

print("STEP 2 — SEARCH OBJECTIVE")
print("-" * 112)
print(SEARCH_OBJECTIVE)


# ================================================================================================================
# STEP 3 — SEARCH CONCEPT FAMILIES
# ================================================================================================================

SEARCH_CONCEPT_FAMILIES = {

    "F1_EVIDENCE_VERIFICATION": [
        "evidence verification",
        "evidence verifier",
        "evidence validation",
        "retrieved evidence verification",
        "RAG evidence verification",
    ],

    "F2_EVIDENCE_SUITABILITY": [
        "evidence suitability",
        "evidence quality",
        "evidence adequacy",
        "evidence sufficiency",
        "evidence support",
    ],

    "F3_CLAIM_EVIDENCE_ALIGNMENT": [
        "claim evidence alignment",
        "claim-evidence verification",
        "claim verification RAG",
        "claim groundedness evidence",
        "claim support verification",
    ],

    "F4_RAG_GROUNDING_FAITHFULNESS": [
        "RAG grounding verification",
        "RAG faithfulness verification",
        "retrieval augmented generation evidence grounding",
        "RAG factuality verification",
        "RAG hallucination evidence verification",
    ],

    "F5_CITATION_VERIFICATION": [
        "citation verification RAG",
        "citation correctness RAG",
        "citation faithfulness RAG",
        "citation evidence verification",
        "source attribution verification RAG",
    ],

    "F6_LEGAL_RAG": [
        "legal RAG evidence verification",
        "legal retrieval augmented generation verification",
        "legal RAG evidence grounding",
        "legal RAG citation verification",
        "legal RAG hallucination verification",
    ],

    "F7_PRE_GENERATION_CONTROL": [
        "evidence filtering before generation RAG",
        "evidence selection before generation RAG",
        "evidence gating RAG",
        "retrieval evidence filtering RAG",
        "pre-generation evidence selection",
    ],

    "F8_MULTILINGUAL_ARABIC_ENGLISH": [
        "Arabic English legal RAG evidence",
        "multilingual legal RAG evidence verification",
        "Arabic legal RAG verification",
        "cross-lingual legal RAG evidence",
    ],
}

print("STEP 3 — SEARCH CONCEPT FAMILIES")
print("-" * 112)

for family, terms in SEARCH_CONCEPT_FAMILIES.items():
    print(f"\n{family}")
    for term in terms:
        print(f"  • {term}")


# ================================================================================================================
# STEP 4 — PRIMARY SEARCH STRATEGY
# ================================================================================================================

print("""
STEP 4 — PRIMARY SEARCH STRATEGY
----------------------------------------------------------------------------------------------------------------

The audit will use concept combinations rather than a single broad query.

Core structure:

    [RAG / Legal RAG]
             +
    [Evidence / Claim / Citation]
             +
    [Verification / Sufficiency / Grounding / Gating]

Priority is given to papers whose task formulation is structurally
close to the proposed F-D+R problem.

The search is NOT limited to papers using the exact phrase
"Evidence Suitability".

Equivalent terminology must be considered because the same research
problem may appear under different names.
""")


# ================================================================================================================
# STEP 5 — INCLUSION CRITERIA
# ================================================================================================================

INCLUSION_CRITERIA = [
    "Addresses retrieval-augmented generation or a closely related retrieval-grounded generation setting.",
    "Evaluates, verifies, selects, filters, or judges retrieved evidence/source material.",
    "Contains a task or mechanism occurring after retrieval or explicitly uses retrieved evidence.",
    "Has a clearly defined evidence/grounding/support/verification target.",
    "Provides enough methodological detail to compare its task with F-D+R.",
    "Published or publicly available as a scholarly research work.",
]

print("STEP 5 — INCLUSION CRITERIA")
print("-" * 112)

for i, criterion in enumerate(INCLUSION_CRITERIA, 1):
    print(f"{i:02d}. {criterion}")


# ================================================================================================================
# STEP 6 — EXCLUSION CRITERIA
# ================================================================================================================

EXCLUSION_CRITERIA = [
    "Pure retrieval-ranking work with no evidence verification or evidence-use decision.",
    "Pure generation work with no explicit evidence/grounding evaluation.",
    "Generic text classification unrelated to retrieved evidence.",
    "Generic NLI work with no relevant evidence-grounding application or methodological connection.",
    "Citation recommendation alone without citation/evidence correctness verification.",
    "Pure hallucination detection that does not evaluate retrieved evidence.",
    "Non-scholarly material unless it is needed only as a trace to a primary scholarly source.",
    "Duplicate versions of the same study.",
]

print("\nSTEP 6 — EXCLUSION CRITERIA")
print("-" * 112)

for i, criterion in enumerate(EXCLUSION_CRITERIA, 1):
    print(f"{i:02d}. {criterion}")


# ================================================================================================================
# STEP 7 — PRIORITY LEVELS
# ================================================================================================================

PRIORITY_LEVELS = {
    "P1_DIRECT": "Directly evaluates retrieved evidence before or during generation.",
    "P2_CLOSE": "Evaluates claim/evidence support, sufficiency, grounding, or citation correctness in RAG.",
    "P3_RELATED": "Methodologically relevant but lacks a major part of the proposed F-D+R formulation.",
    "P4_BACKGROUND": "Provides useful conceptual/background context but is not a direct competitor.",
}

print("""
STEP 7 — PRIORITY LEVELS
----------------------------------------------------------------------------------------------------------------
""")

for level, definition in PRIORITY_LEVELS.items():
    print(f"{level:15s}: {definition}")


# ================================================================================================================
# STEP 8 — REQUIRED COMPARISON FIELDS
# ================================================================================================================

COMPARISON_FIELDS = [
    "Paper",
    "Year",
    "Venue",
    "Domain",
    "Language setting",
    "Dataset / benchmark",
    "Retrieval stage",
    "Evidence-verification stage",
    "Input to verifier",
    "Target / label",
    "Human grounding",
    "Support distinction",
    "Sufficiency distinction",
    "Contradiction / risk",
    "Uncertainty",
    "Pre-generation gating",
    "Legal-domain coverage",
    "Arabic-English / multilingual coverage",
    "Reference-answer dependency",
    "Main limitation relative to F-D+R",
    "Overlap level",
]

print("""
STEP 8 — REQUIRED COMPARISON FIELDS
----------------------------------------------------------------------------------------------------------------
Every P1 and P2 paper must eventually be represented using the same
comparison structure.
""")

for i, field in enumerate(COMPARISON_FIELDS, 1):
    print(f"{i:02d}. {field}")


# ================================================================================================================
# STEP 9 — ANCHOR PAPER HANDLING
# ================================================================================================================

ANCHOR_PAPER = (
    "The Cross-Lingual Cost: Retrieval Biases in RAG over Arabic-English Corpora"
)

print("""
STEP 9 — ANCHOR PAPER
----------------------------------------------------------------------------------------------------------------
""")

print(f"Anchor: {ANCHOR_PAPER}")

print("""
The anchor paper is retained as a high-priority reference because it
shares the Arabic-English Legal benchmark/corpus lineage.

Its role is NOT to establish the new F-D+R gap.

Its role is to establish the already-published boundary:

✓ Cross-lingual retrieval bias
✓ Legal retrieval evaluation
✓ Retrieval bottleneck analysis

The audit must therefore search beyond retrieval and specifically
investigate post-retrieval evidence verification.
""")


# ================================================================================================================
# STEP 10 — BIAS CONTROL
# ================================================================================================================

print("""
STEP 10 — LITERATURE SELECTION BIAS CONTROL
----------------------------------------------------------------------------------------------------------------

The audit MUST NOT:

✗ Select only papers that support the proposed novelty.
✗ Ignore papers using different terminology for the same task.
✗ Treat absence of the exact phrase "Evidence Suitability" as evidence of novelty.
✗ Declare novelty because no paper was found in one search query.
✗ Upgrade a related paper to "different" without comparing task boundaries.

The audit MUST:

✓ Search multiple terminology families.
✓ Record close negative findings.
✓ Preserve papers that challenge the proposed gap.
✓ Compare methodology rather than titles alone.
✓ Record uncertainty when task equivalence cannot be established.
""")


# ================================================================================================================
# STEP 11 — BOUNDED SEARCH RULE
# ================================================================================================================

print("""
STEP 11 — BOUNDED SEARCH RULE
----------------------------------------------------------------------------------------------------------------

The audit is finite.

Initial target:

• Anchor paper: 1
• P1 direct candidates: up to 10
• P2 close candidates: up to 15
• P3 related candidates: up to 10

The objective is NOT to collect the largest possible bibliography.

The objective is to identify the closest competing formulations and
determine whether the proposed F-D+R contribution remains defensible.

If the same highly relevant papers repeatedly appear across search
families, additional broad searching should stop and effort should
shift to methodological comparison.
""")


# ================================================================================================================
# STEP 12 — DECISION STATES
# ================================================================================================================

DECISION_STATES = {
    "GAP_DEFENSIBLE":
        "No prior work identified with substantially equivalent task formulation and control boundary.",
    "GAP_PARTIALLY_OVERLAPPING":
        "Related components exist, but the complete F-D+R formulation remains distinguishable.",
    "GAP_NOT_DEFENSIBLE":
        "A prior work or tightly connected body of work substantially covers the proposed contribution.",
    "GAP_INCONCLUSIVE":
        "Available evidence is insufficient to make a defensible novelty decision.",
}

print("""
STEP 12 — DECISION STATES
----------------------------------------------------------------------------------------------------------------
""")

for state, meaning in DECISION_STATES.items():
    print(f"{state:25s} → {meaning}")


# ================================================================================================================
# STEP 13 — FIREWALL CONFIRMATION
# ================================================================================================================

print("""
STEP 13 — FIREWALL CONFIRMATION
----------------------------------------------------------------------------------------------------------------

✓ Literature search only
✓ No dataset modification
✓ No model loading
✓ No model inference
✓ No training
✓ No prompt modification
✓ No architecture implementation
✓ No new adjudication
✓ No Gold generation
✓ No Y+ / Y− generation
✓ No Preference generation
✓ 199 cohort remains frozen
✓ 400 validation remains closed
✓ Q82_01–Q82_10 remains frozen
""")


# ================================================================================================================
# STEP 14 — OUTPUT OF CELL 1
# ================================================================================================================

print("""
STEP 14 — EXPECTED OUTPUT
----------------------------------------------------------------------------------------------------------------

Cell 1 establishes ONLY the search protocol.

It does NOT yet make a novelty decision.

Expected next artifact:

LITERATURE_GAP_MATRIX_V001

The matrix will be populated from the actual literature search and
will preserve both supporting and challenging evidence.

================================================================================================================
CELL 1 v001 — SEARCH PROTOCOL FROZEN
================================================================================================================

NEXT ACTION:
Execute the bounded literature search using the frozen concept families.

Do NOT modify this search protocol after seeing favorable/unfavorable
results without recording a new protocol version and scientific reason.

================================================================================================================
""")
```

```python
# ================================================================================================================
# CELL 5 v001 — LITERATURE CANDIDATE REGISTRY QC GATE
# ================================================================================================================

print("""
================================================================================================================
CELL 5 v001 — LITERATURE CANDIDATE REGISTRY QC GATE
================================================================================================================
""")

# ================================================================================================================
# STEP 0 — SCIENTIFIC VALUE GATE
# ================================================================================================================

print("""
STEP 0 — SCIENTIFIC VALUE GATE
----------------------------------------------------------------------------------------------------------------
Mandatory question:

ما فائدتها للرسالة العلمية أو لبناء الـdataset؟

Answer:
To verify that the literature candidate registry is structurally
complete, traceable, and free from obvious duplicate inflation before
any methodological comparison or novelty/gap conclusion.

✓ DIRECT VALUE TO THE THESIS
✓ PROTECTS NOVELTY AUDIT INTEGRITY
✓ NO DATASET MODIFICATION
✓ NO MODEL
✓ NO TRAINING
✓ NO ARCHITECTURE IMPLEMENTATION
""")


# ================================================================================================================
# STEP 1 — EXPERIMENT IDENTITY
# ================================================================================================================

EXPERIMENT_ID = "FDR-NOVELTY-GAP-001"
CELL_ID = "CELL_5"
CELL_VERSION = "V001"

print("STEP 1 — EXPERIMENT IDENTITY")
print("-" * 112)
print(f"Experiment ID : {EXPERIMENT_ID}")
print(f"Cell          : {CELL_ID}")
print(f"Version       : {CELL_VERSION}")


# ================================================================================================================
# STEP 2 — INPUT CONTRACT
# ================================================================================================================

INPUT_ARTIFACT = "LITERATURE_CANDIDATE_REGISTRY_V001"

print("\nSTEP 2 — INPUT CONTRACT")
print("-" * 112)
print(f"Required input : {INPUT_ARTIFACT}")

print("""
This QC gate assumes that the registry has been populated from the
actual scholarly search performed under the frozen Cell 1 / Cell 3
protocol.

No new literature discovery is performed here.
""")


# ================================================================================================================
# STEP 3 — REQUIRED SCHEMA
# ================================================================================================================

REQUIRED_COLUMNS = [
    "candidate_id",
    "search_family",
    "search_layer",
    "search_query",
    "title",
    "authors",
    "year",
    "venue",
    "primary_source",
    "doi_or_identifier",
    "url",
    "retrieval_date",
    "initial_relevance",
    "possible_competitor",
    "duplicate_group",
    "version_status",
    "discovery_reason",
    "notes",
]

print("\nSTEP 3 — REQUIRED SCHEMA")
print("-" * 112)

for i, column in enumerate(REQUIRED_COLUMNS, 1):
    print(f"{i:02d}. {column}")


# ================================================================================================================
# STEP 4 — STRUCTURAL QC RULES
# ================================================================================================================

STRUCTURAL_QC = [
    "candidate_id is unique within the registry.",
    "title is present for every candidate.",
    "search_family is present for every candidate.",
    "search_layer is present for every candidate.",
    "search_query is present for every candidate.",
    "primary_source is present for every candidate.",
    "discovery_reason is present for every candidate.",
    "initial_relevance uses only the frozen controlled values.",
    "possible_competitor uses only the frozen controlled values.",
    "version_status uses only the frozen controlled values.",
]

print("\nSTEP 4 — STRUCTURAL QC RULES")
print("-" * 112)

for i, rule in enumerate(STRUCTURAL_QC, 1):
    print(f"{i:02d}. {rule}")


# ================================================================================================================
# STEP 5 — DUPLICATE QC
# ================================================================================================================

print("""
STEP 5 — DUPLICATE QC
----------------------------------------------------------------------------------------------------------------

Duplicate inspection order:

1. DOI / stable identifier
2. Exact title
3. Normalized title
4. Normalized title + author overlap
5. Conference / journal relationship
6. Preprint / publication relationship

QC interpretation:

DUPLICATE
→ Same scholarly contribution represented more than once.

VERSION
→ Same study represented through different publication stages.

UNIQUE
→ Distinct scholarly contribution.

UNCERTAIN
→ Requires source inspection.

Important:
A duplicate/version must NOT be counted as an independent contribution
when estimating the breadth of prior work.
""")


# ================================================================================================================
# STEP 6 — SEARCH COVERAGE QC
# ================================================================================================================

SEARCH_FAMILY_IDS = [
    "F1",
    "F2",
    "F3",
    "F4",
    "F5",
    "F6",
    "F7",
    "F8",
]

print("\nSTEP 6 — SEARCH COVERAGE QC")
print("-" * 112)

for family_id in SEARCH_FAMILY_IDS:
    print(f"{family_id} → REQUIRED TO BE REPRESENTED OR EXPLICITLY RECORDED AS NO QUALIFYING RESULT")

print("""
Absence of candidates from a search family is itself a result and
must not be silently omitted.

A search family with zero qualifying candidates must be recorded as:

NO_QUALIFYING_CANDIDATE_FOUND

This is NOT evidence of novelty by itself.
""")


# ================================================================================================================
# STEP 7 — SEARCH-LAYER QC
# ================================================================================================================

print("""
STEP 7 — SEARCH-LAYER QC
----------------------------------------------------------------------------------------------------------------

The registry must preserve both:

LAYER_A_DIRECT
→ RAG / Legal RAG evidence-oriented competitors.

LAYER_B_METHODOLOGICAL
→ Claim/evidence/support/sufficiency/verification work that may
  represent a methodological competitor outside RAG.

If Layer B is absent, the audit is considered INCOMPLETE.
""")


# ================================================================================================================
# STEP 8 — COMPETITOR PRESERVATION QC
# ================================================================================================================

print("""
STEP 8 — COMPETITOR PRESERVATION QC
----------------------------------------------------------------------------------------------------------------

Any candidate marked:

possible_competitor = YES

must remain in the registry even if it appears unfavorable to the
proposed F-D+R gap.

No candidate may be removed because it threatens the novelty claim.

The purpose of the audit is falsification, not confirmation.
""")


# ================================================================================================================
# STEP 9 — BOUNDED COUNT CHECK
# ================================================================================================================

MAX_P1 = 10
MAX_P2 = 15
MAX_P3 = 10

print("\nSTEP 9 — BOUNDED COUNT CHECK")
print("-" * 112)

print(f"P1_DIRECT maximum : {MAX_P1}")
print(f"P2_CLOSE maximum  : {MAX_P2}")
print(f"P3_RELATED maximum: {MAX_P3}")

print("""
Exceeding a target is NOT automatically an error.

If more candidates appear genuinely competitive, the registry must
not silently truncate them.

Instead, the reason for expansion must be recorded as a protocol
extension before inclusion.
""")


# ================================================================================================================
# STEP 10 — PROHIBITED DECISIONS
# ================================================================================================================

print("""
STEP 10 — PROHIBITED DECISIONS
----------------------------------------------------------------------------------------------------------------

This QC gate MUST NOT determine:

✗ novelty
✗ research gap
✗ methodological equivalence
✗ superiority
✗ whether F-D+R is publishable
✗ whether an architecture should be implemented

It only determines whether the registry is sufficiently clean for
the next comparison stage.
""")


# ================================================================================================================
# STEP 11 — QC DECISION STATES
# ================================================================================================================

QC_STATES = {
    "PASS":
        "Registry is structurally adequate for methodological comparison.",

    "PASS_WITH_FLAGS":
        "Registry is usable, but bounded issues require documentation.",

    "FAIL":
        "Registry has structural/provenance problems that must be repaired before comparison.",
}

print("\nSTEP 11 — QC DECISION STATES")
print("-" * 112)

for state, meaning in QC_STATES.items():
    print(f"{state:18s} → {meaning}")


# ================================================================================================================
# STEP 12 — REQUIRED OUTPUT
# ================================================================================================================

print("""
STEP 12 — REQUIRED OUTPUT
----------------------------------------------------------------------------------------------------------------

LITERATURE_CANDIDATE_REGISTRY_QC_V001

Required summary:

• total candidate records
• unique candidate records
• duplicate/version groups
• candidates by search family
• candidates by search layer
• candidates by initial relevance
• possible competitor count
• missing-field count
• unresolved duplicate count
• QC decision
""")


# ================================================================================================================
# STEP 13 — FIREWALL
# ================================================================================================================

print("""
STEP 13 — FIREWALL
----------------------------------------------------------------------------------------------------------------

✓ Registry QC only
✓ No new literature search
✓ No methodological comparison
✓ No novelty decision

BLOCKED:
✗ dataset modification
✗ model inference
✗ training
✗ architecture implementation
✗ prompt revision
✗ new adjudication
✗ Gold generation
✗ Y+ / Y− generation
✗ Preference generation
✗ reopening 199 cases
✗ reopening 400 validation
✗ expanding Q82


================================================================================================================
CELL 5 v001 — LITERATURE CANDIDATE REGISTRY QC GATE
================================================================================================================

NEXT ACTION:
Run QC against the ACTUAL populated LITERATURE_CANDIDATE_REGISTRY_V001.

If PASS:
→ proceed to methodological comparison of the strongest candidates.

If PASS_WITH_FLAGS:
→ document bounded flags before comparison.

If FAIL:
→ repair the registry only; do not proceed to novelty analysis.

================================================================================================================
""")
```

```python
# ================================================================================================================
# CELL 6 v001 — METHODOLOGICAL COMPETITOR COMPARISON PROTOCOL
# ================================================================================================================

print("""
================================================================================================================
CELL 6 v001 — METHODOLOGICAL COMPETITOR COMPARISON PROTOCOL
================================================================================================================
""")


# ================================================================================================================
# STEP 0 — SCIENTIFIC VALUE GATE
# ================================================================================================================

print("""
STEP 0 — SCIENTIFIC VALUE GATE
----------------------------------------------------------------------------------------------------------------
Mandatory question:

ما فائدتها للرسالة العلمية أو لبناء الـdataset؟

Answer:
To compare the strongest literature candidates against the actual
F-D+R formulation using one fixed methodological framework.

This determines whether the proposed contribution is:
• distinct,
• partially overlapping but defensible,
• or substantially covered by prior work.

✓ DIRECT VALUE TO THE THESIS
✓ REQUIRED BEFORE ARCHITECTURE EXPERIMENT
✓ NO DATASET MODIFICATION
✓ NO MODEL
✓ NO TRAINING
✓ NO ARCHITECTURE IMPLEMENTATION
""")


# ================================================================================================================
# STEP 1 — EXPERIMENT IDENTITY
# ================================================================================================================

EXPERIMENT_ID = "FDR-NOVELTY-GAP-001"
CELL_ID = "CELL_6"
CELL_VERSION = "V001"

print("STEP 1 — EXPERIMENT IDENTITY")
print("-" * 112)
print(f"Experiment ID : {EXPERIMENT_ID}")
print(f"Cell          : {CELL_ID}")
print(f"Version       : {CELL_VERSION}")


# ================================================================================================================
# STEP 2 — INPUT CONTRACT
# ================================================================================================================

INPUT_ARTIFACT = [
    "LITERATURE_CANDIDATE_REGISTRY_V001",
    "LITERATURE_CANDIDATE_REGISTRY_QC_V001",
    "F-D+R frozen formulation from Cell 0",
]

print("\nSTEP 2 — INPUT CONTRACT")
print("-" * 112)

for item in INPUT_ARTIFACT:
    print(f"✓ {item}")

print("""
Only candidates that survive the Registry QC may enter the detailed
comparison stage.

No new candidate may be introduced here without returning to the
frozen discovery protocol.
""")


# ================================================================================================================
# STEP 3 — F-D+R REFERENCE FORMULATION
# ================================================================================================================

FDR_FORMULATION = {
    "stage": "POST_RETRIEVAL / PRE-GENERATION",
    "input": "QUESTION + CANDIDATE_EVIDENCE",
    "purpose": "EVIDENCE-SUITABILITY VERIFICATION",
    "dimensions": [
        "CASE_RELEVANCE",
        "LEGAL_SUPPORT",
        "SUFFICIENCY",
        "EVIDENCE_SUITABILITY",
        "CONTRADICTION / RISK",
        "UNCERTAINTY",
    ],
    "output_role": "EVIDENCE_ELIGIBILITY / RISK_GATE",
}

print("\nSTEP 3 — F-D+R REFERENCE FORMULATION")
print("-" * 112)

for key, value in FDR_FORMULATION.items():
    print(f"{key:20s}: {value}")


# ================================================================================================================
# STEP 4 — REQUIRED PAPER-LEVEL COMPARISON DIMENSIONS
# ================================================================================================================

COMPARISON_DIMENSIONS = [
    "Problem / task",
    "Position in pipeline",
    "Input to verification/judgment mechanism",
    "Evidence representation",
    "Evidence target",
    "Support assessment",
    "Sufficiency assessment",
    "Contradiction / risk assessment",
    "Uncertainty handling",
    "Human-grounded supervision/evaluation",
    "Label structure",
    "Reference-answer dependency",
    "Pre-generation evidence gating",
    "Legal-domain applicability",
    "Arabic-English / multilingual applicability",
    "Retrieval dependence",
    "Generation dependence",
    "Case-level evaluation",
]

print("""
STEP 4 — REQUIRED PAPER-LEVEL COMPARISON DIMENSIONS
----------------------------------------------------------------------------------------------------------------
""")

for i, dimension in enumerate(COMPARISON_DIMENSIONS, 1):
    print(f"{i:02d}. {dimension}")


# ================================================================================================================
# STEP 5 — EQUIVALENCE TEST
# ================================================================================================================

print("""
STEP 5 — METHODOLOGICAL EQUIVALENCE TEST
----------------------------------------------------------------------------------------------------------------

A paper MUST NOT be considered equivalent merely because it contains
the words:

"evidence"
"grounding"
"verification"
"citation"
"faithfulness"
"support"

Potential equivalence requires substantial alignment across:

1. verification stage
2. verifier input
3. evidence target
4. support logic
5. sufficiency logic
6. risk / contradiction treatment
7. output decision role
8. pre-generation control boundary

Similarity of terminology alone is insufficient.
""")


# ================================================================================================================
# STEP 6 — OVERLAP CLASSIFICATION
# ================================================================================================================

OVERLAP_LEVELS = {
    "O0_NONE":
        "No meaningful methodological overlap.",

    "O1_BACKGROUND":
        "Conceptually related but not a competing formulation.",

    "O2_PARTIAL":
        "Shares important components but lacks substantial F-D+R dimensions.",

    "O3_CLOSE":
        "Strong methodological overlap; requires careful distinction.",

    "O4_SUBSTANTIAL":
        "Prior work substantially covers the proposed contribution.",
}

print("\nSTEP 6 — OVERLAP CLASSIFICATION")
print("-" * 112)

for level, meaning in OVERLAP_LEVELS.items():
    print(f"{level:18s} → {meaning}")


# ================================================================================================================
# STEP 7 — CRITICAL CHALLENGE TEST
# ================================================================================================================

print("""
STEP 7 — CRITICAL CHALLENGE TEST
----------------------------------------------------------------------------------------------------------------

For every HIGH candidate, answer:

Q1. Could this paper already be described as a verifier of retrieved
    evidence before generation?

Q2. Does it use essentially the same verifier input:
        QUESTION + CANDIDATE_EVIDENCE ?

Q3. Does it distinguish relevance from actual evidential support?

Q4. Does it explicitly address evidence sufficiency?

Q5. Does it handle contradiction/risk?

Q6. Does it model or preserve uncertainty?

Q7. Does its output control evidence eligibility before generation?

Q8. Is the formulation human-grounded?

Q9. Is it applicable to legal / multilingual settings?

Q10. If adapted to Legal RAG, would the adaptation be merely an
     application change rather than a methodological contribution?

A "YES" answer to several of these questions increases overlap
but does NOT automatically establish equivalence.
""")


# ================================================================================================================
# STEP 8 — F-D+R DISTINCTIVENESS TEST
# ================================================================================================================

print("""
STEP 8 — F-D+R DISTINCTIVENESS TEST
----------------------------------------------------------------------------------------------------------------

For each close candidate, explicitly identify what F-D+R adds, if anything.

Permitted conclusions:

✓ NEW TASK BOUNDARY
✓ NEW COMBINATION OF EXISTING COMPONENTS
✓ NEW HUMAN-GROUNDED LABELING FORMULATION
✓ NEW PRE-GENERATION CONTROL ROLE
✓ NEW DOMAIN / LANGUAGE VALIDATION
✓ NO CLEAR ADDITION

Important:
"Different dataset" alone is NOT sufficient to establish methodological
novelty.

"Legal domain" alone is NOT sufficient.

"Arabic-English" alone is NOT sufficient.

"Better performance" alone is NOT sufficient without a distinct task
or methodological contribution.
""")


# ================================================================================================================
# STEP 9 — EVIDENCE STANDARD
# ================================================================================================================

print("""
STEP 9 — EVIDENCE STANDARD
----------------------------------------------------------------------------------------------------------------

Claims about a paper MUST be supported by the paper itself.

Priority:

1. Full paper
2. Official publisher/proceedings page
3. Official supplementary material
4. Reliable scholarly index only for bibliographic confirmation

Title/abstract-only evidence may be used for provisional screening,
but NOT for a final equivalence or novelty conclusion.
""")


# ================================================================================================================
# STEP 10 — COMPARISON RECORD
# ================================================================================================================

COMPARISON_RECORD = [
    "candidate_id",
    "paper",
    "priority",
    "task_summary",
    "pipeline_stage",
    "verifier_input",
    "evidence_target",
    "support",
    "sufficiency",
    "contradiction_risk",
    "uncertainty",
    "human_grounding",
    "reference_answer_dependency",
    "pre_generation_gate",
    "legal_scope",
    "multilingual_scope",
    "overlap_level",
    "fdr_distinctiveness",
    "critical_challenge",
    "evidence_source",
    "confidence",
    "notes",
]

print("""
STEP 10 — COMPARISON RECORD
----------------------------------------------------------------------------------------------------------------
""")

for i, field in enumerate(COMPARISON_RECORD, 1):
    print(f"{i:02d}. {field}")


# ================================================================================================================
# STEP 11 — CONFIDENCE CONTROL
# ================================================================================================================

CONFIDENCE_LEVELS = [
    "HIGH",
    "MEDIUM",
    "LOW",
]

print("""
STEP 11 — CONFIDENCE CONTROL
----------------------------------------------------------------------------------------------------------------

Every methodological comparison must carry a confidence level.

HIGH
→ Directly supported by the full paper.

MEDIUM
→ Supported by substantial methodological evidence but one element
  remains incompletely documented.

LOW
→ Based mainly on abstract/limited accessible evidence.

LOW confidence MUST NOT be used alone to declare the gap
non-defensible.
""")


# ================================================================================================================
# STEP 12 — ANTI-OVERCLAIM FIREWALL
# ================================================================================================================

print("""
STEP 12 — ANTI-OVERCLAIM FIREWALL
----------------------------------------------------------------------------------------------------------------

The comparison MUST NOT claim:

✗ "Nobody has done this."
✗ "F-D+R is the first."
✗ "F-D+R is novel."
✗ "This paper is irrelevant" without methodological evidence.
✗ superiority over prior work without comparative experiment.

Instead use bounded language:

✓ "No substantially equivalent formulation was identified in the
   audited literature."

✓ "The work overlaps with X but differs in Y."

✓ "The available evidence supports / does not support a defensible
   methodological distinction."

✓ "Novelty remains inconclusive."
""")


# ================================================================================================================
# STEP 13 — DECISION STATES
# ================================================================================================================

DECISION_STATES = {
    "GAP_DEFENSIBLE":
        "No substantially equivalent formulation identified.",

    "GAP_PARTIALLY_OVERLAPPING":
        "Important prior components exist, but the complete formulation remains distinguishable.",

    "GAP_NOT_DEFENSIBLE":
        "Prior work substantially covers the proposed methodological contribution.",

    "GAP_INCONCLUSIVE":
        "Available evidence does not support a sufficiently confident decision.",
}

print("\nSTEP 13 — DECISION STATES")
print("-" * 112)

for state, definition in DECISION_STATES.items():
    print(f"{state:28s} → {definition}")


# ================================================================================================================
# STEP 14 — FIREWALL
# ================================================================================================================

print("""
STEP 14 — FIREWALL
----------------------------------------------------------------------------------------------------------------

✓ Literature comparison only
✓ Human-readable paper evidence
✓ Fixed F-D+R formulation
✓ No dataset modification
✓ No model inference
✓ No training
✓ No architecture implementation
✓ No prompt revision
✓ No new adjudication
✓ No Gold generation
✓ No Y+ / Y−
✓ No Preference
✓ 199 cohort frozen
✓ 400 validation closed
✓ Q82 frozen


================================================================================================================
CELL 6 v001 — METHODOLOGICAL COMPARISON PROTOCOL FROZEN
================================================================================================================

NEXT ACTION:
Compare ONLY the strongest literature candidates that survive
LITERATURE_CANDIDATE_REGISTRY_QC_V001.

Do NOT yet make the final novelty/gap decision.

================================================================================================================
""")
```

```python
# ================================================================================================================
# CELL 7 v001 — CRITICAL CHALLENGE TEST EXECUTION GATE
# ================================================================================================================
#
# PURPOSE
# -------
# Execute the frozen Critical Challenge Test from CELL 6 against ONLY literature candidates
# that have already been classified as HIGH methodological relevance.
#
# SCIENTIFIC VALUE GATE
# ---------------------
# ما فائدتها للرسالة العلمية أو لبناء الـdataset؟
#
# This cell has direct scientific value because it determines whether the proposed
# F-D+R contribution is methodologically distinct from the closest existing work.
#
# It does NOT:
#   - modify the dataset
#   - create Gold / Y+ / Y- / Preference labels
#   - perform model inference
#   - revise prompts
#   - modify architecture
#   - reopen the 199-case cohort
#   - reopen the 400-case validation
#   - expand Q82
#   - make population or causal claims
#
# INPUT
# -----
# LITERATURE_CANDIDATE_REGISTRY_V001
#
# OUTPUT
# ------
# CRITICAL_CHALLENGE_RESULTS_V001
#
# IMPORTANT
# ---------
# This cell DOES NOT infer answers to Q1-Q10 from titles or abstracts.
# If full-paper evidence is unavailable, the corresponding item remains:
#
#     NOT_ESTABLISHED
#
# No candidate may be declared a methodological competitor merely from terminology overlap.
#
# ================================================================================================================

print("=" * 112)
print("CELL 7 v001 — CRITICAL CHALLENGE TEST EXECUTION GATE")
print("=" * 112)

# ------------------------------------------------------------------------------------------------
# STEP 0 — EXPERIMENT IDENTITY
# ------------------------------------------------------------------------------------------------

EXPERIMENT_ID = "FDR-NOVELTY-GAP-001"
CELL_ID = "CELL7_CRITICAL_CHALLENGE_TEST_V001"
CELL_VERSION = "V001"

STATUS = "INITIALIZED_READ_ONLY"

print("\n[STEP 0] EXPERIMENT IDENTITY")
print("-" * 112)
print(f"Experiment ID : {EXPERIMENT_ID}")
print(f"Cell ID       : {CELL_ID}")
print(f"Version       : {CELL_VERSION}")
print(f"Status        : {STATUS}")

# ------------------------------------------------------------------------------------------------
# STEP 1 — SCIENTIFIC VALUE GATE
# ------------------------------------------------------------------------------------------------

SCIENTIFIC_VALUE_GATE = {
    "question":
        "ما فائدتها للرسالة العلمية أو لبناء الـdataset؟",
    "answer":
        "Determine whether the closest methodological literature actually overlaps "
        "with the proposed F-D+R task boundary before freezing the novelty/gap decision.",
    "direct_value": True
}

print("\n[STEP 1] SCIENTIFIC VALUE GATE")
print("-" * 112)
print(f"Question       : {SCIENTIFIC_VALUE_GATE['question']}")
print(f"Direct value   : {SCIENTIFIC_VALUE_GATE['direct_value']}")
print(f"Scientific use : {SCIENTIFIC_VALUE_GATE['answer']}")

assert SCIENTIFIC_VALUE_GATE["direct_value"] is True

# ------------------------------------------------------------------------------------------------
# STEP 2 — FROZEN RESEARCH QUESTION
# ------------------------------------------------------------------------------------------------

RESEARCH_QUESTION = (
    "Is post-retrieval Evidence-Suitability Verification, using "
    "QUESTION + CANDIDATE_EVIDENCE to determine evidence suitability and "
    "pre-generation evidence eligibility/risk, sufficiently distinct from "
    "existing methodological work?"
)

print("\n[STEP 2] FROZEN RESEARCH QUESTION")
print("-" * 112)
print(RESEARCH_QUESTION)

# ------------------------------------------------------------------------------------------------
# STEP 3 — FROZEN F-D+R FORMULATION
# ------------------------------------------------------------------------------------------------

FDR_FORMULATION = {
    "stage": "POST_RETRIEVAL / PRE-GENERATION",
    "input": "QUESTION + CANDIDATE_EVIDENCE",
    "purpose": "EVIDENCE-SUITABILITY VERIFICATION",
    "dimensions": [
        "CASE_RELEVANCE",
        "LEGAL_SUPPORT",
        "SUFFICIENCY",
        "EVIDENCE_SUITABILITY",
        "CONTRADICTION_RISK",
        "UNCERTAINTY"
    ],
    "output_role": "EVIDENCE_ELIGIBILITY / RISK_GATE"
}

print("\n[STEP 3] FROZEN F-D+R FORMULATION")
print("-" * 112)

for key, value in FDR_FORMULATION.items():
    print(f"{key:<20}: {value}")

# ------------------------------------------------------------------------------------------------
# STEP 4 — CRITICAL CHALLENGE QUESTIONS
# ------------------------------------------------------------------------------------------------
#
# These questions are frozen from CELL 6.
#
# Each question must be answered from actual methodological evidence in the paper.
# Valid states:
#
#   YES
#   NO
#   PARTIAL
#   NOT_ESTABLISHED
#
# No "YES" may be assigned solely from title/abstract terminology.
# ------------------------------------------------------------------------------------------------

CRITICAL_CHALLENGE_QUESTIONS = {
    "Q1": "Does the work verify retrieved evidence before generation?",
    "Q2": "Is the verifier input essentially QUESTION + CANDIDATE_EVIDENCE?",
    "Q3": "Does it distinguish relevance/topic similarity from actual evidential support?",
    "Q4": "Does it explicitly evaluate evidence sufficiency?",
    "Q5": "Does it explicitly model contradiction or evidence risk?",
    "Q6": "Does it explicitly represent uncertainty?",
    "Q7": "Does the verifier output control evidence eligibility before generation?",
    "Q8": "Is the verification/assessment human-grounded?",
    "Q9": "Does the work operate in a legal and/or multilingual Arabic-English setting?",
    "Q10": "If adapted to Legal RAG, would the contribution become only domain/application adaptation?"
}

print("\n[STEP 4] CRITICAL CHALLENGE QUESTIONS")
print("-" * 112)

for qid, question in CRITICAL_CHALLENGE_QUESTIONS.items():
    print(f"{qid}: {question}")

# ------------------------------------------------------------------------------------------------
# STEP 5 — RESULT VOCABULARY
# ------------------------------------------------------------------------------------------------

VALID_STATES = {
    "YES",
    "NO",
    "PARTIAL",
    "NOT_ESTABLISHED"
}

OVERLAP_LEVELS = {
    "O0_NONE",
    "O1_BACKGROUND",
    "O2_PARTIAL",
    "O3_CLOSE",
    "O4_SUBSTANTIAL"
}

DISTINCTIVENESS_STATES = {
    "DISTINCTIVE",
    "PARTIALLY_DISTINCTIVE",
    "NOT_DISTINCTIVE",
    "INCONCLUSIVE"
}

CONFIDENCE_LEVELS = {
    "HIGH",
    "MEDIUM",
    "LOW"
}

print("\n[STEP 5] CONTROLLED RESULT VOCABULARY")
print("-" * 112)
print("Challenge states :", sorted(VALID_STATES))
print("Overlap levels   :", sorted(OVERLAP_LEVELS))
print("Distinctiveness  :", sorted(DISTINCTIVENESS_STATES))
print("Confidence       :", sorted(CONFIDENCE_LEVELS))

# ------------------------------------------------------------------------------------------------
# STEP 6 — INPUT REGISTRY CHECK
# ------------------------------------------------------------------------------------------------
#
# The literature registry may not yet be populated.
# Therefore:
#
#   - absence of the registry = EMPTY / WAITING
#   - it is NOT a failure of the scientific protocol
#   - it is NOT evidence that the gap is defensible
#
# This prevents an empty registry from being converted into a false novelty claim.
# ------------------------------------------------------------------------------------------------

print("\n[STEP 6] LITERATURE CANDIDATE REGISTRY CHECK")
print("-" * 112)

registry_name = "LITERATURE_CANDIDATE_REGISTRY_V001"

registry_available = registry_name in globals()

print(f"Registry expected : {registry_name}")
print(f"Registry available: {registry_available}")

if not registry_available:
    CRITICAL_CHALLENGE_STATUS = "WAITING_FOR_POPULATED_LITERATURE_REGISTRY"
    print("\n⚠ Registry is not populated in the current runtime.")
    print("  No literature candidate will be fabricated.")
    print("  No Critical Challenge result will be inferred.")
else:
    registry_obj = globals()[registry_name]

    try:
        registry_count = len(registry_obj)
    except Exception:
        registry_count = None

    print(f"Registry records  : {registry_count}")

# ------------------------------------------------------------------------------------------------
# STEP 7 — HIGH-RELEVANCE CANDIDATE SELECTION
# ------------------------------------------------------------------------------------------------
#
# Only candidates explicitly marked HIGH methodological relevance are eligible.
#
# This cell does NOT upgrade MEDIUM/LOW candidates automatically.
# ------------------------------------------------------------------------------------------------

HIGH_CANDIDATES = []

if registry_available:

    registry_obj = globals()[registry_name]

    # Support either pandas DataFrame or list-of-dicts representation.
    if hasattr(registry_obj, "to_dict"):
        registry_records = registry_obj.to_dict("records")
    elif isinstance(registry_obj, list):
        registry_records = registry_obj
    else:
        registry_records = []

    for record in registry_records:

        relevance = str(
            record.get("initial_relevance", "")
        ).strip().upper()

        if relevance == "HIGH":
            HIGH_CANDIDATES.append(record)

    print("\n[STEP 7] HIGH-RELEVANCE CANDIDATE SELECTION")
    print("-" * 112)
    print(f"Eligible HIGH candidates : {len(HIGH_CANDIDATES)}")

else:
    print("\n[STEP 7] HIGH-RELEVANCE CANDIDATE SELECTION")
    print("-" * 112)
    print("Skipped — literature registry not yet populated.")

# ------------------------------------------------------------------------------------------------
# STEP 8 — CRITICAL CHALLENGE RESULT SCHEMA
# ------------------------------------------------------------------------------------------------

RESULT_COLUMNS = [
    "candidate_id",
    "title",
    "year",
    "venue",
    "priority",
    "Q1_pre_generation_verification",
    "Q2_question_plus_evidence",
    "Q3_support_vs_relevance",
    "Q4_sufficiency",
    "Q5_contradiction_risk",
    "Q6_uncertainty",
    "Q7_pre_generation_gate",
    "Q8_human_grounding",
    "Q9_legal_multilingual",
    "Q10_domain_adaptation_only",
    "overall_overlap_level",
    "fdr_distinctiveness",
    "confidence",
    "evidence_source",
    "evidence_status",
    "notes"
]

print("\n[STEP 8] RESULT SCHEMA")
print("-" * 112)

for i, column in enumerate(RESULT_COLUMNS, start=1):
    print(f"{i:02d}. {column}")

# ------------------------------------------------------------------------------------------------
# STEP 9 — EMPTY EXECUTION FRAME
# ------------------------------------------------------------------------------------------------
#
# IMPORTANT:
# This frame intentionally remains empty until actual paper evidence is entered.
#
# We do NOT use:
#   - semantic similarity
#   - title matching
#   - abstract keyword matching
#   - model-generated judgments
#   - inferred methodology
#
# The actual paper evidence must be recorded manually/from primary source review.
# ------------------------------------------------------------------------------------------------

try:
    import pandas as pd

    CRITICAL_CHALLENGE_RESULTS_V001 = pd.DataFrame(
        columns=RESULT_COLUMNS
    )

except Exception:
    CRITICAL_CHALLENGE_RESULTS_V001 = []

print("\n[STEP 9] EXECUTION FRAME")
print("-" * 112)
print("✓ Empty controlled execution frame created.")
print("✓ No candidate judgment inferred.")
print("✓ No paper methodology inferred from title/abstract.")
print("✓ No model inference executed.")

# ------------------------------------------------------------------------------------------------
# STEP 10 — EVIDENCE REQUIREMENT
# ------------------------------------------------------------------------------------------------

EVIDENCE_REQUIREMENT = {
    "primary_standard": "FULL_PAPER",
    "secondary_standard": "OFFICIAL_PUBLISHER_OR_PROCEEDINGS_PAGE",
    "supplementary_standard": "SUPPLEMENTARY_MATERIAL",
    "discovery_only": [
        "GOOGLE_SCHOLAR",
        "SEMANTIC_SCHOLAR",
        "ABSTRACT_ONLY_INDEX"
    ],
    "abstract_only_allowed_for_final_challenge": False
}

print("\n[STEP 10] EVIDENCE STANDARD")
print("-" * 112)

for key, value in EVIDENCE_REQUIREMENT.items():
    print(f"{key:<28}: {value}")

# ------------------------------------------------------------------------------------------------
# STEP 11 — ANTI-OVERCLAIM FIREWALL
# ------------------------------------------------------------------------------------------------

FIREWALL = {
    "dataset_modification": False,
    "gold_creation": False,
    "y_positive_creation": False,
    "y_negative_creation": False,
    "preference_creation": False,
    "model_training": False,
    "model_inference": False,
    "prompt_revision": False,
    "architecture_change": False,
    "199_cohort_reopen": False,
    "400_validation_reopen": False,
    "Q82_expansion": False,
    "population_claim": False,
    "causal_claim": False,
    "novelty_claim_without_primary_evidence": False
}

print("\n[STEP 11] ANTI-OVERCLAIM FIREWALL")
print("-" * 112)

for key, value in FIREWALL.items():
    print(f"{key:<40}: {'BLOCKED' if value is False else 'ALLOWED'}")

assert all(value is False for value in FIREWALL.values())

# ------------------------------------------------------------------------------------------------
# STEP 12 — EXECUTION STATUS
# ------------------------------------------------------------------------------------------------

if not registry_available:
    EXECUTION_DECISION = "WAITING_FOR_POPULATED_LITERATURE_REGISTRY"
elif len(HIGH_CANDIDATES) == 0:
    EXECUTION_DECISION = "NO_HIGH_CANDIDATES_AVAILABLE"
else:
    EXECUTION_DECISION = "READY_FOR_PRIMARY_SOURCE_CRITICAL_CHALLENGE_REVIEW"

print("\n[STEP 12] EXECUTION STATUS")
print("-" * 112)
print(f"Decision : {EXECUTION_DECISION}")

# ------------------------------------------------------------------------------------------------
# STEP 13 — STOP RULES
# ------------------------------------------------------------------------------------------------

STOP_RULES = {
    "NO_REGISTRY":
        "Do not fabricate candidates. Stop until the literature registry is populated.",
    "NO_PRIMARY_EVIDENCE":
        "Do not assign final Q1-Q10 judgments from title/abstract alone.",
    "INSUFFICIENT_EVIDENCE":
        "Mark NOT_ESTABLISHED rather than inferring YES/NO.",
    "NOVELTY":
        "Do not declare the F-D+R gap defensible or non-defensible from an empty/incomplete audit.",
    "ARCHITECTURE":
        "Do not implement FDR-ARCH-CMP-001 before the literature-gap decision is frozen."
}

print("\n[STEP 13] STOP RULES")
print("-" * 112)

for key, rule in STOP_RULES.items():
    print(f"{key:<22}: {rule}")

# ------------------------------------------------------------------------------------------------
# STEP 14 — READ-BACK / FIREWALL CHECK
# ------------------------------------------------------------------------------------------------

READ_BACK = {
    "cell": CELL_ID,
    "status": "READ_ONLY",
    "registry_required": True,
    "high_candidates_only": True,
    "full_paper_required": True,
    "automatic_judgment": False,
    "novelty_claim": False,
    "architecture_execution": False,
    "dataset_modification": False
}

print("\n[STEP 14] READ-BACK / FIREWALL CHECK")
print("-" * 112)

for key, value in READ_BACK.items():
    print(f"{key:<28}: {value}")

assert READ_BACK["status"] == "READ_ONLY"
assert READ_BACK["high_candidates_only"] is True
assert READ_BACK["full_paper_required"] is True
assert READ_BACK["automatic_judgment"] is False
assert READ_BACK["novelty_claim"] is False
assert READ_BACK["architecture_execution"] is False
assert READ_BACK["dataset_modification"] is False

# ------------------------------------------------------------------------------------------------
# FINAL STATUS
# ------------------------------------------------------------------------------------------------

print("\n" + "=" * 112)
print("CELL 7 v001 — FINAL STATUS")
print("=" * 112)

if not registry_available:

    print("""
✓ CELL 7 PROTOCOL INITIALIZED
✓ READ-ONLY FIREWALL PASS
✓ CRITICAL CHALLENGE TEST FROZEN
✓ NO LITERATURE JUDGMENT FABRICATED
✓ NO NOVELTY CLAIM
✓ NO DATASET / GOLD / MODEL / ARCHITECTURE MODIFICATION

STATUS:
WAITING_FOR_POPULATED_LITERATURE_REGISTRY

NEXT SCIENTIFIC ACTION:
Populate the bounded literature candidate registry from actual scholarly sources,
then apply Q1–Q10 only to HIGH-relevance candidates using primary-source evidence.
""")

elif len(HIGH_CANDIDATES) == 0:

    print("""
✓ CELL 7 EXECUTION FRAME READY
✓ No HIGH-relevance candidates currently available.
✓ No automatic upgrading of MEDIUM/LOW candidates.
✓ No novelty conclusion permitted.

STATUS:
NO_HIGH_CANDIDATES_AVAILABLE

NEXT SCIENTIFIC ACTION:
Complete the bounded literature search/registry before executing the Critical Challenge Test.
""")

else:

    print("""
✓ CELL 7 EXECUTION FRAME READY
✓ HIGH-relevance candidates detected.
✓ Q1–Q10 ready for primary-source evidence review.
✓ No automatic judgments generated.
✓ No novelty conclusion generated.

STATUS:
READY_FOR_PRIMARY_SOURCE_CRITICAL_CHALLENGE_REVIEW
""")

print("=" * 112)
```

```python
# ================================================================================================================
# CELL 10 v002 — LITERATURE REGISTRY QC + PERSISTENCE GATE
# ================================================================================================================
#
# PURPOSE
# -------
# Perform the final structural / provenance QC of the already-materialized
# literature_registry and persist the validated registry as the single
# controlled checkpoint for the downstream literature-gap audit.
#
# SCIENTIFIC VALUE GATE
# ---------------------
# ما فائدتها للرسالة العلمية أو لبناء الـdataset؟
#
# To create a reproducible, auditable, integrity-checked literature registry
# that can be independently reloaded before methodological comparison.
#
# ✓ Direct value to the thesis
# ✓ Protects literature-gap audit reproducibility
# ✓ No new literature discovery
# ✓ No novelty judgment
# ✓ No Q1–Q10 judgment
# ✓ No dataset modification
# ✓ No model inference
# ✓ No training
# ✓ No architecture implementation
#
# INPUT
# -----
# literature_registry
#     → created successfully by CELL 9 v002
#
# OUTPUT
# ------
# LITERATURE_REGISTRY_QC_V002
# LITERATURE_REGISTRY_PERSISTENT_CHECKPOINT_READY
#
# Files:
#   • Parquet registry
#   • CSV registry
#   • JSON manifest
#
# ================================================================================================================

import os
import json
import hashlib
import pandas as pd
from datetime import datetime

print("=" * 112)
print("CELL 10 v002 — LITERATURE REGISTRY QC + PERSISTENCE GATE")
print("=" * 112)

# ================================================================================================================
# STEP 0 — SCIENTIFIC VALUE GATE
# ================================================================================================================

print("\n[STEP 0] SCIENTIFIC VALUE GATE")
print("-" * 112)

SCIENTIFIC_VALUE_GATE = True

print("Question:")
print("ما فائدتها للرسالة العلمية أو لبناء الـdataset؟")
print()
print("Answer:")
print("To create a reproducible, auditable, integrity-checked literature")
print("registry that can be independently reloaded before methodological comparison.")
print()
print("✓ Direct value to the thesis")
print("✓ Reproducibility / audit integrity")
print("✓ No new scientific evidence")
print("✓ No dataset modification")

if not SCIENTIFIC_VALUE_GATE:
    raise RuntimeError(
        "CELL 10 STOP: Scientific Value Gate not satisfied."
    )

# ================================================================================================================
# STEP 1 — EXPERIMENT IDENTITY
# ================================================================================================================

print("\n[STEP 1] EXPERIMENT IDENTITY")
print("-" * 112)

EXPERIMENT_ID = "FDR-NOVELTY-GAP-001"
CELL_ID = "CELL10_LITERATURE_REGISTRY_QC_PERSISTENCE"
VERSION = "V002"

print(f"Experiment ID : {EXPERIMENT_ID}")
print(f"Cell ID       : {CELL_ID}")
print(f"Version       : {VERSION}")

# ================================================================================================================
# STEP 2 — READ-ONLY FIREWALL
# ================================================================================================================

print("\n[STEP 2] READ-ONLY FIREWALL")
print("-" * 112)

FIREWALL = {
    "new_literature_search": False,
    "new_literature_selection": False,
    "novelty_judgment": False,
    "q1_q10_judgment": False,
    "dataset_modification": False,
    "gold_generation": False,
    "y_plus_generation": False,
    "y_minus_generation": False,
    "preference_generation": False,
    "new_adjudication": False,
    "model_loading": False,
    "model_inference": False,
    "model_training": False,
    "architecture_implementation": False,
    "prompt_revision": False,
    "cohort_modification": False,
    "validation_reopen": False,
}

for key, value in FIREWALL.items():
    if value is not False:
        raise RuntimeError(
            f"CELL 10 FIREWALL VIOLATION: {key}"
        )

print("✓ Read-only scientific firewall PASS")

# ================================================================================================================
# STEP 3 — REQUIRED INPUT
# ================================================================================================================

print("\n[STEP 3] REQUIRED INPUT")
print("-" * 112)

if "literature_registry" not in globals():
    raise RuntimeError(
        "CELL 10 STOP: literature_registry is not available. "
        "Run CELL 9 v002 first in the same runtime."
    )

if not isinstance(literature_registry, pd.DataFrame):
    raise TypeError(
        "CELL 10 STOP: literature_registry must be a pandas DataFrame."
    )

print("✓ literature_registry found in current runtime")
print(f"Rows    : {len(literature_registry)}")
print(f"Columns : {len(literature_registry.columns)}")

# ================================================================================================================
# STEP 4 — FROZEN SCHEMA
# ================================================================================================================

print("\n[STEP 4] FROZEN REGISTRY SCHEMA")
print("-" * 112)

REGISTRY_COLUMNS = [
    "candidate_id",
    "search_family",
    "search_layer",
    "search_query",
    "title",
    "authors",
    "year",
    "venue",
    "primary_source",
    "doi_or_identifier",
    "url",
    "retrieval_date",
    "initial_relevance",
    "possible_competitor",
    "duplicate_group",
    "version_status",
    "discovery_reason",
    "notes",
]

EXPECTED_IDS = {
    "ANCHOR-001",
    "LIT-001",
    "LIT-002",
    "LIT-003",
    "LIT-004",
}

assert list(literature_registry.columns) == REGISTRY_COLUMNS, (
    "CELL 10 FAIL: Registry schema does not exactly match the frozen 18-column schema."
)

assert len(literature_registry) == 5, (
    "CELL 10 FAIL: Expected exactly 5 approved records."
)

print("✓ Exact 18-column schema PASS")
print("✓ Expected row count = 5 PASS")

# ================================================================================================================
# STEP 5 — IDENTITY / UNIQUENESS QC
# ================================================================================================================

print("\n[STEP 5] IDENTITY / UNIQUENESS QC")
print("-" * 112)

actual_ids = set(
    literature_registry["candidate_id"].astype(str)
)

assert actual_ids == EXPECTED_IDS, (
    f"CELL 10 FAIL: Candidate ID mismatch.\n"
    f"Expected: {EXPECTED_IDS}\n"
    f"Actual  : {actual_ids}"
)

assert literature_registry["candidate_id"].is_unique, (
    "CELL 10 FAIL: candidate_id values are not unique."
)

print("✓ Exact candidate-ID set PASS")
print("✓ Candidate-ID uniqueness PASS")

# ================================================================================================================
# STEP 6 — REQUIRED-FIELD QC
# ================================================================================================================

print("\n[STEP 6] REQUIRED-FIELD COMPLETENESS QC")
print("-" * 112)

missing_counts = literature_registry[REGISTRY_COLUMNS].isna().sum()

if missing_counts.sum() > 0:
    print("Missing values:")
    print(missing_counts[missing_counts > 0])
    raise RuntimeError(
        "CELL 10 FAIL: Missing required registry values."
    )

empty_counts = (
    literature_registry[REGISTRY_COLUMNS]
    .astype(str)
    .apply(lambda col: col.str.strip().eq("").sum())
)

if empty_counts.sum() > 0:
    print("Empty-string values:")
    print(empty_counts[empty_counts > 0])
    raise RuntimeError(
        "CELL 10 FAIL: Empty required registry values."
    )

print("✓ No missing required values")
print("✓ No empty required values")

# ================================================================================================================
# STEP 7 — CONTROLLED-VALUE QC
# ================================================================================================================

print("\n[STEP 7] CONTROLLED-VALUE QC")
print("-" * 112)

ALLOWED_RELEVANCE = {
    "HIGH",
    "MEDIUM",
    "LOW",
    "UNCERTAIN",
}

ALLOWED_COMPETITOR = {
    True,
    False,
}

ALLOWED_LAYERS = {
    "A_ANCHOR",
    "B_DIRECT_METHODOLOGICAL",
}

assert set(
    literature_registry["initial_relevance"].astype(str)
).issubset(ALLOWED_RELEVANCE), (
    "CELL 10 FAIL: Invalid initial_relevance value."
)

assert set(
    literature_registry["search_layer"].astype(str)
).issubset(ALLOWED_LAYERS), (
    "CELL 10 FAIL: Invalid search_layer value."
)

assert all(
    value in ALLOWED_COMPETITOR
    for value in literature_registry["possible_competitor"].tolist()
), (
    "CELL 10 FAIL: Invalid possible_competitor value."
)

print("✓ initial_relevance controlled vocabulary PASS")
print("✓ search_layer controlled vocabulary PASS")
print("✓ possible_competitor controlled vocabulary PASS")

# ================================================================================================================
# STEP 8 — ANCHOR INTEGRITY QC
# ================================================================================================================

print("\n[STEP 8] ANCHOR INTEGRITY QC")
print("-" * 112)

anchor = literature_registry[
    literature_registry["candidate_id"] == "ANCHOR-001"
]

assert len(anchor) == 1, (
    "CELL 10 FAIL: Exactly one ANCHOR-001 record is required."
)

ANCHOR_URL = "https://aclanthology.org/2025.arabicnlp-main.6/"

assert anchor.iloc[0]["url"] == ANCHOR_URL, (
    "CELL 10 FAIL: Anchor URL does not match the verified ACL Anthology URL."
)

assert anchor.iloc[0]["search_layer"] == "A_ANCHOR", (
    "CELL 10 FAIL: Anchor record has incorrect search layer."
)

print("✓ Exactly one anchor PASS")
print("✓ Verified Anchor URL PASS")
print("✓ Anchor layer PASS")

# ================================================================================================================
# STEP 9 — LAYER / COMPETITOR QC
# ================================================================================================================

print("\n[STEP 9] LAYER / COMPETITOR QC")
print("-" * 112)

anchor_count = (
    literature_registry["search_layer"] == "A_ANCHOR"
).sum()

methodological_count = (
    literature_registry["search_layer"] == "B_DIRECT_METHODOLOGICAL"
).sum()

high_count = (
    literature_registry["initial_relevance"] == "HIGH"
).sum()

competitor_count = (
    literature_registry["possible_competitor"] == True
).sum()

assert anchor_count == 1
assert methodological_count == 4
assert high_count == 5
assert competitor_count == 5

print(f"Anchor records       : {anchor_count}")
print(f"Methodological       : {methodological_count}")
print(f"HIGH relevance       : {high_count}")
print(f"Possible competitors : {competitor_count}")

print("✓ Layer distribution PASS")
print("✓ Relevance distribution PASS")
print("✓ Competitor preservation PASS")

# ================================================================================================================
# STEP 10 — DUPLICATE / VERSION GROUP QC
# ================================================================================================================

print("\n[STEP 10] DUPLICATE / VERSION GROUP QC")
print("-" * 112)

assert literature_registry["duplicate_group"].notna().all()
assert literature_registry["duplicate_group"].astype(str).str.strip().ne("").all()

duplicate_group_counts = (
    literature_registry["duplicate_group"]
    .astype(str)
    .value_counts()
)

print("Duplicate/version groups:")
for group, count in duplicate_group_counts.items():
    print(f"  {group}: {count}")

assert duplicate_group_counts.max() == 1, (
    "CELL 10 FAIL: Multiple records share a duplicate_group. "
    "This registry is expected to contain one record per approved study."
)

print("✓ No duplicate inflation detected")

# ================================================================================================================
# STEP 11 — SEARCH-FAMILY COVERAGE RECORD
# ================================================================================================================

print("\n[STEP 11] SEARCH-FAMILY COVERAGE")
print("-" * 112)

family_counts = (
    literature_registry["search_family"]
    .astype(str)
    .value_counts()
)

for family, count in family_counts.items():
    print(f"  {family}: {count}")

print()
print("NOTE:")
print("Cell 10 does not fabricate zero-result search-family records.")
print("The registry contains only the approved discovered candidates.")
print("Coverage/saturation interpretation remains a downstream audit decision.")

# ================================================================================================================
# STEP 12 — REGISTRY CONTENT FINGERPRINT
# ================================================================================================================

print("\n[STEP 12] REGISTRY CONTENT FINGERPRINT")
print("-" * 112)

# Deterministic representation for hashing.
# Sorting by candidate_id prevents row-order differences from changing
# the registry fingerprint.

registry_for_hash = (
    literature_registry[REGISTRY_COLUMNS]
    .copy()
    .sort_values("candidate_id")
    .reset_index(drop=True)
)

# Normalize representation only for deterministic fingerprinting.
registry_csv_bytes = registry_for_hash.to_csv(
    index=False,
    lineterminator="\n"
).encode("utf-8")

registry_sha256 = hashlib.sha256(
    registry_csv_bytes
).hexdigest()

print(f"Registry SHA-256 : {registry_sha256}")

# ================================================================================================================
# STEP 13 — PERSISTENCE PATH
# ================================================================================================================

print("\n[STEP 13] PERSISTENCE PATH")
print("-" * 112)

from google.colab import drive

DRIVE_ROOT = "/content/drive"

if not os.path.exists(DRIVE_ROOT):
    print("Google Drive mount point not found. Attempting mount...")
    drive.mount(DRIVE_ROOT)

PROJECT_ROOT = (
    "/content/drive/MyDrive/X-AutoAlign_Legal_RAG"
)

CHECKPOINT_DIR = (
    f"{PROJECT_ROOT}/04_AUDIT/"
    f"NOVELTY_GAP_AUDIT/CHECKPOINTS"
)

os.makedirs(CHECKPOINT_DIR, exist_ok=True)

PARQUET_PATH = (
    f"{CHECKPOINT_DIR}/"
    f"LITERATURE_REGISTRY_V002.parquet"
)

CSV_PATH = (
    f"{CHECKPOINT_DIR}/"
    f"LITERATURE_REGISTRY_V002.csv"
)

MANIFEST_PATH = (
    f"{CHECKPOINT_DIR}/"
    f"LITERATURE_REGISTRY_V002_MANIFEST.json"
)

print(f"Project root : {PROJECT_ROOT}")
print(f"Checkpoint   : {CHECKPOINT_DIR}")

# ================================================================================================================
# STEP 14 — PERSIST REGISTRY
# ================================================================================================================

print("\n[STEP 14] PERSISTING VALIDATED REGISTRY")
print("-" * 112)

literature_registry.to_parquet(
    PARQUET_PATH,
    index=False
)

literature_registry.to_csv(
    CSV_PATH,
    index=False
)

print("✓ Parquet written")
print("✓ CSV written")

# ================================================================================================================
# STEP 15 — MANIFEST
# ================================================================================================================

print("\n[STEP 15] WRITING MANIFEST")
print("-" * 112)

manifest = {
    "experiment_id": EXPERIMENT_ID,
    "cell_id": CELL_ID,
    "version": VERSION,
    "artifact_type": "LITERATURE_REGISTRY_PERSISTENT_CHECKPOINT",
    "status": "VALIDATED_AND_PERSISTED",
    "scientific_scope": "BOUNDED_LITERATURE_RESEARCH_GAP_AUDIT",
    "source_cell": "CELL 9 v002",
    "row_count": int(len(literature_registry)),
    "column_count": int(len(literature_registry.columns)),
    "candidate_ids": sorted(actual_ids),
    "anchor_count": int(anchor_count),
    "methodological_count": int(methodological_count),
    "high_relevance_count": int(high_count),
    "possible_competitor_count": int(competitor_count),
    "registry_sha256": registry_sha256,
    "parquet_path": PARQUET_PATH,
    "csv_path": CSV_PATH,
    "created_at": datetime.now().isoformat(),
    "firewall": FIREWALL,
    "next_authorized_stage": "CELL 11",
}

with open(
    MANIFEST_PATH,
    "w",
    encoding="utf-8"
) as f:
    json.dump(
        manifest,
        f,
        ensure_ascii=False,
        indent=2
    )

print("✓ Manifest written")

# ================================================================================================================
# STEP 16 — FILE EXISTENCE QC
# ================================================================================================================

print("\n[STEP 16] PERSISTENCE FILE QC")
print("-" * 112)

for path in [
    PARQUET_PATH,
    CSV_PATH,
    MANIFEST_PATH,
]:
    assert os.path.exists(path), (
        f"CELL 10 FAIL: Expected artifact does not exist: {path}"
    )

    size = os.path.getsize(path)

    assert size > 0, (
        f"CELL 10 FAIL: Empty artifact: {path}"
    )

    print(f"✓ {os.path.basename(path)} | {size:,} bytes")

# ================================================================================================================
# STEP 17 — INDEPENDENT PARQUET RELOAD
# ================================================================================================================

print("\n[STEP 17] INDEPENDENT PARQUET RELOAD")
print("-" * 112)

reloaded_registry = pd.read_parquet(
    PARQUET_PATH
)

assert list(reloaded_registry.columns) == REGISTRY_COLUMNS
assert len(reloaded_registry) == 5

reloaded_ids = set(
    reloaded_registry["candidate_id"].astype(str)
)

assert reloaded_ids == EXPECTED_IDS

print("✓ Independent Parquet reload PASS")
print(f"Reloaded rows    : {len(reloaded_registry)}")
print(f"Reloaded columns : {len(reloaded_registry.columns)}")

# ================================================================================================================
# STEP 18 — RELOAD HASH VERIFICATION
# ================================================================================================================

print("\n[STEP 18] RELOAD HASH VERIFICATION")
print("-" * 112)

reloaded_for_hash = (
    reloaded_registry[REGISTRY_COLUMNS]
    .copy()
    .sort_values("candidate_id")
    .reset_index(drop=True)
)

reloaded_csv_bytes = reloaded_for_hash.to_csv(
    index=False,
    lineterminator="\n"
).encode("utf-8")

reloaded_sha256 = hashlib.sha256(
    reloaded_csv_bytes
).hexdigest()

print(f"Original SHA-256 : {registry_sha256}")
print(f"Reloaded SHA-256 : {reloaded_sha256}")

assert registry_sha256 == reloaded_sha256, (
    "CELL 10 FAIL: SHA-256 mismatch after independent reload."
)

print("✓ SHA-256 integrity PASS")

# ================================================================================================================
# STEP 19 — MANIFEST RELOAD / INTEGRITY CHECK
# ================================================================================================================

print("\n[STEP 19] MANIFEST INTEGRITY CHECK")
print("-" * 112)

with open(
    MANIFEST_PATH,
    "r",
    encoding="utf-8"
) as f:
    reloaded_manifest = json.load(f)

assert reloaded_manifest["artifact_type"] == (
    "LITERATURE_REGISTRY_PERSISTENT_CHECKPOINT"
)

assert reloaded_manifest["status"] == (
    "VALIDATED_AND_PERSISTED"
)

assert reloaded_manifest["row_count"] == 5
assert reloaded_manifest["column_count"] == 18
assert set(reloaded_manifest["candidate_ids"]) == EXPECTED_IDS

assert reloaded_manifest["registry_sha256"] == registry_sha256

print("✓ Manifest structure PASS")
print("✓ Manifest row/column count PASS")
print("✓ Manifest candidate-ID PASS")
print("✓ Manifest SHA-256 PASS")

# ================================================================================================================
# STEP 20 — FINAL QC SUMMARY
# ================================================================================================================

print("\n[STEP 20] FINAL QC SUMMARY")
print("-" * 112)

QC_RESULTS = {
    "schema": True,
    "row_count": len(literature_registry) == 5,
    "candidate_ids": actual_ids == EXPECTED_IDS,
    "unique_ids": literature_registry["candidate_id"].is_unique,
    "required_fields": True,
    "controlled_values": True,
    "anchor_integrity": True,
    "layer_distribution": (
        anchor_count == 1 and
        methodological_count == 4
    ),
    "high_relevance": high_count == 5,
    "competitor_preservation": competitor_count == 5,
    "duplicate_control": True,
    "parquet_persistence": os.path.exists(PARQUET_PATH),
    "csv_persistence": os.path.exists(CSV_PATH),
    "manifest_persistence": os.path.exists(MANIFEST_PATH),
    "independent_reload": True,
    "sha256_integrity": registry_sha256 == reloaded_sha256,
    "manifest_integrity": True,
}

for check_name, result in QC_RESULTS.items():
    print(f"  {check_name:<28} : {'PASS' if result else 'FAIL'}")

assert all(QC_RESULTS.values()), (
    "CELL 10 FAIL: One or more final QC checks failed."
)

# ================================================================================================================
# STEP 21 — FINAL STATUS
# ================================================================================================================

CELL_10_STATUS = (
    "LITERATURE_REGISTRY_PERSISTENT_CHECKPOINT_READY"
)

print("\n" + "=" * 112)
print("CELL 10 FINAL STATUS")
print("=" * 112)

print(f"STATUS : {CELL_10_STATUS}")
print()
print("✓ Literature registry QC PASS")
print("✓ Exactly 5 approved records")
print("✓ 18-column frozen schema")
print("✓ Anchor integrity PASS")
print("✓ Methodological-layer integrity PASS")
print("✓ Required-field completeness PASS")
print("✓ Duplicate control PASS")
print("✓ Parquet persistence PASS")
print("✓ CSV persistence PASS")
print("✓ Manifest persistence PASS")
print("✓ Independent reload PASS")
print("✓ SHA-256 integrity PASS")
print("✓ Manifest integrity PASS")
print()
print("NO NEW LITERATURE DISCOVERY")
print("NO NOVELTY JUDGMENT")
print("NO Q1–Q10 JUDGMENT")
print("NO DATASET MODIFICATION")
print("NO MODEL INFERENCE")
print("NO TRAINING")
print("NO ARCHITECTURE IMPLEMENTATION")
print()
print("NEXT AUTHORIZED ACTION:")
print("Proceed ONLY to CELL 11.")
print("=" * 112)
```

```python
# ================================================================================================================
# CELL 11 v001 — METHODOLOGICAL COMPETITOR COMPARISON
# ================================================================================================================
#
# PURPOSE
# -------
# Compare the approved HIGH-relevance literature candidates against the
# frozen F-D+R formulation using one fixed methodological framework.
#
# SCIENTIFIC VALUE GATE
# ---------------------
# ما فائدتها للرسالة العلمية أو لبناء الـdataset؟
#
# To determine, using source-grounded methodological evidence, whether the
# proposed F-D+R formulation remains distinguishable from the closest
# prior work before architecture experimentation.
#
# IMPORTANT
# ---------
# ✓ Uses ONLY the persistent literature registry from CELL 10
# ✓ No new literature discovery
# ✓ No modification of the search protocol
# ✓ No dataset modification
# ✓ No model inference
# ✓ No training
# ✓ No architecture implementation
# ✓ No prompt revision
# ✓ No new adjudication
# ✓ No novelty claim is generated automatically
#
# This cell creates the CONTROLLED COMPARISON FRAME only.
# Actual paper-level evidence must be supplied from the audited primary
# sources before a final methodological conclusion is accepted.
#
# ================================================================================================================

import os
import json
import pandas as pd

print("=" * 112)
print("CELL 11 v001 — METHODOLOGICAL COMPETITOR COMPARISON")
print("=" * 112)

# ================================================================================================================
# STEP 0 — SCIENTIFIC VALUE GATE
# ================================================================================================================

print("\n[STEP 0] SCIENTIFIC VALUE GATE")
print("-" * 112)

SCIENTIFIC_VALUE_GATE = True

print("Question:")
print("ما فائدتها للرسالة العلمية أو لبناء الـdataset؟")
print()
print("Answer:")
print("To compare the strongest audited literature candidates against the")
print("frozen F-D+R formulation before any architecture-level experiment.")
print()
print("✓ Direct value to the thesis")
print("✓ Required for a defensible research-gap decision")
print("✓ No dataset modification")
print("✓ No model inference")
print("✓ No architecture implementation")

if not SCIENTIFIC_VALUE_GATE:
    raise RuntimeError(
        "CELL 11 STOP: Scientific Value Gate not satisfied."
    )

# ================================================================================================================
# STEP 1 — EXPERIMENT IDENTITY
# ================================================================================================================

print("\n[STEP 1] EXPERIMENT IDENTITY")
print("-" * 112)

EXPERIMENT_ID = "FDR-NOVELTY-GAP-001"
CELL_ID = "CELL11_METHODOLOGICAL_COMPETITOR_COMPARISON"
VERSION = "V001"

print(f"Experiment ID : {EXPERIMENT_ID}")
print(f"Cell ID       : {CELL_ID}")
print(f"Version       : {VERSION}")

# ================================================================================================================
# STEP 2 — SCIENTIFIC FIREWALL
# ================================================================================================================

print("\n[STEP 2] SCIENTIFIC FIREWALL")
print("-" * 112)

FIREWALL = {
    "new_literature_search": False,
    "new_candidate_addition": False,
    "search_protocol_modification": False,
    "novelty_automatic_decision": False,
    "dataset_modification": False,
    "gold_generation": False,
    "y_plus_generation": False,
    "y_minus_generation": False,
    "preference_generation": False,
    "new_adjudication": False,
    "model_loading": False,
    "model_inference": False,
    "model_training": False,
    "architecture_implementation": False,
    "prompt_revision": False,
    "cohort_modification": False,
    "validation_reopen": False,
    "q82_expansion": False,
}

for key, value in FIREWALL.items():
    if value is not False:
        raise RuntimeError(
            f"CELL 11 FIREWALL VIOLATION: {key}"
        )

print("✓ Scientific firewall PASS")

# ================================================================================================================
# STEP 3 — REQUIRED PERSISTENT INPUT
# ================================================================================================================

print("\n[STEP 3] REQUIRED PERSISTENT INPUT")
print("-" * 112)

PROJECT_ROOT = "/content/drive/MyDrive/X-AutoAlign_Legal_RAG"

CHECKPOINT_DIR = (
    f"{PROJECT_ROOT}/04_AUDIT/"
    f"NOVELTY_GAP_AUDIT/CHECKPOINTS"
)

PARQUET_PATH = (
    f"{CHECKPOINT_DIR}/LITERATURE_REGISTRY_V002.parquet"
)

MANIFEST_PATH = (
    f"{CHECKPOINT_DIR}/LITERATURE_REGISTRY_V002_MANIFEST.json"
)

if not os.path.exists(PARQUET_PATH):
    raise RuntimeError(
        "CELL 11 STOP: Persistent literature registry checkpoint not found.\n"
        "Expected:\n"
        f"{PARQUET_PATH}\n"
        "Run CELL 10 v002 first."
    )

if not os.path.exists(MANIFEST_PATH):
    raise RuntimeError(
        "CELL 11 STOP: Literature registry manifest not found.\n"
        "Run CELL 10 v002 first."
    )

print("✓ Persistent registry checkpoint found")
print("✓ Manifest found")

# ================================================================================================================
# STEP 4 — LOAD AND VERIFY PERSISTENT REGISTRY
# ================================================================================================================

print("\n[STEP 4] LOAD PERSISTENT REGISTRY")
print("-" * 112)

literature_registry = pd.read_parquet(
    PARQUET_PATH
)

with open(
    MANIFEST_PATH,
    "r",
    encoding="utf-8"
) as f:
    registry_manifest = json.load(f)

EXPECTED_IDS = {
    "ANCHOR-001",
    "LIT-001",
    "LIT-002",
    "LIT-003",
    "LIT-004",
}

REGISTRY_COLUMNS = [
    "candidate_id",
    "search_family",
    "search_layer",
    "search_query",
    "title",
    "authors",
    "year",
    "venue",
    "primary_source",
    "doi_or_identifier",
    "url",
    "retrieval_date",
    "initial_relevance",
    "possible_competitor",
    "duplicate_group",
    "version_status",
    "discovery_reason",
    "notes",
]

assert list(literature_registry.columns) == REGISTRY_COLUMNS
assert len(literature_registry) == 5

assert set(
    literature_registry["candidate_id"].astype(str)
) == EXPECTED_IDS

assert registry_manifest["row_count"] == 5
assert registry_manifest["column_count"] == 18
assert set(registry_manifest["candidate_ids"]) == EXPECTED_IDS

print("✓ Registry schema verified")
print("✓ Registry row count verified")
print("✓ Candidate IDs verified")
print("✓ Manifest verified")

# ================================================================================================================
# STEP 5 — SELECT ONLY APPROVED HIGH-RELEVANCE CANDIDATES
# ================================================================================================================

print("\n[STEP 5] ELIGIBLE COMPARISON SET")
print("-" * 112)

eligible_registry = literature_registry[
    literature_registry["initial_relevance"] == "HIGH"
].copy()

eligible_registry = (
    eligible_registry
    .sort_values("candidate_id")
    .reset_index(drop=True)
)

print(f"Eligible HIGH candidates : {len(eligible_registry)}")

assert len(eligible_registry) == 5

print("\nEligible candidates:")
for cid in eligible_registry["candidate_id"]:
    print(f"  • {cid}")

print("✓ Only approved HIGH-relevance candidates retained")

# ================================================================================================================
# STEP 6 — FROZEN F-D+R REFERENCE FORMULATION
# ================================================================================================================

print("\n[STEP 6] FROZEN F-D+R REFERENCE FORMULATION")
print("-" * 112)

FDR_REFERENCE = {
    "stage": "POST_RETRIEVAL / PRE-GENERATION",
    "input": "QUESTION + CANDIDATE_EVIDENCE",
    "purpose": "EVIDENCE-SUITABILITY VERIFICATION",
    "dimensions": [
        "CASE_RELEVANCE",
        "LEGAL_SUPPORT",
        "SUFFICIENCY",
        "EVIDENCE_SUITABILITY",
        "CONTRADICTION_RISK",
        "UNCERTAINTY",
    ],
    "output_role": "EVIDENCE_ELIGIBILITY / RISK_GATE",
}

for key, value in FDR_REFERENCE.items():
    print(f"{key:<20}: {value}")

print()
print("✓ F-D+R formulation remains frozen")
print("✓ No reformulation performed")

# ================================================================================================================
# STEP 7 — FIXED PAPER-LEVEL COMPARISON SCHEMA
# ================================================================================================================

print("\n[STEP 7] FIXED COMPARISON SCHEMA")
print("-" * 112)

COMPARISON_COLUMNS = [
    "candidate_id",
    "paper",
    "priority",
    "task_summary",
    "pipeline_stage",
    "verifier_input",
    "evidence_target",
    "support",
    "sufficiency",
    "contradiction_risk",
    "uncertainty",
    "human_grounding",
    "label_structure",
    "reference_answer_dependency",
    "pre_generation_gate",
    "legal_scope",
    "multilingual_scope",
    "retrieval_dependence",
    "generation_dependence",
    "case_level_evaluation",
    "overlap_level",
    "fdr_distinctiveness",
    "critical_challenge",
    "evidence_source",
    "confidence",
    "notes",
]

print(f"Comparison fields : {len(COMPARISON_COLUMNS)}")

# ================================================================================================================
# STEP 8 — CONTROLLED VOCABULARIES
# ================================================================================================================

print("\n[STEP 8] CONTROLLED VOCABULARIES")
print("-" * 112)

ALLOWED_PRIORITY = {
    "ANCHOR",
    "P1_DIRECT",
    "P2_CLOSE",
    "P3_RELATED",
}

ALLOWED_OVERLAP = {
    "O0_NONE",
    "O1_BACKGROUND",
    "O2_PARTIAL",
    "O3_CLOSE",
    "O4_SUBSTANTIAL",
    "NOT_YET_ASSESSED",
}

ALLOWED_DISTINCTIVENESS = {
    "NEW_TASK_BOUNDARY",
    "NEW_COMBINATION",
    "NEW_HUMAN_GROUNDED_LABELING",
    "NEW_PRE_GENERATION_CONTROL_ROLE",
    "NEW_DOMAIN_LANGUAGE_VALIDATION",
    "NO_CLEAR_ADDITION",
    "NOT_YET_ASSESSED",
}

ALLOWED_CONFIDENCE = {
    "HIGH",
    "MEDIUM",
    "LOW",
    "NOT_YET_ASSESSED",
}

print("✓ Priority vocabulary frozen")
print("✓ Overlap vocabulary frozen")
print("✓ Distinctiveness vocabulary frozen")
print("✓ Confidence vocabulary frozen")

# ================================================================================================================
# STEP 9 — COMPARISON FRAME INITIALIZATION
# ================================================================================================================
#
# IMPORTANT:
# Empty fields are intentional.
# This cell MUST NOT invent methodological facts about papers.
#
# Paper-level claims require evidence from the primary paper.
#
# ================================================================================================================

print("\n[STEP 9] INITIALIZING SOURCE-GROUNDED COMPARISON FRAME")
print("-" * 112)

comparison_records = []

for _, row in eligible_registry.iterrows():

    candidate_id = row["candidate_id"]

    if candidate_id == "ANCHOR-001":
        priority = "ANCHOR"
    else:
        priority = "P1_DIRECT"

    comparison_records.append({
        "candidate_id": candidate_id,
        "paper": row["title"],
        "priority": priority,

        "task_summary": None,
        "pipeline_stage": None,
        "verifier_input": None,
        "evidence_target": None,
        "support": None,
        "sufficiency": None,
        "contradiction_risk": None,
        "uncertainty": None,
        "human_grounding": None,
        "label_structure": None,
        "reference_answer_dependency": None,
        "pre_generation_gate": None,
        "legal_scope": None,
        "multilingual_scope": None,
        "retrieval_dependence": None,
        "generation_dependence": None,
        "case_level_evaluation": None,

        "overlap_level": "NOT_YET_ASSESSED",
        "fdr_distinctiveness": "NOT_YET_ASSESSED",
        "critical_challenge": None,

        "evidence_source": row["primary_source"],
        "confidence": "NOT_YET_ASSESSED",
        "notes": (
            "Comparison record initialized from the validated registry. "
            "Methodological fields require source-grounded inspection."
        ),
    })

literature_gap_matrix = pd.DataFrame(
    comparison_records,
    columns=COMPARISON_COLUMNS
)

assert list(literature_gap_matrix.columns) == COMPARISON_COLUMNS
assert len(literature_gap_matrix) == 5

print(f"Comparison records initialized : {len(literature_gap_matrix)}")
print("✓ No methodological values fabricated")

# ================================================================================================================
# STEP 10 — CRITICAL CHALLENGE FRAME
# ================================================================================================================

print("\n[STEP 10] CRITICAL CHALLENGE FRAME")
print("-" * 112)

CRITICAL_CHALLENGE_QUESTIONS = {
    "Q1": "Does the work verify retrieved evidence before generation?",
    "Q2": "Is the verifier input essentially QUESTION + CANDIDATE_EVIDENCE?",
    "Q3": "Does it distinguish relevance from actual evidential support?",
    "Q4": "Does it explicitly evaluate evidence sufficiency?",
    "Q5": "Does it explicitly model contradiction or evidence risk?",
    "Q6": "Does it explicitly represent uncertainty?",
    "Q7": "Does its output control evidence eligibility before generation?",
    "Q8": "Is the verification/assessment human-grounded?",
    "Q9": "Does it operate in a legal and/or multilingual Arabic-English setting?",
    "Q10": "If adapted to Legal RAG, would the adaptation be merely an application change?",
}

for q, text in CRITICAL_CHALLENGE_QUESTIONS.items():
    print(f"{q}: {text}")

print()
print("✓ Q1–Q10 framework frozen")
print("✓ No answers assigned automatically")

# ================================================================================================================
# STEP 11 — SOURCE EVIDENCE REQUIREMENT
# ================================================================================================================

print("\n[STEP 11] SOURCE EVIDENCE REQUIREMENT")
print("-" * 112)

SOURCE_STANDARD = {
    "HIGH": "Full paper / primary scholarly source directly supports the claim.",
    "MEDIUM": "Substantial methodological evidence supports the claim, but one element remains incomplete.",
    "LOW": "Claim relies mainly on abstract or limited accessible evidence.",
}

for level, definition in SOURCE_STANDARD.items():
    print(f"{level:<8}: {definition}")

print()
print("✓ Final equivalence/novelty claims require primary-source evidence")
print("✓ Abstract-only evidence cannot establish substantial equivalence")

# ================================================================================================================
# STEP 12 — ANTI-OVERCLAIM GATE
# ================================================================================================================

print("\n[STEP 12] ANTI-OVERCLAIM GATE")
print("-" * 112)

FORBIDDEN_CONCLUSIONS = [
    "Nobody has done this.",
    "F-D+R is the first.",
    "F-D+R is novel.",
    "This paper is irrelevant.",
    "This paper is equivalent.",
    "F-D+R is superior.",
]

for statement in FORBIDDEN_CONCLUSIONS:
    print(f"✗ BLOCKED: {statement}")

print()
print("Permitted bounded language:")
print("✓ No substantially equivalent formulation was identified in the audited evidence.")
print("✓ The work overlaps with F-D+R in X but differs in Y.")
print("✓ Available evidence supports / does not support a methodological distinction.")
print("✓ Novelty remains inconclusive.")

# ================================================================================================================
# STEP 13 — NO-AUTOMATIC-JUDGMENT ASSERTION
# ================================================================================================================

print("\n[STEP 13] NO-AUTOMATIC-JUDGMENT ASSERTION")
print("-" * 112)

assert (
    literature_gap_matrix["overlap_level"]
    .eq("NOT_YET_ASSESSED")
    .all()
)

assert (
    literature_gap_matrix["fdr_distinctiveness"]
    .eq("NOT_YET_ASSESSED")
    .all()
)

assert (
    literature_gap_matrix["confidence"]
    .eq("NOT_YET_ASSESSED")
    .all()
)

print("✓ No overlap level fabricated")
print("✓ No distinctiveness judgment fabricated")
print("✓ No confidence judgment fabricated")

# ================================================================================================================
# STEP 14 — SAVE COMPARISON FRAME IN RAM ONLY
# ================================================================================================================

print("\n[STEP 14] RAM ARTIFACT")
print("-" * 112)

LITERATURE_GAP_MATRIX_STATUS = (
    "COMPARISON_FRAME_INITIALIZED_SOURCE_EVIDENCE_REQUIRED"
)

print(f"Artifact : LITERATURE_GAP_MATRIX_V002")
print(f"Status   : {LITERATURE_GAP_MATRIX_STATUS}")

print()
print("The matrix is intentionally NOT persisted by this cell.")
print("Persistence should occur only after source-grounded comparison entries")
print("have been completed and independently checked.")

# ================================================================================================================
# STEP 15 — FINAL READ-BACK
# ================================================================================================================

print("\n[STEP 15] READ-BACK")
print("-" * 112)

print(f"Experiment ID                    : {EXPERIMENT_ID}")
print(f"Cell                              : {CELL_ID}")
print(f"Registry candidates               : {len(eligible_registry)}")
print(f"Comparison records                : {len(literature_gap_matrix)}")
print(f"F-D+R stage                       : {FDR_REFERENCE['stage']}")
print(f"F-D+R input                       : {FDR_REFERENCE['input']}")
print(f"F-D+R target                      : {FDR_REFERENCE['purpose']}")
print(f"Comparison judgment               : NOT PERFORMED")
print(f"Novelty judgment                  : NOT PERFORMED")
print(f"Gap decision                      : NOT PERFORMED")
print(f"Dataset modified                  : False")
print(f"Model used                        : False")
print(f"Architecture modified             : False")
print(f"199 cohort reopened               : False")
print(f"400 validation reopened           : False")
print(f"Q82 expanded                      : False")

# ================================================================================================================
# FINAL STATUS
# ================================================================================================================

print("\n" + "=" * 112)
print("CELL 11 FINAL STATUS")
print("=" * 112)

CELL_11_STATUS = (
    "COMPARISON_FRAME_READY_FOR_SOURCE_GROUNDED_METHODOLOGICAL_REVIEW"
)

print(f"STATUS : {CELL_11_STATUS}")
print()
print("✓ Persistent registry successfully consumed")
print("✓ Five approved HIGH-relevance records loaded")
print("✓ Fixed F-D+R formulation preserved")
print("✓ Fixed comparison schema created")
print("✓ Q1–Q10 challenge framework created")
print("✓ Anti-overclaim firewall active")
print("✓ No methodological judgments fabricated")
print("✓ No novelty decision made")
print()
print("NEXT AUTHORIZED ACTION:")
print("Populate the comparison fields ONLY from the audited primary scholarly sources.")
print()
print("DO NOT:")
print("✗ perform new broad literature search")
print("✗ modify F-D+R formulation")
print("✗ modify dataset")
print("✗ reopen 199 cohort")
print("✗ reopen 400 validation")
print("✗ expand Q82")
print("✗ implement architecture alternatives")
print("✗ train models")
print("✗ make an automatic novelty decision")
print("=" * 112)
```

```python
# ================================================================================================================
# X-AutoAlign Legal RAG
# F-D+R — BOUNDED LITERATURE / NOVELTY GAP AUDIT
# CELL 13 v004 — PRIMARY-SOURCE IDENTITY VERIFICATION
#              + CANONICAL REGISTRY RESTORE GATE
# ================================================================================================================

from pathlib import Path
import pandas as pd
import hashlib
import json
import os


print("=" * 112)
print("X-AutoAlign Legal RAG")
print("F-D+R — BOUNDED LITERATURE / NOVELTY GAP AUDIT")
print("CELL 13 v004 — PRIMARY-SOURCE IDENTITY VERIFICATION")
print("             + CANONICAL REGISTRY RESTORE GATE")
print("=" * 112)


# ================================================================================================================
# [STEP 0] SCIENTIFIC VALUE GATE
# ================================================================================================================

print("\n" + "=" * 112)
print("[STEP 0] SCIENTIFIC VALUE GATE")
print("=" * 112)

print("""
Question:
  ما فائدتها للرسالة العلمية أو لبناء الـdataset؟

Answer:
  هذه الخلية ضرورية لتثبيت هوية كل Primary Source قبل استخراج
  Q1-Q10 methodological evidence.

  الـcanonical literature registry تم استعادته وتصحيحه في
  CELL 13 v005، ولذلك يجب أن تستخدم هذه الخلية النسخة المصححة
  V003 وليس الـobsolete V002.

  الهدف:
    • prevent Candidate-ID attribution errors
    • preserve source traceability
    • preserve reproducibility
    • preserve literature audit integrity
    • guarantee correct Q1-Q10 attribution

Decision:
  PROCEED ONLY WITH CANONICAL REGISTRY RESTORE
  + EXACT SOURCE IDENTITY VERIFICATION
""")


# ================================================================================================================
# [STEP 1] EXPERIMENT / CELL ID
# ================================================================================================================

print("=" * 112)
print("[STEP 1] EXPERIMENT / CELL ID")
print("=" * 112)

EXPERIMENT_ID = "FDR-NOVELTY-GAP-001"
CELL_ID = "CELL13_PRIMARY_SOURCE_IDENTITY_VERIFICATION_V004"
CELL_STATUS = "READ_ONLY"

print(f"Experiment ID : {EXPERIMENT_ID}")
print(f"Cell ID       : {CELL_ID}")
print(f"Status        : {CELL_STATUS}")


# ================================================================================================================
# [STEP 2] SCIENTIFIC FIREWALL
# ================================================================================================================

print("\n" + "=" * 112)
print("[STEP 2] SCIENTIFIC FIREWALL")
print("=" * 112)

FIREWALL = {
    "dataset_modification": "BLOCKED",
    "199_cohort_modification": "BLOCKED",
    "400_validation_modification": "BLOCKED",
    "Q82_modification": "BLOCKED",
    "new_adjudication": "BLOCKED",
    "gold_generation": "BLOCKED",
    "y_positive_generation": "BLOCKED",
    "y_negative_generation": "BLOCKED",
    "preference_generation": "BLOCKED",
    "model_training": "BLOCKED",
    "model_inference": "BLOCKED",
    "prompt_revision": "BLOCKED",
    "architecture_modification": "BLOCKED",
    "new_legal_extraction": "BLOCKED",
    "validation_reopen": "BLOCKED",
    "automatic_novelty_judgment": "BLOCKED",
    "automatic_overlap_judgment": "BLOCKED",
}

for key, value in FIREWALL.items():
    print(f"{key:<42}: {value}")

assert all(v == "BLOCKED" for v in FIREWALL.values())

print("✓ Firewall PASS")


# ================================================================================================================
# [STEP 3] CANONICAL REGISTRY CHECKPOINT
# ================================================================================================================

print("\n" + "=" * 112)
print("[STEP 3] CANONICAL REGISTRY CHECKPOINT")
print("=" * 112)

CHECKPOINT_DIR = Path(
    "/content/drive/MyDrive/X-AutoAlign_Legal_RAG/"
    "04_AUDIT/NOVELTY_GAP_AUDIT/CHECKPOINTS"
)

# V003 is the corrected canonical registry produced by CELL 13 v005.
# V002 is intentionally NOT used here because it contains the
# LIT-003 / LIT-004 identity inversion.

REGISTRY_VERSION = "V003"

REGISTRY_PARQUET = (
    CHECKPOINT_DIR / f"LITERATURE_REGISTRY_{REGISTRY_VERSION}.parquet"
)

REGISTRY_CSV = (
    CHECKPOINT_DIR / f"LITERATURE_REGISTRY_{REGISTRY_VERSION}.csv"
)

REGISTRY_MANIFEST = (
    CHECKPOINT_DIR / f"LITERATURE_REGISTRY_{REGISTRY_VERSION}_MANIFEST.json"
)

print("Checkpoint directory:")
print(f"  {CHECKPOINT_DIR}")

print("\nOperational canonical registry:")
print(f"  {REGISTRY_PARQUET}")

print("\nOperational CSV:")
print(f"  {REGISTRY_CSV}")

print("\nOperational manifest:")
print(f"  {REGISTRY_MANIFEST}")


# ================================================================================================================
# [STEP 4] CANONICAL REGISTRY RESTORE GATE
# ================================================================================================================

print("\n" + "=" * 112)
print("[STEP 4] CANONICAL REGISTRY RESTORE GATE")
print("=" * 112)

if not CHECKPOINT_DIR.exists():
    raise RuntimeError(
        "\nCELL 13 STOP: Canonical registry checkpoint directory does not exist.\n"
        f"Expected:\n  {CHECKPOINT_DIR}"
    )

print("✓ Checkpoint directory exists")

if not REGISTRY_PARQUET.exists():
    raise RuntimeError(
        "\nCELL 13 STOP: Corrected canonical registry V003 does not exist.\n"
        f"Expected:\n  {REGISTRY_PARQUET}\n"
        "\nDo NOT fall back to V002."
    )

print("✓ Corrected canonical Parquet V003 exists")

if not REGISTRY_CSV.exists():
    raise RuntimeError(
        "\nCELL 13 STOP: Canonical CSV V003 does not exist.\n"
        f"Expected:\n  {REGISTRY_CSV}"
    )

print("✓ Corrected canonical CSV V003 exists")


# ================================================================================================================
# [STEP 5] LOAD CANONICAL REGISTRY
# ================================================================================================================

print("\n" + "=" * 112)
print("[STEP 5] LOAD CANONICAL REGISTRY")
print("=" * 112)

registry = pd.read_parquet(REGISTRY_PARQUET)

print(f"Canonical registry rows : {registry.shape[0]}")
print(f"Canonical registry cols : {registry.shape[1]}")

if registry.shape[0] != 5:
    raise RuntimeError(
        "\nCELL 13 STOP: Canonical registry must contain exactly 5 records.\n"
        f"Observed rows: {registry.shape[0]}"
    )

print("✓ Exactly 5 canonical registry records loaded")


# ================================================================================================================
# [STEP 6] CANONICAL CANDIDATE-ID CHECK
# ================================================================================================================

print("\n" + "=" * 112)
print("[STEP 6] CANONICAL CANDIDATE-ID CHECK")
print("=" * 112)

EXPECTED_CANDIDATE_IDS = [
    "ANCHOR-001",
    "LIT-001",
    "LIT-002",
    "LIT-003",
    "LIT-004",
]

if "candidate_id" not in registry.columns:
    raise RuntimeError(
        "\nCELL 13 STOP: Required column 'candidate_id' is missing."
    )

observed_ids = registry["candidate_id"].astype(str).tolist()

print("Observed Candidate IDs:")
for cid in observed_ids:
    print(f"  {cid}")

if observed_ids != EXPECTED_CANDIDATE_IDS:
    raise RuntimeError(
        "\nCELL 13 STOP: Candidate IDs are not in the exact canonical order.\n"
        f"Observed : {observed_ids}\n"
        f"Expected : {EXPECTED_CANDIDATE_IDS}"
    )

if registry["candidate_id"].astype(str).nunique() != 5:
    raise RuntimeError(
        "\nCELL 13 STOP: Candidate IDs are not unique."
    )

print("✓ Candidate IDs exact")
print("✓ Candidate IDs unique")


# ================================================================================================================
# [STEP 7] CANONICAL TITLE PROTECTION
# ================================================================================================================

print("\n" + "=" * 112)
print("[STEP 7] CANONICAL TITLE PROTECTION")
print("=" * 112)

EXPECTED_TITLES = {
    "ANCHOR-001":
        "The Cross-Lingual Cost: Retrieval Biases in RAG over Arabic-English Corpora",

    "LIT-001":
        "SURE-RAG: Sufficiency and Uncertainty-Aware Evidence Verification for Selective Retrieval-Augmented Generation",

    "LIT-002":
        "Do LLMs Know When Evidence is Insufficient? An Evidence Sufficiency Benchmark for Answer-Abstention Calibration in Retrieval-Augmented Generation",

    "LIT-003":
        "From noisy feedback to evidence-aware issue specifications: an agent-governed retrieval-augmented generation approach",

    "LIT-004":
        "AdaKG-RAG: Adaptive KG-guided retrieval-augmented generation with hypothesis-and-verification retrieval for multi-hop question answering",
}

if "title" not in registry.columns:
    raise RuntimeError(
        "\nCELL 13 STOP: Required column 'title' is missing."
    )

for candidate_id in EXPECTED_CANDIDATE_IDS:

    row = registry.loc[
        registry["candidate_id"].astype(str) == candidate_id
    ]

    if len(row) != 1:
        raise RuntimeError(
            f"\nCELL 13 STOP: Expected exactly one registry row for {candidate_id}\n"
            f"Observed rows: {len(row)}"
        )

    observed_title = str(row.iloc[0]["title"]).strip()
    expected_title = EXPECTED_TITLES[candidate_id]

    if observed_title != expected_title:
        raise RuntimeError(
            f"\nCELL 13 STOP: Canonical registry title mismatch for {candidate_id}\n"
            f"Observed:\n  {observed_title}\n"
            f"Expected:\n  {expected_title}\n"
        )

    print(f"✓ {candidate_id} canonical title exact")


# ================================================================================================================
# [STEP 8] CANONICAL IDENTITY TABLE
# ================================================================================================================

print("\n" + "=" * 112)
print("[STEP 8] CANONICAL IDENTITY TABLE")
print("=" * 112)

identity_rows = []

for candidate_id in EXPECTED_CANDIDATE_IDS:

    row = registry.loc[
        registry["candidate_id"].astype(str) == candidate_id
    ].iloc[0]

    identity_rows.append({
        "candidate_id": candidate_id,
        "title": str(row["title"]).strip()
    })

identity_df = pd.DataFrame(identity_rows)

print(identity_df.to_string(index=False))

print("\n✓ Canonical identity table materialized")


# ================================================================================================================
# [STEP 9] PRIMARY-SOURCE DIRECTORY
# ================================================================================================================

print("\n" + "=" * 112)
print("[STEP 9] PRIMARY-SOURCE DIRECTORY")
print("=" * 112)

SOURCE_DIR = Path(
    "/content/drive/MyDrive/F-D+R/"
    "DATASET_CONSTRUCTION/SOURCES"
)

print("Primary-source directory:")
print(f"  {SOURCE_DIR}")

if not SOURCE_DIR.exists():
    raise RuntimeError(
        "\nCELL 13 STOP: Primary-source directory does not exist.\n"
        f"Expected:\n  {SOURCE_DIR}"
    )

print("✓ Primary-source directory exists")


# ================================================================================================================
# [STEP 10] EXACT PRIMARY-SOURCE MAP
# ================================================================================================================

print("\n" + "=" * 112)
print("[STEP 10] EXACT PRIMARY-SOURCE MAP")
print("=" * 112)

EXPECTED_SOURCE_FILES = {
    "ANCHOR-001":
        "The Cross Lingual Cost Retrieval Biases in RAG over Arabic English Corpora.pdf",

    "LIT-001":
        "SURE RAG Sufficiency and Uncertainty-Aware.pdf",

    "LIT-002":
        "Do LLMs Know When Evidence is Insufficient An Evidence Sufficiency.pdf",

    "LIT-003":
        "From noisy feedback to evidence aware issue specifications an agent governed retrieval augmented generation approach.pdf",

    "LIT-004":
        "AdaKG RAG Adaptive KG guided retrieval augmented generation.pdf",
}

resolved_sources = {}

for candidate_id in EXPECTED_CANDIDATE_IDS:

    expected_filename = EXPECTED_SOURCE_FILES[candidate_id]
    source_path = SOURCE_DIR / expected_filename

    print(f"\n{candidate_id}")
    print("  Expected source:")
    print(f"    {expected_filename}")

    if not source_path.exists():
        raise RuntimeError(
            f"\nCELL 13 STOP: Primary source missing for {candidate_id}\n"
            f"Expected:\n  {source_path}"
        )

    resolved_sources[candidate_id] = source_path

    print("  ✓ Source exists")


# ================================================================================================================
# [STEP 11] PRIMARY-SOURCE TITLE ↔ CANDIDATE-ID CONSISTENCY
# ================================================================================================================

print("\n" + "=" * 112)
print("[STEP 11] PRIMARY-SOURCE TITLE ↔ CANDIDATE-ID CONSISTENCY")
print("=" * 112)

# Exact filename-to-ID mapping is the protected identity mechanism.
# No semantic matching or automatic source reassignment is allowed.

for candidate_id in EXPECTED_CANDIDATE_IDS:

    path = resolved_sources[candidate_id]

    print(f"✓ {candidate_id} → {path.name}")


# Explicit critical inversion protection
if resolved_sources["LIT-003"].name != EXPECTED_SOURCE_FILES["LIT-003"]:
    raise RuntimeError(
        "\nCELL 13 STOP: LIT-003 source identity mismatch."
    )

if resolved_sources["LIT-004"].name != EXPECTED_SOURCE_FILES["LIT-004"]:
    raise RuntimeError(
        "\nCELL 13 STOP: LIT-004 source identity mismatch."
    )

print("\n✓ LIT-003 / LIT-004 inversion protection PASS")


# ================================================================================================================
# [STEP 12] PRIMARY-SOURCE SHA-256 VERIFICATION
# ================================================================================================================

print("\n" + "=" * 112)
print("[STEP 12] PRIMARY-SOURCE SHA-256 VERIFICATION")
print("=" * 112)

EXPECTED_SHA256 = {
    "ANCHOR-001":
        "536236facb16659b38b032110c20bc6f3fb9716769cf9e4028ea176f680bbd93",

    "LIT-001":
        "1c13f9889408bdf9e54d1e26e4c7577f016d7c4d30d8c49b314b3189e0a93ff9",

    "LIT-002":
        "210b04fa8f6a1a6d484b271681283a2db058e1e763bf617845218a07c217c08b",

    "LIT-003":
        "48001d7bcef76e9b5e56f75c064d8419962d439522f14e33dcecbd86c1635c83",

    "LIT-004":
        "245410fe0f625d3d0879850f2c84db3bcb23bdf74cc03be65360de289e5dbd4a",
}


def sha256_file(path, chunk_size=1024 * 1024):
    h = hashlib.sha256()

    with open(path, "rb") as f:
        while True:
            chunk = f.read(chunk_size)
            if not chunk:
                break
            h.update(chunk)

    return h.hexdigest()


source_hash_rows = []

for candidate_id in EXPECTED_CANDIDATE_IDS:

    source_path = resolved_sources[candidate_id]

    print(f"\n{candidate_id}")
    print(f"  File: {source_path.name}")

    observed_hash = sha256_file(source_path)
    expected_hash = EXPECTED_SHA256[candidate_id]

    print(f"  Expected SHA-256 : {expected_hash}")
    print(f"  Observed SHA-256 : {observed_hash}")

    if observed_hash != expected_hash:
        raise RuntimeError(
            f"\nCELL 13 STOP: SHA-256 mismatch for {candidate_id}\n"
            f"File:\n  {source_path}\n"
            f"Expected:\n  {expected_hash}\n"
            f"Observed:\n  {observed_hash}"
        )

    source_hash_rows.append({
        "candidate_id": candidate_id,
        "source_filename": source_path.name,
        "sha256": observed_hash,
        "identity_status": "PASS"
    })

    print("  ✓ SHA-256 PASS")


source_hash_df = pd.DataFrame(source_hash_rows)


# ================================================================================================================
# [STEP 13] REGISTRY PRIMARY-SOURCE METADATA QC
# ================================================================================================================

print("\n" + "=" * 112)
print("[STEP 13] REGISTRY PRIMARY-SOURCE METADATA QC")
print("=" * 112)

print("""
Interpretation rule:
  registry.primary_source = source / publication provenance metadata

  registry.primary_source is NOT required to equal the local PDF filename.

  Primary-source file identity is already protected independently by:
    Candidate ID
      +
    Exact Primary-Source Map
      +
    SHA-256 verification

Therefore:
  No filename-equality assertion is applied to registry.primary_source.
""")

if "primary_source" not in registry.columns:
    raise RuntimeError(
        "\nCELL 13 STOP: Required registry column 'primary_source' is missing."
    )

registry_primary_source_values = []

for candidate_id in EXPECTED_CANDIDATE_IDS:

    row = registry.loc[
        registry["candidate_id"].astype(str) == candidate_id
    ]

    if len(row) != 1:
        raise RuntimeError(
            f"\nCELL 13 STOP: Expected exactly one registry row for {candidate_id}."
        )

    primary_source_value = str(
        row.iloc[0]["primary_source"]
    ).strip()

    if (
        primary_source_value == ""
        or primary_source_value.lower() in {"nan", "none", "null"}
    ):
        raise RuntimeError(
            f"\nCELL 13 STOP: Empty primary_source metadata for {candidate_id}."
        )

    registry_primary_source_values.append({
        "candidate_id": candidate_id,
        "registry_primary_source": primary_source_value,
        "metadata_status": "PASS"
    })

    print(
        f"✓ {candidate_id} → registry.primary_source = "
        f"{primary_source_value}"
    )


registry_primary_source_df = pd.DataFrame(
    registry_primary_source_values
)

print("\n✓ Registry primary-source metadata present and non-empty")
print("✓ No incorrect filename-equality assumption applied")
print("✓ Primary-source identity remains protected by exact map + SHA-256")


# ================================================================================================================
# [STEP 14] MANIFEST VERIFICATION
# ================================================================================================================

print("\n" + "=" * 112)
print("[STEP 14] MANIFEST VERIFICATION")
print("=" * 112)

if not REGISTRY_MANIFEST.exists():

    print("ℹ V003 manifest not found at expected path.")
    print(
        "✓ Continuing because the registry and all five primary-source "
        "SHA checks passed."
    )

else:

    with open(REGISTRY_MANIFEST, "r", encoding="utf-8") as f:
        manifest = json.load(f)

    print("✓ V003 manifest loaded")

    manifest_sha_candidates = []

    for key in [
        "parquet_sha256",
        "registry_parquet_sha256",
        "sha256",
    ]:
        if key in manifest:
            manifest_sha_candidates.append(
                str(manifest[key]).strip()
            )

    actual_registry_sha = sha256_file(REGISTRY_PARQUET)

    if manifest_sha_candidates:

        if actual_registry_sha not in manifest_sha_candidates:
            raise RuntimeError(
                "\nCELL 13 STOP: V003 registry SHA-256 does not match manifest.\n"
                f"Observed:\n  {actual_registry_sha}\n"
                f"Manifest candidates:\n  {manifest_sha_candidates}"
            )

        print("✓ V003 registry SHA-256 matches manifest")

    else:

        print(
            "ℹ Manifest loaded but no registry SHA field was available "
            "for comparison."
        )


# ================================================================================================================
# [STEP 15] FINAL CANONICAL IDENTITY ASSERTION
# ================================================================================================================

print("\n" + "=" * 112)
print("[STEP 15] FINAL CANONICAL IDENTITY ASSERTION")
print("=" * 112)

FINAL_CANONICAL_MAP = {
    "ANCHOR-001":
        "The Cross-Lingual Cost: Retrieval Biases in RAG over Arabic-English Corpora",

    "LIT-001":
        "SURE-RAG: Sufficiency and Uncertainty-Aware Evidence Verification for Selective Retrieval-Augmented Generation",

    "LIT-002":
        "Do LLMs Know When Evidence is Insufficient? An Evidence Sufficiency Benchmark for Answer-Abstention Calibration in Retrieval-Augmented Generation",

    "LIT-003":
        "From noisy feedback to evidence-aware issue specifications: an agent-governed retrieval-augmented generation approach",

    "LIT-004":
        "AdaKG-RAG: Adaptive KG-guided retrieval-augmented generation with hypothesis-and-verification retrieval for multi-hop question answering",
}

for candidate_id, expected_title in FINAL_CANONICAL_MAP.items():

    observed_title = str(
        registry.loc[
            registry["candidate_id"].astype(str) == candidate_id,
            "title"
        ].iloc[0]
    ).strip()

    assert observed_title == expected_title

    print(f"✓ {candidate_id}")
    print(f"    {observed_title}")

print("\n✓ FINAL CANONICAL IDENTITY ASSERTION PASS")


# ================================================================================================================
# [STEP 16] READ-ONLY ACTIVATION
# ================================================================================================================

print("\n" + "=" * 112)
print("[STEP 16] READ-ONLY ACTIVATION")
print("=" * 112)

# Activate the verified V003 registry in RAM.
# No source file, dataset, cohort, adjudication, or annotation is modified.

CANONICAL_REGISTRY = registry.copy(deep=True)
CANONICAL_REGISTRY_VERSION = "V003"
CANONICAL_REGISTRY_PATH = str(REGISTRY_PARQUET)

print(f"Active registry version : {CANONICAL_REGISTRY_VERSION}")
print(f"Active registry path    : {CANONICAL_REGISTRY_PATH}")
print("✓ V003 activated in RAM")
print("✓ READ-ONLY")
print("✓ No dataset modified")
print("✓ No adjudication modified")
print("✓ No Gold modified")


# ================================================================================================================
# [STEP 17] FINAL SOURCE IDENTITY TABLE
# ================================================================================================================

print("\n" + "=" * 112)
print("[STEP 17] FINAL SOURCE IDENTITY TABLE")
print("=" * 112)

final_identity_table = []

for candidate_id in EXPECTED_CANDIDATE_IDS:

    registry_row = registry.loc[
        registry["candidate_id"].astype(str) == candidate_id
    ].iloc[0]

    source_row = source_hash_df.loc[
        source_hash_df["candidate_id"] == candidate_id
    ].iloc[0]

    metadata_row = registry_primary_source_df.loc[
        registry_primary_source_df["candidate_id"] == candidate_id
    ].iloc[0]

    final_identity_table.append({
        "candidate_id": candidate_id,
        "registry_title": str(registry_row["title"]).strip(),
        "registry_primary_source": metadata_row["registry_primary_source"],
        "primary_source_file": source_row["source_filename"],
        "sha256": source_row["sha256"],
        "identity_status": "PASS"
    })

final_identity_df = pd.DataFrame(final_identity_table)

print(final_identity_df.to_string(index=False))


# ================================================================================================================
# [STEP 18] FINAL SAFETY GATE
# ================================================================================================================

print("\n" + "=" * 112)
print("[STEP 18] FINAL SAFETY GATE")
print("=" * 112)

SAFETY_CHECKS = {
    "canonical_registry_version": CANONICAL_REGISTRY_VERSION == "V003",

    "registry_rows": len(CANONICAL_REGISTRY) == 5,

    "candidate_ids_unique": (
        CANONICAL_REGISTRY["candidate_id"].astype(str).nunique() == 5
    ),

    "titles_exact": True,

    "primary_sources_resolved": len(resolved_sources) == 5,

    "sha256_verified": len(source_hash_df) == 5,

    "primary_source_metadata_present": (
        len(registry_primary_source_df) == 5
    ),

    "read_only": CELL_STATUS == "READ_ONLY",
}

for name, result in SAFETY_CHECKS.items():
    print(f"{name:<42}: {'PASS' if result else 'FAIL'}")

if not all(SAFETY_CHECKS.values()):
    raise RuntimeError(
        "\nCELL 13 STOP: Final safety gate failed."
    )

print("\n✓ ALL FINAL SAFETY CHECKS PASS")


# ================================================================================================================
# [STEP 19] FINAL STATUS
# ================================================================================================================

print("\n" + "=" * 112)
print("[STEP 19] FINAL STATUS")
print("=" * 112)

FINAL_STATUS = "PRIMARY_SOURCE_IDENTITY_VERIFIED_V003"

print(f"""
Experiment ID:
  {EXPERIMENT_ID}

Cell:
  {CELL_ID}

Final Status:
  {FINAL_STATUS}

Canonical Registry:
  V003

Registry:
  {REGISTRY_PARQUET}

Primary Sources:
  5 / 5 verified

SHA-256:
  5 / 5 verified

Registry primary_source metadata:
  5 / 5 present

LIT-003:
  AGR — correctly mapped

LIT-004:
  AdaKG-RAG — correctly mapped

Dataset modification:
  NONE

199 Cohort modification:
  NONE

400 Validation modification:
  NONE

Q82 modification:
  NONE

Adjudication:
  NONE

Gold:
  NONE

Model inference/training:
  NONE

Prompt revision:
  NONE

Architecture modification:
  NONE
""")

print("=" * 112)
print("✓ CELL 13 v004 COMPLETE — PRIMARY SOURCE IDENTITY VERIFIED")
print("=" * 112)

print("""
NEXT:
  Proceed to the next bounded novelty-gap audit cell.

IMPORTANT:
  • Do NOT revert to LITERATURE_REGISTRY_V002.
  • CELL 13 v005 remains preserved as the canonical restoration/recovery record.
  • No new adjudication or dataset modification is authorized.
  • Q1-Q10 extraction may proceed only from the verified primary-source identities.
""")
```

```python
# ================================================================================================================
# X-AutoAlign Legal RAG
# F-D+R — BOUNDED LITERATURE / NOVELTY GAP AUDIT
#
# CELL 13 v005 — CANONICAL LITERATURE REGISTRY RESTORATION
#
# PURPOSE:
#   Controlled restoration of the canonical literature registry after Cell 13 v004
#   detected that the persistent LITERATURE_REGISTRY_V002 checkpoint contains
#   a LIT-003 / LIT-004 attribution inversion.
#
# IMPORTANT:
#   This is NOT a literature expansion step.
#   This is NOT a novelty judgment.
#   This is NOT a content review.
#   This is NOT a dataset modification.
#
#   The correction is restricted to the literature-audit registry checkpoint.
#
# CANONICAL ORDER:
#   ANCHOR-001 → Cross-Lingual Cost
#   LIT-001    → SURE-RAG
#   LIT-002    → Evidence Sufficiency Benchmark
#   LIT-003    → AGR
#   LIT-004    → AdaKG-RAG
#
# NEXT:
#   After PASS:
#       Cell 13 v004
#           ↓
#       Cell 14
#
# 13.1 / 13.2:
#   CANCELLED
# ================================================================================================================


import os
import json
import hashlib
import pandas as pd


print("=" * 112)
print("X-AutoAlign Legal RAG")
print("F-D+R — BOUNDED LITERATURE / NOVELTY GAP AUDIT")
print("CELL 13 v005 — CANONICAL LITERATURE REGISTRY RESTORATION")
print("=" * 112)


# ================================================================================================================
# STEP 0 — SCIENTIFIC VALUE GATE
# ================================================================================================================

print("\n" + "=" * 112)
print("[STEP 0] SCIENTIFIC VALUE GATE")
print("=" * 112)

print("Question:")
print("  ما فائدتها للرسالة العلمية أو لبناء الـdataset؟")
print("")
print("Answer:")
print("  هذه الخلية ضرورية لأن Cell 13 v004 أثبتت أن الـpersistent")
print("  LITERATURE_REGISTRY_V002 يحتوي على خطأ في attribution بين")
print("  LIT-003 و LIT-004.")
print("")
print("  استمرار هذا الخطأ إلى Cell 14 قد يؤدي إلى إسناد Q1-Q10")
print("  methodological evidence إلى البحث الخطأ.")
print("")
print("  لذلك تقوم هذه الخلية بإعادة بناء الـcanonical literature registry")
print("  من السجل الموثق سابقًا، ثم تتحقق من الهوية قبل تحديث الـcheckpoint.")
print("")
print("Scientific value:")
print("  • correct literature traceability")
print("  • correct Q1-Q10 attribution")
print("  • reproducibility")
print("  • audit integrity")
print("  • prevention of candidate/source misattribution")
print("")
print("Decision:")
print("  PROCEED ONLY WITH CONTROLLED CANONICAL REGISTRY RESTORATION")


# ================================================================================================================
# STEP 1 — EXPERIMENT / CELL ID
# ================================================================================================================

print("\n" + "=" * 112)
print("[STEP 1] EXPERIMENT / CELL ID")
print("=" * 112)

EXPERIMENT_ID = "FDR-NOVELTY-GAP-001"
CELL_ID = "CELL13_CANONICAL_REGISTRY_RESTORATION_V005"

print("Experiment ID :", EXPERIMENT_ID)
print("Cell ID       :", CELL_ID)
print("Status        : CONTROLLED_READ_ONLY_REPAIR")


# ================================================================================================================
# STEP 2 — FIREWALL
# ================================================================================================================

print("\n" + "=" * 112)
print("[STEP 2] SCIENTIFIC FIREWALL")
print("=" * 112)

FIREWALL = {
    "dataset_modification": "BLOCKED",
    "199_cohort_modification": "BLOCKED",
    "400_validation_modification": "BLOCKED",
    "Q82_modification": "BLOCKED",
    "new_adjudication": "BLOCKED",
    "gold_generation": "BLOCKED",
    "y_positive_generation": "BLOCKED",
    "y_negative_generation": "BLOCKED",
    "preference_generation": "BLOCKED",
    "model_training": "BLOCKED",
    "model_inference": "BLOCKED",
    "prompt_revision": "BLOCKED",
    "architecture_modification": "BLOCKED",
    "new_legal_extraction": "BLOCKED",
    "validation_reopen": "BLOCKED",
    "automatic_novelty_judgment": "BLOCKED",
    "automatic_overlap_judgment": "BLOCKED",
}

for key, value in FIREWALL.items():
    print(f"{key:<42}: {value}")

assert all(value == "BLOCKED" for value in FIREWALL.values())

print("✓ Firewall PASS")


# ================================================================================================================
# STEP 3 — PATHS
# ================================================================================================================

print("\n" + "=" * 112)
print("[STEP 3] AUDIT PATHS")
print("=" * 112)

CHECKPOINT_DIR = (
    "/content/drive/MyDrive/"
    "X-AutoAlign_Legal_RAG/"
    "04_AUDIT/"
    "NOVELTY_GAP_AUDIT/"
    "CHECKPOINTS"
)

OLD_PARQUET = os.path.join(
    CHECKPOINT_DIR,
    "LITERATURE_REGISTRY_V002.parquet"
)

OLD_CSV = os.path.join(
    CHECKPOINT_DIR,
    "LITERATURE_REGISTRY_V002.csv"
)

OLD_MANIFEST = os.path.join(
    CHECKPOINT_DIR,
    "LITERATURE_REGISTRY_V002_MANIFEST.json"
)

CORRECTED_PARQUET = os.path.join(
    CHECKPOINT_DIR,
    "LITERATURE_REGISTRY_V003.parquet"
)

CORRECTED_CSV = os.path.join(
    CHECKPOINT_DIR,
    "LITERATURE_REGISTRY_V003.csv"
)

CORRECTED_MANIFEST = os.path.join(
    CHECKPOINT_DIR,
    "LITERATURE_REGISTRY_V003_MANIFEST.json"
)

SOURCE_DIR = (
    "/content/drive/MyDrive/"
    "F-D+R/"
    "DATASET_CONSTRUCTION/"
    "SOURCES"
)

print("Checkpoint directory:")
print(f"  {CHECKPOINT_DIR}")
print("")
print("Existing registry:")
print(f"  {OLD_PARQUET}")
print("")
print("Corrected registry:")
print(f"  {CORRECTED_PARQUET}")


# ================================================================================================================
# STEP 4 — DIRECTORY GATE
# ================================================================================================================

print("\n" + "=" * 112)
print("[STEP 4] DIRECTORY GATE")
print("=" * 112)

if not os.path.isdir(CHECKPOINT_DIR):
    raise RuntimeError(
        "\nCELL 13 v005 STOP: Checkpoint directory does not exist.\n"
        f"{CHECKPOINT_DIR}"
    )

if not os.path.isdir(SOURCE_DIR):
    raise RuntimeError(
        "\nCELL 13 v005 STOP: Primary-source directory does not exist.\n"
        f"{SOURCE_DIR}"
    )

print("✓ Checkpoint directory exists")
print("✓ Primary-source directory exists")


# ================================================================================================================
# STEP 5 — PRESERVE THE PREVIOUS CHECKPOINT
# ================================================================================================================

print("\n" + "=" * 112)
print("[STEP 5] PREVIOUS CHECKPOINT PRESERVATION")
print("=" * 112)

if os.path.isfile(OLD_PARQUET):
    print("✓ Existing V002 Parquet found")
else:
    raise RuntimeError(
        "\nCELL 13 v005 STOP: Expected previous V002 checkpoint is missing."
    )

if os.path.isfile(OLD_CSV):
    print("✓ Existing V002 CSV found")
else:
    print("⚠ V002 CSV not found — Parquet remains the required audit artifact")

if os.path.isfile(OLD_MANIFEST):
    print("✓ Existing V002 manifest found")
else:
    print("⚠ V002 manifest not found — proceeding only with Parquet verification")


# ================================================================================================================
# STEP 6 — READ PREVIOUS CHECKPOINT FOR DIAGNOSTIC CONFIRMATION
# ================================================================================================================

print("\n" + "=" * 112)
print("[STEP 6] PREVIOUS CHECKPOINT DIAGNOSTIC CONFIRMATION")
print("=" * 112)

previous_registry = pd.read_parquet(OLD_PARQUET)

print("Previous V002 rows :", len(previous_registry))
print("Previous V002 cols :", len(previous_registry.columns))

if len(previous_registry) != 5:
    raise RuntimeError(
        "\nCELL 13 v005 STOP: Previous V002 registry does not contain exactly 5 records."
    )

if "candidate_id" not in previous_registry.columns:
    raise RuntimeError(
        "\nCELL 13 v005 STOP: candidate_id missing from previous V002 registry."
    )

if "title" not in previous_registry.columns:
    raise RuntimeError(
        "\nCELL 13 v005 STOP: title missing from previous V002 registry."
    )

print("")
print("Previous V002 Candidate → Title state:")

for _, row in previous_registry.iterrows():
    print(
        f"  {str(row['candidate_id']):<12} → "
        f"{str(row['title']).strip()}"
    )

print("")
print("✓ Previous checkpoint loaded for controlled diagnosis")
print("✓ No modification performed")


# ================================================================================================================
# STEP 7 — CANONICAL REGISTRY SPECIFICATION
# ================================================================================================================

print("\n" + "=" * 112)
print("[STEP 7] CANONICAL REGISTRY SPECIFICATION")
print("=" * 112)

CANONICAL_TITLES = {
    "ANCHOR-001":
        "The Cross-Lingual Cost: Retrieval Biases in RAG over Arabic-English Corpora",

    "LIT-001":
        "SURE-RAG: Sufficiency and Uncertainty-Aware Evidence Verification for Selective Retrieval-Augmented Generation",

    "LIT-002":
        "Do LLMs Know When Evidence is Insufficient? An Evidence Sufficiency Benchmark for Answer-Abstention Calibration in Retrieval-Augmented Generation",

    "LIT-003":
        "From noisy feedback to evidence-aware issue specifications: an agent-governed retrieval-augmented generation approach",

    "LIT-004":
        "AdaKG-RAG: Adaptive KG-guided retrieval-augmented generation with hypothesis-and-verification retrieval for multi-hop question answering",
}

EXPECTED_IDS = [
    "ANCHOR-001",
    "LIT-001",
    "LIT-002",
    "LIT-003",
    "LIT-004",
]

print("Canonical Candidate → Title specification:")

for candidate_id in EXPECTED_IDS:
    print(
        f"  {candidate_id:<12} → {CANONICAL_TITLES[candidate_id]}"
    )

print("")
print("✓ Canonical specification loaded")


# ================================================================================================================
# STEP 8 — EXPLICIT INVERSION DETECTION
# ================================================================================================================

print("\n" + "=" * 112)
print("[STEP 8] LIT-003 / LIT-004 INVERSION DIAGNOSTIC")
print("=" * 112)

previous_titles = {
    str(row["candidate_id"]): str(row["title"]).strip()
    for _, row in previous_registry.iterrows()
}

lit003_previous = previous_titles.get("LIT-003")
lit004_previous = previous_titles.get("LIT-004")

print("Previous V002:")
print(f"  LIT-003 → {lit003_previous}")
print(f"  LIT-004 → {lit004_previous}")
print("")
print("Expected canonical:")
print(f"  LIT-003 → {CANONICAL_TITLES['LIT-003']}")
print(f"  LIT-004 → {CANONICAL_TITLES['LIT-004']}")

if (
    lit003_previous == CANONICAL_TITLES["LIT-004"]
    and
    lit004_previous == CANONICAL_TITLES["LIT-003"]
):
    print("")
    print("✓ Confirmed V002 LIT-003 / LIT-004 inversion")
else:
    print("")
    print("⚠ V002 is not exactly the known inversion state")
    print("The restoration will still proceed only if the canonical specification")
    print("can be reconstructed without ambiguity.")


# ================================================================================================================
# STEP 9 — CONTROLLED CANONICAL RESTORATION
# ================================================================================================================

print("\n" + "=" * 112)
print("[STEP 9] CONTROLLED CANONICAL RESTORATION")
print("=" * 112)

# IMPORTANT:
# We do NOT modify individual V002 rows in place.
# We reconstruct the canonical registry from the previously established
# 18-field registry specification and canonical Candidate-ID/title mapping.
#
# Metadata for all non-identity fields is retained from the corresponding
# Candidate ID in V002.
#
# Only the Candidate-ID ↔ Title identity mapping is corrected.

required_columns = [
    "candidate_id",
    "search_family",
    "search_layer",
    "search_query",
    "title",
    "authors",
    "year",
    "venue",
    "primary_source",
    "doi_or_identifier",
    "url",
    "retrieval_date",
    "initial_relevance",
    "possible_competitor",
    "duplicate_group",
    "version_status",
    "discovery_reason",
    "notes",
]

missing_columns = [
    col for col in required_columns
    if col not in previous_registry.columns
]

if missing_columns:
    raise RuntimeError(
        "\nCELL 13 v005 STOP: V002 registry schema mismatch.\n"
        f"Missing columns:\n  {missing_columns}"
    )

restored_rows = []

for candidate_id in EXPECTED_IDS:

    matching = previous_registry[
        previous_registry["candidate_id"].astype(str) == candidate_id
    ]

    if len(matching) != 1:
        raise RuntimeError(
            f"\nCELL 13 v005 STOP: Expected exactly one V002 record for {candidate_id}."
        )

    row = matching.iloc[0].copy()

    # Controlled identity restoration.
    row["title"] = CANONICAL_TITLES[candidate_id]

    restored_rows.append(row)


canonical_registry = pd.DataFrame(restored_rows)

canonical_registry = canonical_registry[required_columns].copy()

print("✓ Canonical registry reconstructed")
print("✓ Exactly 5 records reconstructed")
print("✓ Non-identity registry metadata retained by Candidate ID")
print("✓ Title identity restored from canonical specification")


# ================================================================================================================
# STEP 10 — RESTORED REGISTRY STRUCTURAL QC
# ================================================================================================================

print("\n" + "=" * 112)
print("[STEP 10] RESTORED REGISTRY STRUCTURAL QC")
print("=" * 112)

if len(canonical_registry) != 5:
    raise RuntimeError(
        "\nCELL 13 v005 STOP: Restored registry row count != 5."
    )

observed_ids = canonical_registry["candidate_id"].astype(str).tolist()

if observed_ids != EXPECTED_IDS:
    raise RuntimeError(
        "\nCELL 13 v005 STOP: Restored Candidate-ID ordering mismatch.\n"
        f"Expected:\n  {EXPECTED_IDS}\n"
        f"Observed:\n  {observed_ids}"
    )

if canonical_registry["candidate_id"].duplicated().any():
    raise RuntimeError(
        "\nCELL 13 v005 STOP: Duplicate Candidate IDs after restoration."
    )

print("✓ Row count = 5")
print("✓ Candidate-ID order exact")
print("✓ Candidate IDs unique")


# ================================================================================================================
# STEP 11 — RESTORED TITLE QC
# ================================================================================================================

print("\n" + "=" * 112)
print("[STEP 11] RESTORED TITLE QC")
print("=" * 112)

for candidate_id in EXPECTED_IDS:

    observed_title = str(
        canonical_registry.loc[
            canonical_registry["candidate_id"].astype(str) == candidate_id,
            "title"
        ].iloc[0]
    ).strip()

    expected_title = CANONICAL_TITLES[candidate_id]

    if observed_title != expected_title:
        raise RuntimeError(
            f"\nCELL 13 v005 STOP: Restored title mismatch for {candidate_id}\n"
            f"Observed:\n  {observed_title}\n"
            f"Expected:\n  {expected_title}"
        )

    print(f"✓ {candidate_id} title exact")


# ================================================================================================================
# STEP 12 — EXPLICIT ORDER PROTECTION
# ================================================================================================================

print("\n" + "=" * 112)
print("[STEP 12] EXPLICIT LIT-003 / LIT-004 ORDER PROTECTION")
print("=" * 112)

lit003_title = canonical_registry.loc[
    canonical_registry["candidate_id"].astype(str) == "LIT-003",
    "title"
].iloc[0]

lit004_title = canonical_registry.loc[
    canonical_registry["candidate_id"].astype(str) == "LIT-004",
    "title"
].iloc[0]

assert lit003_title == CANONICAL_TITLES["LIT-003"]
assert lit004_title == CANONICAL_TITLES["LIT-004"]

print("✓ LIT-003 = From noisy feedback... [AGR]")
print("✓ LIT-004 = AdaKG-RAG")
print("✓ Inversion eliminated")


# ================================================================================================================
# STEP 13 — PRIMARY-SOURCE MAP
# ================================================================================================================

print("\n" + "=" * 112)
print("[STEP 13] PRIMARY-SOURCE MAP")
print("=" * 112)

SOURCE_MAP = {
    "ANCHOR-001":
        "The Cross Lingual Cost Retrieval Biases in RAG over Arabic English Corpora.pdf",

    "LIT-001":
        "SURE RAG Sufficiency and Uncertainty-Aware.pdf",

    "LIT-002":
        "Do LLMs Know When Evidence is Insufficient An Evidence Sufficiency.pdf",

    "LIT-003":
        "From noisy feedback to evidence aware issue specifications an agent governed retrieval augmented generation approach.pdf",

    "LIT-004":
        "AdaKG RAG Adaptive KG guided retrieval augmented generation.pdf",
}

for candidate_id, filename in SOURCE_MAP.items():
    path = os.path.join(SOURCE_DIR, filename)

    if not os.path.isfile(path):
        raise RuntimeError(
            f"\nCELL 13 v005 STOP: Primary source missing for {candidate_id}\n"
            f"{path}"
        )

    if os.path.getsize(path) <= 0:
        raise RuntimeError(
            f"\nCELL 13 v005 STOP: Primary source is empty for {candidate_id}\n"
            f"{path}"
        )

    print(f"✓ {candidate_id:<12} → {filename}")


# ================================================================================================================
# STEP 14 — REGISTRY ↔ PRIMARY-SOURCE IDENTITY CHECK
# ================================================================================================================

print("\n" + "=" * 112)
print("[STEP 14] REGISTRY ↔ PRIMARY-SOURCE IDENTITY CHECK")
print("=" * 112)

# File names are intentionally explicit.
# No fuzzy matching is used.

IDENTITY_CHECKS = {

    "ANCHOR-001": {
        "title": CANONICAL_TITLES["ANCHOR-001"],
        "filename": SOURCE_MAP["ANCHOR-001"],
    },

    "LIT-001": {
        "title": CANONICAL_TITLES["LIT-001"],
        "filename": SOURCE_MAP["LIT-001"],
    },

    "LIT-002": {
        "title": CANONICAL_TITLES["LIT-002"],
        "filename": SOURCE_MAP["LIT-002"],
    },

    "LIT-003": {
        "title": CANONICAL_TITLES["LIT-003"],
        "filename": SOURCE_MAP["LIT-003"],
    },

    "LIT-004": {
        "title": CANONICAL_TITLES["LIT-004"],
        "filename": SOURCE_MAP["LIT-004"],
    },
}

for candidate_id in EXPECTED_IDS:

    registry_title = str(
        canonical_registry.loc[
            canonical_registry["candidate_id"].astype(str) == candidate_id,
            "title"
        ].iloc[0]
    ).strip()

    expected_title = IDENTITY_CHECKS[candidate_id]["title"]

    if registry_title != expected_title:
        raise RuntimeError(
            f"\nCELL 13 v005 STOP: Registry identity failure for {candidate_id}"
        )

    print(f"✓ {candidate_id:<12} registry identity PASS")


# ================================================================================================================
# STEP 15 — WRITE CORRECTED CHECKPOINT
# ================================================================================================================

print("\n" + "=" * 112)
print("[STEP 15] WRITE CORRECTED CHECKPOINT")
print("=" * 112)

canonical_registry.to_parquet(
    CORRECTED_PARQUET,
    index=False
)

canonical_registry.to_csv(
    CORRECTED_CSV,
    index=False
)

print(f"✓ Wrote:")
print(f"  {CORRECTED_PARQUET}")
print(f"  {CORRECTED_CSV}")


# ================================================================================================================
# STEP 16 — HASH CORRECTED CHECKPOINT
# ================================================================================================================

print("\n" + "=" * 112)
print("[STEP 16] CHECKPOINT HASH")
print("=" * 112)


def sha256_file(path, chunk_size=1024 * 1024):

    h = hashlib.sha256()

    with open(path, "rb") as f:
        while True:

            chunk = f.read(chunk_size)

            if not chunk:
                break

            h.update(chunk)

    return h.hexdigest()


corrected_parquet_sha256 = sha256_file(CORRECTED_PARQUET)
corrected_csv_sha256 = sha256_file(CORRECTED_CSV)

print("Corrected Parquet SHA-256:")
print(f"  {corrected_parquet_sha256}")

print("")
print("Corrected CSV SHA-256:")
print(f"  {corrected_csv_sha256}")


# ================================================================================================================
# STEP 17 — MANIFEST
# ================================================================================================================

print("\n" + "=" * 112)
print("[STEP 17] CORRECTED CHECKPOINT MANIFEST")
print("=" * 112)

manifest = {
    "experiment_id": EXPERIMENT_ID,
    "cell_id": CELL_ID,

    "registry_version": "V003",

    "source_checkpoint": "LITERATURE_REGISTRY_V002.parquet",

    "source_checkpoint_status":
        "FOUND_WITH_LIT003_LIT004_INVERSION",

    "restoration_reason":
        "Controlled restoration of canonical Candidate-ID/title attribution",

    "canonical_candidate_ids": EXPECTED_IDS,

    "canonical_titles": CANONICAL_TITLES,

    "lit003_identity":
        "From noisy feedback to evidence-aware issue specifications: an agent-governed retrieval-augmented generation approach",

    "lit004_identity":
        "AdaKG-RAG: Adaptive KG-guided retrieval-augmented generation with hypothesis-and-verification retrieval for multi-hop question answering",

    "primary_source_map": SOURCE_MAP,

    "corrected_parquet":
        CORRECTED_PARQUET,

    "corrected_csv":
        CORRECTED_CSV,

    "corrected_parquet_sha256":
        corrected_parquet_sha256,

    "corrected_csv_sha256":
        corrected_csv_sha256,

    "scientific_scope":
        "Literature audit registry identity only",

    "dataset_modification":
        "BLOCKED",

    "content_judgment":
        "NOT_PERFORMED",

    "novelty_judgment":
        "NOT_PERFORMED",

    "overlap_judgment":
        "NOT_PERFORMED",

    "q1_q10_evidence_extraction":
        "NOT_PERFORMED",

    "status":
        "CANONICAL_REGISTRY_RESTORED_V003",
}

with open(CORRECTED_MANIFEST, "w", encoding="utf-8") as f:
    json.dump(
        manifest,
        f,
        ensure_ascii=False,
        indent=2
    )

print(f"✓ Manifest written:")
print(f"  {CORRECTED_MANIFEST}")


# ================================================================================================================
# STEP 18 — INDEPENDENT RELOAD
# ================================================================================================================

print("\n" + "=" * 112)
print("[STEP 18] INDEPENDENT RELOAD VERIFICATION")
print("=" * 112)

reloaded_registry = pd.read_parquet(CORRECTED_PARQUET)

if len(reloaded_registry) != 5:
    raise RuntimeError(
        "\nCELL 13 v005 STOP: Reloaded corrected registry row count != 5."
    )

if reloaded_registry["candidate_id"].duplicated().any():
    raise RuntimeError(
        "\nCELL 13 v005 STOP: Duplicate Candidate IDs after reload."
    )

for candidate_id in EXPECTED_IDS:

    row = reloaded_registry[
        reloaded_registry["candidate_id"].astype(str) == candidate_id
    ]

    if len(row) != 1:
        raise RuntimeError(
            f"\nCELL 13 v005 STOP: Reload failed for {candidate_id}"
        )

    observed_title = str(row.iloc[0]["title"]).strip()

    if observed_title != CANONICAL_TITLES[candidate_id]:
        raise RuntimeError(
            f"\nCELL 13 v005 STOP: Reloaded title mismatch for {candidate_id}\n"
            f"Observed:\n  {observed_title}\n"
            f"Expected:\n  {CANONICAL_TITLES[candidate_id]}"
        )

    print(f"✓ {candidate_id:<12} reload identity PASS")


# ================================================================================================================
# STEP 19 — MANIFEST RELOAD
# ================================================================================================================

print("\n" + "=" * 112)
print("[STEP 19] MANIFEST RELOAD VERIFICATION")
print("=" * 112)

with open(CORRECTED_MANIFEST, "r", encoding="utf-8") as f:
    reloaded_manifest = json.load(f)

if reloaded_manifest["registry_version"] != "V003":
    raise RuntimeError(
        "\nCELL 13 v005 STOP: Manifest registry version mismatch."
    )

if reloaded_manifest["corrected_parquet_sha256"] != corrected_parquet_sha256:
    raise RuntimeError(
        "\nCELL 13 v005 STOP: Manifest Parquet hash mismatch."
    )

print("✓ Manifest reload PASS")
print("✓ Manifest hash consistency PASS")


# ================================================================================================================
# STEP 20 — ACTIVATE CORRECTED REGISTRY IN RAM
# ================================================================================================================

print("\n" + "=" * 112)
print("[STEP 20] ACTIVATE CORRECTED REGISTRY")
print("=" * 112)

literature_registry = reloaded_registry.copy(deep=True)

print("✓ literature_registry now points to independently verified V003")
print("✓ RAM registry is canonical")
print("✓ No dataset object modified")


# ================================================================================================================
# STEP 21 — FINAL CANONICAL STATE
# ================================================================================================================

print("\n" + "=" * 112)
print("[STEP 21] FINAL CANONICAL STATE")
print("=" * 112)

print("Candidate ID → Canonical Title:")
print("")

for candidate_id in EXPECTED_IDS:

    row = literature_registry[
        literature_registry["candidate_id"].astype(str) == candidate_id
    ].iloc[0]

    print(
        f"{candidate_id:<12} → {str(row['title']).strip()}"
    )


# ================================================================================================================
# STEP 22 — FINAL SAFETY ASSERTIONS
# ================================================================================================================

print("\n" + "=" * 112)
print("[STEP 22] FINAL SAFETY ASSERTIONS")
print("=" * 112)

assert len(literature_registry) == 5

assert literature_registry[
    "candidate_id"
].astype(str).tolist() == EXPECTED_IDS

assert literature_registry[
    literature_registry["candidate_id"].astype(str) == "LIT-003"
]["title"].iloc[0] == CANONICAL_TITLES["LIT-003"]

assert literature_registry[
    literature_registry["candidate_id"].astype(str) == "LIT-004"
]["title"].iloc[0] == CANONICAL_TITLES["LIT-004"]

assert os.path.isfile(CORRECTED_PARQUET)
assert os.path.isfile(CORRECTED_CSV)
assert os.path.isfile(CORRECTED_MANIFEST)

print("✓ 5 records = PASS")
print("✓ Candidate IDs = PASS")
print("✓ Canonical titles = PASS")
print("✓ LIT-003 / LIT-004 ordering = PASS")
print("✓ Corrected Parquet persistence = PASS")
print("✓ Corrected CSV persistence = PASS")
print("✓ Manifest persistence = PASS")
print("✓ Independent reload = PASS")
print("✓ Dataset firewall = PASS")
print("✓ Literature-content firewall = PASS")


# ================================================================================================================
# STEP 23 — FINAL DECISION
# ================================================================================================================

print("\n" + "=" * 112)
print("[STEP 23] FINAL DECISION")
print("=" * 112)

FINAL_STATUS = "CANONICAL_REGISTRY_RESTORED_V003"

print("STATUS :", FINAL_STATUS)
print("")
print("Canonical registry:")
print("  ANCHOR-001 → Cross-Lingual Cost")
print("  LIT-001    → SURE-RAG")
print("  LIT-002    → Evidence Sufficiency Benchmark")
print("  LIT-003    → AGR")
print("  LIT-004    → AdaKG-RAG")
print("")
print("✓ V002 inversion isolated")
print("✓ Canonical registry reconstructed")
print("✓ Corrected checkpoint persisted")
print("✓ Independent reload verified")
print("✓ LIT-003 / LIT-004 attribution protected")
print("")
print("NO:")
print("  • new literature")
print("  • Q1-Q10 extraction")
print("  • novelty judgment")
print("  • overlap judgment")
print("  • dataset modification")
print("  • model inference")
print("  • architecture modification")
print("")
print("NEXT:")
print("  CELL 13 v004 — PRIMARY-SOURCE IDENTITY VERIFICATION")
print("  using the independently verified canonical V003 registry.")
print("")
print("After Cell 13 v004 PASS:")
print("  CELL 14 — SOURCE-GROUNDED Q1-Q10 METHODOLOGICAL EVIDENCE EXTRACTION")


print("\n" + "=" * 112)
print("CELL 13 v005 COMPLETE")
print("=" * 112)
```

```python
# ================================================================================================================
# CELL 13A V002 — CANONICAL ARTIFACT & PATH LOCK GATE
# ================================================================================================================
#
# Experiment : FDR-NOVELTY-GAP-001
#
# Purpose:
#   Establish ONE authoritative downstream artifact/path lock BEFORE rebuilding Cell 14.
#
# V002 DESIGN CORRECTION:
#   The canonical LITERATURE_REGISTRY_V003.parquet contains 18 columns and does NOT contain
#   primary_source_filename.
#
#   Therefore:
#       - DO NOT infer filename from registry.
#       - DO NOT modify registry.
#       - DO NOT use fuzzy matching.
#       - DO NOT search for alternative artifacts.
#
#   Instead, V002 consumes:
#       1. Canonical Registry V003
#       2. Canonical Primary-Source Map explicitly documented and verified by CELL 13 V005
#
# Scientific role:
#   Freeze artifact identity, exact paths, filenames, hashes, Q01-Q10 frame,
#   and downstream dependency contracts.
#
# Firewall:
#   DATASET                 = UNTOUCHED
#   199 VERIFIED COHORT     = UNTOUCHED
#   400 HUMAN VALIDATION    = UNTOUCHED
#   GOLD / Y+ / Y-          = BLOCKED
#   PREFERENCE              = BLOCKED
#   MODEL INFERENCE         = BLOCKED
#   MODEL TRAINING          = BLOCKED
#   PROMPT REVISION         = BLOCKED
#   ARCHITECTURE CHANGE     = BLOCKED
#   NOVELTY JUDGMENT        = BLOCKED
#   EQUIVALENCE JUDGMENT    = BLOCKED
#
# FINAL:
#   PASS -> CELL 14 MAY START
#   FAIL -> STOP. NO PATH GUESSING.
# ================================================================================================================

from pathlib import Path
import pandas as pd
import json
import hashlib
from datetime import datetime


# ================================================================================================================
# STEP 0 — SCIENTIFIC VALUE GATE
# ================================================================================================================

print("=" * 120)
print("CELL 13A V002 — CANONICAL ARTIFACT & PATH LOCK GATE")
print("=" * 120)

SCIENTIFIC_VALUE = True

print("\n[STEP 0] SCIENTIFIC VALUE GATE")
print("-" * 120)

print("Purpose:")
print("  Establish one authoritative artifact/path source-of-truth before rebuilding Cell 14 onward.")
print()
print("Direct scientific value:")
print("  Protects reproducibility, provenance, auditability, and downstream artifact integrity.")
print()
print(f"SCIENTIFIC_VALUE = {SCIENTIFIC_VALUE}")

if not SCIENTIFIC_VALUE:
    raise RuntimeError("STOP — Scientific value gate failed.")


# ================================================================================================================
# STEP 1 — PROJECT ROOTS
# ================================================================================================================

print("\n[STEP 1] PROJECT ROOTS")
print("-" * 120)

X_ROOT = Path("/content/drive/MyDrive/X-AutoAlign_Legal_RAG")
FDR_ROOT = Path("/content/drive/MyDrive/F-D+R")

CHECKPOINT_ROOT = (
    X_ROOT
    / "04_AUDIT"
    / "NOVELTY_GAP_AUDIT"
    / "CHECKPOINTS"
)

SOURCE_ROOT = (
    FDR_ROOT
    / "DATASET_CONSTRUCTION"
    / "SOURCES"
)

print(f"X_ROOT          : {X_ROOT}")
print(f"FDR_ROOT        : {FDR_ROOT}")
print(f"CHECKPOINT_ROOT : {CHECKPOINT_ROOT}")
print(f"SOURCE_ROOT     : {SOURCE_ROOT}")


# ================================================================================================================
# STEP 2 — ROOT EXISTENCE GATE
# ================================================================================================================

print("\n[STEP 2] ROOT EXISTENCE GATE")
print("-" * 120)

ROOTS = {
    "X_ROOT": X_ROOT,
    "FDR_ROOT": FDR_ROOT,
    "CHECKPOINT_ROOT": CHECKPOINT_ROOT,
    "SOURCE_ROOT": SOURCE_ROOT,
}

for name, path in ROOTS.items():

    exists = path.exists()

    print(f"{name:<20} : {'PASS' if exists else 'FAIL'}")

    if not exists:
        raise FileNotFoundError(
            f"STOP — Required root does not exist:\n{name} -> {path}"
        )


# ================================================================================================================
# STEP 3 — EXACT CANONICAL REGISTRY DISCOVERY
# ================================================================================================================

print("\n[STEP 3] EXACT CANONICAL REGISTRY DISCOVERY")
print("-" * 120)

REGISTRY_FILENAME = "LITERATURE_REGISTRY_V003.parquet"

registry_matches = sorted(
    p
    for p in CHECKPOINT_ROOT.rglob(REGISTRY_FILENAME)
    if p.is_file() and p.name == REGISTRY_FILENAME
)

print(f"Requested exact filename : {REGISTRY_FILENAME}")
print(f"Exact matches found      : {len(registry_matches)}")

for p in registry_matches:
    print(f"  • {p}")

if len(registry_matches) != 1:

    raise RuntimeError(
        "STOP — Canonical Registry resolution is not unique.\n"
        "No alternative path or filename may be guessed."
    )

REGISTRY_PATH = registry_matches[0]

print("\n✓ Canonical Registry locked:")
print(f"  {REGISTRY_PATH}")


# ================================================================================================================
# STEP 4 — REGISTRY STRUCTURAL LOCK
# ================================================================================================================

print("\n[STEP 4] REGISTRY STRUCTURAL LOCK")
print("-" * 120)

registry_df = pd.read_parquet(REGISTRY_PATH)

print(f"Rows    : {len(registry_df)}")
print(f"Columns : {len(registry_df.columns)}")

EXPECTED_REGISTRY_COLUMNS = [
    "candidate_id",
    "search_family",
    "search_layer",
    "search_query",
    "title",
    "authors",
    "year",
    "venue",
    "primary_source",
    "doi_or_identifier",
    "url",
    "retrieval_date",
    "initial_relevance",
    "possible_competitor",
    "duplicate_group",
    "version_status",
    "discovery_reason",
    "notes",
]

actual_columns = registry_df.columns.tolist()

print("\nExpected canonical registry columns:")
for c in EXPECTED_REGISTRY_COLUMNS:
    print(f"  • {c}")

print("\nActual registry columns:")
for c in actual_columns:
    print(f"  • {c}")

if actual_columns != EXPECTED_REGISTRY_COLUMNS:

    raise RuntimeError(
        "STOP — Canonical Registry schema mismatch.\n"
        f"Expected:\n{EXPECTED_REGISTRY_COLUMNS}\n\n"
        f"Actual:\n{actual_columns}"
    )

if len(registry_df) != 5:

    raise RuntimeError(
        f"STOP — Expected 5 canonical literature records; found {len(registry_df)}."
    )

print("\n✓ Registry schema = PASS")
print("✓ Registry row count = PASS")


# ================================================================================================================
# STEP 5 — CANONICAL IDENTITY LOCK
# ================================================================================================================

print("\n[STEP 5] CANONICAL IDENTITY LOCK")
print("-" * 120)

EXPECTED_CANONICAL_IDS = [
    "ANCHOR-001",
    "LIT-001",
    "LIT-002",
    "LIT-003",
    "LIT-004",
]

actual_ids = (
    registry_df["candidate_id"]
    .astype(str)
    .tolist()
)

print("Expected:")
print(EXPECTED_CANONICAL_IDS)

print("\nActual:")
print(actual_ids)

if actual_ids != EXPECTED_CANONICAL_IDS:

    raise RuntimeError(
        "STOP — Canonical literature identity/order mismatch."
    )

print("\n✓ Canonical identity/order = PASS")


# ================================================================================================================
# STEP 6 — CELL 13 V005 CANONICAL PRIMARY-SOURCE MAP
# ================================================================================================================
#
# IMPORTANT:
#   This map is NOT inferred from filenames.
#   It is the exact canonical source map established in CELL 13 V005.
#
#   We deliberately keep this map explicit here because V003 Registry does not
#   contain primary_source_filename.
#
#   Any future filename/path change must cause FAIL rather than automatic repair.
# ================================================================================================================

print("\n[STEP 6] CELL 13 V005 CANONICAL PRIMARY-SOURCE MAP")
print("-" * 120)

CELL13_CANONICAL_SOURCE_MAP = {
    "ANCHOR-001":
        "The Cross Lingual Cost Retrieval Biases in RAG over Arabic English Corpora.pdf",

    "LIT-001":
        "SURE RAG Sufficiency and Uncertainty-Aware.pdf",

    "LIT-002":
        "Do LLMs Know When Evidence is Insufficient An Evidence Sufficiency.pdf",

    "LIT-003":
        "From noisy feedback to evidence aware issue specifications an agent governed retrieval augmented generation approach.pdf",

    "LIT-004":
        "AdaKG RAG Adaptive KG guided retrieval augmented generation.pdf",
}

if list(CELL13_CANONICAL_SOURCE_MAP.keys()) != EXPECTED_CANONICAL_IDS:

    raise RuntimeError(
        "STOP — Cell 13 V005 canonical source-map identity/order mismatch."
    )

print("Canonical source map:")
for cid in EXPECTED_CANONICAL_IDS:
    print(f"  {cid} -> {CELL13_CANONICAL_SOURCE_MAP[cid]}")


# ================================================================================================================
# STEP 7 — CELL 13 V005 SOURCE SHA LOCK
# ================================================================================================================
#
# These SHA-256 values are the source identities already verified in the canonical
# Cell 13 lineage.
#
# No SHA is calculated into a replacement identity.
# Actual SHA must EXACTLY match.
# ================================================================================================================

print("\n[STEP 7] CELL 13 V005 PRIMARY-SOURCE SHA LOCK")
print("-" * 120)

CELL13_CANONICAL_SOURCE_SHA256 = {

    "ANCHOR-001":
        "536236facb16659b38b032110c20bc6f3fb9716769cf9e4028ea176f680bbd93",

    "LIT-001":
        "1c13f9889408bdf9e54d1e26e4c7577f016d7c4d30d8c49b314b3189e0a93ff9",

    "LIT-002":
        "210b04fa8f6a1a6d484b271681283a2db058e1e763bf617845218a07c217c08b",

    "LIT-003":
        "48001d7bcef76e9b5e56f75c064d8419962d439522f14e33dcecbd86c1635c83",

    "LIT-004":
        "245410fe0f625d3d0879850f2c84db3bcb23bdf74cc03be65360de289e5dbd4a",
}

if list(CELL13_CANONICAL_SOURCE_SHA256.keys()) != EXPECTED_CANONICAL_IDS:

    raise RuntimeError(
        "STOP — Cell 13 V005 source SHA-map identity/order mismatch."
    )

print("Canonical SHA map loaded for exactly 5 sources.")


# ================================================================================================================
# STEP 8 — REGISTRY PRIMARY_SOURCE CONSISTENCY CHECK
# ================================================================================================================

print("\n[STEP 8] REGISTRY PRIMARY_SOURCE CONSISTENCY CHECK")
print("-" * 120)

for _, row in registry_df.iterrows():

    cid = str(row["candidate_id"])
    registry_primary_source = str(row["primary_source"]).strip()

    print(f"\n{cid}")
    print(f"  Registry primary_source : {registry_primary_source}")
    print(f"  Cell 13 V005 filename   : {CELL13_CANONICAL_SOURCE_MAP[cid]}")

    # Registry primary_source is metadata, not filename identity.
    # We only require that it is present and non-empty here.
    if (
        not registry_primary_source
        or registry_primary_source.lower() == "nan"
    ):

        raise RuntimeError(
            f"STOP — Missing registry primary_source metadata for {cid}."
        )

print("\n✓ Registry primary_source metadata = PASS")


# ================================================================================================================
# STEP 9 — EXACT SOURCE FILE RESOLUTION
# ================================================================================================================

print("\n[STEP 9] EXACT PRIMARY-SOURCE FILE RESOLUTION")
print("-" * 120)

locked_sources = []


for cid in EXPECTED_CANONICAL_IDS:

    filename = CELL13_CANONICAL_SOURCE_MAP[cid]

    exact_matches = sorted(
        p
        for p in SOURCE_ROOT.rglob(filename)
        if p.is_file() and p.name == filename
    )

    print(f"\n{cid}")
    print(f"  Exact filename : {filename}")
    print(f"  Matches        : {len(exact_matches)}")

    for p in exact_matches:
        print(f"    • {p}")

    if len(exact_matches) != 1:

        raise RuntimeError(
            f"STOP — Exact source resolution failed for {cid}.\n"
            f"Filename: {filename}\n"
            f"Expected exactly 1 match; found {len(exact_matches)}.\n"
            "NO FALLBACK / NO FUZZY MATCHING / NO PATH GUESSING."
        )

    source_path = exact_matches[0]

    locked_sources.append({
        "candidate_id": cid,
        "primary_source_filename": filename,
        "source_path": str(source_path),
        "expected_sha256": CELL13_CANONICAL_SOURCE_SHA256[cid],
    })

    print(f"  ✓ Locked path : {source_path}")


# ================================================================================================================
# STEP 10 — SHA-256 VERIFICATION
# ================================================================================================================

print("\n[STEP 10] PRIMARY-SOURCE SHA-256 VERIFICATION")
print("-" * 120)


def sha256_file(path, chunk_size=1024 * 1024):

    h = hashlib.sha256()

    with open(path, "rb") as f:

        while True:

            chunk = f.read(chunk_size)

            if not chunk:
                break

            h.update(chunk)

    return h.hexdigest()


for item in locked_sources:

    path = Path(item["source_path"])

    actual_sha = sha256_file(path)
    expected_sha = item["expected_sha256"]

    item["actual_sha256"] = actual_sha
    item["sha256_match"] = (
        actual_sha == expected_sha
    )

    print(f"\n{item['candidate_id']}")
    print(f"  Expected SHA : {expected_sha}")
    print(f"  Actual SHA   : {actual_sha}")
    print(
        f"  Match        : "
        f"{'PASS' if item['sha256_match'] else 'FAIL'}"
    )

    if not item["sha256_match"]:

        raise RuntimeError(
            f"STOP — SHA-256 mismatch for {item['candidate_id']}.\n"
            f"Do NOT substitute another source file."
        )


# ================================================================================================================
# STEP 11 — Q01-Q10 FRAME LOCK
# ================================================================================================================

print("\n[STEP 11] Q01-Q10 FRAME LOCK")
print("-" * 120)

QUESTIONS = [
    ("Q01", "Does the work perform post-retrieval evidence verification?"),
    ("Q02", "What is the input to the verification/control mechanism?"),
    ("Q03", "Is evidence relevance assessed independently from evidence sufficiency/support?"),
    ("Q04", "Is legal support or domain-specific evidentiary support explicitly assessed?"),
    ("Q05", "Is evidence sufficiency explicitly assessed?"),
    ("Q06", "Is contradiction, conflict, or risk explicitly assessed?"),
    ("Q07", "Does the mechanism operate before generation or answering?"),
    ("Q08", "Does the mechanism produce an evidence eligibility, risk, or abstention control?"),
    ("Q09", "Is human-grounded evidence suitability part of the evaluated methodology?"),
    ("Q10", "Is the mechanism equivalent to the proposed F-D+R evidence-suitability verification stage?"),
]

question_frame = [
    {
        "question_id": qid,
        "question_text": qtext
    }
    for qid, qtext in QUESTIONS
]

question_frame_json = json.dumps(
    question_frame,
    ensure_ascii=False,
    sort_keys=True,
    separators=(",", ":")
)

QUESTION_FRAME_SHA256 = hashlib.sha256(
    question_frame_json.encode("utf-8")
).hexdigest()

print(f"Question count : {len(question_frame)}")
print(f"Frame SHA-256  : {QUESTION_FRAME_SHA256}")

if len(question_frame) != 10:

    raise RuntimeError(
        "STOP — Q01-Q10 frame count mismatch."
    )

print("✓ Q01-Q10 frame = LOCKED")


# ================================================================================================================
# STEP 12 — CANONICAL REGISTRY SHA
# ================================================================================================================

print("\n[STEP 12] CANONICAL REGISTRY SHA-256")
print("-" * 120)

REGISTRY_SHA256 = sha256_file(REGISTRY_PATH)

print(f"Registry SHA-256:")
print(f"  {REGISTRY_SHA256}")


# ================================================================================================================
# STEP 13 — CANONICAL ARTIFACT LOCK TABLE
# ================================================================================================================

print("\n[STEP 13] CANONICAL ARTIFACT LOCK TABLE")
print("-" * 120)

artifact_rows = [
    {
        "artifact_role": "CANONICAL_LITERATURE_REGISTRY",
        "candidate_id": "ALL",
        "artifact_name": REGISTRY_PATH.name,
        "path": str(REGISTRY_PATH),
        "sha256": REGISTRY_SHA256,
        "status": "LOCKED",
        "source_of_identity": "CELL13_V005_CANONICAL_REGISTRY",
    }
]

for item in locked_sources:

    artifact_rows.append({
        "artifact_role": "PRIMARY_SOURCE_PDF",
        "candidate_id": item["candidate_id"],
        "artifact_name": item["primary_source_filename"],
        "path": item["source_path"],
        "sha256": item["actual_sha256"],
        "status": "LOCKED",
        "source_of_identity": "CELL13_V005_CANONICAL_SOURCE_MAP",
    })

artifact_lock_df = pd.DataFrame(artifact_rows)

print(
    artifact_lock_df.to_string(index=False)
)


# ================================================================================================================
# STEP 14 — DOWNSTREAM DEPENDENCY CONTRACT
# ================================================================================================================

print("\n[STEP 14] DOWNSTREAM DEPENDENCY CONTRACT")
print("-" * 120)

DOWNSTREAM_CONTRACTS = {

    "CELL14": {
        "role": "Evidence Matrix Initialization",
        "required_locked_inputs": [
            "CANONICAL_LITERATURE_REGISTRY",
            "PRIMARY_SOURCE_PDF",
            "Q01_Q10_FRAME",
        ],
        "status": "FUTURE_OUTPUT",
    },

    "CELL15": {
        "role": "Source-Grounded Extraction",
        "required_locked_inputs": [
            "CELL14_OUTPUT",
            "PRIMARY_SOURCE_PDF",
        ],
        "status": "FUTURE_OUTPUT",
    },

    "CELL16": {
        "role": "Extraction QC",
        "required_locked_inputs": [
            "CELL15_OUTPUT",
        ],
        "status": "FUTURE_OUTPUT",
    },

    "CELL16A": {
        "role": "Human Source Review",
        "required_locked_inputs": [
            "CELL16_OUTPUT",
            "PRIMARY_SOURCE_PDF",
        ],
        "status": "FUTURE_OUTPUT",
    },

    "CELL17": {
        "role": "QID / Question Text Integrity QC",
        "required_locked_inputs": [
            "CELL16A_OUTPUT",
        ],
        "status": "FUTURE_OUTPUT",
    },

    "CELL18": {
        "role": "Source-Grounded Methodological Comparison Frame",
        "required_locked_inputs": [
            "CELL14_OUTPUT",
            "CELL16A_OUTPUT",
            "CELL17_OUTPUT",
        ],
        "status": "FUTURE_OUTPUT",
    },

    "CELL19": {
        "role": "Source-Grounded Methodological Review",
        "required_locked_inputs": [
            "CELL13A_LOCK",
            "CELL14_OUTPUT",
            "CELL16A_OUTPUT",
            "CELL17_OUTPUT",
            "CELL18_OUTPUT",
            "PRIMARY_SOURCE_PDF",
        ],
        "status": "FUTURE_OUTPUT",
    },
}

for cell_id, contract in DOWNSTREAM_CONTRACTS.items():

    print(f"\n{cell_id}")
    print(f"  Role   : {contract['role']}")
    print(f"  Inputs : {contract['required_locked_inputs']}")
    print(f"  Status : {contract['status']}")


# ================================================================================================================
# STEP 15 — NO-GUESSING FIREWALL
# ================================================================================================================

print("\n[STEP 15] NO-GUESSING FIREWALL")
print("-" * 120)

NO_GUESSING_FIREWALL = {

    "registry_exact_filename_only": True,

    "registry_exactly_one_match_required": True,

    "source_filename_must_come_from_cell13_v005_map": True,

    "source_exact_filename_matching_only": True,

    "source_exactly_one_match_required": True,

    "source_sha256_required": True,

    "source_sha256_exact_match_required": True,

    "no_fuzzy_matching": True,

    "no_partial_matching": True,

    "no_alternate_version_fallback": True,

    "no_manual_path_substitution": True,

    "no_missing_artifact_substitution": True,

    "no_automatic_artifact_repair": True,

    "future_cells_must_consume_cell13a_lock": True,
}

for rule, value in NO_GUESSING_FIREWALL.items():

    print(
        f"{rule:<55} : "
        f"{value}"
    )

if not all(NO_GUESSING_FIREWALL.values()):

    raise RuntimeError(
        "STOP — No-guessing firewall is not fully enabled."
    )


# ================================================================================================================
# STEP 16 — PROJECT FIREWALL
# ================================================================================================================

print("\n[STEP 16] PROJECT FIREWALL")
print("-" * 120)

PROJECT_FIREWALL = {

    "DATASET": "UNTOUCHED",

    "VERIFIED_199_COHORT": "UNTOUCHED",

    "HUMAN_VALIDATION_400": "UNTOUCHED",

    "GOLD_LABELS": "UNTOUCHED",

    "Y_PLUS": "BLOCKED",

    "Y_MINUS": "BLOCKED",

    "ABSTAIN_LABELS": "BLOCKED",

    "PREFERENCE_LABELS": "BLOCKED",

    "MODEL_INFERENCE": "BLOCKED",

    "MODEL_TRAINING": "BLOCKED",

    "PROMPT_REVISION": "BLOCKED",

    "ARCHITECTURE_MODIFICATION": "BLOCKED",

    "NEW_LEGAL_EXTRACTION": "BLOCKED",

    "NOVELTY_JUDGMENT": "BLOCKED",

    "EQUIVALENCE_JUDGMENT": "BLOCKED",
}

for key, value in PROJECT_FIREWALL.items():

    print(f"{key:<35} : {value}")


# ================================================================================================================
# STEP 17 — PERSIST CANONICAL LOCK
# ================================================================================================================

print("\n[STEP 17] PERSIST CANONICAL LOCK")
print("-" * 120)

LOCK_DIR = (
    CHECKPOINT_ROOT
    / "CELL13A_V002_CANONICAL_ARTIFACT_PATH_LOCK"
)

LOCK_DIR.mkdir(
    parents=True,
    exist_ok=True
)

MANIFEST_PATH = (
    LOCK_DIR
    / "CELL13A_CANONICAL_ARTIFACT_PATH_LOCK_V002.json"
)

ARTIFACT_CSV_PATH = (
    LOCK_DIR
    / "CELL13A_CANONICAL_ARTIFACT_LOCK_V002.csv"
)

manifest = {

    "experiment_id":
        "FDR-NOVELTY-GAP-001",

    "cell_id":
        "CELL13A_CANONICAL_ARTIFACT_PATH_LOCK_V002",

    "status":
        "CANONICAL_ARTIFACT_PATH_LOCK_PASS",

    "created_utc":
        datetime.utcnow().isoformat() + "Z",

    "design_correction":
        "V002 uses the Cell 13 V005 canonical primary-source map because "
        "LITERATURE_REGISTRY_V003 does not contain primary_source_filename.",

    "roots": {

        "x_root":
            str(X_ROOT),

        "fdr_root":
            str(FDR_ROOT),

        "checkpoint_root":
            str(CHECKPOINT_ROOT),

        "source_root":
            str(SOURCE_ROOT),
    },

    "canonical_registry": {

        "filename":
            REGISTRY_PATH.name,

        "path":
            str(REGISTRY_PATH),

        "sha256":
            REGISTRY_SHA256,

        "canonical_ids":
            EXPECTED_CANONICAL_IDS,

        "schema_columns":
            EXPECTED_REGISTRY_COLUMNS,

        "row_count":
            len(registry_df),
    },

    "cell13_v005_canonical_source_map":
        CELL13_CANONICAL_SOURCE_MAP,

    "cell13_v005_canonical_source_sha256":
        CELL13_CANONICAL_SOURCE_SHA256,

    "locked_primary_sources":
        locked_sources,

    "question_frame": {

        "count":
            len(question_frame),

        "sha256":
            QUESTION_FRAME_SHA256,

        "questions":
            question_frame,
    },

    "downstream_contracts":
        DOWNSTREAM_CONTRACTS,

    "no_guessing_firewall":
        NO_GUESSING_FIREWALL,

    "project_firewall":
        PROJECT_FIREWALL,

    "upstream_baseline": {

        "cell":
            "CELL13 V005",

        "status":
            "TRUSTED_UPSTREAM_BASELINE",

        "modification_allowed":
            False,
    },

    "cell14_start_condition": {

        "required":
            "CELL13A_CANONICAL_ARTIFACT_PATH_LOCK_PASS",

        "if_pass":
            "CELL14_MAY_START",

        "if_fail":
            "STOP_NO_PATH_GUESSING",
    },
}

with open(
    MANIFEST_PATH,
    "w",
    encoding="utf-8"
) as f:

    json.dump(
        manifest,
        f,
        ensure_ascii=False,
        indent=2
    )

artifact_lock_df.to_csv(
    ARTIFACT_CSV_PATH,
    index=False,
    encoding="utf-8-sig"
)

print(f"Manifest:")
print(f"  {MANIFEST_PATH}")

print(f"\nArtifact CSV:")
print(f"  {ARTIFACT_CSV_PATH}")


# ================================================================================================================
# STEP 18 — READ-BACK + INTEGRITY VERIFICATION
# ================================================================================================================

print("\n[STEP 18] READ-BACK + INTEGRITY VERIFICATION")
print("-" * 120)

with open(
    MANIFEST_PATH,
    "r",
    encoding="utf-8"
) as f:

    manifest_rb = json.load(f)


assert (
    manifest_rb["status"]
    == "CANONICAL_ARTIFACT_PATH_LOCK_PASS"
)

assert (
    manifest_rb["canonical_registry"]["path"]
    == str(REGISTRY_PATH)
)

assert (
    manifest_rb["canonical_registry"]["sha256"]
    == REGISTRY_SHA256
)

assert (
    manifest_rb["canonical_registry"]["canonical_ids"]
    == EXPECTED_CANONICAL_IDS
)

assert (
    manifest_rb["question_frame"]["sha256"]
    == QUESTION_FRAME_SHA256
)

assert (
    manifest_rb["cell13_v005_canonical_source_map"]
    == CELL13_CANONICAL_SOURCE_MAP
)

assert (
    manifest_rb["cell13_v005_canonical_source_sha256"]
    == CELL13_CANONICAL_SOURCE_SHA256
)


for item in manifest_rb["locked_primary_sources"]:

    path = Path(item["source_path"])

    assert path.exists(), (
        f"READ-BACK FAIL — Missing locked source: {path}"
    )

    assert (
        path.name
        == item["primary_source_filename"]
    ), (
        "READ-BACK FAIL — Filename drift detected."
    )

    current_sha = sha256_file(path)

    assert (
        current_sha
        == item["actual_sha256"]
    ), (
        f"READ-BACK FAIL — SHA drift detected: {path}"
    )

    assert (
        current_sha
        == item["expected_sha256"]
    ), (
        f"READ-BACK FAIL — Canonical SHA mismatch: {path}"
    )


print("✓ Manifest status          = PASS")
print("✓ Registry path            = PASS")
print("✓ Registry SHA             = PASS")
print("✓ Canonical IDs            = PASS")
print("✓ Registry schema          = PASS")
print("✓ Cell 13 source map       = PASS")
print("✓ Primary source paths     = PASS")
print("✓ Primary source SHA       = PASS")
print("✓ Q01-Q10 frame            = PASS")
print("✓ No path drift            = PASS")


# ================================================================================================================
# STEP 19 — FINAL PRE-CELL14 GO / NO-GO
# ================================================================================================================

print("\n" + "=" * 120)
print("FINAL PRE-CELL14 GO / NO-GO GATE")
print("=" * 120)

GO_GATES = {

    "scientific_value":
        SCIENTIFIC_VALUE,

    "registry_unique":
        len(registry_matches) == 1,

    "registry_schema_locked":
        actual_columns == EXPECTED_REGISTRY_COLUMNS,

    "registry_row_count":
        len(registry_df) == 5,

    "canonical_identity_locked":
        actual_ids == EXPECTED_CANONICAL_IDS,

    "cell13_source_map_locked":
        list(CELL13_CANONICAL_SOURCE_MAP.keys())
        == EXPECTED_CANONICAL_IDS,

    "cell13_source_sha_map_locked":
        list(CELL13_CANONICAL_SOURCE_SHA256.keys())
        == EXPECTED_CANONICAL_IDS,

    "primary_source_count":
        len(locked_sources) == 5,

    "primary_source_exact_match":
        all(
            Path(item["source_path"]).exists()
            for item in locked_sources
        ),

    "primary_source_sha_match":
        all(
            item["sha256_match"]
            for item in locked_sources
        ),

    "q01_q10_locked":
        len(question_frame) == 10,

    "no_guessing_firewall":
        all(NO_GUESSING_FIREWALL.values()),

    "manifest_exists":
        MANIFEST_PATH.exists(),

    "artifact_csv_exists":
        ARTIFACT_CSV_PATH.exists(),

    "manifest_readback":
        True,
}


for gate, result in GO_GATES.items():

    print(
        f"{gate:<45} : "
        f"{'PASS' if result else 'FAIL'}"
    )


FINAL_PASS = all(
    GO_GATES.values()
)

print("\n" + "-" * 120)

if FINAL_PASS:

    print("✅ CELL13A V002 — CANONICAL ARTIFACT & PATH LOCK = PASS")
    print()
    print("🔒 CELL 13 V005 remains the trusted upstream baseline.")
    print("🔒 LITERATURE_REGISTRY_V003 is locked.")
    print("🔒 Cell 13 V005 canonical source map is locked.")
    print("🔒 All 5 primary source paths are exact-match locked.")
    print("🔒 All 5 primary source SHA-256 values are verified.")
    print("🔒 Q01-Q10 frame is locked.")
    print("🔒 Future Cells 14–19 must consume this lock.")
    print("🔒 No artifact/path guessing is permitted.")
    print()
    print("🚦 GO — CELL 14 MAY START.")
    print()
    print("LOCK MANIFEST:")
    print(f"  {MANIFEST_PATH}")

else:

    print("❌ CELL13A V002 — FAIL")
    print()
    print("🛑 CELL 14 MUST NOT START.")
    print("🛑 NO PATH GUESSING.")
    print("🛑 NO ALTERNATE ARTIFACT SUBSTITUTION.")
    print("🛑 NO MANUAL REPAIR INSIDE DOWNSTREAM CELLS.")


print("=" * 120)
```

```python
# ================================================================================================================
# CELL 14 V002 — CANONICAL LITERATURE EVIDENCE MATRIX INITIALIZATION
# ================================================================================================================
#
# REBUILD POLICY:
#   V001 FAILED at STEP 17 due to a QC assertion implementation issue.
#
#   V002 is REBUILT from:
#       CELL13A V002 CANONICAL ARTIFACT & PATH LOCK
#
#   V001 is NOT patched.
#   V001 is preserved as historical failed execution.
#
# ================================================================================================================

from pathlib import Path
import pandas as pd
import json
import hashlib
from datetime import datetime, timezone


print("=" * 120)
print("CELL 14 V002 — REBUILT CANONICAL LITERATURE EVIDENCE MATRIX")
print("=" * 120)


# ================================================================================================================
# STEP 0 — SCIENTIFIC VALUE GATE
# ================================================================================================================

print("\n[STEP 0] SCIENTIFIC VALUE GATE")
print("-" * 120)

SCIENTIFIC_VALUE = True

print("Purpose:")
print("  Initialize and independently verify the frozen 5 × 10 methodological evidence matrix.")

print("\nScientific value:")
print("  Establishes a reproducible evidence-review frame and verifies that the persisted artifact")
print("  preserves the extraction/judgment firewall after serialization and reload.")

print(f"\nSCIENTIFIC_VALUE = {SCIENTIFIC_VALUE}")

if not SCIENTIFIC_VALUE:
    raise RuntimeError("STOP — Scientific value gate failed.")


# ================================================================================================================
# STEP 1 — CELL13A V002 LOCK
# ================================================================================================================

print("\n[STEP 1] CELL13A V002 CANONICAL LOCK RESOLUTION")
print("-" * 120)

X_ROOT = Path(
    "/content/drive/MyDrive/X-AutoAlign_Legal_RAG"
)

LOCK_PATH = (
    X_ROOT
    / "04_AUDIT"
    / "NOVELTY_GAP_AUDIT"
    / "CHECKPOINTS"
    / "CELL13A_V002_CANONICAL_ARTIFACT_PATH_LOCK"
    / "CELL13A_CANONICAL_ARTIFACT_PATH_LOCK_V002.json"
)

print(f"Lock path : {LOCK_PATH}")

if not LOCK_PATH.exists():
    raise RuntimeError(
        "STOP — CELL13A V002 canonical lock does not exist."
    )

print("✓ CELL13A V002 lock found.")


# ================================================================================================================
# STEP 2 — LOAD + VALIDATE LOCK
# ================================================================================================================

print("\n[STEP 2] LOAD + VALIDATE CELL13A V002 LOCK")
print("-" * 120)

with open(
    LOCK_PATH,
    "r",
    encoding="utf-8"
) as f:
    lock = json.load(f)

EXPECTED_LOCK_ID = (
    "CELL13A_CANONICAL_ARTIFACT_PATH_LOCK_V002"
)

EXPECTED_LOCK_STATUS = (
    "CANONICAL_ARTIFACT_PATH_LOCK_PASS"
)

if lock.get("cell_id") != EXPECTED_LOCK_ID:
    raise RuntimeError(
        "STOP — Wrong Cell13A lock identity."
    )

if lock.get("status") != EXPECTED_LOCK_STATUS:
    raise RuntimeError(
        "STOP — Cell13A lock is not PASS."
    )

print("✓ Lock identity = PASS")
print("✓ Lock status   = PASS")


# ================================================================================================================
# STEP 3 — RESOLVE REGISTRY ONLY FROM LOCK
# ================================================================================================================

print("\n[STEP 3] LOCKED CANONICAL REGISTRY")
print("-" * 120)

registry_info = lock["canonical_registry"]

REGISTRY_PATH = Path(
    registry_info["path"]
)

EXPECTED_REGISTRY_FILENAME = (
    registry_info["filename"]
)

EXPECTED_REGISTRY_SHA = (
    registry_info["sha256"]
)

EXPECTED_IDS = (
    registry_info["canonical_ids"]
)

EXPECTED_REGISTRY_COLUMNS = (
    registry_info["schema_columns"]
)

EXPECTED_REGISTRY_ROWS = (
    registry_info["row_count"]
)

print(f"Registry : {REGISTRY_PATH}")
print(f"Rows     : {EXPECTED_REGISTRY_ROWS}")
print(f"SHA      : {EXPECTED_REGISTRY_SHA}")

if not REGISTRY_PATH.exists():
    raise RuntimeError(
        "STOP — Locked Registry does not exist."
    )

if REGISTRY_PATH.name != EXPECTED_REGISTRY_FILENAME:
    raise RuntimeError(
        "STOP — Registry filename drift."
    )


# ================================================================================================================
# STEP 4 — HASH FUNCTION
# ================================================================================================================

def sha256_file(
    path,
    chunk_size=1024 * 1024
):

    h = hashlib.sha256()

    with open(path, "rb") as f:

        while True:

            chunk = f.read(chunk_size)

            if not chunk:
                break

            h.update(chunk)

    return h.hexdigest()


# ================================================================================================================
# STEP 5 — REGISTRY SHA
# ================================================================================================================

print("\n[STEP 5] CANONICAL REGISTRY SHA-256")
print("-" * 120)

actual_registry_sha = sha256_file(
    REGISTRY_PATH
)

print(f"Expected : {EXPECTED_REGISTRY_SHA}")
print(f"Actual   : {actual_registry_sha}")

if actual_registry_sha != EXPECTED_REGISTRY_SHA:
    raise RuntimeError(
        "STOP — Registry SHA-256 mismatch."
    )

print("✓ Registry SHA = PASS")


# ================================================================================================================
# STEP 6 — LOAD REGISTRY + STRUCTURAL QC
# ================================================================================================================

print("\n[STEP 6] REGISTRY STRUCTURAL QC")
print("-" * 120)

registry_df = pd.read_parquet(
    REGISTRY_PATH
)

print(f"Rows    : {len(registry_df)}")
print(f"Columns : {len(registry_df.columns)}")

if len(registry_df) != EXPECTED_REGISTRY_ROWS:
    raise RuntimeError(
        "STOP — Registry row count mismatch."
    )

if registry_df.columns.tolist() != EXPECTED_REGISTRY_COLUMNS:
    raise RuntimeError(
        "STOP — Registry schema mismatch."
    )

actual_ids = (
    registry_df["candidate_id"]
    .astype(str)
    .tolist()
)

print(f"\nExpected IDs : {EXPECTED_IDS}")
print(f"Actual IDs   : {actual_ids}")

if actual_ids != EXPECTED_IDS:
    raise RuntimeError(
        "STOP — Canonical identity/order mismatch."
    )

print("✓ Registry schema = PASS")
print("✓ Canonical identity = PASS")


# ================================================================================================================
# STEP 7 — PRIMARY SOURCE LOCK VERIFICATION
# ================================================================================================================

print("\n[STEP 7] PRIMARY-SOURCE LOCK VERIFICATION")
print("-" * 120)

locked_sources = lock.get(
    "locked_primary_sources"
)

if locked_sources is None:
    raise RuntimeError(
        "STOP — Primary-source lock missing."
    )

if len(locked_sources) != 5:
    raise RuntimeError(
        "STOP — Expected exactly 5 locked primary sources."
    )

for item in locked_sources:

    cid = item["candidate_id"]
    filename = item["primary_source_filename"]
    path = Path(item["source_path"])
    expected_sha = item["expected_sha256"]

    print(f"\n{cid}")
    print(f"  Filename : {filename}")
    print(f"  Path     : {path}")

    if not path.exists():
        raise RuntimeError(
            f"STOP — Missing locked source: {cid}"
        )

    if path.name != filename:
        raise RuntimeError(
            f"STOP — Filename drift: {cid}"
        )

    actual_sha = sha256_file(path)

    print(f"  Expected SHA : {expected_sha}")
    print(f"  Actual SHA   : {actual_sha}")

    if actual_sha != expected_sha:
        raise RuntimeError(
            f"STOP — Source SHA mismatch: {cid}"
        )

    print("  ✓ PASS")


# ================================================================================================================
# STEP 8 — Q01-Q10 LOCK
# ================================================================================================================

print("\n[STEP 8] Q01-Q10 FRAME LOCK")
print("-" * 120)

question_info = lock["question_frame"]

LOCKED_QUESTIONS = question_info["questions"]
LOCKED_Q_COUNT = question_info["count"]
LOCKED_Q_SHA = question_info["sha256"]

question_json = json.dumps(
    LOCKED_QUESTIONS,
    ensure_ascii=False,
    sort_keys=True,
    separators=(",", ":")
)

actual_q_sha = hashlib.sha256(
    question_json.encode("utf-8")
).hexdigest()

print(f"Question count : {LOCKED_Q_COUNT}")
print(f"Expected SHA   : {LOCKED_Q_SHA}")
print(f"Actual SHA     : {actual_q_sha}")

if LOCKED_Q_COUNT != 10:
    raise RuntimeError(
        "STOP — Q01-Q10 count mismatch."
    )

if actual_q_sha != LOCKED_Q_SHA:
    raise RuntimeError(
        "STOP — Q01-Q10 SHA mismatch."
    )

print("✓ Q01-Q10 frame = PASS")


# ================================================================================================================
# STEP 9 — INITIALIZE 5 × 10 MATRIX
# ================================================================================================================

print("\n[STEP 9] INITIALIZE FROZEN 5 × 10 EVIDENCE MATRIX")
print("-" * 120)

source_lookup = {
    x["candidate_id"]: x
    for x in locked_sources
}

rows = []

for candidate_id in EXPECTED_IDS:

    matches = registry_df[
        registry_df["candidate_id"].astype(str)
        == candidate_id
    ]

    if len(matches) != 1:
        raise RuntimeError(
            f"STOP — Candidate resolution failed: {candidate_id}"
        )

    rec = matches.iloc[0]
    source = source_lookup[candidate_id]

    for q in LOCKED_QUESTIONS:

        rows.append({

            "candidate_id":
                candidate_id,

            "question_id":
                q["question_id"],

            "title":
                str(rec["title"]),

            "authors":
                str(rec["authors"]),

            "year":
                rec["year"],

            "venue":
                str(rec["venue"]),

            "primary_source":
                str(rec["primary_source"]),

            "primary_source_filename":
                source["primary_source_filename"],

            "primary_source_path":
                source["source_path"],

            "primary_source_sha256":
                source["actual_sha256"],

            "question_text":
                q["question_text"],

            # Extraction — intentionally EMPTY
            "source_excerpt": "",
            "source_location": "",
            "source_text_status": "",
            "source_text_unit_start": "",
            "source_text_unit_end": "",
            "extraction_method": "",
            "extraction_interpretation": "",
            "extraction_status": "",
            "source_paraphrase": "",
            "methodological_note": "",

            # Judgments — intentionally NOT ASSESSED
            "equivalence_judgment":
                "NOT_ASSESSED",

            "novelty_judgment":
                "NOT_ASSESSED",

            "overlap_class":
                "NOT_ASSESSED",
        })


evidence_matrix_df = pd.DataFrame(
    rows
)

print(f"Rows generated : {len(evidence_matrix_df)}")

if len(evidence_matrix_df) != 50:
    raise RuntimeError(
        "STOP — Expected 50 evidence slots."
    )

print("✓ 50 rows = PASS")


# ================================================================================================================
# STEP 10 — STRUCTURAL QC
# ================================================================================================================

print("\n[STEP 10] MATRIX STRUCTURAL QC")
print("-" * 120)

expected_qids = [
    q["question_id"]
    for q in LOCKED_QUESTIONS
]

actual_candidate_ids = (
    evidence_matrix_df[
        "candidate_id"
    ]
    .drop_duplicates()
    .tolist()
)

actual_qids = (
    evidence_matrix_df[
        "question_id"
    ]
    .drop_duplicates()
    .tolist()
)

if actual_candidate_ids != EXPECTED_IDS:
    raise RuntimeError(
        "STOP — Candidate identity mismatch."
    )

if actual_qids != expected_qids:
    raise RuntimeError(
        "STOP — Question identity mismatch."
    )

candidate_counts = (
    evidence_matrix_df
    .groupby("candidate_id")
    .size()
)

question_counts = (
    evidence_matrix_df
    .groupby("question_id")
    .size()
)

print("\nRows per candidate:")
print(candidate_counts.to_string())

print("\nRows per question:")
print(question_counts.to_string())

if not (candidate_counts == 10).all():
    raise RuntimeError(
        "STOP — Candidate cardinality mismatch."
    )

if not (question_counts == 5).all():
    raise RuntimeError(
        "STOP — Question cardinality mismatch."
    )

print("\n✓ 5 × 10 cardinality = PASS")


# ================================================================================================================
# STEP 11 — QUESTION TEXT INTEGRITY
# ================================================================================================================

print("\n[STEP 11] QUESTION TEXT INTEGRITY")
print("-" * 120)

for q in LOCKED_QUESTIONS:

    qid = q["question_id"]
    expected_text = q["question_text"]

    observed = (
        evidence_matrix_df[
            evidence_matrix_df["question_id"] == qid
        ]["question_text"]
        .drop_duplicates()
        .tolist()
    )

    if observed != [expected_text]:
        raise RuntimeError(
            f"STOP — Question text mismatch: {qid}"
        )

print("✓ All Q01-Q10 texts = PASS")


# ================================================================================================================
# STEP 12 — EXTRACTION / JUDGMENT FIREWALL
# ================================================================================================================

print("\n[STEP 12] EXTRACTION / JUDGMENT FIREWALL")
print("-" * 120)

EXTRACTION_FIELDS = [
    "source_excerpt",
    "source_location",
    "source_text_status",
    "source_text_unit_start",
    "source_text_unit_end",
    "extraction_method",
    "extraction_interpretation",
    "extraction_status",
    "source_paraphrase",
    "methodological_note",
]

for field in EXTRACTION_FIELDS:

    non_empty = (
        evidence_matrix_df[field]
        .fillna("")
        .astype(str)
        .str.strip()
        .ne("")
        .sum()
    )

    print(
        f"{field:<30} : "
        f"non-empty = {non_empty}"
    )

    if non_empty != 0:
        raise RuntimeError(
            f"STOP — Extraction field populated: {field}"
        )


for field in [
    "equivalence_judgment",
    "novelty_judgment",
    "overlap_class",
]:

    counts = (
        evidence_matrix_df[field]
        .fillna("")
        .astype(str)
        .value_counts()
        .to_dict()
    )

    print(
        f"{field:<30} : {counts}"
    )

    if counts != {"NOT_ASSESSED": 50}:
        raise RuntimeError(
            f"STOP — Judgment field violation: {field}"
        )

print("\n✓ Extraction firewall = PASS")
print("✓ Judgment firewall   = PASS")


# ================================================================================================================
# STEP 13 — DUPLICATE SLOT QC
# ================================================================================================================

print("\n[STEP 13] DUPLICATE SLOT QC")
print("-" * 120)

duplicate_count = int(
    evidence_matrix_df
    .duplicated(
        subset=[
            "candidate_id",
            "question_id"
        ],
        keep=False
    )
    .sum()
)

print(
    f"Duplicate (candidate_id, question_id) rows : "
    f"{duplicate_count}"
)

if duplicate_count != 0:
    raise RuntimeError(
        "STOP — Duplicate evidence slots."
    )

print("✓ Slot uniqueness = PASS")


# ================================================================================================================
# STEP 14 — SCHEMA LOCK
# ================================================================================================================

print("\n[STEP 14] MATRIX SCHEMA LOCK")
print("-" * 120)

EXPECTED_MATRIX_COLUMNS = [
    "candidate_id",
    "question_id",
    "title",
    "authors",
    "year",
    "venue",
    "primary_source",
    "primary_source_filename",
    "primary_source_path",
    "primary_source_sha256",
    "question_text",
    "source_excerpt",
    "source_location",
    "source_text_status",
    "source_text_unit_start",
    "source_text_unit_end",
    "extraction_method",
    "extraction_interpretation",
    "extraction_status",
    "source_paraphrase",
    "methodological_note",
    "equivalence_judgment",
    "novelty_judgment",
    "overlap_class",
]

if evidence_matrix_df.columns.tolist() != EXPECTED_MATRIX_COLUMNS:
    raise RuntimeError(
        "STOP — Matrix schema mismatch."
    )

print(
    f"Columns : {len(EXPECTED_MATRIX_COLUMNS)}"
)

print("✓ Matrix schema = PASS")


# ================================================================================================================
# STEP 15 — PERSIST V002
# ================================================================================================================

print("\n[STEP 15] PERSIST CELL 14 V002")
print("-" * 120)

CELL14_DIR = (
    X_ROOT
    / "04_AUDIT"
    / "NOVELTY_GAP_AUDIT"
    / "CHECKPOINTS"
    / "CELL14_REBUILT_V002_EVIDENCE_MATRIX"
)

CELL14_DIR.mkdir(
    parents=True,
    exist_ok=True
)

OUTPUT_PARQUET = (
    CELL14_DIR
    / "LITERATURE_EVIDENCE_MATRIX_V002.parquet"
)

OUTPUT_CSV = (
    CELL14_DIR
    / "LITERATURE_EVIDENCE_MATRIX_V002.csv"
)

OUTPUT_MANIFEST = (
    CELL14_DIR
    / "LITERATURE_EVIDENCE_MATRIX_V002_MANIFEST.json"
)

evidence_matrix_df.to_parquet(
    OUTPUT_PARQUET,
    index=False
)

evidence_matrix_df.to_csv(
    OUTPUT_CSV,
    index=False,
    encoding="utf-8-sig"
)

print(f"Parquet : {OUTPUT_PARQUET}")
print(f"CSV     : {OUTPUT_CSV}")


# ================================================================================================================
# STEP 16 — OUTPUT HASHES
# ================================================================================================================

print("\n[STEP 16] OUTPUT SHA-256")
print("-" * 120)

OUTPUT_PARQUET_SHA = sha256_file(
    OUTPUT_PARQUET
)

OUTPUT_CSV_SHA = sha256_file(
    OUTPUT_CSV
)

print(f"Parquet SHA : {OUTPUT_PARQUET_SHA}")
print(f"CSV SHA     : {OUTPUT_CSV_SHA}")


# ================================================================================================================
# STEP 17 — MANIFEST
# ================================================================================================================

print("\n[STEP 17] BUILD V002 MANIFEST")
print("-" * 120)

manifest = {

    "experiment_id":
        "FDR-NOVELTY-GAP-001",

    "cell_id":
        "CELL14_CANONICAL_LITERATURE_EVIDENCE_MATRIX_V002",

    "status":
        "CELL14_V002_EVIDENCE_MATRIX_INITIALIZED_PASS",

    "created_utc":
        datetime.now(timezone.utc).isoformat(),

    "rebuild_reason":
        "V001 failed at read-back QC due to a boolean assertion implementation issue. "
        "V002 is a clean rebuild from CELL13A V002; V001 was not patched.",

    "upstream_lock": {

        "cell_id":
            lock["cell_id"],

        "status":
            lock["status"],

        "path":
            str(LOCK_PATH),

        "registry_path":
            str(REGISTRY_PATH),

        "registry_sha256":
            actual_registry_sha,

        "q01_q10_sha256":
            actual_q_sha,
    },

    "matrix": {

        "rows":
            len(evidence_matrix_df),

        "candidate_count":
            5,

        "question_count":
            10,

        "cardinality":
            "5 × 10 = 50",

        "schema":
            EXPECTED_MATRIX_COLUMNS,
    },

    "output": {

        "parquet":
            str(OUTPUT_PARQUET),

        "parquet_sha256":
            OUTPUT_PARQUET_SHA,

        "csv":
            str(OUTPUT_CSV),

        "csv_sha256":
            OUTPUT_CSV_SHA,
    },

    "firewall": {

        "source_extraction":
            "BLOCKED_AT_CELL14",

        "novelty_judgment":
            "NOT_ASSESSED",

        "equivalence_judgment":
            "NOT_ASSESSED",

        "overlap_class":
            "NOT_ASSESSED",

        "dataset_modification":
            "BLOCKED",

        "199_cohort_modification":
            "BLOCKED",

        "400_validation_modification":
            "BLOCKED",

        "model_inference":
            "BLOCKED",

        "model_training":
            "BLOCKED",

        "prompt_revision":
            "BLOCKED",

        "architecture_modification":
            "BLOCKED",
    },

    "path_policy": {

        "upstream_paths_source":
            "CELL13A_V002_LOCK",

        "legacy_cell14_artifact_consumed":
            False,

        "fallback_path_used":
            False,

        "fuzzy_matching_used":
            False,

        "path_guessing_used":
            False,

        "artifact_substitution_used":
            False,
    },

    "next_cell_contract":
        "CELL15_MUST_CONSUME_CELL14_V002_AND_CELL13A_V002_LOCK",
}

with open(
    OUTPUT_MANIFEST,
    "w",
    encoding="utf-8"
) as f:

    json.dump(
        manifest,
        f,
        ensure_ascii=False,
        indent=2
    )

print(f"Manifest : {OUTPUT_MANIFEST}")


# ================================================================================================================
# STEP 18 — INDEPENDENT READ-BACK
# ================================================================================================================

print("\n[STEP 18] INDEPENDENT READ-BACK")
print("-" * 120)

rb_df = pd.read_parquet(
    OUTPUT_PARQUET
)

print(f"Read-back rows    : {len(rb_df)}")
print(f"Read-back columns : {len(rb_df.columns)}")

if len(rb_df) != 50:
    raise RuntimeError(
        "STOP — Read-back row count mismatch."
    )

if rb_df.columns.tolist() != EXPECTED_MATRIX_COLUMNS:
    raise RuntimeError(
        "STOP — Read-back schema mismatch."
    )


# Identity

rb_ids = (
    rb_df["candidate_id"]
    .drop_duplicates()
    .tolist()
)

rb_qids = (
    rb_df["question_id"]
    .drop_duplicates()
    .tolist()
)

if rb_ids != EXPECTED_IDS:
    raise RuntimeError(
        "STOP — Read-back candidate identity mismatch."
    )

if rb_qids != expected_qids:
    raise RuntimeError(
        "STOP — Read-back QID mismatch."
    )

print("✓ Read-back identity = PASS")


# --------------------------------------------------------------------------------
# CRITICAL V002 CORRECTION:
#
# Do NOT use:
#
#     .all().all() is not True
#
# Instead explicitly convert the scalar boolean to Python bool.
# --------------------------------------------------------------------------------

readback_extraction_results = {}

for field in EXTRACTION_FIELDS:

    is_empty = (
        rb_df[field]
        .fillna("")
        .astype(str)
        .str.strip()
        .eq("")
        .all()
    )

    is_empty = bool(is_empty)

    readback_extraction_results[field] = is_empty

    print(
        f"{field:<30} : "
        f"{'EMPTY / PASS' if is_empty else 'NON-EMPTY / FAIL'}"
    )

    if not is_empty:
        raise RuntimeError(
            f"STOP — Read-back extraction firewall failure: {field}"
        )


print("\n✓ Read-back extraction firewall = PASS")


# Judgment read-back

for field in [
    "equivalence_judgment",
    "novelty_judgment",
    "overlap_class",
]:

    all_not_assessed = bool(
        rb_df[field]
        .fillna("")
        .astype(str)
        .eq("NOT_ASSESSED")
        .all()
    )

    print(
        f"{field:<30} : "
        f"{'NOT_ASSESSED / PASS' if all_not_assessed else 'FAIL'}"
    )

    if not all_not_assessed:
        raise RuntimeError(
            f"STOP — Read-back judgment firewall failure: {field}"
        )


print("✓ Read-back judgment firewall = PASS")


# Slot uniqueness read-back

rb_duplicate_count = int(
    rb_df
    .duplicated(
        subset=[
            "candidate_id",
            "question_id"
        ],
        keep=False
    )
    .sum()
)

print(
    f"\nRead-back duplicate slots : "
    f"{rb_duplicate_count}"
)

if rb_duplicate_count != 0:
    raise RuntimeError(
        "STOP — Read-back duplicate evidence slots."
    )

print("✓ Read-back slot uniqueness = PASS")


# ================================================================================================================
# STEP 19 — MANIFEST READ-BACK
# ================================================================================================================

print("\n[STEP 19] MANIFEST READ-BACK")
print("-" * 120)

with open(
    OUTPUT_MANIFEST,
    "r",
    encoding="utf-8"
) as f:

    manifest_rb = json.load(f)

if manifest_rb["status"] != (
    "CELL14_V002_EVIDENCE_MATRIX_INITIALIZED_PASS"
):
    raise RuntimeError(
        "STOP — V002 manifest status mismatch."
    )

if manifest_rb["upstream_lock"]["cell_id"] != (
    "CELL13A_CANONICAL_ARTIFACT_PATH_LOCK_V002"
):
    raise RuntimeError(
        "STOP — Wrong upstream lock in manifest."
    )

if manifest_rb["output"]["parquet_sha256"] != (
    OUTPUT_PARQUET_SHA
):
    raise RuntimeError(
        "STOP — Manifest Parquet SHA mismatch."
    )

if manifest_rb["output"]["csv_sha256"] != (
    OUTPUT_CSV_SHA
):
    raise RuntimeError(
        "STOP — Manifest CSV SHA mismatch."
    )

print("✓ Manifest status = PASS")
print("✓ Upstream lock = PASS")
print("✓ Output SHA records = PASS")


# ================================================================================================================
# STEP 20 — FINAL GO / NO-GO
# ================================================================================================================

print("\n" + "=" * 120)
print("CELL 14 V002 — FINAL GO / NO-GO")
print("=" * 120)

GO_GATES = {

    "scientific_value":
        SCIENTIFIC_VALUE,

    "cell13a_lock_exists":
        LOCK_PATH.exists(),

    "cell13a_lock_pass":
        lock["status"]
        == EXPECTED_LOCK_STATUS,

    "registry_sha_match":
        actual_registry_sha
        == EXPECTED_REGISTRY_SHA,

    "registry_schema_match":
        registry_df.columns.tolist()
        == EXPECTED_REGISTRY_COLUMNS,

    "registry_identity_match":
        actual_ids
        == EXPECTED_IDS,

    "primary_sources_verified":
        len(locked_sources) == 5
        and all(
            sha256_file(
                Path(x["source_path"])
            )
            == x["expected_sha256"]
            for x in locked_sources
        ),

    "q01_q10_locked":
        actual_q_sha
        == LOCKED_Q_SHA,

    "matrix_50_rows":
        len(evidence_matrix_df)
        == 50,

    "matrix_unique_slots":
        duplicate_count
        == 0,

    "extraction_firewall_prewrite":
        all(
            evidence_matrix_df[field]
            .fillna("")
            .astype(str)
            .str.strip()
            .eq("")
            .all()
            for field in EXTRACTION_FIELDS
        ),

    "judgment_firewall_prewrite":
        all(
            evidence_matrix_df[field]
            .fillna("")
            .astype(str)
            .eq("NOT_ASSESSED")
            .all()
            for field in [
                "equivalence_judgment",
                "novelty_judgment",
                "overlap_class",
            ]
        ),

    "readback_rows":
        len(rb_df) == 50,

    "readback_extraction_firewall":
        all(
            readback_extraction_results.values()
        ),

    "readback_judgment_firewall":
        all(
            bool(
                rb_df[field]
                .fillna("")
                .astype(str)
                .eq("NOT_ASSESSED")
                .all()
            )
            for field in [
                "equivalence_judgment",
                "novelty_judgment",
                "overlap_class",
            ]
        ),

    "readback_unique_slots":
        rb_duplicate_count == 0,

    "manifest_exists":
        OUTPUT_MANIFEST.exists(),

    "manifest_readback":
        manifest_rb["status"]
        == "CELL14_V002_EVIDENCE_MATRIX_INITIALIZED_PASS",
}


for gate, result in GO_GATES.items():

    print(
        f"{gate:<45} : "
        f"{'PASS' if result else 'FAIL'}"
    )


FINAL_PASS = all(
    GO_GATES.values()
)


print("\n" + "-" * 120)

if FINAL_PASS:

    print("✅ CELL 14 V002 — REBUILT EVIDENCE MATRIX = PASS")
    print()
    print("🔒 Rebuilt from CELL13A V002 canonical lock.")
    print("🔒 No legacy Cell14 artifact consumed.")
    print("🔒 5 canonical candidates = LOCKED.")
    print("🔒 Q01-Q10 = LOCKED.")
    print("🔒 50 unique evidence slots = VERIFIED.")
    print("🔒 Extraction fields = EMPTY after read-back.")
    print("🔒 Novelty = NOT_ASSESSED.")
    print("🔒 Equivalence = NOT_ASSESSED.")
    print("🔒 Overlap = NOT_ASSESSED.")
    print("🔒 No fallback / guessing / substitution.")
    print()
    print("🚦 GO — CELL 15 MAY BE BUILT FROM CELL 14 V002.")

else:

    print("❌ CELL 14 V002 — FAIL")
    print()
    print("🛑 STOP.")
    print("🛑 DO NOT PATCH V002.")
    print("🛑 DO NOT START CELL 15.")
    print("🛑 Identify trusted baseline and REBUILD a new version.")

print("=" * 120)
```

```python
# ================================================================================================================
# CELL 15 — CANONICAL SOURCE-GROUNDED EXTRACTION
# ================================================================================================================
#
# UPSTREAM:
#   CELL13A V002 CANONICAL LOCK
#   CELL14 V002 CANONICAL LITERATURE EVIDENCE MATRIX
#
# IMPORTANT:
#   This Cell 15 is built DIRECTLY from the actual Cell14 V002 schema.
#
#   NO legacy Cell15 artifact is consumed.
#   NO schema is inferred.
#   NO column is invented.
#   NO Cell14 modification.
#   NO Cell13A modification.
#   NO human judgment.
#   NO novelty/equivalence/overlap judgment.
#   NO Gold / Y+ / Y- / Preference.
#
# ================================================================================================================

from pathlib import Path
import pandas as pd
import json
import hashlib
from datetime import datetime, timezone
import sys


print("=" * 120)
print("CELL 15 — CANONICAL SOURCE-GROUNDED EXTRACTION")
print("=" * 120)


# ================================================================================================================
# STEP 0 — SCIENTIFIC VALUE GATE
# ================================================================================================================

print("\n[STEP 0] SCIENTIFIC VALUE GATE")
print("-" * 120)

SCIENTIFIC_VALUE = True

print("""
Purpose:
Create the source-grounded extraction layer for the frozen
50-slot methodological evidence-review matrix.

The extraction is strictly source-grounded:

CELL14 V002
    ↓
PRIMARY SOURCE PDF
    ↓
ORIGINAL PAGE TEXT
    ↓
BOUNDED LEXICAL LOCATOR
    ↓
EXACT SOURCE EXCERPT

No semantic reconstruction.
No interpretation.
No human judgment.
No novelty judgment.
No equivalence judgment.
No overlap judgment.
""")

print(f"SCIENTIFIC_VALUE = {SCIENTIFIC_VALUE}")

if not SCIENTIFIC_VALUE:
    raise RuntimeError(
        "STOP — Scientific value gate failed."
    )


# ================================================================================================================
# STEP 1 — HARD FIREWALL
# ================================================================================================================

print("\n[STEP 1] HARD FIREWALL")
print("-" * 120)

FIREWALL = {
    "dataset_modification": "BLOCKED",
    "199_cohort_modification": "BLOCKED",
    "400_validation_modification": "BLOCKED",
    "Q01_Q10_modification": "BLOCKED",
    "gold_generation": "BLOCKED",
    "y_plus_generation": "BLOCKED",
    "y_minus_generation": "BLOCKED",
    "preference_generation": "BLOCKED",
    "human_judgment": "BLOCKED",
    "model_inference": "BLOCKED",
    "prompt_revision": "BLOCKED",
    "architecture_modification": "BLOCKED",
    "semantic_reconstruction": "BLOCKED",
    "external_legal_knowledge": "BLOCKED",
    "legacy_cell15_consumption": "BLOCKED",
}

for key, value in FIREWALL.items():
    print(f"{key:<35} : {value}")


# ================================================================================================================
# STEP 2 — TRUSTED UPSTREAM PATHS
# ================================================================================================================

print("\n[STEP 2] TRUSTED UPSTREAM PATHS")
print("-" * 120)

X_ROOT = Path(
    "/content/drive/MyDrive/X-AutoAlign_Legal_RAG"
)

CHECKPOINT_ROOT = (
    X_ROOT
    / "04_AUDIT"
    / "NOVELTY_GAP_AUDIT"
    / "CHECKPOINTS"
)

CELL13A_MANIFEST = (
    CHECKPOINT_ROOT
    / "CELL13A_V002_CANONICAL_ARTIFACT_PATH_LOCK"
    / "CELL13A_CANONICAL_ARTIFACT_PATH_LOCK_V002.json"
)

CELL14_DIR = (
    CHECKPOINT_ROOT
    / "CELL14_REBUILT_V002_EVIDENCE_MATRIX"
)

CELL14_PARQUET = (
    CELL14_DIR
    / "LITERATURE_EVIDENCE_MATRIX_V002.parquet"
)

CELL14_CSV = (
    CELL14_DIR
    / "LITERATURE_EVIDENCE_MATRIX_V002.csv"
)

CELL14_MANIFEST = (
    CELL14_DIR
    / "LITERATURE_EVIDENCE_MATRIX_V002_MANIFEST.json"
)

# One official Cell15 output.
CELL15_DIR = (
    CHECKPOINT_ROOT
    / "CELL15_CANONICAL_SOURCE_GROUNDED_EXTRACTION"
)

print("Cell13A manifest:")
print(CELL13A_MANIFEST)

print("\nCell14 Parquet:")
print(CELL14_PARQUET)

print("\nCell14 CSV:")
print(CELL14_CSV)

print("\nCell14 manifest:")
print(CELL14_MANIFEST)

print("\nOfficial Cell15 output:")
print(CELL15_DIR)


# ================================================================================================================
# STEP 3 — OUTPUT SAFETY
# ================================================================================================================

print("\n[STEP 3] OUTPUT SAFETY GATE")
print("-" * 120)

if CELL15_DIR.exists():
    raise RuntimeError(
        "STOP — Official Cell15 output already exists.\n"
        "No overwrite. No patching. No reuse."
    )

print("✓ Official Cell15 output does not exist.")


# ================================================================================================================
# STEP 4 — UPSTREAM EXISTENCE
# ================================================================================================================

print("\n[STEP 4] UPSTREAM EXISTENCE GATE")
print("-" * 120)

required_upstream = [
    CELL13A_MANIFEST,
    CELL14_PARQUET,
    CELL14_CSV,
    CELL14_MANIFEST,
]

for path in required_upstream:

    if not path.exists():
        raise RuntimeError(
            f"STOP — Required trusted upstream artifact missing:\n{path}"
        )

    print(f"✓ FOUND: {path.name}")


# ================================================================================================================
# STEP 5 — CELL13A MANIFEST
# ================================================================================================================

print("\n[STEP 5] CELL13A V002 MANIFEST")
print("-" * 120)

with open(
    CELL13A_MANIFEST,
    "r",
    encoding="utf-8"
) as f:

    cell13a_lock = json.load(f)

EXPECTED_CELL13A_ID = (
    "CELL13A_CANONICAL_ARTIFACT_PATH_LOCK_V002"
)

EXPECTED_CELL13A_STATUS = (
    "CANONICAL_ARTIFACT_PATH_LOCK_PASS"
)

if cell13a_lock.get("cell_id") != EXPECTED_CELL13A_ID:
    raise RuntimeError(
        "STOP — Cell13A identity mismatch."
    )

if cell13a_lock.get("status") != EXPECTED_CELL13A_STATUS:
    raise RuntimeError(
        "STOP — Cell13A status is not PASS."
    )

print("✓ Cell13A identity = PASS")
print("✓ Cell13A status   = PASS")


# ================================================================================================================
# STEP 6 — CELL14 MANIFEST VERIFICATION
# ================================================================================================================

print("\n[STEP 6] CELL14 V002 MANIFEST VERIFICATION")
print("-" * 120)

with open(
    CELL14_MANIFEST,
    "r",
    encoding="utf-8"
) as f:

    cell14_manifest = json.load(f)

EXPECTED_CELL14_ID = (
    "CELL14_CANONICAL_LITERATURE_EVIDENCE_MATRIX_V002"
)

EXPECTED_CELL14_STATUS = (
    "CELL14_V002_EVIDENCE_MATRIX_INITIALIZED_PASS"
)

if cell14_manifest.get("cell_id") != EXPECTED_CELL14_ID:
    raise RuntimeError(
        "STOP — Cell14 manifest identity mismatch."
    )

if cell14_manifest.get("status") != EXPECTED_CELL14_STATUS:
    raise RuntimeError(
        "STOP — Cell14 manifest status mismatch."
    )

print("✓ Cell14 identity = PASS")
print("✓ Cell14 status   = PASS")


# ================================================================================================================
# STEP 7 — HASH FUNCTION
# ================================================================================================================

def sha256_file(path, chunk_size=1024 * 1024):

    h = hashlib.sha256()

    with open(path, "rb") as f:

        while True:

            chunk = f.read(chunk_size)

            if not chunk:
                break

            h.update(chunk)

    return h.hexdigest()


# ================================================================================================================
# STEP 8 — CELL14 SHA VERIFICATION
# ================================================================================================================

print("\n[STEP 8] CELL14 V002 SHA-256")
print("-" * 120)

EXPECTED_CELL14_SHA = (
    "4179472725911ff4fbbc07bd8366153f120a34551b7d8dd27e43724640d95b4d"
)

actual_cell14_sha = sha256_file(
    CELL14_PARQUET
)

print("Expected:")
print(EXPECTED_CELL14_SHA)

print("\nActual:")
print(actual_cell14_sha)

if actual_cell14_sha != EXPECTED_CELL14_SHA:
    raise RuntimeError(
        "STOP — Cell14 V002 SHA mismatch."
    )

print("✓ Cell14 V002 SHA = PASS")


# ================================================================================================================
# STEP 9 — LOAD ACTUAL CELL14 V002
# ================================================================================================================

print("\n[STEP 9] LOAD CELL14 V002")
print("-" * 120)

df14 = pd.read_parquet(
    CELL14_PARQUET
)

print("Rows    :", len(df14))
print("Columns :", len(df14.columns))

if len(df14) != 50:
    raise RuntimeError(
        "STOP — Cell14 V002 does not contain exactly 50 rows."
    )

print("✓ 50 rows = PASS")


# ================================================================================================================
# STEP 10 — ACTUAL CELL14 SCHEMA CONTRACT
# ================================================================================================================

print("\n[STEP 10] ACTUAL CELL14 V002 SCHEMA CONTRACT")
print("-" * 120)

# EXACT schema from the Cell14 V002 code supplied by the user.
CELL14_COLUMNS = [
    "candidate_id",
    "question_id",
    "title",
    "authors",
    "year",
    "venue",
    "primary_source",
    "primary_source_filename",
    "primary_source_path",
    "primary_source_sha256",
    "question_text",
    "source_excerpt",
    "source_location",
    "source_text_status",
    "source_text_unit_start",
    "source_text_unit_end",
    "extraction_method",
    "extraction_interpretation",
    "extraction_status",
    "source_paraphrase",
    "methodological_note",
    "equivalence_judgment",
    "novelty_judgment",
    "overlap_class",
]

print("Expected columns:", len(CELL14_COLUMNS))
print("Actual columns  :", len(df14.columns))

if df14.columns.tolist() != CELL14_COLUMNS:
    raise RuntimeError(
        "STOP — Actual Cell14 V002 schema does not match "
        "the schema produced by Cell14 V002."
    )

print("✓ 24-column schema = PASS")


# ================================================================================================================
# STEP 11 — FROZEN CANDIDATES
# ================================================================================================================

print("\n[STEP 11] CANONICAL CANDIDATES")
print("-" * 120)

EXPECTED_IDS = [
    "ANCHOR-001",
    "LIT-001",
    "LIT-002",
    "LIT-003",
    "LIT-004",
]

actual_ids = (
    df14["candidate_id"]
    .drop_duplicates()
    .tolist()
)

print("Expected:", EXPECTED_IDS)
print("Actual  :", actual_ids)

if actual_ids != EXPECTED_IDS:
    raise RuntimeError(
        "STOP — Candidate identity/order mismatch."
    )

print("✓ Candidate identity = PASS")


# ================================================================================================================
# STEP 12 — FROZEN QUESTIONS
# ================================================================================================================

print("\n[STEP 12] FROZEN Q01-Q10")
print("-" * 120)

question_frame = cell13a_lock["question_frame"]

LOCKED_QUESTIONS = question_frame["questions"]

if question_frame["count"] != 10:
    raise RuntimeError(
        "STOP — Expected exactly 10 locked questions."
    )

expected_qids = [
    q["question_id"]
    for q in LOCKED_QUESTIONS
]

actual_qids = (
    df14["question_id"]
    .drop_duplicates()
    .tolist()
)

print("Expected:", expected_qids)
print("Actual  :", actual_qids)

if actual_qids != expected_qids:
    raise RuntimeError(
        "STOP — Q01-Q10 identity/order mismatch."
    )

for q in LOCKED_QUESTIONS:

    qid = q["question_id"]
    expected_text = q["question_text"]

    observed = (
        df14[
            df14["question_id"] == qid
        ]["question_text"]
        .drop_duplicates()
        .tolist()
    )

    if observed != [expected_text]:
        raise RuntimeError(
            f"STOP — Question text mismatch: {qid}"
        )

print("✓ Q01-Q10 identity = PASS")
print("✓ Q01-Q10 text integrity = PASS")


# ================================================================================================================
# STEP 13 — CELL14 FIREWALL VERIFICATION
# ================================================================================================================

print("\n[STEP 13] CELL14 FIREWALL VERIFICATION")
print("-" * 120)

EXTRACTION_FIELDS = [
    "source_excerpt",
    "source_location",
    "source_text_status",
    "source_text_unit_start",
    "source_text_unit_end",
    "extraction_method",
    "extraction_interpretation",
    "extraction_status",
    "source_paraphrase",
    "methodological_note",
]

for field in EXTRACTION_FIELDS:

    non_empty = int(
        df14[field]
        .fillna("")
        .astype(str)
        .str.strip()
        .ne("")
        .sum()
    )

    print(
        f"{field:<30} : {non_empty}"
    )

    if non_empty != 0:
        raise RuntimeError(
            f"STOP — Cell14 extraction firewall violated: {field}"
        )


for field in [
    "equivalence_judgment",
    "novelty_judgment",
    "overlap_class",
]:

    values = (
        df14[field]
        .fillna("")
        .astype(str)
        .value_counts()
        .to_dict()
    )

    print(
        f"{field:<30} : {values}"
    )

    if values != {"NOT_ASSESSED": 50}:
        raise RuntimeError(
            f"STOP — Cell14 judgment firewall violated: {field}"
        )

print("✓ Cell14 extraction firewall = PASS")
print("✓ Cell14 judgment firewall = PASS")


# ================================================================================================================
# STEP 14 — PRIMARY SOURCE MAP
# ================================================================================================================

print("\n[STEP 14] PRIMARY SOURCE MAP")
print("-" * 120)

locked_sources = cell13a_lock[
    "locked_primary_sources"
]

if len(locked_sources) != 5:
    raise RuntimeError(
        "STOP — Expected exactly 5 locked primary sources."
    )

source_lookup = {
    x["candidate_id"]: x
    for x in locked_sources
}

RESOLVED_SOURCES = {}

for candidate_id in EXPECTED_IDS:

    source = source_lookup.get(candidate_id)

    if source is None:
        raise RuntimeError(
            f"STOP — No locked source for {candidate_id}."
        )

    path = Path(
        source["source_path"]
    )

    expected_filename = source[
        "primary_source_filename"
    ]

    expected_sha = source[
        "expected_sha256"
    ]

    if not path.exists():
        raise RuntimeError(
            f"STOP — Locked source missing: {candidate_id}"
        )

    if path.name != expected_filename:
        raise RuntimeError(
            f"STOP — Source filename drift: {candidate_id}"
        )

    actual_sha = sha256_file(path)

    print(f"\n{candidate_id}")
    print("Filename :", path.name)
    print("Expected :", expected_sha)
    print("Actual   :", actual_sha)

    if actual_sha != expected_sha:
        raise RuntimeError(
            f"STOP — Source SHA mismatch: {candidate_id}"
        )

    RESOLVED_SOURCES[candidate_id] = path

print("\n✓ 5 / 5 canonical sources verified")


# ================================================================================================================
# STEP 15 — FROZEN LEXICAL LOCATOR
# ================================================================================================================

print("\n[STEP 15] FROZEN LEXICAL LOCATOR")
print("-" * 120)

LEXICAL_TERMS = {

    "Q01": [
        "verification",
        "verifier",
        "verify",
        "evidence",
        "retrieval",
    ],

    "Q02": [
        "input",
        "query",
        "question",
        "context",
        "evidence",
    ],

    "Q03": [
        "relevance",
        "sufficiency",
        "support",
    ],

    "Q04": [
        "legal",
        "domain",
        "support",
        "evidence",
    ],

    "Q05": [
        "sufficiency",
        "sufficient",
        "insufficient",
    ],

    "Q06": [
        "contradiction",
        "conflict",
        "risk",
        "uncertainty",
    ],

    "Q07": [
        "generation",
        "answer",
        "answering",
        "before",
    ],

    "Q08": [
        "abstain",
        "abstention",
        "eligibility",
        "risk",
        "control",
    ],

    "Q09": [
        "human",
        "annotation",
        "annotator",
        "human evaluation",
    ],

    "Q10": [
        "evidence",
        "verification",
        "sufficiency",
        "relevance",
    ],
}

if set(LEXICAL_TERMS.keys()) != set(expected_qids):
    raise RuntimeError(
        "STOP — Locator frame does not cover exactly Q01-Q10."
    )

print("✓ Locator frame covers exactly Q01-Q10")


# ================================================================================================================
# STEP 16 — SOURCE-GROUNDED EXTRACTION FUNCTION
# ================================================================================================================

print("\n[STEP 16] SOURCE-GROUNDED EXTRACTION FUNCTION")
print("-" * 120)

try:
    import fitz
except Exception as e:
    raise RuntimeError(
        "STOP — PyMuPDF/fitz unavailable."
    ) from e

print("✓ fitz import = PASS")


def normalize_for_locator(text):

    return " ".join(
        str(text).split()
    ).lower()


def extract_source_grounded(
    pdf_path,
    terms
):

    doc = fitz.open(
        pdf_path
    )

    try:

        for page_index in range(
            len(doc)
        ):

            page = doc[
                page_index
            ]

            raw_text = page.get_text(
                "text"
            )

            if not raw_text:
                continue

            normalized = normalize_for_locator(
                raw_text
            )

            hits = [
                term
                for term in terms
                if term.lower() in normalized
            ]

            if hits:

                return {
                    "page_number": page_index + 1,
                    "excerpt": raw_text,
                    "hits": hits,
                    "status": "SOURCE_LOCATED",
                }

        return {
            "page_number": "",
            "excerpt": "",
            "hits": [],
            "status": "NO_LEXICAL_HIT",
        }

    finally:

        doc.close()


# ================================================================================================================
# STEP 17 — WORKING COPY
# ================================================================================================================

print("\n[STEP 17] CREATE WORKING COPY")
print("-" * 120)

WORKING = df14.copy()

# Preserve the exact 24-column schema.
# No new column is added.

for field in EXTRACTION_FIELDS:

    if field in [
        "source_excerpt",
        "source_location",
        "source_text_status",
        "source_text_unit_start",
        "source_text_unit_end",
        "extraction_method",
        "extraction_interpretation",
        "extraction_status",
        "source_paraphrase",
        "methodological_note",
    ]:

        WORKING[field] = ""


# ================================================================================================================
# STEP 18 — SOURCE-GROUNDED EXTRACTION
# ================================================================================================================

print("\n[STEP 18] SOURCE-GROUNDED EXTRACTION")
print("-" * 120)

for idx, row in WORKING.iterrows():

    candidate_id = str(
        row["candidate_id"]
    )

    question_id = str(
        row["question_id"]
    )

    pdf_path = RESOLVED_SOURCES[
        candidate_id
    ]

    result = extract_source_grounded(
        pdf_path,
        LEXICAL_TERMS[question_id]
    )

    if result["status"] == "SOURCE_LOCATED":

        WORKING.at[
            idx,
            "source_excerpt"
        ] = result["excerpt"]

        WORKING.at[
            idx,
            "source_location"
        ] = (
            f"{pdf_path.name} | "
            f"page {result['page_number']}"
        )

        WORKING.at[
            idx,
            "source_text_status"
        ] = "ORIGINAL_PAGE_TEXT"

        WORKING.at[
            idx,
            "source_text_unit_start"
        ] = str(
            result["page_number"]
        )

        WORKING.at[
            idx,
            "source_text_unit_end"
        ] = str(
            result["page_number"]
        )

        WORKING.at[
            idx,
            "extraction_method"
        ] = "BOUNDED_LEXICAL_PAGE_LOCATOR"

        WORKING.at[
            idx,
            "extraction_interpretation"
        ] = "NONE"

        WORKING.at[
            idx,
            "extraction_status"
        ] = "SOURCE_LOCATED"

        WORKING.at[
            idx,
            "source_paraphrase"
        ] = ""

        WORKING.at[
            idx,
            "methodological_note"
        ] = ""

    else:

        WORKING.at[
            idx,
            "extraction_status"
        ] = "NO_LEXICAL_HIT"

print("✓ Extraction completed")


# ================================================================================================================
# STEP 19 — STRUCTURAL QC
# ================================================================================================================

print("\n[STEP 19] STRUCTURAL QC")
print("-" * 120)

if len(WORKING) != 50:
    raise RuntimeError(
        "STOP — Row count changed."
    )

if WORKING.columns.tolist() != CELL14_COLUMNS:
    raise RuntimeError(
        "STOP — Cell15 changed the canonical Cell14 schema."
    )

duplicate_count = int(
    WORKING
    .duplicated(
        subset=[
            "candidate_id",
            "question_id"
        ],
        keep=False
    )
    .sum()
)

if duplicate_count != 0:
    raise RuntimeError(
        "STOP — Duplicate candidate/question slots."
    )

print("✓ 50 rows")
print("✓ 24 columns")
print("✓ 50 unique slots")


# ================================================================================================================
# STEP 20 — SOURCE PROVENANCE QC
# ================================================================================================================

print("\n[STEP 20] SOURCE PROVENANCE QC")
print("-" * 120)

located = 0
no_hit = 0

for idx, row in WORKING.iterrows():

    status = str(
        row["extraction_status"]
    )

    candidate_id = str(
        row["candidate_id"]
    )

    pdf_path = RESOLVED_SOURCES[
        candidate_id
    ]

    if status == "SOURCE_LOCATED":

        located += 1

        page_number = int(
            row["source_text_unit_start"]
        )

        excerpt = str(
            row["source_excerpt"]
        )

        doc = fitz.open(
            pdf_path
        )

        try:

            original_page_text = doc[
                page_number - 1
            ].get_text("text")

        finally:

            doc.close()

        if excerpt not in original_page_text:

            raise RuntimeError(
                "STOP — Stored excerpt is not an exact "
                "substring of the original PDF page."
            )

    elif status == "NO_LEXICAL_HIT":

        no_hit += 1

    else:

        raise RuntimeError(
            f"STOP — Invalid extraction status: {status}"
        )

print("SOURCE_LOCATED :", located)
print("NO_LEXICAL_HIT :", no_hit)

if located + no_hit != 50:
    raise RuntimeError(
        "STOP — Extraction status accounting mismatch."
    )

print("✓ Provenance QC = PASS")


# ================================================================================================================
# STEP 21 — JUDGMENT FIREWALL
# ================================================================================================================

print("\n[STEP 21] JUDGMENT FIREWALL")
print("-" * 120)

for field in [
    "equivalence_judgment",
    "novelty_judgment",
    "overlap_class",
]:

    values = (
        WORKING[field]
        .fillna("")
        .astype(str)
        .value_counts()
        .to_dict()
    )

    print(
        f"{field:<30} : {values}"
    )

    if values != {"NOT_ASSESSED": 50}:
        raise RuntimeError(
            f"STOP — Judgment field changed: {field}"
        )

print("✓ No judgment generated")


# ================================================================================================================
# STEP 22 — UPSTREAM IMMUTABILITY
# ================================================================================================================

print("\n[STEP 22] UPSTREAM IMMUTABILITY")
print("-" * 120)

IMMUTABLE_FIELDS = [
    "candidate_id",
    "question_id",
    "title",
    "authors",
    "year",
    "venue",
    "primary_source",
    "primary_source_filename",
    "primary_source_path",
    "primary_source_sha256",
    "question_text",
]

for field in IMMUTABLE_FIELDS:

    if not (
        WORKING[field]
        .fillna("")
        .astype(str)
        .tolist()
        ==
        df14[field]
        .fillna("")
        .astype(str)
        .tolist()
    ):

        raise RuntimeError(
            f"STOP — Upstream field modified: {field}"
        )

print("✓ Upstream identity/content fields unchanged")


# ================================================================================================================
# STEP 23 — PERSIST CELL15
# ================================================================================================================

print("\n[STEP 23] PERSIST CELL15")
print("-" * 120)

CELL15_DIR.mkdir(
    parents=True,
    exist_ok=False
)

OUTPUT_PARQUET = (
    CELL15_DIR
    / "LITERATURE_EVIDENCE_MATRIX_CELL15.parquet"
)

OUTPUT_CSV = (
    CELL15_DIR
    / "LITERATURE_EVIDENCE_MATRIX_CELL15.csv"
)

OUTPUT_MANIFEST = (
    CELL15_DIR
    / "CELL15_MANIFEST.json"
)

WORKING.to_parquet(
    OUTPUT_PARQUET,
    index=False
)

WORKING.to_csv(
    OUTPUT_CSV,
    index=False,
    encoding="utf-8-sig"
)

print("✓ Parquet written")
print("✓ CSV written")


# ================================================================================================================
# STEP 24 — HASH OUTPUTS
# ================================================================================================================

print("\n[STEP 24] OUTPUT SHA-256")
print("-" * 120)

OUTPUT_PARQUET_SHA = sha256_file(
    OUTPUT_PARQUET
)

OUTPUT_CSV_SHA = sha256_file(
    OUTPUT_CSV
)

print("Parquet SHA :", OUTPUT_PARQUET_SHA)
print("CSV SHA     :", OUTPUT_CSV_SHA)


# ================================================================================================================
# STEP 25 — INDEPENDENT READ-BACK
# ================================================================================================================

print("\n[STEP 25] INDEPENDENT READ-BACK")
print("-" * 120)

READBACK = pd.read_parquet(
    OUTPUT_PARQUET
)

print("Rows    :", len(READBACK))
print("Columns :", len(READBACK.columns))

if len(READBACK) != 50:
    raise RuntimeError(
        "STOP — Read-back row count mismatch."
    )

if READBACK.columns.tolist() != CELL14_COLUMNS:
    raise RuntimeError(
        "STOP — Read-back schema mismatch."
    )

print("✓ Read-back schema = PASS")


# ================================================================================================================
# STEP 26 — READ-BACK EXTRACTION FIREWALL
# ================================================================================================================

print("\n[STEP 26] READ-BACK EXTRACTION QC")
print("-" * 120)

for field in EXTRACTION_FIELDS:

    if field == "source_excerpt":

        # SOURCE_LOCATED rows must contain source text.
        located_mask = (
            READBACK["extraction_status"]
            == "SOURCE_LOCATED"
        )

        if located_mask.any():

            empty_located = (
                READBACK.loc[
                    located_mask,
                    field
                ]
                .fillna("")
                .astype(str)
                .str.strip()
                .eq("")
                .sum()
            )

            if empty_located != 0:
                raise RuntimeError(
                    "STOP — SOURCE_LOCATED row has empty source_excerpt."
                )

    # NO_LEXICAL_HIT rows may legitimately have empty extraction fields.


print("✓ Read-back extraction state = PASS")


# ================================================================================================================
# STEP 27 — READ-BACK JUDGMENT FIREWALL
# ================================================================================================================

print("\n[STEP 27] READ-BACK JUDGMENT FIREWALL")
print("-" * 120)

for field in [
    "equivalence_judgment",
    "novelty_judgment",
    "overlap_class",
]:

    valid = bool(
        READBACK[field]
        .fillna("")
        .astype(str)
        .eq("NOT_ASSESSED")
        .all()
    )

    print(
        f"{field:<30} : "
        f"{'PASS' if valid else 'FAIL'}"
    )

    if not valid:
        raise RuntimeError(
            f"STOP — Read-back judgment firewall failed: {field}"
        )

print("✓ Read-back judgment firewall = PASS")


# ================================================================================================================
# STEP 28 — FINAL OUTPUT HASH STABILITY
# ================================================================================================================

print("\n[STEP 28] FINAL HASH STABILITY")
print("-" * 120)

FINAL_PARQUET_SHA = sha256_file(
    OUTPUT_PARQUET
)

FINAL_CSV_SHA = sha256_file(
    OUTPUT_CSV
)

if FINAL_PARQUET_SHA != OUTPUT_PARQUET_SHA:
    raise RuntimeError(
        "STOP — Parquet SHA changed."
    )

if FINAL_CSV_SHA != OUTPUT_CSV_SHA:
    raise RuntimeError(
        "STOP — CSV SHA changed."
    )

print("✓ Parquet SHA stable")
print("✓ CSV SHA stable")


# ================================================================================================================
# STEP 29 — CELL15 MANIFEST
# ================================================================================================================

print("\n[STEP 29] CELL15 MANIFEST")
print("-" * 120)

manifest = {

    "experiment_id":
        "FDR-NOVELTY-GAP-001",

    "cell_id":
        "CELL15_CANONICAL_SOURCE_GROUNDED_EXTRACTION",

    "status":
        "CELL15_SOURCE_GROUNDED_EXTRACTION_PASS",

    "created_utc":
        datetime.now(
            timezone.utc
        ).isoformat(),

    "upstream": {

        "cell13a_manifest":
            str(CELL13A_MANIFEST),

        "cell14_parquet":
            str(CELL14_PARQUET),

        "cell14_parquet_sha256":
            actual_cell14_sha,

        "cell14_schema_columns":
            CELL14_COLUMNS,

        "cell14_rows":
            50,
    },

    "population": {

        "rows":
            50,

        "candidates":
            5,

        "questions":
            10,

        "unique_slots":
            50,
    },

    "extraction": {

        "source_located":
            located,

        "no_lexical_hit":
            no_hit,

        "semantic_reconstruction":
            False,

        "paraphrase":
            False,

        "human_judgment":
            False,

        "gold_generation":
            False,

    },

    "firewall":
        FIREWALL,

    "output": {

        "parquet":
            str(OUTPUT_PARQUET),

        "parquet_sha256":
            OUTPUT_PARQUET_SHA,

        "csv":
            str(OUTPUT_CSV),

        "csv_sha256":
            OUTPUT_CSV_SHA,
    },

    "legacy_cell15_consumed":
        False,

    "upstream_modified":
        False,
}


with open(
    OUTPUT_MANIFEST,
    "w",
    encoding="utf-8"
) as f:

    json.dump(
        manifest,
        f,
        ensure_ascii=False,
        indent=2
    )

print("✓ Cell15 manifest written")


# ================================================================================================================
# STEP 30 — MANIFEST READ-BACK
# ================================================================================================================

print("\n[STEP 30] MANIFEST READ-BACK")
print("-" * 120)

with open(
    OUTPUT_MANIFEST,
    "r",
    encoding="utf-8"
) as f:

    manifest_rb = json.load(f)

if manifest_rb["status"] != (
    "CELL15_SOURCE_GROUNDED_EXTRACTION_PASS"
):

    raise RuntimeError(
        "STOP — Cell15 manifest status mismatch."
    )

if manifest_rb["upstream"]["cell14_parquet_sha256"] != (
    actual_cell14_sha
):

    raise RuntimeError(
        "STOP — Cell15 manifest upstream SHA mismatch."
    )

if manifest_rb["upstream"]["cell14_schema_columns"] != (
    CELL14_COLUMNS
):

    raise RuntimeError(
        "STOP — Cell15 manifest schema mismatch."
    )

print("✓ Manifest status = PASS")
print("✓ Upstream SHA = PASS")
print("✓ Upstream schema = PASS")


# ================================================================================================================
# STEP 31 — FINAL GO / NO-GO
# ================================================================================================================

print("\n" + "=" * 120)
print("CELL 15 — FINAL GO / NO-GO")
print("=" * 120)

print()
print("Scientific Value              : PASS")
print("Cell13A V002                  : VERIFIED")
print("Cell14 V002                   : VERIFIED")
print("Cell14 SHA                    : PASS")
print("Cell14 Schema                 : 24 / 24 VERIFIED")
print("Candidates                    : 5 / 5")
print("Questions                     : 10 / 10")
print("Evidence Slots                : 50 / 50")
print("Primary Sources               : 5 / 5 VERIFIED")
print("PyMuPDF / fitz                : PASS")
print("Source-Grounded Extraction    : PASS")
print("Exact Page Provenance         : PASS")
print("Upstream Immutability         : PASS")
print("Judgment Firewall             : PASS")
print("Read-Back                     : PASS")
print("SHA Stability                 : PASS")
print("Legacy Cell15 Consumption     : BLOCKED")
print("No Patching                   : PASS")
print()

print("FINAL STATUS:")
print()
print("CELL 15 = PASS")
print()
print("GO — CELL 16 MAY START.")
print()
print("=" * 120)
print("CELL 15 — END")
print("=" * 120)
```

```python
# ================================================================================================================
# CELL 16 — SOURCE-GROUNDED EXTRACTION QC
# ================================================================================================================
#
# UPSTREAM:
#   CELL 15 — CANONICAL SOURCE-GROUNDED EXTRACTION
#
# PURPOSE:
#   Audit the mechanically extracted source-grounded evidence produced by Cell 15.
#
# THIS CELL DOES:
#   ✓ Read Cell15 canonical artifact
#   ✓ Verify Cell15 manifest
#   ✓ Verify row/schema/identity integrity
#   ✓ Verify SOURCE_LOCATED / NO_LEXICAL_HIT accounting
#   ✓ Verify source excerpts remain non-empty where located
#   ✓ Verify excerpts are exact substrings of the original PDF page text
#   ✓ Produce a human source-review queue
#
# THIS CELL DOES NOT:
#   ✗ Perform novelty judgment
#   ✗ Perform equivalence judgment
#   ✗ Perform overlap classification
#   ✗ Determine legal support
#   ✗ Determine methodological support
#   ✗ Generate Gold
#   ✗ Generate Y+ / Y-
#   ✗ Generate preference labels
#   ✗ Perform adjudication
#   ✗ Modify Cell15
#   ✗ Modify Cell14
#   ✗ Modify dataset
#
# NO PATCHING
# NO GUESSING
# NO LEGACY CELL16 CONSUMPTION
# ================================================================================================================

from pathlib import Path
import pandas as pd
import json
import hashlib
import re
from datetime import datetime, timezone


print("=" * 120)
print("CELL 16 — SOURCE-GROUNDED EXTRACTION QC")
print("=" * 120)


# ================================================================================================================
# STEP 0 — SCIENTIFIC VALUE GATE
# ================================================================================================================

print("\n[STEP 0] SCIENTIFIC VALUE GATE")
print("-" * 120)

SCIENTIFIC_VALUE = True

print("""
Scientific purpose:

Independently QC the source-grounded extraction produced by the current
canonical Cell 15 before any human methodological comparison.

The unit remains:

(candidate_id, question_id, source-grounded evidence)

This cell verifies extraction integrity only.

A NO_LEXICAL_HIT result is NOT interpreted as:
  • evidence absence
  • methodological absence
  • non-equivalence
  • novelty
  • negative evidence
""")

print("Scientific value:", "PASS" if SCIENTIFIC_VALUE else "FAIL")

if not SCIENTIFIC_VALUE:
    raise RuntimeError(
        "STOP — Scientific value gate failed."
    )


# ================================================================================================================
# STEP 1 — HARD FIREWALL
# ================================================================================================================

print("\n[STEP 1] HARD FIREWALL")
print("-" * 120)

FIREWALL = {
    "dataset_modification": "BLOCKED",
    "199_cohort_modification": "BLOCKED",
    "400_validation_reopen": "BLOCKED",
    "Q01_Q10_modification": "BLOCKED",
    "gold_generation": "BLOCKED",
    "y_plus_generation": "BLOCKED",
    "y_minus_generation": "BLOCKED",
    "preference_generation": "BLOCKED",
    "adjudication": "BLOCKED",
    "novelty_judgment": "BLOCKED",
    "equivalence_judgment": "BLOCKED",
    "overlap_classification": "BLOCKED",
    "legal_support_judgment": "BLOCKED",
    "methodological_inference": "BLOCKED",
    "architecture_modification": "BLOCKED",
    "prompt_revision": "BLOCKED",
    "model_training": "BLOCKED",
    "model_inference": "BLOCKED",
    "legacy_cell16_consumption": "BLOCKED",
    "cell15_modification": "BLOCKED",
}

for key, value in FIREWALL.items():
    print(f"{key:<35} : {value}")

if not all(v == "BLOCKED" for v in FIREWALL.values()):
    raise RuntimeError(
        "STOP — Firewall configuration is not fully closed."
    )


# ================================================================================================================
# STEP 2 — CURRENT CELL15 PATH
# ================================================================================================================

print("\n[STEP 2] CURRENT CELL15 PATH")
print("-" * 120)

PROJECT_ROOT = Path(
    "/content/drive/MyDrive/X-AutoAlign_Legal_RAG"
)

CHECKPOINT_ROOT = (
    PROJECT_ROOT
    / "04_AUDIT"
    / "NOVELTY_GAP_AUDIT"
    / "CHECKPOINTS"
)

CELL15_DIR = (
    CHECKPOINT_ROOT
    / "CELL15_CANONICAL_SOURCE_GROUNDED_EXTRACTION"
)

INPUT_PARQUET = (
    CELL15_DIR
    / "LITERATURE_EVIDENCE_MATRIX_CELL15.parquet"
)

INPUT_CSV = (
    CELL15_DIR
    / "LITERATURE_EVIDENCE_MATRIX_CELL15.csv"
)

INPUT_MANIFEST = (
    CELL15_DIR
    / "CELL15_MANIFEST.json"
)

CELL16_DIR = (
    CHECKPOINT_ROOT
    / "CELL16_SOURCE_GROUNDED_EXTRACTION_QC"
)

print("Cell15 directory:")
print(CELL15_DIR)

print("\nCell15 Parquet:")
print(INPUT_PARQUET)

print("\nCell15 CSV:")
print(INPUT_CSV)

print("\nCell15 Manifest:")
print(INPUT_MANIFEST)

print("\nCell16 output directory:")
print(CELL16_DIR)


# ================================================================================================================
# STEP 3 — OUTPUT SAFETY
# ================================================================================================================

print("\n[STEP 3] OUTPUT SAFETY")
print("-" * 120)

if CELL16_DIR.exists():
    raise RuntimeError(
        "STOP — Official Cell16 output directory already exists.\n"
        "No overwrite. No patching. No reuse."
    )

print("✓ Cell16 output directory does not exist")


# ================================================================================================================
# STEP 4 — CELL15 EXISTENCE
# ================================================================================================================

print("\n[STEP 4] CELL15 EXISTENCE GATE")
print("-" * 120)

for path in [
    INPUT_PARQUET,
    INPUT_CSV,
    INPUT_MANIFEST,
]:

    if not path.exists():
        raise RuntimeError(
            f"STOP — Required Cell15 artifact missing:\n{path}"
        )

    print("✓ FOUND:", path.name)


# ================================================================================================================
# STEP 5 — CELL15 MANIFEST
# ================================================================================================================

print("\n[STEP 5] CELL15 MANIFEST")
print("-" * 120)

with open(
    INPUT_MANIFEST,
    "r",
    encoding="utf-8"
) as f:

    cell15_manifest = json.load(f)

EXPECTED_CELL15_ID = (
    "CELL15_CANONICAL_SOURCE_GROUNDED_EXTRACTION"
)

EXPECTED_CELL15_STATUS = (
    "CELL15_SOURCE_GROUNDED_EXTRACTION_PASS"
)

if cell15_manifest.get("cell_id") != EXPECTED_CELL15_ID:
    raise RuntimeError(
        "STOP — Cell15 manifest identity mismatch."
    )

if cell15_manifest.get("status") != EXPECTED_CELL15_STATUS:
    raise RuntimeError(
        "STOP — Cell15 manifest status is not PASS."
    )

print("✓ Cell15 identity = PASS")
print("✓ Cell15 status   = PASS")


# ================================================================================================================
# STEP 6 — HASH FUNCTION
# ================================================================================================================

def sha256_file(path, chunk_size=1024 * 1024):

    h = hashlib.sha256()

    with open(path, "rb") as f:

        while True:

            chunk = f.read(chunk_size)

            if not chunk:
                break

            h.update(chunk)

    return h.hexdigest()


# ================================================================================================================
# STEP 7 — CELL15 INPUT HASH
# ================================================================================================================

print("\n[STEP 7] CELL15 INPUT HASH")
print("-" * 120)

CURRENT_CELL15_SHA = sha256_file(
    INPUT_PARQUET
)

MANIFEST_CELL15_UPSTREAM_SHA = (
    cell15_manifest
    .get("output", {})
    .get("parquet_sha256")
)

print("Actual Cell15 Parquet SHA:")
print(CURRENT_CELL15_SHA)

print("\nManifest-recorded SHA:")
print(MANIFEST_CELL15_UPSTREAM_SHA)

if (
    MANIFEST_CELL15_UPSTREAM_SHA is not None
    and CURRENT_CELL15_SHA != MANIFEST_CELL15_UPSTREAM_SHA
):

    raise RuntimeError(
        "STOP — Cell15 Parquet SHA does not match its manifest."
    )

print("✓ Cell15 input SHA = PASS")


# ================================================================================================================
# STEP 8 — LOAD CELL15
# ================================================================================================================

print("\n[STEP 8] LOAD CURRENT CELL15")
print("-" * 120)

df = pd.read_parquet(
    INPUT_PARQUET
)

print("Rows    :", len(df))
print("Columns :", len(df.columns))

if len(df) != 50:
    raise RuntimeError(
        f"STOP — Current Cell15 must contain 50 rows; found {len(df)}."
    )

print("✓ 50 rows")


# ================================================================================================================
# STEP 9 — ACTUAL CELL15 SCHEMA
# ================================================================================================================

print("\n[STEP 9] CELL15 SCHEMA")
print("-" * 120)

EXPECTED_COLUMNS = [
    "candidate_id",
    "question_id",
    "title",
    "authors",
    "year",
    "venue",
    "primary_source",
    "primary_source_filename",
    "primary_source_path",
    "primary_source_sha256",
    "question_text",
    "source_excerpt",
    "source_location",
    "source_text_status",
    "source_text_unit_start",
    "source_text_unit_end",
    "extraction_method",
    "extraction_interpretation",
    "extraction_status",
    "source_paraphrase",
    "methodological_note",
    "equivalence_judgment",
    "novelty_judgment",
    "overlap_class",
]

if df.columns.tolist() != EXPECTED_COLUMNS:
    raise RuntimeError(
        "STOP — Current Cell15 schema differs from the verified 24-column contract."
    )

print("✓ 24 / 24 columns verified")


# ================================================================================================================
# STEP 10 — SLOT INTEGRITY
# ================================================================================================================

print("\n[STEP 10] SLOT INTEGRITY")
print("-" * 120)

EXPECTED_CANDIDATES = [
    "ANCHOR-001",
    "LIT-001",
    "LIT-002",
    "LIT-003",
    "LIT-004",
]

EXPECTED_QUESTIONS = [
    f"Q{i:02d}"
    for i in range(1, 11)
]

actual_candidates = (
    df["candidate_id"]
    .drop_duplicates()
    .tolist()
)

actual_questions = (
    df["question_id"]
    .drop_duplicates()
    .tolist()
)

print("Candidates:")
print(actual_candidates)

print("\nQuestions:")
print(actual_questions)

if actual_candidates != EXPECTED_CANDIDATES:
    raise RuntimeError(
        "STOP — Candidate identity/order changed."
    )

if actual_questions != EXPECTED_QUESTIONS:
    raise RuntimeError(
        "STOP — Question identity/order changed."
    )

duplicate_slots = int(
    df[
        ["candidate_id", "question_id"]
    ].duplicated().sum()
)

if duplicate_slots != 0:
    raise RuntimeError(
        f"STOP — Duplicate slots detected: {duplicate_slots}"
    )

print("\n✓ Candidate identity = PASS")
print("✓ Q01-Q10 identity = PASS")
print("✓ 50 unique slots = PASS")


# ================================================================================================================
# STEP 11 — EXTRACTION STATUS AUDIT
# ================================================================================================================

print("\n[STEP 11] EXTRACTION STATUS AUDIT")
print("-" * 120)

status_counts = (
    df["extraction_status"]
    .astype(str)
    .value_counts()
    .sort_index()
)

print(status_counts.to_string())

ALLOWED_STATUSES = {
    "SOURCE_LOCATED",
    "NO_LEXICAL_HIT",
}

if not set(status_counts.index).issubset(
    ALLOWED_STATUSES
):

    raise RuntimeError(
        "STOP — Unexpected extraction status detected."
    )

SOURCE_LOCATED = (
    df["extraction_status"]
    .astype(str)
    .eq("SOURCE_LOCATED")
)

NO_LEXICAL_HIT = (
    df["extraction_status"]
    .astype(str)
    .eq("NO_LEXICAL_HIT")
)

located_count = int(
    SOURCE_LOCATED.sum()
)

no_hit_count = int(
    NO_LEXICAL_HIT.sum()
)

print("\nSOURCE_LOCATED :", located_count)
print("NO_LEXICAL_HIT :", no_hit_count)

if located_count + no_hit_count != 50:
    raise RuntimeError(
        "STOP — Extraction status accounting does not equal 50."
    )

# CRITICAL:
# We derive the expected counts from the ACTUAL current Cell15.
# We do not import the legacy 45/5 assumption.
print("\n✓ Current Cell15 extraction accounting = PASS")


# ================================================================================================================
# STEP 12 — SOURCE LOCATED EXCERPT QC
# ================================================================================================================

print("\n[STEP 12] SOURCE_LOCATED EXCERPT QC")
print("-" * 120)

located_df = df.loc[
    SOURCE_LOCATED
].copy()

def clean_text(value):

    if pd.isna(value):
        return ""

    return re.sub(
        r"\s+",
        " ",
        str(value)
    ).strip()


located_df["_qc_excerpt"] = (
    located_df["source_excerpt"]
    .map(clean_text)
)

empty_excerpt_count = int(
    located_df["_qc_excerpt"]
    .eq("")
    .sum()
)

print("SOURCE_LOCATED rows :", len(located_df))
print("Empty excerpts      :", empty_excerpt_count)

if empty_excerpt_count != 0:
    raise RuntimeError(
        "STOP — SOURCE_LOCATED row contains empty source_excerpt."
    )

print("✓ All SOURCE_LOCATED rows contain source text")


# ================================================================================================================
# STEP 13 — PLACEHOLDER QC
# ================================================================================================================

print("\n[STEP 13] PLACEHOLDER QC")
print("-" * 120)

PLACEHOLDER_PATTERNS = [
    r"^N/?A$",
    r"^NONE$",
    r"^NOT FOUND$",
    r"^NO LEXICAL HIT$",
    r"^PLACEHOLDER$",
]

placeholder_mask = pd.Series(
    False,
    index=located_df.index
)

for pattern in PLACEHOLDER_PATTERNS:

    placeholder_mask |= (
        located_df["_qc_excerpt"]
        .str.upper()
        .str.match(
            pattern,
            na=False
        )
    )

placeholder_count = int(
    placeholder_mask.sum()
)

print("Placeholder excerpts:", placeholder_count)

if placeholder_count != 0:
    raise RuntimeError(
        "STOP — Placeholder detected in SOURCE_LOCATED excerpt."
    )

print("✓ No placeholder excerpts")


# ================================================================================================================
# STEP 14 — EXACT SOURCE PROVENANCE
# ================================================================================================================

print("\n[STEP 14] EXACT SOURCE PROVENANCE")
print("-" * 120)

try:
    import fitz
except Exception as e:
    raise RuntimeError(
        "STOP — PyMuPDF/fitz unavailable."
    ) from e

print("✓ fitz import = PASS")


for idx, row in located_df.iterrows():

    source_path = Path(
        str(row["primary_source_path"])
    )

    if not source_path.exists():
        raise RuntimeError(
            "STOP — Source path missing:\n"
            f"{source_path}"
        )

    page_start = str(
        row["source_text_unit_start"]
    ).strip()

    page_end = str(
        row["source_text_unit_end"]
    ).strip()

    if not page_start or not page_end:
        raise RuntimeError(
            "STOP — SOURCE_LOCATED row lacks source-text unit boundary."
        )

    if page_start != page_end:
        raise RuntimeError(
            "STOP — Current Cell15 source unit is not page-bounded."
        )

    try:
        page_number = int(
            float(page_start)
        )
    except Exception:
        raise RuntimeError(
            "STOP — Source-text unit boundary is not numeric."
        )

    excerpt = str(
        row["source_excerpt"]
    )

    doc = fitz.open(
        source_path
    )

    try:

        if page_number < 1 or page_number > len(doc):
            raise RuntimeError(
                f"STOP — Invalid page number {page_number} "
                f"for {source_path.name}"
            )

        original_page_text = doc[
            page_number - 1
        ].get_text("text")

    finally:

        doc.close()

    if excerpt not in original_page_text:
        raise RuntimeError(
            "STOP — source_excerpt is not an exact substring "
            f"of original PDF page.\n"
            f"Candidate: {row['candidate_id']}\n"
            f"Question : {row['question_id']}\n"
            f"Page     : {page_number}"
        )

print("✓ All SOURCE_LOCATED excerpts verified against original PDF pages")


# ================================================================================================================
# STEP 15 — EXTRACTION METHOD FIREWALL
# ================================================================================================================

print("\n[STEP 15] EXTRACTION METHOD FIREWALL")
print("-" * 120)

located_methods = (
    located_df["extraction_method"]
    .fillna("")
    .astype(str)
    .unique()
    .tolist()
)

print("Observed extraction methods:")
for method in located_methods:
    print(" •", method)

if any(
    "SEMANTIC" in method.upper()
    or "PARAPHRASE" in method.upper()
    or "LLM" in method.upper()
    for method in located_methods
):

    raise RuntimeError(
        "STOP — Semantic/LLM-based extraction detected."
    )

print("✓ No semantic/LLM extraction detected")


# ================================================================================================================
# STEP 16 — JUDGMENT FIREWALL
# ================================================================================================================

print("\n[STEP 16] JUDGMENT FIREWALL")
print("-" * 120)

for field in [
    "equivalence_judgment",
    "novelty_judgment",
    "overlap_class",
]:

    values = (
        df[field]
        .fillna("")
        .astype(str)
        .value_counts()
        .to_dict()
    )

    print(
        f"{field:<30}: {values}"
    )

    if values != {
        "NOT_ASSESSED": 50
    }:

        raise RuntimeError(
            f"STOP — Judgment firewall violated: {field}"
        )

print("✓ No equivalence judgment")
print("✓ No novelty judgment")
print("✓ No overlap classification")


# ================================================================================================================
# STEP 17 — HUMAN SOURCE REVIEW QUEUE
# ================================================================================================================

print("\n[STEP 17] HUMAN SOURCE REVIEW QUEUE")
print("-" * 120)

review_rows = []

for idx, row in df.iterrows():

    status = str(
        row["extraction_status"]
    )

    if status == "NO_LEXICAL_HIT":

        review_rows.append({
            "row_index": idx,
            "candidate_id": row["candidate_id"],
            "question_id": row["question_id"],
            "primary_source": row["primary_source"],
            "review_reason": "NO_LEXICAL_HIT",
            "review_status": "PENDING_SOURCE_REVIEW",
        })

print("NO_LEXICAL_HIT review rows:", len(review_rows))

if len(review_rows) != no_hit_count:
    raise RuntimeError(
        "STOP — Review queue accounting mismatch."
    )

print("✓ No-hit rows explicitly surfaced")
print("✓ No-hit rows are NOT treated as negative evidence")


# ================================================================================================================
# STEP 18 — Q-BY-Q COVERAGE
# ================================================================================================================

print("\n[STEP 18] Q01-Q10 COVERAGE")
print("-" * 120)

coverage = (
    df.assign(
        _located=SOURCE_LOCATED.astype(int),
        _no_hit=NO_LEXICAL_HIT.astype(int),
    )
    .groupby("question_id")
    .agg(
        slots=("question_id", "size"),
        source_located=("_located", "sum"),
        no_lexical_hit=("_no_hit", "sum"),
    )
    .reindex(EXPECTED_QUESTIONS)
)

print(
    coverage.to_string()
)

if not coverage["slots"].eq(5).all():
    raise RuntimeError(
        "STOP — Every question must have exactly 5 candidate slots."
    )

print("\n✓ Every frozen question has exactly 5 slots")


# ================================================================================================================
# STEP 19 — CANDIDATE × QUESTION COVERAGE
# ================================================================================================================

print("\n[STEP 19] CANDIDATE × QUESTION COVERAGE")
print("-" * 120)

candidate_q = (
    df.assign(
        _located=SOURCE_LOCATED.astype(int)
    )
    .pivot_table(
        index="candidate_id",
        columns="question_id",
        values="_located",
        aggfunc="sum",
        fill_value=0,
    )
    .reindex(
        index=EXPECTED_CANDIDATES,
        columns=EXPECTED_QUESTIONS,
        fill_value=0,
    )
)

print(
    candidate_q.to_string()
)

if candidate_q.shape != (5, 10):
    raise RuntimeError(
        "STOP — Candidate × question matrix changed."
    )

print("\n✓ 5 × 10 matrix preserved")


# ================================================================================================================
# STEP 20 — NO AUTOMATIC TARGET INTERPRETATION
# ================================================================================================================

print("\n[STEP 20] INTERPRETATION FIREWALL")
print("-" * 120)

print("""
This Cell 16 intentionally does NOT introduce a new lexical/topic
target classifier.

Reason:

Cell15 has already produced the source-grounded extraction.

At this stage, automatically deciding whether an excerpt is:
  • relevant
  • sufficient
  • legally supportive
  • equivalent
  • novel
  • contradictory

would constitute methodological interpretation.

Such judgments require the separately controlled human-review stage.
""")

print("✓ No automatic methodological interpretation")


# ================================================================================================================
# STEP 21 — QC STATUS
# ================================================================================================================

print("\n[STEP 21] FINAL QC STATUS")
print("-" * 120)

mechanical_qc_pass = (
    len(df) == 50
    and len(df.columns) == 24
    and located_count + no_hit_count == 50
    and empty_excerpt_count == 0
    and placeholder_count == 0
    and candidate_q.shape == (5, 10)
)

if not mechanical_qc_pass:
    raise RuntimeError(
        "STOP — Mechanical extraction QC failed."
    )

QC_STATUS = (
    "EXTRACTION_QC_PASS_PENDING_HUMAN_SOURCE_REVIEW"
)

print("✓ 50 frozen slots preserved")
print(f"✓ {located_count} SOURCE_LOCATED preserved")
print(f"✓ {no_hit_count} NO_LEXICAL_HIT preserved")
print("✓ Exact source provenance verified")
print("✓ No automatic methodological judgment")
print()
print("STATUS:", QC_STATUS)


# ================================================================================================================
# STEP 22 — CREATE OUTPUT DIRECTORY
# ================================================================================================================

print("\n[STEP 22] PERSIST QC OUTPUTS")
print("-" * 120)

CELL16_DIR.mkdir(
    parents=True,
    exist_ok=False
)

OUTPUT_REVIEW = (
    CELL16_DIR
    / "CELL16_SOURCE_REVIEW_QUEUE.csv"
)

OUTPUT_COVERAGE = (
    CELL16_DIR
    / "CELL16_EXTRACTION_COVERAGE.csv"
)

OUTPUT_REPORT = (
    CELL16_DIR
    / "CELL16_EXTRACTION_QC_REPORT.json"
)

review_df = pd.DataFrame(
    review_rows
)

review_df.to_csv(
    OUTPUT_REVIEW,
    index=False,
    encoding="utf-8-sig"
)

coverage.to_csv(
    OUTPUT_COVERAGE,
    encoding="utf-8-sig"
)

print("✓ Review queue written")
print("✓ Coverage table written")


# ================================================================================================================
# STEP 23 — REPORT
# ================================================================================================================

print("\n[STEP 23] QC REPORT")
print("-" * 120)

report = {

    "cell":
        "CELL16",

    "status":
        QC_STATUS,

    "scientific_value":
        True,

    "input":
        str(INPUT_PARQUET),

    "input_sha256":
        CURRENT_CELL15_SHA,

    "rows":
        50,

    "columns":
        24,

    "candidate_count":
        5,

    "question_count":
        10,

    "unique_slots":
        50,

    "source_located":
        located_count,

    "no_lexical_hit":
        no_hit_count,

    "human_source_review_queue":
        len(review_df),

    "mechanical_qc_pass":
        bool(mechanical_qc_pass),

    "methodological_judgment":
        "NOT_PERFORMED",

    "novelty_judgment":
        "NOT_PERFORMED",

    "equivalence_judgment":
        "NOT_PERFORMED",

    "overlap_classification":
        "NOT_PERFORMED",

    "firewall":
        FIREWALL,

    "outputs": {
        "review_queue":
            str(OUTPUT_REVIEW),
        "coverage":
            str(OUTPUT_COVERAGE),
    },

    "generated_utc":
        datetime.now(
            timezone.utc
        ).isoformat(),
}

with open(
    OUTPUT_REPORT,
    "w",
    encoding="utf-8"
) as f:

    json.dump(
        report,
        f,
        ensure_ascii=False,
        indent=2
    )

print("✓ QC report written")


# ================================================================================================================
# STEP 24 — READ-BACK
# ================================================================================================================

print("\n[STEP 24] READ-BACK")
print("-" * 120)

review_rb = pd.read_csv(
    OUTPUT_REVIEW,
    encoding="utf-8-sig"
)

coverage_rb = pd.read_csv(
    OUTPUT_COVERAGE,
    index_col=0,
    encoding="utf-8-sig"
)

with open(
    OUTPUT_REPORT,
    "r",
    encoding="utf-8"
) as f:

    report_rb = json.load(f)

if len(review_rb) != no_hit_count:
    raise RuntimeError(
        "STOP — Review queue read-back mismatch."
    )

if report_rb["rows"] != 50:
    raise RuntimeError(
        "STOP — Report row count mismatch."
    )

if report_rb["source_located"] != located_count:
    raise RuntimeError(
        "STOP — Report SOURCE_LOCATED mismatch."
    )

if report_rb["no_lexical_hit"] != no_hit_count:
    raise RuntimeError(
        "STOP — Report NO_LEXICAL_HIT mismatch."
    )

if report_rb["mechanical_qc_pass"] is not True:
    raise RuntimeError(
        "STOP — Report mechanical QC state mismatch."
    )

print("✓ Review queue read-back = PASS")
print("✓ Coverage read-back = PASS")
print("✓ Report read-back = PASS")


# ================================================================================================================
# STEP 25 — FINAL GO / NO-GO
# ================================================================================================================

print("\n" + "=" * 120)
print("CELL 16 — FINAL GO / NO-GO")
print("=" * 120)

print()
print("Scientific Value                 : PASS")
print("Cell15 Input                    : VERIFIED")
print("Cell15 SHA                      : PASS")
print("Rows                            : 50 / 50")
print("Columns                         : 24 / 24")
print("Candidates                      : 5 / 5")
print("Questions                       : 10 / 10")
print("Unique Slots                    : 50 / 50")
print(f"SOURCE_LOCATED                  : {located_count} / 50")
print(f"NO_LEXICAL_HIT                 : {no_hit_count} / 50")
print("Exact Source Provenance         : PASS")
print("Judgment Firewall              : PASS")
print("Methodological Interpretation  : BLOCKED")
print("Novelty Judgment               : BLOCKED")
print("Equivalence Judgment           : BLOCKED")
print("Overlap Classification         : BLOCKED")
print("Legacy Cell16 Consumption      : BLOCKED")
print("Read-Back                      : PASS")
print()

print("FINAL STATUS:")
print()
print("CELL 16 = EXTRACTION_QC_PASS_PENDING_HUMAN_SOURCE_REVIEW")
print()
print("NEXT AUTHORIZED STAGE:")
print("Human source-level review of the explicitly surfaced review queue.")
print()
print("=" * 120)
print("CELL 16 — END")
print("=" * 120)
```

```python
# ================================================================================================================
# CELL 16A V001 — HUMAN SOURCE CORRESPONDENCE REVIEW
# ================================================================================================================
#
# SCIENTIFIC ROLE
# ---------------
# Human source-level review of the explicitly surfaced Cell16 review queue.
#
# This cell does NOT determine:
#   • novelty
#   • equivalence
#   • overlap
#   • legal support
#   • sufficiency
#   • Gold
#   • Y+
#   • Y-
#   • preference
#   • methodological superiority
#
# It only prepares a controlled human-review artifact for source correspondence.
#
# NO PATCHING.
# NO GUESSING.
# NO LEGACY CELL16A.
# ================================================================================================================

from pathlib import Path
import pandas as pd
import json
import hashlib
import re
import shutil
from datetime import datetime


# ================================================================================================================
# STEP 0 — SCIENTIFIC VALUE GATE
# ================================================================================================================

SCIENTIFIC_VALUE = True

SCIENTIFIC_PURPOSE = """
Humanly inspect the explicitly surfaced Cell16 review queue to determine,
for each queued slot, whether the primary source contains material
corresponding to the frozen Q01-Q10 extraction target.

This is scientifically necessary because Cell16 verified extraction
integrity and provenance, but intentionally did not make methodological
interpretations.

Human review is therefore the controlled boundary between:
    source-grounded extraction
and
    later methodological comparison.

NO automatic correspondence decision is permitted.
"""

print("=" * 112)
print("CELL 16A V001 — HUMAN SOURCE CORRESPONDENCE REVIEW")
print("=" * 112)
print()
print("[STEP 0] SCIENTIFIC VALUE GATE")
print("-" * 112)
print(SCIENTIFIC_PURPOSE)
print("Scientific value: PASS")


# ================================================================================================================
# STEP 1 — HARD FIREWALL
# ================================================================================================================

FIREWALL = {
    "dataset_modification": "BLOCKED",
    "199_cohort_modification": "BLOCKED",
    "400_validation_reopen": "BLOCKED",
    "Q01_Q10_modification": "BLOCKED",
    "gold_generation": "BLOCKED",
    "y_plus_generation": "BLOCKED",
    "y_minus_generation": "BLOCKED",
    "preference_generation": "BLOCKED",
    "automatic_correspondence": "BLOCKED",
    "automatic_relevance": "BLOCKED",
    "automatic_sufficiency": "BLOCKED",
    "legal_support_judgment": "BLOCKED",
    "novelty_judgment": "BLOCKED",
    "equivalence_judgment": "BLOCKED",
    "overlap_classification": "BLOCKED",
    "methodological_inference": "BLOCKED",
    "architecture_modification": "BLOCKED",
    "prompt_revision": "BLOCKED",
    "model_training": "BLOCKED",
    "model_inference": "BLOCKED",
    "legacy_cell16_consumption": "BLOCKED",
    "legacy_cell16a_consumption": "BLOCKED",
    "cell15_modification": "BLOCKED",
    "cell16_modification": "BLOCKED",
}

print()
print("[STEP 1] HARD FIREWALL")
print("-" * 112)

for k, v in FIREWALL.items():
    print(f"{k:<38}: {v}")


# ================================================================================================================
# STEP 2 — CANONICAL ROOTS
# ================================================================================================================

ROOT = Path("/content/drive/MyDrive/X-AutoAlign_Legal_RAG")

CELL15_DIR = (
    ROOT
    / "04_AUDIT"
    / "NOVELTY_GAP_AUDIT"
    / "CHECKPOINTS"
    / "CELL15_CANONICAL_SOURCE_GROUNDED_EXTRACTION"
)

CELL16_DIR = (
    ROOT
    / "04_AUDIT"
    / "NOVELTY_GAP_AUDIT"
    / "CHECKPOINTS"
    / "CELL16_SOURCE_GROUNDED_EXTRACTION_QC"
)

CELL16A_DIR = (
    ROOT
    / "04_AUDIT"
    / "NOVELTY_GAP_AUDIT"
    / "CHECKPOINTS"
    / "CELL16A_HUMAN_SOURCE_CORRESPONDENCE_REVIEW_V001"
)

CELL15_PARQUET = (
    CELL15_DIR /
    "LITERATURE_EVIDENCE_MATRIX_CELL15.parquet"
)

CELL15_MANIFEST = (
    CELL15_DIR /
    "CELL15_MANIFEST.json"
)

print()
print("[STEP 2] CANONICAL PATHS")
print("-" * 112)

print("Cell15:")
print(CELL15_PARQUET)

print("Cell16:")
print(CELL16_DIR)

print("Cell16A output:")
print(CELL16A_DIR)


# ================================================================================================================
# STEP 3 — OUTPUT SAFETY
# ================================================================================================================

print()
print("[STEP 3] OUTPUT SAFETY")
print("-" * 112)

if CELL16A_DIR.exists():
    raise RuntimeError(
        "STOP — Cell16A output directory already exists. "
        "No patching or overwrite is permitted. "
        "Create a new V002 only after reviewing this artifact."
    )

print("✓ Cell16A output directory does not exist")


# ================================================================================================================
# STEP 4 — CELL15 EXISTENCE / INTEGRITY
# ================================================================================================================

print()
print("[STEP 4] CELL15 EXISTENCE GATE")
print("-" * 112)

required_cell15 = [
    CELL15_PARQUET,
    CELL15_MANIFEST,
]

for p in required_cell15:
    if not p.exists():
        raise FileNotFoundError(f"STOP — Required Cell15 artifact missing: {p}")
    print(f"✓ FOUND: {p.name}")


# ================================================================================================================
# STEP 5 — CELL15 HASH LOCK
# ================================================================================================================

EXPECTED_CELL15_SHA = (
    "4f919b7d611272e8604b3ff62e90ed09cf9e8f44bd1e21a90feb5f29509ecea3"
)

def sha256_file(path):
    h = hashlib.sha256()
    with open(path, "rb") as f:
        for block in iter(lambda: f.read(1024 * 1024), b""):
            h.update(block)
    return h.hexdigest()


actual_cell15_sha = sha256_file(CELL15_PARQUET)

print()
print("[STEP 5] CELL15 HASH LOCK")
print("-" * 112)
print("Expected SHA:")
print(EXPECTED_CELL15_SHA)
print("Actual SHA:")
print(actual_cell15_sha)

if actual_cell15_sha != EXPECTED_CELL15_SHA:
    raise RuntimeError(
        "STOP — Cell15 SHA mismatch. "
        "Do not patch or continue."
    )

print("✓ Cell15 SHA = PASS")


# ================================================================================================================
# STEP 6 — LOAD CURRENT CELL15
# ================================================================================================================

print()
print("[STEP 6] LOAD CURRENT CELL15")
print("-" * 112)

df15 = pd.read_parquet(CELL15_PARQUET)

print("Rows   :", len(df15))
print("Columns:", len(df15.columns))

if len(df15) != 50:
    raise RuntimeError(
        f"STOP — Expected 50 Cell15 rows, found {len(df15)}."
    )

print("✓ 50 rows")


# ================================================================================================================
# STEP 7 — FROZEN IDENTITY CHECK
# ================================================================================================================

EXPECTED_CANDIDATES = [
    "ANCHOR-001",
    "LIT-001",
    "LIT-002",
    "LIT-003",
    "LIT-004",
]

EXPECTED_QUESTIONS = [
    "Q01", "Q02", "Q03", "Q04", "Q05",
    "Q06", "Q07", "Q08", "Q09", "Q10",
]

expected_slots = {
    (c, q)
    for c in EXPECTED_CANDIDATES
    for q in EXPECTED_QUESTIONS
}

actual_slots = set(
    zip(df15["candidate_id"], df15["question_id"])
)

print()
print("[STEP 7] FROZEN SLOT IDENTITY")
print("-" * 112)

if actual_slots != expected_slots:
    raise RuntimeError(
        "STOP — Cell15 candidate/question identity does not match "
        "the frozen 5×10 structure."
    )

print("✓ Candidate identity = PASS")
print("✓ Q01-Q10 identity = PASS")
print("✓ 50 unique slots = PASS")


# ================================================================================================================
# STEP 8 — CONFIRM CELL16 OUTPUT EXISTS
# ================================================================================================================

print()
print("[STEP 8] CELL16 OUTPUT GATE")
print("-" * 112)

if not CELL16_DIR.exists():
    raise RuntimeError(
        "STOP — Current Cell16 output directory not found."
    )

print("✓ Cell16 output directory found")


# ================================================================================================================
# STEP 9 — DISCOVER CELL16 REVIEW QUEUE
# ================================================================================================================
#
# Controlled discovery is used here instead of guessing a filename.
# We only accept a uniquely identifiable review-queue artifact.
# ================================================================================================================

print()
print("[STEP 9] CELL16 REVIEW QUEUE DISCOVERY")
print("-" * 112)

candidate_queue_files = sorted(
    [
        p for p in CELL16_DIR.iterdir()
        if p.is_file()
        and "review" in p.name.lower()
        and "queue" in p.name.lower()
        and p.suffix.lower() in [".csv", ".parquet"]
    ]
)

if len(candidate_queue_files) == 0:
    raise RuntimeError(
        "STOP — No uniquely identifiable Cell16 review queue found."
    )

if len(candidate_queue_files) > 1:
    print("Candidate review-queue files:")
    for p in candidate_queue_files:
        print("  •", p.name)

    raise RuntimeError(
        "STOP — Multiple Cell16 review-queue candidates found. "
        "No filename guessing is permitted."
    )

CELL16_QUEUE = candidate_queue_files[0]

print("✓ Review queue:")
print(CELL16_QUEUE)


# ================================================================================================================
# STEP 10 — LOAD CELL16 REVIEW QUEUE
# ================================================================================================================

print()
print("[STEP 10] LOAD CELL16 REVIEW QUEUE")
print("-" * 112)

if CELL16_QUEUE.suffix.lower() == ".csv":
    queue = pd.read_csv(CELL16_QUEUE)
elif CELL16_QUEUE.suffix.lower() == ".parquet":
    queue = pd.read_parquet(CELL16_QUEUE)
else:
    raise RuntimeError("STOP — Unsupported review queue format.")

print("Rows   :", len(queue))
print("Columns:", len(queue.columns))

if len(queue) != 3:
    raise RuntimeError(
        f"STOP — Current Cell16 explicitly surfaced {len(queue)} review rows, "
        "but Cell16A expects exactly the explicit 3-row queue."
    )

print("✓ Exactly 3 explicitly surfaced review rows")


# ================================================================================================================
# STEP 11 — VERIFY QUEUE AGAINST CURRENT CELL15
# ================================================================================================================

print()
print("[STEP 11] REVIEW QUEUE ↔ CELL15 INTEGRITY")
print("-" * 112)

required_queue_columns = [
    "candidate_id",
    "question_id",
]

for col in required_queue_columns:
    if col not in queue.columns:
        raise RuntimeError(
            f"STOP — Required Cell16 queue column missing: {col}"
        )

queue_slots = set(
    zip(queue["candidate_id"], queue["question_id"])
)

cell15_no_hit_slots = set(
    zip(
        df15.loc[
            df15["extraction_status"] == "NO_LEXICAL_HIT",
            "candidate_id"
        ],
        df15.loc[
            df15["extraction_status"] == "NO_LEXICAL_HIT",
            "question_id"
        ],
    )
)

if queue_slots != cell15_no_hit_slots:
    raise RuntimeError(
        "STOP — Cell16 review queue does not exactly match "
        "the current Cell15 NO_LEXICAL_HIT population."
    )

print("✓ Queue exactly matches current NO_LEXICAL_HIT population")


# ================================================================================================================
# STEP 12 — CURRENT FROZEN QUESTION TEXT
# ================================================================================================================

FROZEN_QUESTIONS = {
    "Q01": "Does the work perform post-retrieval evidence verification?",
    "Q02": "What is the input to the verification/control mechanism?",
    "Q03": "Is evidence relevance assessed independently from evidence sufficiency/support?",
    "Q04": "Is legal support or domain-specific evidentiary support explicitly assessed?",
    "Q05": "Is evidence sufficiency explicitly assessed?",
    "Q06": "Is contradiction, conflict, or risk explicitly assessed?",
    "Q07": "Does the mechanism operate before generation or answering?",
    "Q08": "Does the mechanism produce an evidence eligibility, risk, or abstention control?",
    "Q09": "Is human-grounded evidence suitability part of the evaluated methodology?",
    "Q10": "Is the mechanism equivalent to the proposed F-D+R evidence-suitability verification stage?",
}


# ================================================================================================================
# STEP 13 — BUILD HUMAN REVIEW ARTIFACT
# ================================================================================================================

print()
print("[STEP 13] BUILD HUMAN REVIEW ARTIFACT")
print("-" * 112)

# Merge ONLY source-grounded information from current Cell15.
# No interpretation is added.

review = queue.copy()

# Remove any pre-existing decision columns from the source queue.
# They are not trusted as human decisions.
decision_columns = [
    "correspondence_decision",
    "reviewer_note",
    "reviewer_id",
    "review_timestamp",
]

for col in decision_columns:
    if col in review.columns:
        review = review.drop(columns=[col])

cell15_cols = [
    "candidate_id",
    "question_id",
    "question_text",
    "primary_source",
    "source_excerpt",
    "source_location",
    "extraction_status",
    "extraction_method",
    "source_text_status",
]

available_cell15_cols = [
    c for c in cell15_cols
    if c in df15.columns
]

cell15_subset = df15[available_cell15_cols].copy()

review = review[
    ["candidate_id", "question_id"]
].merge(
    cell15_subset,
    on=["candidate_id", "question_id"],
    how="left",
    validate="one_to_one",
)

# Enforce frozen question text.
review["question_text"] = review["question_id"].map(FROZEN_QUESTIONS)

# Human decisions MUST start PENDING.
review["correspondence_decision"] = "PENDING"
review["reviewer_note"] = ""
review["reviewer_id"] = ""
review["review_timestamp"] = ""

# Review scope is explicit.
review["review_scope"] = (
    "SOURCE_LEVEL_CORRESPONDENCE_ONLY"
)

review["methodological_judgment_allowed"] = False


# ================================================================================================================
# STEP 14 — REVIEW DECISION VOCABULARY
# ================================================================================================================

review["allowed_human_decisions"] = (
    "CORRESPONDS | DOES_NOT_CORRESPOND | UNCERTAIN"
)

print("Human decision states:")
print("  PENDING")
print("  CORRESPONDS")
print("  DOES_NOT_CORRESPOND")
print("  UNCERTAIN")

print()
print("✓ No automatic correspondence decisions assigned")


# ================================================================================================================
# STEP 15 — HUMAN REVIEW FIREWALL
# ================================================================================================================

print()
print("[STEP 15] HUMAN REVIEW FIREWALL")
print("-" * 112)

# Every decision must be PENDING at construction time.
if not (review["correspondence_decision"] == "PENDING").all():
    raise RuntimeError(
        "STOP — Automatic correspondence decision detected."
    )

# No methodological fields are created.
for forbidden in [
    "novelty_judgment",
    "equivalence_judgment",
    "overlap_class",
    "legal_support_judgment",
    "sufficiency_judgment",
    "risk_judgment",
    "gold_label",
    "y_plus",
    "y_minus",
]:
    if forbidden in review.columns:
        raise RuntimeError(
            f"STOP — Forbidden methodological field present: {forbidden}"
        )

print("✓ All correspondence decisions = PENDING")
print("✓ No methodological judgments")
print("✓ No Gold/Y+/Y-/preference")
print("✓ Human-only decision boundary preserved")


# ================================================================================================================
# STEP 16 — NO-LEXICAL-HIT FIREWALL
# ================================================================================================================

print()
print("[STEP 16] NO_LEXICAL_HIT FIREWALL")
print("-" * 112)

if not (review["extraction_status"] == "NO_LEXICAL_HIT").all():
    raise RuntimeError(
        "STOP — Cell16A queue contains a non-NO_LEXICAL_HIT row."
    )

print("✓ All 3 rows are NO_LEXICAL_HIT")
print("✓ NO_LEXICAL_HIT is preserved")
print("✓ NO_LEXICAL_HIT is NOT interpreted as evidence absence")
print("✓ Human review remains open")


# ================================================================================================================
# STEP 17 — HUMAN REVIEW INSTRUCTIONS
# ================================================================================================================

review_instructions = pd.DataFrame({
    "instruction_id": [
        "H01",
        "H02",
        "H03",
        "H04",
        "H05",
    ],
    "instruction": [
        "Inspect the identified primary source directly.",
        "Use the frozen question text exactly as displayed.",
        "Determine only whether source material corresponds to the extraction target.",
        "Do not judge novelty, equivalence, legal support, sufficiency, contradiction, or methodological superiority.",
        "If correspondence cannot be established confidently, select UNCERTAIN.",
    ]
})


# ================================================================================================================
# STEP 18 — PERSIST OUTPUTS
# ================================================================================================================

print()
print("[STEP 18] PERSIST CELL16A OUTPUTS")
print("-" * 112)

CELL16A_DIR.mkdir(parents=False, exist_ok=False)

REVIEW_CSV = CELL16A_DIR / "CELL16A_HUMAN_REVIEW_QUEUE_V001.csv"
REVIEW_PARQUET = CELL16A_DIR / "CELL16A_HUMAN_REVIEW_QUEUE_V001.parquet"
INSTRUCTIONS_CSV = CELL16A_DIR / "CELL16A_HUMAN_REVIEW_INSTRUCTIONS_V001.csv"
MANIFEST_JSON = CELL16A_DIR / "CELL16A_MANIFEST_V001.json"

review.to_csv(REVIEW_CSV, index=False)
review.to_parquet(REVIEW_PARQUET, index=False)
review_instructions.to_csv(INSTRUCTIONS_CSV, index=False)

print("✓ Review CSV written")
print("✓ Review Parquet written")
print("✓ Review instructions written")


# ================================================================================================================
# STEP 19 — MANIFEST
# ================================================================================================================

manifest = {
    "cell": "CELL 16A",
    "version": "V001",
    "status": "HUMAN_REVIEW_QUEUE_READY",
    "scientific_value": True,

    "purpose": (
        "Human source-level correspondence review of the explicitly "
        "surfaced Cell16 review queue."
    ),

    "input_cell15": str(CELL15_PARQUET),
    "input_cell15_sha256": actual_cell15_sha,

    "input_cell16_directory": str(CELL16_DIR),
    "input_cell16_review_queue": str(CELL16_QUEUE),

    "rows": int(len(review)),
    "review_scope": "NO_LEXICAL_HIT_ONLY",

    "candidate_ids": sorted(review["candidate_id"].unique().tolist()),
    "question_ids": sorted(review["question_id"].unique().tolist()),

    "initial_correspondence_decisions": {
        "PENDING": int(
            (review["correspondence_decision"] == "PENDING").sum()
        ),
        "CORRESPONDS": 0,
        "DOES_NOT_CORRESPOND": 0,
        "UNCERTAIN": 0,
    },

    "automatic_methodological_judgment": False,
    "automatic_correspondence": False,

    "forbidden_operations": [
        "dataset_modification",
        "199_cohort_modification",
        "400_validation_reopen",
        "Q01_Q10_modification",
        "gold_generation",
        "y_plus_generation",
        "y_minus_generation",
        "preference_generation",
        "automatic_correspondence",
        "novelty_judgment",
        "equivalence_judgment",
        "overlap_classification",
        "legal_support_judgment",
        "methodological_inference",
        "architecture_modification",
        "prompt_revision",
        "model_training",
        "model_inference",
        "legacy_cell16_consumption",
        "legacy_cell16a_consumption",
        "cell15_modification",
        "cell16_modification",
    ],

    "allowed_human_decisions": [
        "PENDING",
        "CORRESPONDS",
        "DOES_NOT_CORRESPOND",
        "UNCERTAIN",
    ],

    "created_utc": datetime.utcnow().isoformat() + "Z",
}

with open(MANIFEST_JSON, "w", encoding="utf-8") as f:
    json.dump(manifest, f, indent=2, ensure_ascii=False)

print("✓ Manifest written")


# ================================================================================================================
# STEP 20 — READ-BACK
# ================================================================================================================

print()
print("[STEP 20] READ-BACK")
print("-" * 112)

rb = pd.read_parquet(REVIEW_PARQUET)

if len(rb) != 3:
    raise RuntimeError(
        "STOP — Review queue read-back row count mismatch."
    )

if not (rb["correspondence_decision"] == "PENDING").all():
    raise RuntimeError(
        "STOP — Read-back contains non-PENDING correspondence decision."
    )

rb_slots = set(zip(rb["candidate_id"], rb["question_id"]))

if rb_slots != queue_slots:
    raise RuntimeError(
        "STOP — Read-back queue slots differ from Cell16 queue."
    )

print("✓ Row count read-back = PASS")
print("✓ Slot identity read-back = PASS")
print("✓ PENDING decision firewall = PASS")


# ================================================================================================================
# STEP 21 — FINAL QC
# ================================================================================================================

print()
print("=" * 112)
print("CELL 16A V001 — FINAL GO / NO-GO")
print("=" * 112)

print()
print(f"Scientific Value                  : PASS")
print(f"Cell15 Input SHA                  : PASS")
print(f"Cell16 Review Queue               : VERIFIED")
print(f"Review Rows                       : {len(review)} / 3")
print(f"NO_LEXICAL_HIT                    : {(review['extraction_status'] == 'NO_LEXICAL_HIT').sum()} / 3")
print(f"Correspondence Decisions          : PENDING / {len(review)}")
print(f"Automatic Correspondence          : BLOCKED")
print(f"Methodological Judgment           : BLOCKED")
print(f"Novelty Judgment                  : BLOCKED")
print(f"Equivalence Judgment              : BLOCKED")
print(f"Overlap Classification            : BLOCKED")
print(f"Gold / Y+ / Y-                    : BLOCKED")
print(f"Dataset Modification              : BLOCKED")
print(f"Legacy Cell16A                    : BLOCKED")
print(f"Read-Back                         : PASS")

print()
print("FINAL STATUS:")
print()
print("CELL 16A V001 = HUMAN_REVIEW_QUEUE_READY")

print()
print("NEXT AUTHORIZED ACTION:")
print()
print("Human reviewer may inspect the 3 explicitly surfaced")
print("NO_LEXICAL_HIT source/question pairs and enter:")
print()
print("  CORRESPONDS")
print("  DOES_NOT_CORRESPOND")
print("  UNCERTAIN")
print()
print("No other methodological judgment is authorized in this cell.")
print()
print("=" * 112)
print("CELL 16A V001 — END")
print("=" * 112)
```

```python
# ================================================================================================================
# CELL 16A V002 — HUMAN SOURCE CORRESPONDENCE REVIEW — COMPLETED
# ================================================================================================================
#
# PURPOSE
# -------
# Independently freeze the completed human source-correspondence decisions
# for the explicitly reviewed Cell16 queue.
#
# THIS IS A NEW ARTIFACT.
# V001 IS NOT MODIFIED.
#
# NO PATCHING
# NO GUESSING
# NO LEGACY CELL16A
# NO METHODOLOGICAL JUDGMENT
# ================================================================================================================

from pathlib import Path
import pandas as pd
import json
import hashlib
from datetime import datetime, timezone


# ================================================================================================================
# STEP 0 — SCIENTIFIC VALUE GATE
# ================================================================================================================

print("=" * 112)
print("CELL 16A V002 — HUMAN SOURCE CORRESPONDENCE REVIEW — COMPLETED")
print("=" * 112)

SCIENTIFIC_VALUE = True

SCIENTIFIC_PURPOSE = """
Freeze the completed human source-correspondence decisions for the
explicit Cell16 review queue without modifying Cell15, Cell16, Q01-Q10,
or introducing any methodological judgment.

The scientific value is to create an auditable, immutable checkpoint
between source-grounded extraction QC and later controlled methodological
comparison.
"""

print()
print("[STEP 0] SCIENTIFIC VALUE GATE")
print("-" * 112)
print(SCIENTIFIC_PURPOSE)
print("Scientific value: PASS")


# ================================================================================================================
# STEP 1 — HARD FIREWALL
# ================================================================================================================

FIREWALL = {
    "cell15_modification": "BLOCKED",
    "cell16_modification": "BLOCKED",
    "cell16a_v001_modification": "BLOCKED",
    "dataset_modification": "BLOCKED",
    "199_cohort_modification": "BLOCKED",
    "400_validation_reopen": "BLOCKED",
    "q01_q10_modification": "BLOCKED",
    "gold_generation": "BLOCKED",
    "y_plus_generation": "BLOCKED",
    "y_minus_generation": "BLOCKED",
    "preference_generation": "BLOCKED",
    "novelty_judgment": "BLOCKED",
    "equivalence_judgment": "BLOCKED",
    "overlap_classification": "BLOCKED",
    "legal_support_judgment": "BLOCKED",
    "sufficiency_judgment": "BLOCKED",
    "contradiction_risk_judgment": "BLOCKED",
    "methodological_inference": "BLOCKED",
    "automatic_correspondence": "BLOCKED",
    "automatic_source_correction": "BLOCKED",
    "model_training": "BLOCKED",
    "model_inference": "BLOCKED",
    "prompt_revision": "BLOCKED",
    "architecture_modification": "BLOCKED",
}

print()
print("[STEP 1] HARD FIREWALL")
print("-" * 112)

for k, v in FIREWALL.items():
    print(f"{k:<40}: {v}")


# ================================================================================================================
# STEP 2 — CANONICAL PATHS
# ================================================================================================================

ROOT = Path(
    "/content/drive/MyDrive/X-AutoAlign_Legal_RAG"
)

CELL15_DIR = (
    ROOT
    / "04_AUDIT"
    / "NOVELTY_GAP_AUDIT"
    / "CHECKPOINTS"
    / "CELL15_CANONICAL_SOURCE_GROUNDED_EXTRACTION"
)

CELL16_DIR = (
    ROOT
    / "04_AUDIT"
    / "NOVELTY_GAP_AUDIT"
    / "CHECKPOINTS"
    / "CELL16_SOURCE_GROUNDED_EXTRACTION_QC"
)

CELL16A_V001_DIR = (
    ROOT
    / "04_AUDIT"
    / "NOVELTY_GAP_AUDIT"
    / "CHECKPOINTS"
    / "CELL16A_HUMAN_SOURCE_CORRESPONDENCE_REVIEW_V001"
)

OUTPUT_DIR = (
    ROOT
    / "04_AUDIT"
    / "NOVELTY_GAP_AUDIT"
    / "CHECKPOINTS"
    / "CELL16A_HUMAN_SOURCE_CORRESPONDENCE_REVIEW_V002"
)

CELL15_PARQUET = (
    CELL15_DIR /
    "LITERATURE_EVIDENCE_MATRIX_CELL15.parquet"
)

CELL15_MANIFEST = (
    CELL15_DIR /
    "CELL15_MANIFEST.json"
)

V001_REVIEW_PARQUET = (
    CELL16A_V001_DIR /
    "CELL16A_HUMAN_REVIEW_QUEUE_V001.parquet"
)

V001_REVIEW_CSV = (
    CELL16A_V001_DIR /
    "CELL16A_HUMAN_REVIEW_QUEUE_V001.csv"
)

print()
print("[STEP 2] CANONICAL PATHS")
print("-" * 112)
print("Cell15 :", CELL15_PARQUET)
print("Cell16 :", CELL16_DIR)
print("V001   :", CELL16A_V001_DIR)
print("V002   :", OUTPUT_DIR)


# ================================================================================================================
# STEP 3 — OUTPUT SAFETY
# ================================================================================================================

print()
print("[STEP 3] OUTPUT SAFETY")
print("-" * 112)

if OUTPUT_DIR.exists():
    raise RuntimeError(
        "STOP — V002 output directory already exists. "
        "No overwrite or patching is permitted."
    )

print("✓ V002 output directory does not exist")


# ================================================================================================================
# STEP 4 — REQUIRED INPUTS
# ================================================================================================================

print()
print("[STEP 4] REQUIRED INPUT ARTIFACTS")
print("-" * 112)

required_inputs = [
    CELL15_PARQUET,
    CELL15_MANIFEST,
    V001_REVIEW_PARQUET,
    V001_REVIEW_CSV,
]

for p in required_inputs:
    if not p.exists():
        raise FileNotFoundError(
            f"STOP — Required trusted artifact missing:\n{p}"
        )
    print(f"✓ FOUND: {p}")


# ================================================================================================================
# STEP 5 — SHA256 UTILITY
# ================================================================================================================

def sha256_file(path):
    h = hashlib.sha256()
    with open(path, "rb") as f:
        for block in iter(lambda: f.read(1024 * 1024), b""):
            h.update(block)
    return h.hexdigest()


# ================================================================================================================
# STEP 6 — CELL15 IMMUTABLE INPUT LOCK
# ================================================================================================================

EXPECTED_CELL15_SHA = (
    "4f919b7d611272e8604b3ff62e90ed09cf9e8f44bd1e21a90feb5f29509ecea3"
)

actual_cell15_sha = sha256_file(CELL15_PARQUET)

print()
print("[STEP 6] CELL15 IMMUTABLE INPUT LOCK")
print("-" * 112)
print("Expected:", EXPECTED_CELL15_SHA)
print("Actual  :", actual_cell15_sha)

if actual_cell15_sha != EXPECTED_CELL15_SHA:
    raise RuntimeError(
        "STOP — Cell15 SHA mismatch. "
        "Trusted baseline cannot be verified."
    )

print("✓ Cell15 SHA = PASS")


# ================================================================================================================
# STEP 7 — LOAD CELL15
# ================================================================================================================

print()
print("[STEP 7] LOAD CELL15")
print("-" * 112)

df15 = pd.read_parquet(CELL15_PARQUET)

if len(df15) != 50:
    raise RuntimeError(
        f"STOP — Cell15 expected 50 rows, found {len(df15)}."
    )

print("Rows   :", len(df15))
print("Columns:", len(df15.columns))
print("✓ Cell15 structure = PASS")


# ================================================================================================================
# STEP 8 — LOAD V001
# ================================================================================================================

print()
print("[STEP 8] LOAD CELL16A V001")
print("-" * 112)

v001 = pd.read_parquet(V001_REVIEW_PARQUET)

if len(v001) != 3:
    raise RuntimeError(
        f"STOP — V001 expected 3 review rows, found {len(v001)}."
    )

print("V001 rows:", len(v001))
print("✓ V001 review queue loaded")


# ================================================================================================================
# STEP 9 — V001 MUST REMAIN PENDING
# ================================================================================================================
#
# This proves that V001 itself is preserved as originally created.
# Human decisions are NOT written back into V001.
# ================================================================================================================

print()
print("[STEP 9] V001 IMMUTABILITY CHECK")
print("-" * 112)

if not (
    v001["correspondence_decision"]
    .astype(str)
    .eq("PENDING")
    .all()
):
    raise RuntimeError(
        "STOP — V001 is no longer in its original PENDING state. "
        "Do not continue."
    )

print("✓ V001 remains PENDING / 3")
print("✓ V001 has not been patched")


# ================================================================================================================
# STEP 10 — FROZEN HUMAN DECISIONS
# ================================================================================================================

print()
print("[STEP 10] FROZEN HUMAN DECISIONS")
print("-" * 112)

HUMAN_DECISIONS = {
    ("ANCHOR-001", "Q05"): "DOES_NOT_CORRESPOND",
    ("ANCHOR-001", "Q06"): "DOES_NOT_CORRESPOND",
    ("ANCHOR-001", "Q08"): "DOES_NOT_CORRESPOND",
}

for key, decision in HUMAN_DECISIONS.items():
    print(f"{key[0]} / {key[1]} -> {decision}")


# ================================================================================================================
# STEP 11 — VERIFY DECISION KEYS AGAINST V001
# ================================================================================================================

print()
print("[STEP 11] DECISION KEY INTEGRITY")
print("-" * 112)

v001_slots = set(
    zip(
        v001["candidate_id"],
        v001["question_id"]
    )
)

decision_slots = set(HUMAN_DECISIONS.keys())

if decision_slots != v001_slots:
    raise RuntimeError(
        "STOP — Human decision keys do not exactly match V001 review queue."
    )

print("✓ Decision slots exactly match V001 queue")


# ================================================================================================================
# STEP 12 — BUILD V002 FROM V001
# ================================================================================================================

print()
print("[STEP 12] BUILD V002")
print("-" * 112)

v002 = v001.copy()

# Human decisions are explicitly supplied here.
# No inference is performed.

for (candidate_id, question_id), decision in HUMAN_DECISIONS.items():

    mask = (
        (v002["candidate_id"] == candidate_id)
        &
        (v002["question_id"] == question_id)
    )

    if mask.sum() != 1:
        raise RuntimeError(
            f"STOP — Expected exactly one row for "
            f"{candidate_id}/{question_id}."
        )

    v002.loc[
        mask,
        "correspondence_decision"
    ] = decision

    v002.loc[
        mask,
        "reviewer_note"
    ] = (
        "Human source-correspondence decision frozen in Cell16A V002. "
        "Decision concerns extraction-target correspondence only; "
        "it is not a novelty, equivalence, support, sufficiency, "
        "or methodological judgment."
    )

    v002.loc[
        mask,
        "review_timestamp"
    ] = datetime.now(timezone.utc).isoformat()

# Reviewer identity is intentionally not fabricated.
v002["reviewer_id"] = "HUMAN_REVIEW_DECISION_FROZEN"

print("✓ V002 constructed")
print("✓ No automatic correspondence inference")


# ================================================================================================================
# STEP 13 — DECISION DISTRIBUTION
# ================================================================================================================

print()
print("[STEP 13] HUMAN DECISION DISTRIBUTION")
print("-" * 112)

distribution = (
    v002["correspondence_decision"]
    .value_counts()
    .to_dict()
)

for decision in [
    "PENDING",
    "CORRESPONDS",
    "DOES_NOT_CORRESPOND",
    "UNCERTAIN",
]:
    print(
        f"{decision:<24}: "
        f"{distribution.get(decision, 0)}"
    )

if distribution.get("PENDING", 0) != 0:
    raise RuntimeError(
        "STOP — Completed V002 still contains PENDING decisions."
    )

if distribution.get("DOES_NOT_CORRESPOND", 0) != 3:
    raise RuntimeError(
        "STOP — Expected exactly 3 DOES_NOT_CORRESPOND decisions."
    )

if (
    distribution.get("CORRESPONDS", 0) != 0
    or distribution.get("UNCERTAIN", 0) != 0
):
    raise RuntimeError(
        "STOP — Unexpected human decision distribution."
    )

print("✓ 3 / 3 human decisions frozen")


# ================================================================================================================
# STEP 14 — SOURCE / EXTRACTION IMMUTABILITY
# ================================================================================================================

print()
print("[STEP 14] SOURCE / EXTRACTION IMMUTABILITY")
print("-" * 112)

for key in HUMAN_DECISIONS.keys():

    candidate_id, question_id = key

    a = v001[
        (v001["candidate_id"] == candidate_id)
        &
        (v001["question_id"] == question_id)
    ].iloc[0]

    b = v002[
        (v002["candidate_id"] == candidate_id)
        &
        (v002["question_id"] == question_id)
    ].iloc[0]

    protected_columns = [
        "candidate_id",
        "question_id",
        "question_text",
        "primary_source",
        "source_excerpt",
        "source_location",
        "extraction_status",
        "extraction_method",
        "source_text_status",
    ]

    for col in protected_columns:
        if col in v001.columns and col in v002.columns:
            if str(a[col]) != str(b[col]):
                raise RuntimeError(
                    f"STOP — Protected field modified: "
                    f"{candidate_id}/{question_id}/{col}"
                )

print("✓ Source identity unchanged")
print("✓ Question identity unchanged")
print("✓ Source excerpt unchanged")
print("✓ Source location unchanged")
print("✓ Extraction status unchanged")
print("✓ Extraction method unchanged")


# ================================================================================================================
# STEP 15 — METHODOLOGICAL JUDGMENT FIREWALL
# ================================================================================================================

print()
print("[STEP 15] METHODOLOGICAL JUDGMENT FIREWALL")
print("-" * 112)

forbidden_columns = [
    "novelty_judgment",
    "equivalence_judgment",
    "overlap_class",
    "legal_support_judgment",
    "sufficiency_judgment",
    "contradiction_risk_judgment",
    "gold_label",
    "y_plus",
    "y_minus",
    "preference_label",
]

for col in forbidden_columns:

    if col in v002.columns:
        # If inherited from V001, it must not be populated.
        nonempty = (
            v002[col]
            .notna()
            &
            v002[col].astype(str).str.strip().ne("")
            &
            v002[col].astype(str).str.upper().ne("NOT_ASSESSED")
        )

        if nonempty.any():
            raise RuntimeError(
                f"STOP — Forbidden methodological field populated: {col}"
            )

print("✓ No novelty judgment")
print("✓ No equivalence judgment")
print("✓ No overlap judgment")
print("✓ No legal-support judgment")
print("✓ No sufficiency judgment")
print("✓ No contradiction/risk judgment")
print("✓ No Gold/Y+/Y-/preference")


# ================================================================================================================
# STEP 16 — V002 OUTPUT SCHEMA
# ================================================================================================================

print()
print("[STEP 16] V002 OUTPUT SCHEMA")
print("-" * 112)

if len(v002.columns) != len(v001.columns):
    raise RuntimeError(
        "STOP — V002 schema width differs from V001."
    )

if list(v002.columns) != list(v001.columns):
    raise RuntimeError(
        "STOP — V002 column schema differs from V001."
    )

print(f"Columns: {len(v002.columns)}")
print("✓ Schema preserved")


# ================================================================================================================
# STEP 17 — PERSIST V002
# ================================================================================================================

print()
print("[STEP 17] PERSIST V002")
print("-" * 112)

OUTPUT_DIR.mkdir(parents=False, exist_ok=False)

V002_PARQUET = (
    OUTPUT_DIR /
    "CELL16A_HUMAN_SOURCE_CORRESPONDENCE_REVIEW_V002.parquet"
)

V002_CSV = (
    OUTPUT_DIR /
    "CELL16A_HUMAN_SOURCE_CORRESPONDENCE_REVIEW_V002.csv"
)

V002_MANIFEST = (
    OUTPUT_DIR /
    "CELL16A_MANIFEST_V002.json"
)

V002_DECISION_SUMMARY = (
    OUTPUT_DIR /
    "CELL16A_HUMAN_DECISION_SUMMARY_V002.csv"
)

v002.to_parquet(V002_PARQUET, index=False)
v002.to_csv(V002_CSV, index=False)

decision_summary = pd.DataFrame([
    {
        "candidate_id": "ANCHOR-001",
        "question_id": "Q05",
        "human_correspondence": "DOES_NOT_CORRESPOND",
        "scope": "EXTRACTION_CORRESPONDENCE_ONLY",
    },
    {
        "candidate_id": "ANCHOR-001",
        "question_id": "Q06",
        "human_correspondence": "DOES_NOT_CORRESPOND",
        "scope": "EXTRACTION_CORRESPONDENCE_ONLY",
    },
    {
        "candidate_id": "ANCHOR-001",
        "question_id": "Q08",
        "human_correspondence": "DOES_NOT_CORRESPOND",
        "scope": "EXTRACTION_CORRESPONDENCE_ONLY",
    },
])

decision_summary.to_csv(
    V002_DECISION_SUMMARY,
    index=False
)

print("✓ V002 Parquet written")
print("✓ V002 CSV written")
print("✓ Decision summary written")


# ================================================================================================================
# STEP 18 — HASH OUTPUT
# ================================================================================================================

print()
print("[STEP 18] OUTPUT HASHES")
print("-" * 112)

v002_parquet_sha = sha256_file(V002_PARQUET)
v002_csv_sha = sha256_file(V002_CSV)

print("V002 Parquet SHA256:")
print(v002_parquet_sha)

print()
print("V002 CSV SHA256:")
print(v002_csv_sha)


# ================================================================================================================
# STEP 19 — MANIFEST
# ================================================================================================================

manifest = {
    "cell": "CELL 16A",
    "version": "V002",
    "status": "HUMAN_REVIEW_COMPLETED",

    "scientific_value": True,

    "purpose": (
        "Freeze completed human source-correspondence decisions "
        "for the explicitly surfaced Cell16 review queue."
    ),

    "v001_preserved": True,
    "v001_modified": False,

    "cell15_path": str(CELL15_PARQUET),
    "cell15_sha256": actual_cell15_sha,

    "v001_parquet": str(V001_REVIEW_PARQUET),
    "v001_csv": str(V001_REVIEW_CSV),

    "v002_parquet": str(V002_PARQUET),
    "v002_csv": str(V002_CSV),

    "review_scope": "EXTRACTION_CORRESPONDENCE_ONLY",

    "review_rows": 3,

    "human_decisions": [
        {
            "candidate_id": "ANCHOR-001",
            "question_id": "Q05",
            "decision": "DOES_NOT_CORRESPOND",
        },
        {
            "candidate_id": "ANCHOR-001",
            "question_id": "Q06",
            "decision": "DOES_NOT_CORRESPOND",
        },
        {
            "candidate_id": "ANCHOR-001",
            "question_id": "Q08",
            "decision": "DOES_NOT_CORRESPOND",
        },
    ],

    "automatic_correspondence": False,
    "automatic_source_correction": False,

    "novelty_judgment": "NOT_PERFORMED",
    "equivalence_judgment": "NOT_PERFORMED",
    "overlap_classification": "NOT_PERFORMED",
    "legal_support_judgment": "NOT_PERFORMED",
    "sufficiency_judgment": "NOT_PERFORMED",
    "contradiction_risk_judgment": "NOT_PERFORMED",

    "dataset_modified": False,
    "cell15_modified": False,
    "cell16_modified": False,
    "cell16a_v001_modified": False,

    "gold_created": False,
    "y_plus_created": False,
    "y_minus_created": False,
    "preference_created": False,

    "v002_parquet_sha256": v002_parquet_sha,
    "v002_csv_sha256": v002_csv_sha,

    "created_utc": datetime.now(timezone.utc).isoformat(),
}


with open(
    V002_MANIFEST,
    "w",
    encoding="utf-8"
) as f:
    json.dump(
        manifest,
        f,
        indent=2,
        ensure_ascii=False
    )

print("✓ Manifest written")


# ================================================================================================================
# STEP 20 — INDEPENDENT READ-BACK
# ================================================================================================================

print()
print("[STEP 20] INDEPENDENT READ-BACK")
print("-" * 112)

rb = pd.read_parquet(V002_PARQUET)

if len(rb) != 3:
    raise RuntimeError(
        "STOP — V002 read-back row count mismatch."
    )

expected_decisions = {
    ("ANCHOR-001", "Q05"): "DOES_NOT_CORRESPOND",
    ("ANCHOR-001", "Q06"): "DOES_NOT_CORRESPOND",
    ("ANCHOR-001", "Q08"): "DOES_NOT_CORRESPOND",
}

for key, expected in expected_decisions.items():

    row = rb[
        (rb["candidate_id"] == key[0])
        &
        (rb["question_id"] == key[1])
    ]

    if len(row) != 1:
        raise RuntimeError(
            f"STOP — Read-back row missing or duplicated: {key}"
        )

    actual = row.iloc[0]["correspondence_decision"]

    if actual != expected:
        raise RuntimeError(
            f"STOP — Read-back decision mismatch for {key}: "
            f"{actual} != {expected}"
        )

print("✓ Row count = 3")
print("✓ Q05 decision = DOES_NOT_CORRESPOND")
print("✓ Q06 decision = DOES_NOT_CORRESPOND")
print("✓ Q08 decision = DOES_NOT_CORRESPOND")
print("✓ Decision read-back = PASS")


# ================================================================================================================
# STEP 21 — OUTPUT HASH RECHECK
# ================================================================================================================

print()
print("[STEP 21] OUTPUT HASH RECHECK")
print("-" * 112)

recheck_sha = sha256_file(V002_PARQUET)

if recheck_sha != v002_parquet_sha:
    raise RuntimeError(
        "STOP — V002 Parquet hash changed after persistence."
    )

print("✓ V002 Parquet hash stable")
print("SHA256:", recheck_sha)


# ================================================================================================================
# STEP 22 — FINAL GO / NO-GO
# ================================================================================================================

print()
print("=" * 112)
print("CELL 16A V002 — FINAL GO / NO-GO")
print("=" * 112)

print()
print("Scientific Value                  : PASS")
print("Cell15 Immutable Input            : PASS")
print("Cell16A V001 Preserved            : PASS")
print("V001 Modified                     : NO")
print("Review Queue                      : 3 / 3")
print("Human Decisions                   : 3 / 3")
print("Q05                               : DOES_NOT_CORRESPOND")
print("Q06                               : DOES_NOT_CORRESPOND")
print("Q08                               : DOES_NOT_CORRESPOND")
print("Automatic Correspondence          : BLOCKED")
print("Automatic Source Correction       : BLOCKED")
print("Novelty Judgment                  : BLOCKED")
print("Equivalence Judgment              : BLOCKED")
print("Overlap Classification            : BLOCKED")
print("Legal Support Judgment            : BLOCKED")
print("Sufficiency Judgment              : BLOCKED")
print("Gold / Y+ / Y- / Preference       : BLOCKED")
print("Dataset Modification              : BLOCKED")
print("Independent Read-Back             : PASS")
print("Output Hash Stability             : PASS")

print()
print("FINAL STATUS:")
print()
print("CELL 16A V002 = HUMAN_REVIEW_COMPLETED")

print()
print("NEXT AUTHORIZED STAGE:")
print()
print(
    "The completed correspondence checkpoint may now serve as the "
    "trusted baseline for the next separately controlled stage."
)

print()
print("=" * 112)
print("CELL 16A V002 — END")
print("=" * 112)
```

```python
# ================================================================================================================
# CELL 17 V002 — FROZEN METHODOLOGICAL COMPARISON FRAME / INTEGRITY GATE
# ================================================================================================================
#
# REBUILD POLICY
# --------------
# CLEAN REBUILD FROM ZERO
# V001 IS NOT PATCHED
# V001 IS NOT USED AS A SOURCE OF TRUTH
# LEGACY CELL17 / CELL17A ARE NOT USED
#
# TRUSTED INPUTS
# --------------
# Cell15 Canonical Source-Grounded Extraction
# Cell16A V002 Human Source Correspondence Review
#
# SCIENTIFIC SCOPE
# ----------------
# Freeze the 5 approved candidates × Q01-Q10 comparison frame.
#
# Cell17 DOES NOT perform:
#   - novelty judgment
#   - equivalence judgment
#   - overlap classification
#   - methodological interpretation
#   - legal-support judgment
#   - sufficiency judgment
#   - contradiction/risk judgment
#   - Gold / Y+ / Y- / preference generation
#
# ================================================================================================================

from pathlib import Path
import pandas as pd
import json
import hashlib
from datetime import datetime, timezone


# ================================================================================================================
# STEP 0 — SCIENTIFIC VALUE GATE
# ================================================================================================================

print("=" * 112)
print("CELL 17 V002 — FROZEN METHODOLOGICAL COMPARISON FRAME / INTEGRITY GATE")
print("=" * 112)

print()
print("[STEP 0] SCIENTIFIC VALUE GATE")
print("-" * 112)

print("""
Scientific purpose:

Create one frozen, auditable comparison frame for the five approved
primary sources × Q01-Q10 before Cell18 performs any methodological
comparison.

This prevents:
    • use of legacy Cell17 artifacts
    • reconstruction of Q01-Q10
    • inferred source identity
    • automatic methodological judgment
    • accidental propagation of unreviewed correspondence decisions

The three completed human correspondence decisions from Cell16A V002
are carried forward exactly as human decisions.

Scientific value: PASS
""")

SCIENTIFIC_VALUE = True


# ================================================================================================================
# STEP 1 — HARD FIREWALL
# ================================================================================================================

print()
print("[STEP 1] HARD FIREWALL")
print("-" * 112)

FIREWALL = {
    "cell15_modification": "BLOCKED",
    "cell16_modification": "BLOCKED",
    "cell16a_v002_modification": "BLOCKED",

    "legacy_cell17_use": "BLOCKED",
    "legacy_cell17a_use": "BLOCKED",
    "legacy_cell16_use": "BLOCKED",
    "legacy_cell16a_use": "BLOCKED",

    "dataset_modification": "BLOCKED",
    "199_cohort_modification": "BLOCKED",
    "400_validation_reopen": "BLOCKED",

    "q01_q10_modification": "BLOCKED",

    "gold_generation": "BLOCKED",
    "y_plus_generation": "BLOCKED",
    "y_minus_generation": "BLOCKED",
    "preference_generation": "BLOCKED",

    "novelty_judgment": "BLOCKED",
    "equivalence_judgment": "BLOCKED",
    "overlap_classification": "BLOCKED",

    "legal_support_judgment": "BLOCKED",
    "sufficiency_judgment": "BLOCKED",
    "contradiction_risk_judgment": "BLOCKED",

    "automatic_correspondence_inference": "BLOCKED",
    "automatic_methodological_inference": "BLOCKED",

    "model_training": "BLOCKED",
    "model_inference": "BLOCKED",
    "prompt_revision": "BLOCKED",
    "architecture_modification": "BLOCKED",
}

for firewall_key, firewall_value in FIREWALL.items():
    print(f"{firewall_key:<44}: {firewall_value}")


# ================================================================================================================
# STEP 2 — CANONICAL PATHS
# ================================================================================================================

print()
print("[STEP 2] CANONICAL PATHS")
print("-" * 112)

PROJECT_ROOT = Path(
    "/content/drive/MyDrive/X-AutoAlign_Legal_RAG"
)

CHECKPOINT_ROOT = (
    PROJECT_ROOT
    / "04_AUDIT"
    / "NOVELTY_GAP_AUDIT"
    / "CHECKPOINTS"
)

CELL15_DIR = (
    CHECKPOINT_ROOT
    / "CELL15_CANONICAL_SOURCE_GROUNDED_EXTRACTION"
)

CELL16_DIR = (
    CHECKPOINT_ROOT
    / "CELL16_SOURCE_GROUNDED_EXTRACTION_QC"
)

CELL16A_V002_DIR = (
    CHECKPOINT_ROOT
    / "CELL16A_HUMAN_SOURCE_CORRESPONDENCE_REVIEW_V002"
)

CELL17_V002_DIR = (
    CHECKPOINT_ROOT
    / "CELL17_FROZEN_METHODOLOGICAL_COMPARISON_FRAME_V002"
)

CELL15_PARQUET_PATH = (
    CELL15_DIR
    / "LITERATURE_EVIDENCE_MATRIX_CELL15.parquet"
)

CELL15_MANIFEST_PATH = (
    CELL15_DIR
    / "CELL15_MANIFEST.json"
)

CELL16A_V002_PARQUET_PATH = (
    CELL16A_V002_DIR
    / "CELL16A_HUMAN_SOURCE_CORRESPONDENCE_REVIEW_V002.parquet"
)

CELL16A_V002_MANIFEST_PATH = (
    CELL16A_V002_DIR
    / "CELL16A_MANIFEST_V002.json"
)

CELL17_V002_PARQUET_PATH = (
    CELL17_V002_DIR
    / "CELL17_FROZEN_METHODOLOGICAL_COMPARISON_FRAME_V002.parquet"
)

CELL17_V002_CSV_PATH = (
    CELL17_V002_DIR
    / "CELL17_FROZEN_METHODOLOGICAL_COMPARISON_FRAME_V002.csv"
)

CELL17_V002_MANIFEST_PATH = (
    CELL17_V002_DIR
    / "CELL17_MANIFEST_V002.json"
)

print("Cell15:")
print(CELL15_PARQUET_PATH)

print()
print("Cell16:")
print(CELL16_DIR)

print()
print("Cell16A V002:")
print(CELL16A_V002_PARQUET_PATH)

print()
print("Cell17 V002 output:")
print(CELL17_V002_DIR)


# ================================================================================================================
# STEP 3 — OUTPUT SAFETY
# ================================================================================================================

print()
print("[STEP 3] OUTPUT SAFETY")
print("-" * 112)

if CELL17_V002_DIR.exists():
    raise RuntimeError(
        "STOP — Cell17 V002 output directory already exists.\n"
        "No overwrite, patching, or reuse is permitted."
    )

print("✓ Fresh Cell17 V002 output directory")


# ================================================================================================================
# STEP 4 — REQUIRED TRUSTED INPUTS
# ================================================================================================================

print()
print("[STEP 4] REQUIRED TRUSTED INPUTS")
print("-" * 112)

required_trusted_inputs = [
    CELL15_PARQUET_PATH,
    CELL15_MANIFEST_PATH,
    CELL16A_V002_PARQUET_PATH,
    CELL16A_V002_MANIFEST_PATH,
]

for trusted_path in required_trusted_inputs:

    if not trusted_path.exists():
        raise FileNotFoundError(
            f"STOP — Trusted input missing:\n{trusted_path}"
        )

    print(f"✓ FOUND: {trusted_path}")


# ================================================================================================================
# STEP 5 — HASH FUNCTION
# ================================================================================================================

def sha256_file(file_path):
    sha = hashlib.sha256()

    with open(file_path, "rb") as file_handle:

        for block in iter(
            lambda: file_handle.read(1024 * 1024),
            b""
        ):
            sha.update(block)

    return sha.hexdigest()


# ================================================================================================================
# STEP 6 — TRUSTED HASH LOCK
# ================================================================================================================

print()
print("[STEP 6] TRUSTED INPUT HASH LOCK")
print("-" * 112)

EXPECTED_CELL15_SHA256 = (
    "4f919b7d611272e8604b3ff62e90ed09cf9e8f44bd1e21a90feb5f29509ecea3"
)

EXPECTED_CELL16A_V002_SHA256 = (
    "f1387bc63a5edf588825b4ad4e792f26f6d54d2204b0f2bdc3c4be45211ef151"
)

cell15_sha_actual = sha256_file(
    CELL15_PARQUET_PATH
)

cell16a_v002_sha_actual = sha256_file(
    CELL16A_V002_PARQUET_PATH
)

print("Cell15 expected :")
print(EXPECTED_CELL15_SHA256)

print("Cell15 actual   :")
print(cell15_sha_actual)

if cell15_sha_actual != EXPECTED_CELL15_SHA256:
    raise RuntimeError(
        "STOP — Cell15 SHA256 mismatch."
    )

print("✓ Cell15 SHA256 = PASS")

print()
print("Cell16A V002 expected :")
print(EXPECTED_CELL16A_V002_SHA256)

print("Cell16A V002 actual   :")
print(cell16a_v002_sha_actual)

if cell16a_v002_sha_actual != EXPECTED_CELL16A_V002_SHA256:
    raise RuntimeError(
        "STOP — Cell16A V002 SHA256 mismatch."
    )

print("✓ Cell16A V002 SHA256 = PASS")


# ================================================================================================================
# STEP 7 — MANIFEST IDENTITY CHECK
# ================================================================================================================

print()
print("[STEP 7] TRUSTED MANIFEST IDENTITY")
print("-" * 112)

with open(
    CELL15_MANIFEST_PATH,
    "r",
    encoding="utf-8"
) as manifest_file:

    cell15_manifest = json.load(manifest_file)


with open(
    CELL16A_V002_MANIFEST_PATH,
    "r",
    encoding="utf-8"
) as manifest_file:

    cell16a_v002_manifest = json.load(manifest_file)


if str(
    cell15_manifest.get("v002_parquet_sha256", "")
) != cell15_sha_actual:

    # Cell15 manifest is expected to identify its canonical parquet
    # through v002_parquet_sha256 in the restored canonical manifest.
    # If the field is absent, use the direct artifact SHA lock only.
    if (
        "v002_parquet_sha256"
        not in cell15_manifest
        and
        "v003_parquet_sha256"
        not in cell15_manifest
        and
        "parquet_sha256"
        not in cell15_manifest
    ):
        print(
            "✓ Cell15 manifest has no conflicting canonical SHA field; "
            "direct artifact SHA lock remains authoritative."
        )
    else:
        candidate_manifest_shas = [
            cell15_manifest.get("v002_parquet_sha256"),
            cell15_manifest.get("v003_parquet_sha256"),
            cell15_manifest.get("parquet_sha256"),
        ]

        candidate_manifest_shas = [
            x for x in candidate_manifest_shas
            if x
        ]

        if candidate_manifest_shas and (
            cell15_sha_actual not in candidate_manifest_shas
        ):
            raise RuntimeError(
                "STOP — Cell15 manifest SHA conflicts with trusted artifact SHA."
            )


manifest_cell16a_sha = (
    cell16a_v002_manifest.get(
        "v002_parquet_sha256",
        ""
    )
)

if manifest_cell16a_sha:

    if manifest_cell16a_sha != cell16a_v002_sha_actual:
        raise RuntimeError(
            "STOP — Cell16A V002 manifest SHA conflicts with artifact SHA."
        )

    print("✓ Cell16A V002 manifest SHA = PASS")

else:

    print(
        "✓ Cell16A V002 manifest contains no conflicting SHA field; "
        "direct artifact SHA lock remains authoritative."
    )


print("✓ Trusted manifest identity check complete")


# ================================================================================================================
# STEP 8 — LOAD TRUSTED DATA
# ================================================================================================================

print()
print("[STEP 8] LOAD TRUSTED DATA")
print("-" * 112)

cell15_df = pd.read_parquet(
    CELL15_PARQUET_PATH
)

cell16a_v002_df = pd.read_parquet(
    CELL16A_V002_PARQUET_PATH
)

print(
    "Cell15 rows:",
    len(cell15_df),
    "| columns:",
    len(cell15_df.columns)
)

print(
    "Cell16A V002 rows:",
    len(cell16a_v002_df),
    "| columns:",
    len(cell16a_v002_df.columns)
)

if len(cell15_df) != 50:
    raise RuntimeError(
        f"STOP — Cell15 expected 50 rows; found {len(cell15_df)}."
    )

if len(cell16a_v002_df) != 3:
    raise RuntimeError(
        f"STOP — Cell16A V002 expected 3 rows; found {len(cell16a_v002_df)}."
    )

print("✓ Trusted data loaded")


# ================================================================================================================
# STEP 9 — FROZEN CANDIDATE REGISTRY
# ================================================================================================================

print()
print("[STEP 9] FROZEN CANDIDATE REGISTRY")
print("-" * 112)

EXPECTED_CANDIDATE_IDS = [
    "ANCHOR-001",
    "LIT-001",
    "LIT-002",
    "LIT-003",
    "LIT-004",
]

actual_candidate_ids = sorted(
    cell15_df[
        "candidate_id"
    ]
    .astype(str)
    .unique()
    .tolist()
)

if actual_candidate_ids != sorted(
    EXPECTED_CANDIDATE_IDS
):
    raise RuntimeError(
        "STOP — Candidate identity mismatch."
    )

print("✓ Exactly 5 frozen candidates")

for candidate_id in EXPECTED_CANDIDATE_IDS:
    print("  •", candidate_id)


# ================================================================================================================
# STEP 10 — FROZEN QUESTION FRAME
# ================================================================================================================

print()
print("[STEP 10] FROZEN Q01-Q10 FRAME")
print("-" * 112)

EXPECTED_QUESTION_TEXT = {

    "Q01":
        "Does the work perform post-retrieval evidence verification?",

    "Q02":
        "What is the input to the verification/control mechanism?",

    "Q03":
        "Is evidence relevance assessed independently from evidence sufficiency/support?",

    "Q04":
        "Is legal support or domain-specific evidentiary support explicitly assessed?",

    "Q05":
        "Is evidence sufficiency explicitly assessed?",

    "Q06":
        "Is contradiction, conflict, or risk explicitly assessed?",

    "Q07":
        "Does the mechanism operate before generation or answering?",

    "Q08":
        "Does the mechanism produce an evidence eligibility, risk, or abstention control?",

    "Q09":
        "Is human-grounded evidence suitability part of the evaluated methodology?",

    "Q10":
        "Is the mechanism equivalent to the proposed F-D+R evidence-suitability verification stage?",
}


actual_question_ids = sorted(
    cell15_df[
        "question_id"
    ]
    .astype(str)
    .unique()
    .tolist()
)

if actual_question_ids != sorted(
    EXPECTED_QUESTION_TEXT.keys()
):
    raise RuntimeError(
        "STOP — QID set does not exactly match frozen Q01-Q10."
    )

for question_id, expected_text in EXPECTED_QUESTION_TEXT.items():

    question_rows = cell15_df[
        cell15_df["question_id"].astype(str)
        == question_id
    ]

    if len(question_rows) != 5:
        raise RuntimeError(
            f"STOP — {question_id} expected 5 candidate rows; "
            f"found {len(question_rows)}."
        )

    observed_texts = (
        question_rows[
            "question_text"
        ]
        .astype(str)
        .unique()
        .tolist()
    )

    if observed_texts != [expected_text]:
        raise RuntimeError(
            f"STOP — Question text mismatch for {question_id}."
        )

    print(
        f"✓ {question_id} exact frozen text"
    )


# ================================================================================================================
# STEP 11 — 5 × 10 SLOT INTEGRITY
# ================================================================================================================

print()
print("[STEP 11] 5 × 10 SLOT INTEGRITY")
print("-" * 112)

expected_slots = {
    (candidate_id, question_id)
    for candidate_id in EXPECTED_CANDIDATE_IDS
    for question_id in EXPECTED_QUESTION_TEXT.keys()
}

actual_slots = set(
    zip(
        cell15_df[
            "candidate_id"
        ].astype(str),

        cell15_df[
            "question_id"
        ].astype(str)
    )
)

if actual_slots != expected_slots:
    raise RuntimeError(
        "STOP — Frozen 5×10 slot set mismatch."
    )

duplicate_mask = cell15_df.duplicated(
    subset=[
        "candidate_id",
        "question_id"
    ],
    keep=False
)

if duplicate_mask.any():

    duplicated_slots = cell15_df.loc[
        duplicate_mask,
        [
            "candidate_id",
            "question_id"
        ]
    ]

    raise RuntimeError(
        "STOP — Duplicate candidate/question slots detected:\n"
        + duplicated_slots.to_string(index=False)
    )

print("✓ 50 unique candidate × question slots")
print("✓ 5 candidates × 10 questions")
print("✓ No duplicates")


# ================================================================================================================
# STEP 12 — VERIFY CELL16A V002 HUMAN DECISIONS
# ================================================================================================================

print()
print("[STEP 12] CELL16A V002 HUMAN DECISION LOCK")
print("-" * 112)

EXPECTED_HUMAN_DECISIONS = {

    ("ANCHOR-001", "Q05"):
        "DOES_NOT_CORRESPOND",

    ("ANCHOR-001", "Q06"):
        "DOES_NOT_CORRESPOND",

    ("ANCHOR-001", "Q08"):
        "DOES_NOT_CORRESPOND",
}


for decision_key, expected_decision in (
    EXPECTED_HUMAN_DECISIONS.items()
):

    candidate_id, question_id = decision_key

    matching_rows = cell16a_v002_df[
        (
            cell16a_v002_df[
                "candidate_id"
            ].astype(str)
            == candidate_id
        )
        &
        (
            cell16a_v002_df[
                "question_id"
            ].astype(str)
            == question_id
        )
    ]

    if len(matching_rows) != 1:
        raise RuntimeError(
            f"STOP — Cell16A V002 expected exactly one row for "
            f"{candidate_id}/{question_id}."
        )

    observed_decision = str(
        matching_rows.iloc[0][
            "correspondence_decision"
        ]
    )

    if observed_decision != expected_decision:
        raise RuntimeError(
            f"STOP — Human decision mismatch for "
            f"{candidate_id}/{question_id}."
        )

    print(
        f"✓ {candidate_id} / {question_id} = "
        f"{observed_decision}"
    )


print("✓ All 3 human decisions independently verified")


# ================================================================================================================
# STEP 13 — VERIFY CELL16A CONTAINS NO PENDING DECISION
# ================================================================================================================

print()
print("[STEP 13] CELL16A COMPLETION CHECK")
print("-" * 112)

if (
    cell16a_v002_df[
        "correspondence_decision"
    ]
    .astype(str)
    .str.upper()
    .eq("PENDING")
    .any()
):

    raise RuntimeError(
        "STOP — Cell16A V002 still contains PENDING decisions."
    )

if not (
    cell16a_v002_df[
        "correspondence_decision"
    ]
    .astype(str)
    .str.upper()
    .eq("DOES_NOT_CORRESPOND")
    .all()
):

    raise RuntimeError(
        "STOP — Cell16A V002 contains an unexpected decision."
    )

print("✓ Cell16A V002 = 3 / 3 completed")
print("✓ No PENDING decisions")


# ================================================================================================================
# STEP 14 — CONSTRUCT FROZEN FRAME
# ================================================================================================================

print()
print("[STEP 14] CONSTRUCT FROZEN COMPARISON FRAME")
print("-" * 112)

comparison_frame = cell15_df.copy()

comparison_frame[
    "human_correspondence_status"
] = "NOT_REVIEWED"

for (
    decision_key,
    human_decision
) in EXPECTED_HUMAN_DECISIONS.items():

    candidate_id, question_id = decision_key

    slot_mask = (
        comparison_frame[
            "candidate_id"
        ].astype(str)
        == candidate_id
    ) & (
        comparison_frame[
            "question_id"
        ].astype(str)
        == question_id
    )

    if slot_mask.sum() != 1:
        raise RuntimeError(
            f"STOP — Cannot uniquely map "
            f"{candidate_id}/{question_id}."
        )

    comparison_frame.loc[
        slot_mask,
        "human_correspondence_status"
    ] = human_decision


# --------------------------------------------------------------------------------
# Status fields for the future comparison stage.
# They are deliberately NOT judgments.
# --------------------------------------------------------------------------------

comparison_frame[
    "methodological_comparison_status"
] = "NOT_ASSESSED"

comparison_frame[
    "novelty_status"
] = "NOT_ASSESSED"

comparison_frame[
    "equivalence_status"
] = "NOT_ASSESSED"

comparison_frame[
    "overlap_status"
] = "NOT_ASSESSED"

print("✓ 50-row frozen comparison frame constructed")
print("✓ 3 human correspondence decisions carried forward")
print("✓ 47 slots remain NOT_REVIEWED")
print("✓ No methodological judgment generated")


# ================================================================================================================
# STEP 15 — HUMAN CORRESPONDENCE DISTRIBUTION
# ================================================================================================================

print()
print("[STEP 15] HUMAN CORRESPONDENCE DISTRIBUTION")
print("-" * 112)

correspondence_counts = (
    comparison_frame[
        "human_correspondence_status"
    ]
    .value_counts(
        dropna=False
    )
)

print(
    correspondence_counts.to_string()
)

reviewed_count = int(
    (
        comparison_frame[
            "human_correspondence_status"
        ]
        != "NOT_REVIEWED"
    )
    .sum()
)

not_reviewed_count = int(
    (
        comparison_frame[
            "human_correspondence_status"
        ]
        == "NOT_REVIEWED"
    )
    .sum()
)

does_not_correspond_count = int(
    (
        comparison_frame[
            "human_correspondence_status"
        ]
        == "DOES_NOT_CORRESPOND"
    )
    .sum()
)

if reviewed_count != 3:
    raise RuntimeError(
        f"STOP — Expected 3 reviewed slots; found {reviewed_count}."
    )

if not_reviewed_count != 47:
    raise RuntimeError(
        f"STOP — Expected 47 NOT_REVIEWED slots; found {not_reviewed_count}."
    )

if does_not_correspond_count != 3:
    raise RuntimeError(
        "STOP — Expected exactly 3 DOES_NOT_CORRESPOND slots."
    )

print()
print("✓ Reviewed correspondence slots : 3")
print("✓ NOT_REVIEWED slots             : 47")
print("✓ DOES_NOT_CORRESPOND            : 3")


# ================================================================================================================
# STEP 16 — METHODOLOGICAL JUDGMENT FIREWALL
# ================================================================================================================

print()
print("[STEP 16] METHODOLOGICAL JUDGMENT FIREWALL")
print("-" * 112)

methodological_status_columns = [
    "methodological_comparison_status",
    "novelty_status",
    "equivalence_status",
    "overlap_status",
]

for status_column in methodological_status_columns:

    if status_column not in comparison_frame.columns:
        raise RuntimeError(
            f"STOP — Missing methodological status column: "
            f"{status_column}"
        )

    if not (
        comparison_frame[
            status_column
        ]
        .astype(str)
        .eq("NOT_ASSESSED")
        .all()
    ):

        raise RuntimeError(
            f"STOP — Methodological field was populated: "
            f"{status_column}"
        )

    print(
        f"✓ {status_column} = NOT_ASSESSED for all 50"
    )


# ================================================================================================================
# STEP 17 — SOURCE / QUESTION / EXTRACTION INTEGRITY
# ================================================================================================================

print()
print("[STEP 17] SOURCE / QUESTION / EXTRACTION INTEGRITY")
print("-" * 112)

required_source_fields = [
    "candidate_id",
    "question_id",
    "question_text",
    "primary_source",
    "source_excerpt",
    "source_location",
    "extraction_status",
    "extraction_method",
]

missing_source_fields = [
    field
    for field in required_source_fields
    if field not in comparison_frame.columns
]

if missing_source_fields:

    raise RuntimeError(
        "STOP — Missing source-grounded fields:\n"
        + "\n".join(missing_source_fields)
    )

print("✓ Candidate identity present")
print("✓ Question identity present")
print("✓ Frozen question text present")
print("✓ Primary source present")
print("✓ Source excerpt present")
print("✓ Source location present")
print("✓ Extraction status present")
print("✓ Extraction method present")


# ================================================================================================================
# STEP 18 — NO SOURCE / QUESTION RECONSTRUCTION
# ================================================================================================================

print()
print("[STEP 18] NO SOURCE / QUESTION RECONSTRUCTION")
print("-" * 112)

# Cell17 is permitted only to carry forward exact source/question values.
# No new source text or question text is created.

source_question_columns = [
    "candidate_id",
    "question_id",
    "question_text",
    "primary_source",
]

for column_name in source_question_columns:

    source_values = (
        cell15_df[column_name]
        .astype(str)
        .tolist()
    )

    frame_values = (
        comparison_frame[column_name]
        .astype(str)
        .tolist()
    )

    if source_values != frame_values:

        raise RuntimeError(
            f"STOP — Source/question field changed during frame construction: "
            f"{column_name}"
        )

    print(
        f"✓ {column_name} unchanged from Cell15"
    )


# ================================================================================================================
# STEP 19 — NO LEGACY CONTAMINATION
# ================================================================================================================

print()
print("[STEP 19] LEGACY CONTAMINATION FIREWALL")
print("-" * 112)

print("✓ Legacy Cell17 not loaded")
print("✓ Legacy Cell17A not loaded")
print("✓ Legacy Cell16 not loaded")
print("✓ Legacy Cell16A not loaded")
print("✓ Cell15 canonical artifact is sole source for 50-slot frame")
print("✓ Cell16A V002 is sole source for completed human decisions")


# ================================================================================================================
# STEP 20 — FINAL STRUCTURAL QC BEFORE WRITE
# ================================================================================================================

print()
print("[STEP 20] FINAL STRUCTURAL QC BEFORE WRITE")
print("-" * 112)

if len(comparison_frame) != 50:
    raise RuntimeError(
        "STOP — Final frame row count is not 50."
    )

if comparison_frame.duplicated(
    subset=[
        "candidate_id",
        "question_id"
    ]
).any():

    raise RuntimeError(
        "STOP — Duplicate candidate/question slot in final frame."
    )

if set(
    comparison_frame[
        "question_id"
    ].astype(str).unique()
) != set(
    EXPECTED_QUESTION_TEXT.keys()
):

    raise RuntimeError(
        "STOP — Final frame QID set mismatch."
    )

if set(
    comparison_frame[
        "candidate_id"
    ].astype(str).unique()
) != set(
    EXPECTED_CANDIDATE_IDS
):

    raise RuntimeError(
        "STOP — Final frame candidate set mismatch."
    )

print("✓ Row count = 50")
print("✓ Unique slots = 50")
print("✓ Candidates = 5")
print("✓ Questions = 10")
print("✓ Structural QC = PASS")


# ================================================================================================================
# STEP 21 — PERSIST V002
# ================================================================================================================

print()
print("[STEP 21] PERSIST CELL17 V002")
print("-" * 112)

CELL17_V002_DIR.mkdir(
    parents=False,
    exist_ok=False
)

comparison_frame.to_parquet(
    CELL17_V002_PARQUET_PATH,
    index=False
)

comparison_frame.to_csv(
    CELL17_V002_CSV_PATH,
    index=False
)

print("✓ Parquet written")
print("✓ CSV written")


# ================================================================================================================
# STEP 22 — OUTPUT HASHES
# ================================================================================================================

print()
print("[STEP 22] OUTPUT HASHES")
print("-" * 112)

cell17_v002_parquet_sha_actual = sha256_file(
    CELL17_V002_PARQUET_PATH
)

cell17_v002_csv_sha_actual = sha256_file(
    CELL17_V002_CSV_PATH
)

print("Cell17 V002 Parquet SHA256:")
print(cell17_v002_parquet_sha_actual)

print()
print("Cell17 V002 CSV SHA256:")
print(cell17_v002_csv_sha_actual)


# ================================================================================================================
# STEP 23 — MANIFEST
# ================================================================================================================

print()
print("[STEP 23] WRITE CELL17 V002 MANIFEST")
print("-" * 112)

cell17_v002_manifest = {

    "cell": "CELL 17",

    "version": "V002",

    "status":
        "FROZEN_COMPARISON_FRAME_PASS",

    "rebuild_policy":
        "CLEAN_REBUILD_FROM_ZERO",

    "legacy_cell17_used":
        False,

    "legacy_cell17a_used":
        False,

    "scientific_value":
        True,

    "scientific_purpose":
        (
            "Freeze a 5-candidate × Q01-Q10 methodological comparison "
            "frame before controlled source-grounded comparison."
        ),

    "trusted_inputs": {

        "cell15": {
            "path": str(
                CELL15_PARQUET_PATH
            ),
            "sha256":
                cell15_sha_actual,
        },

        "cell16a_v002": {
            "path": str(
                CELL16A_V002_PARQUET_PATH
            ),
            "sha256":
                cell16a_v002_sha_actual,
        },
    },

    "candidate_count": 5,

    "question_count": 10,

    "slot_count": 50,

    "candidate_ids":
        EXPECTED_CANDIDATE_IDS,

    "question_ids":
        list(
            EXPECTED_QUESTION_TEXT.keys()
        ),

    "human_correspondence": {

        "reviewed_slots": 3,

        "not_reviewed_slots": 47,

        "decisions": [

            {
                "candidate_id":
                    "ANCHOR-001",

                "question_id":
                    "Q05",

                "decision":
                    "DOES_NOT_CORRESPOND",
            },

            {
                "candidate_id":
                    "ANCHOR-001",

                "question_id":
                    "Q06",

                "decision":
                    "DOES_NOT_CORRESPOND",
            },

            {
                "candidate_id":
                    "ANCHOR-001",

                "question_id":
                    "Q08",

                "decision":
                    "DOES_NOT_CORRESPOND",
            },
        ],
    },

    "methodological_comparison":
        "NOT_ASSESSED",

    "novelty":
        "NOT_ASSESSED",

    "equivalence":
        "NOT_ASSESSED",

    "overlap":
        "NOT_ASSESSED",

    "automatic_methodological_inference":
        False,

    "automatic_correspondence_inference":
        False,

    "dataset_modified":
        False,

    "cohort_199_modified":
        False,

    "validation_400_reopened":
        False,

    "gold_created":
        False,

    "y_plus_created":
        False,

    "y_minus_created":
        False,

    "preference_created":
        False,

    "output": {

        "parquet":
            str(
                CELL17_V002_PARQUET_PATH
            ),

        "csv":
            str(
                CELL17_V002_CSV_PATH
            ),

        "parquet_sha256":
            cell17_v002_parquet_sha_actual,

        "csv_sha256":
            cell17_v002_csv_sha_actual,
    },

    "created_utc":
        datetime.now(
            timezone.utc
        ).isoformat(),
}


with open(
    CELL17_V002_MANIFEST_PATH,
    "w",
    encoding="utf-8"
) as manifest_file:

    json.dump(
        cell17_v002_manifest,
        manifest_file,
        indent=2,
        ensure_ascii=False
    )

print("✓ Manifest written")


# ================================================================================================================
# STEP 24 — INDEPENDENT READ-BACK
# ================================================================================================================

print()
print("[STEP 24] INDEPENDENT READ-BACK")
print("-" * 112)

readback_frame = pd.read_parquet(
    CELL17_V002_PARQUET_PATH
)

if len(readback_frame) != 50:

    raise RuntimeError(
        "STOP — Read-back row count mismatch."
    )

readback_slots = set(
    zip(
        readback_frame[
            "candidate_id"
        ].astype(str),

        readback_frame[
            "question_id"
        ].astype(str)
    )
)

if readback_slots != expected_slots:

    raise RuntimeError(
        "STOP — Read-back slot set mismatch."
    )


for (
    decision_key,
    expected_decision
) in EXPECTED_HUMAN_DECISIONS.items():

    candidate_id, question_id = decision_key

    readback_rows = readback_frame[
        (
            readback_frame[
                "candidate_id"
            ].astype(str)
            == candidate_id
        )
        &
        (
            readback_frame[
                "question_id"
            ].astype(str)
            == question_id
        )
    ]

    if len(readback_rows) != 1:

        raise RuntimeError(
            f"STOP — Read-back row mismatch for "
            f"{candidate_id}/{question_id}."
        )

    observed = str(
        readback_rows.iloc[0][
            "human_correspondence_status"
        ]
    )

    if observed != expected_decision:

        raise RuntimeError(
            f"STOP — Read-back decision mismatch for "
            f"{candidate_id}/{question_id}."
        )


if not (
    readback_frame[
        "methodological_comparison_status"
    ]
    .astype(str)
    .eq("NOT_ASSESSED")
    .all()
):

    raise RuntimeError(
        "STOP — Methodological comparison status changed "
        "during persistence/read-back."
    )


if not (
    readback_frame[
        "novelty_status"
    ]
    .astype(str)
    .eq("NOT_ASSESSED")
    .all()
):

    raise RuntimeError(
        "STOP — Novelty status changed during read-back."
    )


if not (
    readback_frame[
        "equivalence_status"
    ]
    .astype(str)
    .eq("NOT_ASSESSED")
    .all()
):

    raise RuntimeError(
        "STOP — Equivalence status changed during read-back."
    )


if not (
    readback_frame[
        "overlap_status"
    ]
    .astype(str)
    .eq("NOT_ASSESSED")
    .all()
):

    raise RuntimeError(
        "STOP — Overlap status changed during read-back."
    )


print("✓ 50 rows read back")
print("✓ 50 unique slots read back")
print("✓ Q05 human decision read back")
print("✓ Q06 human decision read back")
print("✓ Q08 human decision read back")
print("✓ 47 NOT_REVIEWED slots preserved")
print("✓ All methodological fields remain NOT_ASSESSED")


# ================================================================================================================
# STEP 25 — FINAL HASH STABILITY
# ================================================================================================================

print()
print("[STEP 25] FINAL HASH STABILITY")
print("-" * 112)

cell17_v002_parquet_sha_recheck = sha256_file(
    CELL17_V002_PARQUET_PATH
)

if (
    cell17_v002_parquet_sha_recheck
    !=
    cell17_v002_parquet_sha_actual
):

    raise RuntimeError(
        "STOP — Cell17 V002 Parquet SHA changed after persistence."
    )

print("✓ Cell17 V002 Parquet SHA stable")
print(
    "SHA256:",
    cell17_v002_parquet_sha_recheck
)


# ================================================================================================================
# STEP 26 — FINAL GO / NO-GO
# ================================================================================================================

print()
print("=" * 112)
print("CELL 17 V002 — FINAL GO / NO-GO")
print("=" * 112)

print()

print(
    "Scientific Value                  : PASS"
)

print(
    "Cell15 Trusted Baseline           : PASS"
)

print(
    "Cell16A V002 Trusted Baseline     : PASS"
)

print(
    "Legacy Cell17 Used                : NO"
)

print(
    "Legacy Cell17A Used               : NO"
)

print(
    "Clean Rebuild                     : PASS"
)

print(
    "Candidates                        : 5 / 5"
)

print(
    "Questions                         : 10 / 10"
)

print(
    "Frozen Comparison Slots           : 50 / 50"
)

print(
    "Human Correspondence Decisions    : 3 / 3"
)

print(
    "NOT_REVIEWED Slots                : 47 / 47"
)

print(
    "Q05                               : DOES_NOT_CORRESPOND"
)

print(
    "Q06                               : DOES_NOT_CORRESPOND"
)

print(
    "Q08                               : DOES_NOT_CORRESPOND"
)

print(
    "Methodological Comparison         : NOT_ASSESSED"
)

print(
    "Novelty Judgment                  : NOT_ASSESSED"
)

print(
    "Equivalence Judgment              : NOT_ASSESSED"
)

print(
    "Overlap Classification            : NOT_ASSESSED"
)

print(
    "Automatic Methodological Inference: BLOCKED"
)

print(
    "Dataset Modification              : BLOCKED"
)

print(
    "199 Cohort Modification            : BLOCKED"
)

print(
    "400 Validation Reopen              : BLOCKED"
)

print(
    "Gold / Y+ / Y- / Preference       : BLOCKED"
)

print(
    "Independent Read-Back             : PASS"
)

print(
    "Output Hash Stability             : PASS"
)

print()
print("FINAL STATUS:")
print()

print(
    "CELL 17 V002 = FROZEN_COMPARISON_FRAME_PASS"
)

print()
print("NEXT AUTHORIZED STAGE:")
print()

print(
    "CELL 18 may begin the separately controlled "
    "source-grounded methodological comparison."
)

print()
print("=" * 112)
print("CELL 17 V002 — END")
print("=" * 112)
```

```python
# ================================================================================================================
# CELL 18 V001 — CONTROLLED SOURCE-GROUNDED METHODOLOGICAL COMPARISON
# ================================================================================================================
#
# SCIENTIFIC ROLE
# ---------------
# Prepare the first controlled methodological comparison artifact from
# the trusted Cell17 V002 frozen comparison frame.
#
# IMPORTANT
# ---------
# This cell DOES NOT automatically decide novelty/equivalence/overlap.
#
# It creates a controlled human-assessment matrix in which every
# methodological comparison dimension is explicitly defined and tied
# to the frozen source-grounded evidence.
#
# NO PATCHING
# NO GUESSING
# NO LEGACY CELL18
# NO LLM INFERENCE
# NO AUTOMATIC NOVELTY
# NO AUTOMATIC EQUIVALENCE
# NO AUTOMATIC OVERLAP
#
# ================================================================================================================

from pathlib import Path
import pandas as pd
import json
import hashlib
from datetime import datetime, timezone


# ================================================================================================================
# STEP 0 — SCIENTIFIC VALUE GATE
# ================================================================================================================

print("=" * 112)
print("CELL 18 V001 — CONTROLLED SOURCE-GROUNDED METHODOLOGICAL COMPARISON")
print("=" * 112)

print()
print("[STEP 0] SCIENTIFIC VALUE GATE")
print("-" * 112)

print("""
Scientific purpose:

Create the first controlled methodological-comparison artifact using
only the frozen Cell17 V002 frame and its source-grounded evidence.

The comparison is restricted to the pre-defined dimensions:

    D01 — Post-retrieval stage
    D02 — Verification/control input
    D03 — Relevance vs sufficiency/support distinction
    D04 — Legal/domain-specific evidentiary support
    D05 — Evidence sufficiency
    D06 — Contradiction/conflict/risk
    D07 — Pre-generation / pre-answer operation
    D08 — Evidence eligibility / risk / abstention control
    D09 — Human-grounded evidence suitability
    D10 — Overall equivalence to the proposed F-D+R stage

This cell prepares the controlled assessment frame.

It does NOT automatically assign:
    • novelty
    • equivalence
    • overlap
    • methodological superiority
    • support
    • contradiction
    • causal claims

Scientific value: PASS
""")


# ================================================================================================================
# STEP 1 — HARD FIREWALL
# ================================================================================================================

print()
print("[STEP 1] HARD FIREWALL")
print("-" * 112)

FIREWALL = {
    "cell15_modification": "BLOCKED",
    "cell16_modification": "BLOCKED",
    "cell16a_modification": "BLOCKED",
    "cell17_modification": "BLOCKED",
    "legacy_cell18_use": "BLOCKED",

    "dataset_modification": "BLOCKED",
    "199_cohort_modification": "BLOCKED",
    "400_validation_reopen": "BLOCKED",

    "q01_q10_modification": "BLOCKED",

    "automatic_novelty_judgment": "BLOCKED",
    "automatic_equivalence_judgment": "BLOCKED",
    "automatic_overlap_classification": "BLOCKED",

    "automatic_methodological_inference": "BLOCKED",
    "automatic_legal_inference": "BLOCKED",

    "gold_generation": "BLOCKED",
    "y_plus_generation": "BLOCKED",
    "y_minus_generation": "BLOCKED",
    "preference_generation": "BLOCKED",

    "model_training": "BLOCKED",
    "model_inference": "BLOCKED",
    "prompt_revision": "BLOCKED",
    "architecture_modification": "BLOCKED",

    "external_legal_knowledge": "BLOCKED",
    "semantic_reconstruction": "BLOCKED",
    "source_reconstruction": "BLOCKED",
}

for key, value in FIREWALL.items():
    print(f"{key:<44}: {value}")


# ================================================================================================================
# STEP 2 — TRUSTED INPUT PATHS
# ================================================================================================================

print()
print("[STEP 2] TRUSTED INPUT PATHS")
print("-" * 112)

PROJECT_ROOT = Path(
    "/content/drive/MyDrive/X-AutoAlign_Legal_RAG"
)

CHECKPOINT_ROOT = (
    PROJECT_ROOT
    / "04_AUDIT"
    / "NOVELTY_GAP_AUDIT"
    / "CHECKPOINTS"
)

CELL17_V002_DIR = (
    CHECKPOINT_ROOT
    / "CELL17_FROZEN_METHODOLOGICAL_COMPARISON_FRAME_V002"
)

CELL17_V002_PARQUET = (
    CELL17_V002_DIR
    / "CELL17_FROZEN_METHODOLOGICAL_COMPARISON_FRAME_V002.parquet"
)

CELL17_V002_MANIFEST = (
    CELL17_V002_DIR
    / "CELL17_MANIFEST_V002.json"
)

CELL18_V001_DIR = (
    CHECKPOINT_ROOT
    / "CELL18_CONTROLLED_METHODOLOGICAL_COMPARISON_V001"
)

CELL18_V001_PARQUET = (
    CELL18_V001_DIR
    / "CELL18_CONTROLLED_METHODOLOGICAL_COMPARISON_V001.parquet"
)

CELL18_V001_CSV = (
    CELL18_V001_DIR
    / "CELL18_CONTROLLED_METHODOLOGICAL_COMPARISON_V001.csv"
)

CELL18_V001_MANIFEST = (
    CELL18_V001_DIR
    / "CELL18_MANIFEST_V001.json"
)

print("Cell17 V002:")
print(CELL17_V002_PARQUET)

print()
print("Cell18 V001 output:")
print(CELL18_V001_DIR)


# ================================================================================================================
# STEP 3 — OUTPUT SAFETY
# ================================================================================================================

print()
print("[STEP 3] OUTPUT SAFETY")
print("-" * 112)

if CELL18_V001_DIR.exists():
    raise RuntimeError(
        "STOP — Cell18 V001 output directory already exists. "
        "No patching or overwrite permitted."
    )

print("✓ Fresh Cell18 V001 output directory")


# ================================================================================================================
# STEP 4 — REQUIRED TRUSTED INPUT
# ================================================================================================================

print()
print("[STEP 4] REQUIRED TRUSTED INPUT")
print("-" * 112)

if not CELL17_V002_PARQUET.exists():
    raise FileNotFoundError(
        f"STOP — Cell17 V002 missing:\n{CELL17_V002_PARQUET}"
    )

if not CELL17_V002_MANIFEST.exists():
    raise FileNotFoundError(
        f"STOP — Cell17 V002 manifest missing:\n{CELL17_V002_MANIFEST}"
    )

print("✓ Cell17 V002 Parquet found")
print("✓ Cell17 V002 manifest found")


# ================================================================================================================
# STEP 5 — SHA256
# ================================================================================================================

def sha256_file(file_path):
    h = hashlib.sha256()

    with open(file_path, "rb") as f:
        for block in iter(
            lambda: f.read(1024 * 1024),
            b""
        ):
            h.update(block)

    return h.hexdigest()


EXPECTED_CELL17_V002_SHA256 = (
    "d5a10c5db8eef344dbe45e54b1c00839c1bbf2058bd96733f63281cb79ff7d72"
)

cell17_v002_sha_actual = sha256_file(
    CELL17_V002_PARQUET
)

print()
print("[STEP 5] CELL17 V002 TRUSTED HASH LOCK")
print("-" * 112)

print("Expected:")
print(EXPECTED_CELL17_V002_SHA256)

print()
print("Actual:")
print(cell17_v002_sha_actual)

if cell17_v002_sha_actual != EXPECTED_CELL17_V002_SHA256:
    raise RuntimeError(
        "STOP — Cell17 V002 SHA256 mismatch."
    )

print("✓ Cell17 V002 SHA256 = PASS")


# ================================================================================================================
# STEP 6 — MANIFEST CHECK
# ================================================================================================================

print()
print("[STEP 6] CELL17 V002 MANIFEST CHECK")
print("-" * 112)

with open(
    CELL17_V002_MANIFEST,
    "r",
    encoding="utf-8"
) as f:

    cell17_manifest = json.load(f)

manifest_sha = (
    cell17_manifest
    .get("output", {})
    .get("parquet_sha256", "")
)

if manifest_sha:

    if manifest_sha != cell17_v002_sha_actual:
        raise RuntimeError(
            "STOP — Cell17 manifest SHA conflicts with actual artifact."
        )

    print("✓ Manifest SHA = PASS")

else:

    print(
        "✓ No conflicting manifest SHA field; "
        "direct artifact SHA remains authoritative."
    )


# ================================================================================================================
# STEP 7 — LOAD CELL17
# ================================================================================================================

print()
print("[STEP 7] LOAD CELL17 V002")
print("-" * 112)

cell17_df = pd.read_parquet(
    CELL17_V002_PARQUET
)

print(
    "Rows   :",
    len(cell17_df)
)

print(
    "Columns:",
    len(cell17_df.columns)
)

if len(cell17_df) != 50:
    raise RuntimeError(
        "STOP — Cell17 V002 must contain exactly 50 slots."
    )

print("✓ Cell17 V002 loaded")


# ================================================================================================================
# STEP 8 — FROZEN CANDIDATES
# ================================================================================================================

print()
print("[STEP 8] FROZEN CANDIDATES")
print("-" * 112)

EXPECTED_CANDIDATES = [
    "ANCHOR-001",
    "LIT-001",
    "LIT-002",
    "LIT-003",
    "LIT-004",
]

actual_candidates = sorted(
    cell17_df[
        "candidate_id"
    ]
    .astype(str)
    .unique()
    .tolist()
)

if actual_candidates != sorted(
    EXPECTED_CANDIDATES
):

    raise RuntimeError(
        "STOP — Candidate registry mismatch."
    )

print("✓ 5 candidates verified")

for candidate_id in EXPECTED_CANDIDATES:
    print("  •", candidate_id)


# ================================================================================================================
# STEP 9 — FROZEN Q01-Q10
# ================================================================================================================

print()
print("[STEP 9] FROZEN Q01-Q10")
print("-" * 112)

EXPECTED_QIDS = [
    "Q01",
    "Q02",
    "Q03",
    "Q04",
    "Q05",
    "Q06",
    "Q07",
    "Q08",
    "Q09",
    "Q10",
]

actual_qids = sorted(
    cell17_df[
        "question_id"
    ]
    .astype(str)
    .unique()
    .tolist()
)

if actual_qids != sorted(EXPECTED_QIDS):
    raise RuntimeError(
        "STOP — Q01-Q10 set mismatch."
    )

print("✓ Q01-Q10 verified")


# ================================================================================================================
# STEP 10 — SLOT INTEGRITY
# ================================================================================================================

print()
print("[STEP 10] 5 × 10 SLOT INTEGRITY")
print("-" * 112)

expected_slots = {
    (candidate_id, question_id)
    for candidate_id in EXPECTED_CANDIDATES
    for question_id in EXPECTED_QIDS
}

actual_slots = set(
    zip(
        cell17_df[
            "candidate_id"
        ].astype(str),

        cell17_df[
            "question_id"
        ].astype(str)
    )
)

if actual_slots != expected_slots:
    raise RuntimeError(
        "STOP — Cell17 slot set differs from frozen 5×10 frame."
    )

if cell17_df.duplicated(
    subset=[
        "candidate_id",
        "question_id"
    ]
).any():

    raise RuntimeError(
        "STOP — Duplicate candidate/question slot detected."
    )

print("✓ 50 unique slots")
print("✓ 5 × 10 integrity PASS")


# ================================================================================================================
# STEP 11 — HUMAN CORRESPONDENCE FIREWALL
# ================================================================================================================

print()
print("[STEP 11] HUMAN CORRESPONDENCE FIREWALL")
print("-" * 112)

required_correspondence_column = (
    "human_correspondence_status"
)

if required_correspondence_column not in cell17_df.columns:
    raise RuntimeError(
        "STOP — Cell17 V002 missing human correspondence status."
    )

correspondence_counts = (
    cell17_df[
        required_correspondence_column
    ]
    .astype(str)
    .value_counts(
        dropna=False
    )
)

print(
    correspondence_counts.to_string()
)

expected_not_reviewed = 47
expected_does_not_correspond = 3

if (
    int(
        correspondence_counts.get(
            "NOT_REVIEWED",
            0
        )
    )
    != expected_not_reviewed
):

    raise RuntimeError(
        "STOP — Expected 47 NOT_REVIEWED slots."
    )

if (
    int(
        correspondence_counts.get(
            "DOES_NOT_CORRESPOND",
            0
        )
    )
    != expected_does_not_correspond
):

    raise RuntimeError(
        "STOP — Expected 3 DOES_NOT_CORRESPOND slots."
    )

print("✓ 47 NOT_REVIEWED")
print("✓ 3 DOES_NOT_CORRESPOND")


# ================================================================================================================
# STEP 12 — DEFINE FROZEN COMPARISON DIMENSIONS
# ================================================================================================================

print()
print("[STEP 12] FROZEN METHODOLOGICAL COMPARISON DIMENSIONS")
print("-" * 112)

COMPARISON_DIMENSIONS = {

    "D01_POST_RETRIEVAL_STAGE": {
        "question_id": "Q01",
        "description":
            "Whether the work performs verification/control after retrieval.",
    },

    "D02_VERIFICATION_INPUT": {
        "question_id": "Q02",
        "description":
            "What enters the verification/control mechanism.",
    },

    "D03_RELEVANCE_VS_SUFFICIENCY": {
        "question_id": "Q03",
        "description":
            "Whether relevance is distinguished from sufficiency/support.",
    },

    "D04_LEGAL_OR_DOMAIN_SUPPORT": {
        "question_id": "Q04",
        "description":
            "Whether legal or domain-specific evidentiary support is explicitly assessed.",
    },

    "D05_EVIDENCE_SUFFICIENCY": {
        "question_id": "Q05",
        "description":
            "Whether evidence sufficiency is explicitly assessed.",
    },

    "D06_CONTRADICTION_CONFLICT_RISK": {
        "question_id": "Q06",
        "description":
            "Whether contradiction, conflict, or risk is explicitly assessed.",
    },

    "D07_PRE_GENERATION_OPERATION": {
        "question_id": "Q07",
        "description":
            "Whether the mechanism operates before generation or answering.",
    },

    "D08_ELIGIBILITY_RISK_ABSTENTION": {
        "question_id": "Q08",
        "description":
            "Whether the mechanism produces an eligibility, risk, or abstention control.",
    },

    "D09_HUMAN_GROUNDED_SUITABILITY": {
        "question_id": "Q09",
        "description":
            "Whether human-grounded evidence suitability is part of the evaluated methodology.",
    },

    "D10_OVERALL_FDR_EQUIVALENCE": {
        "question_id": "Q10",
        "description":
            "Whether the mechanism is equivalent to the proposed F-D+R evidence-suitability verification stage.",
    },
}

print(
    "✓ 10 frozen comparison dimensions"
)

for dimension_id, definition in COMPARISON_DIMENSIONS.items():

    print(
        f"{dimension_id:<38} "
        f"{definition['question_id']:<5} "
        f"{definition['description']}"
    )


# ================================================================================================================
# STEP 13 — FROZEN COMPARISON SCALE
# ================================================================================================================

print()
print("[STEP 13] FROZEN COMPARISON SCALE")
print("-" * 112)

OVERLAP_SCALE = {
    "O0_NONE":
        "No meaningful methodological overlap established.",

    "O1_DISTANT":
        "Related concept or objective, but structurally different.",

    "O2_PARTIAL":
        "Partial methodological overlap, with material structural differences.",

    "O3_STRONG":
        "Strong overlap across major dimensions, but not fully equivalent.",

    "O4_SUBSTANTIAL":
        "Substantial methodological equivalence across the defined stage/input/target/control dimensions.",
}

for scale_id, definition in OVERLAP_SCALE.items():
    print(
        f"{scale_id:<18}: {definition}"
    )

print()
print(
    "✓ Scale defined; no score assigned automatically."
)


# ================================================================================================================
# STEP 14 — BUILD CONTROLLED ASSESSMENT MATRIX
# ================================================================================================================

print()
print("[STEP 14] BUILD CONTROLLED ASSESSMENT MATRIX")
print("-" * 112)

comparison_df = cell17_df.copy()

# --------------------------------------------------------------------------------
# Eligibility for methodological comparison
# --------------------------------------------------------------------------------

comparison_df[
    "comparison_eligibility"
] = "ELIGIBLE_PENDING_HUMAN_ASSESSMENT"

comparison_df.loc[
    comparison_df[
        "human_correspondence_status"
    ].astype(str)
    ==
    "DOES_NOT_CORRESPOND",
    "comparison_eligibility"
] = "BLOCKED_BY_HUMAN_CORRESPONDENCE"


# --------------------------------------------------------------------------------
# Human assessment fields
# --------------------------------------------------------------------------------

comparison_fields = [
    "post_retrieval_stage_assessment",
    "verification_input_assessment",
    "relevance_sufficiency_assessment",
    "legal_domain_support_assessment",
    "evidence_sufficiency_assessment",
    "contradiction_conflict_risk_assessment",
    "pre_generation_assessment",
    "eligibility_risk_abstention_assessment",
    "human_grounded_suitability_assessment",
    "overall_fdr_equivalence_assessment",
    "overlap_class",
]

for field_name in comparison_fields:

    comparison_df[field_name] = "PENDING_HUMAN_ASSESSMENT"


# --------------------------------------------------------------------------------
# Evidence basis status
# --------------------------------------------------------------------------------

comparison_df[
    "source_grounding_status"
] = "SOURCE_GROUNDED_EVIDENCE_AVAILABLE"


# --------------------------------------------------------------------------------
# Assessment provenance
# --------------------------------------------------------------------------------

comparison_df[
    "assessment_method"
] = "HUMAN_SOURCE_GROUNDED_COMPARISON"


comparison_df[
    "automatic_judgment"
] = False


# --------------------------------------------------------------------------------
# Block the 3 correspondence failures explicitly.
# --------------------------------------------------------------------------------

blocked_mask = (
    comparison_df[
        "comparison_eligibility"
    ]
    ==
    "BLOCKED_BY_HUMAN_CORRESPONDENCE"
)

for field_name in comparison_fields:

    comparison_df.loc[
        blocked_mask,
        field_name
    ] = "NOT_ASSESSED_DUE_TO_CORRESPONDENCE"


print("✓ 50 comparison records constructed")
print("✓ 47 eligible pending human assessment")
print("✓ 3 blocked by prior human correspondence decision")
print("✓ No methodological judgment automatically generated")


# ================================================================================================================
# STEP 15 — PROTECTED SOURCE EVIDENCE
# ================================================================================================================

print()
print("[STEP 15] PROTECTED SOURCE EVIDENCE")
print("-" * 112)

protected_source_columns = [
    "candidate_id",
    "question_id",
    "question_text",
    "primary_source",
    "source_excerpt",
    "source_location",
    "extraction_status",
    "extraction_method",
]

for column_name in protected_source_columns:

    if column_name not in comparison_df.columns:

        raise RuntimeError(
            f"STOP — Required source-grounded field missing: "
            f"{column_name}"
        )

print("✓ Source identity protected")
print("✓ Question identity protected")
print("✓ Source excerpt protected")
print("✓ Source location protected")
print("✓ Extraction provenance protected")


# ================================================================================================================
# STEP 16 — NO AUTOMATIC METHODOLOGICAL JUDGMENT
# ================================================================================================================

print()
print("[STEP 16] NO AUTOMATIC METHODOLOGICAL JUDGMENT")
print("-" * 112)

for field_name in comparison_fields:

    valid_values = {
        "PENDING_HUMAN_ASSESSMENT",
        "NOT_ASSESSED_DUE_TO_CORRESPONDENCE",
    }

    observed_values = set(
        comparison_df[
            field_name
        ].astype(str)
    )

    if not observed_values.issubset(
        valid_values
    ):

        raise RuntimeError(
            f"STOP — Unexpected automatic value in "
            f"{field_name}: {observed_values}"
        )

    print(
        f"✓ {field_name}: no automatic judgment"
    )


# ================================================================================================================
# STEP 17 — Q10 SPECIAL FIREWALL
# ================================================================================================================
#
# Q10 asks equivalence explicitly.
# It MUST remain human assessed.
# It cannot be derived from Q01-Q09 automatically.
# ================================================================================================================

print()
print("[STEP 17] Q10 EQUIVALENCE FIREWALL")
print("-" * 112)

q10_mask = (
    comparison_df[
        "question_id"
    ].astype(str)
    ==
    "Q10"
)

if not (
    comparison_df.loc[
        q10_mask,
        "overall_fdr_equivalence_assessment"
    ]
    .isin([
        "PENDING_HUMAN_ASSESSMENT",
        "NOT_ASSESSED_DUE_TO_CORRESPONDENCE",
    ])
    .all()
):

    raise RuntimeError(
        "STOP — Q10 equivalence was automatically assigned."
    )

print(
    "✓ Q10 equivalence remains human-assessed only"
)


# ================================================================================================================
# STEP 18 — STRUCTURAL INTEGRITY
# ================================================================================================================

print()
print("[STEP 18] STRUCTURAL INTEGRITY")
print("-" * 112)

if len(comparison_df) != 50:

    raise RuntimeError(
        "STOP — Expected exactly 50 comparison records."
    )

if comparison_df.duplicated(
    subset=[
        "candidate_id",
        "question_id"
    ]
).any():

    raise RuntimeError(
        "STOP — Duplicate candidate/question slot."
    )

print("✓ 50 rows")
print("✓ 50 unique candidate × question slots")


# ================================================================================================================
# STEP 19 — ELIGIBILITY DISTRIBUTION
# ================================================================================================================

print()
print("[STEP 19] COMPARISON ELIGIBILITY")
print("-" * 112)

eligibility_counts = (
    comparison_df[
        "comparison_eligibility"
    ]
    .value_counts()
)

print(
    eligibility_counts.to_string()
)

if int(
    eligibility_counts.get(
        "ELIGIBLE_PENDING_HUMAN_ASSESSMENT",
        0
    )
) != 47:

    raise RuntimeError(
        "STOP — Expected 47 eligible slots."
    )

if int(
    eligibility_counts.get(
        "BLOCKED_BY_HUMAN_CORRESPONDENCE",
        0
    )
) != 3:

    raise RuntimeError(
        "STOP — Expected 3 correspondence-blocked slots."
    )

print()
print("✓ 47 eligible")
print("✓ 3 blocked")


# ================================================================================================================
# STEP 20 — PERSIST
# ================================================================================================================

print()
print("[STEP 20] PERSIST CELL18 V001")
print("-" * 112)

CELL18_V001_DIR.mkdir(
    parents=False,
    exist_ok=False
)

comparison_df.to_parquet(
    CELL18_V001_PARQUET,
    index=False
)

comparison_df.to_csv(
    CELL18_V001_CSV,
    index=False
)

print("✓ Parquet written")
print("✓ CSV written")


# ================================================================================================================
# STEP 21 — OUTPUT HASHES
# ================================================================================================================

print()
print("[STEP 21] OUTPUT HASHES")
print("-" * 112)

cell18_parquet_sha = sha256_file(
    CELL18_V001_PARQUET
)

cell18_csv_sha = sha256_file(
    CELL18_V001_CSV
)

print("Cell18 V001 Parquet SHA256:")
print(cell18_parquet_sha)

print()
print("Cell18 V001 CSV SHA256:")
print(cell18_csv_sha)


# ================================================================================================================
# STEP 22 — MANIFEST
# ================================================================================================================

print()
print("[STEP 22] WRITE MANIFEST")
print("-" * 112)

manifest = {

    "cell":
        "CELL 18",

    "version":
        "V001",

    "status":
        "CONTROLLED_COMPARISON_FRAME_PASS",

    "scientific_value":
        True,

    "purpose":
        (
            "Prepare a controlled source-grounded methodological "
            "comparison matrix before human methodological assessment."
        ),

    "trusted_input": {

        "cell17_v002_path":
            str(CELL17_V002_PARQUET),

        "cell17_v002_sha256":
            cell17_v002_sha_actual,
    },

    "candidate_count":
        5,

    "question_count":
        10,

    "slot_count":
        50,

    "eligible_for_human_comparison":
        47,

    "blocked_by_human_correspondence":
        3,

    "comparison_dimensions":
        list(COMPARISON_DIMENSIONS.keys()),

    "overlap_scale":
        list(OVERLAP_SCALE.keys()),

    "automatic_novelty_judgment":
        False,

    "automatic_equivalence_judgment":
        False,

    "automatic_overlap_classification":
        False,

    "automatic_methodological_inference":
        False,

    "dataset_modified":
        False,

    "199_cohort_modified":
        False,

    "400_validation_reopened":
        False,

    "gold_created":
        False,

    "y_plus_created":
        False,

    "y_minus_created":
        False,

    "preference_created":
        False,

    "output": {

        "parquet":
            str(CELL18_V001_PARQUET),

        "csv":
            str(CELL18_V001_CSV),

        "parquet_sha256":
            cell18_parquet_sha,

        "csv_sha256":
            cell18_csv_sha,
    },

    "created_utc":
        datetime.now(
            timezone.utc
        ).isoformat(),
}


with open(
    CELL18_V001_MANIFEST,
    "w",
    encoding="utf-8"
) as f:

    json.dump(
        manifest,
        f,
        indent=2,
        ensure_ascii=False
    )

print("✓ Manifest written")


# ================================================================================================================
# STEP 23 — INDEPENDENT READ-BACK
# ================================================================================================================

print()
print("[STEP 23] INDEPENDENT READ-BACK")
print("-" * 112)

readback = pd.read_parquet(
    CELL18_V001_PARQUET
)

if len(readback) != 50:

    raise RuntimeError(
        "STOP — Read-back row count mismatch."
    )

readback_slots = set(
    zip(
        readback[
            "candidate_id"
        ].astype(str),

        readback[
            "question_id"
        ].astype(str)
    )
)

if readback_slots != expected_slots:

    raise RuntimeError(
        "STOP — Read-back slot set mismatch."
    )


# Verify three blocked decisions.

for candidate_id, question_id in [
    ("ANCHOR-001", "Q05"),
    ("ANCHOR-001", "Q06"),
    ("ANCHOR-001", "Q08"),
]:

    row = readback[
        (
            readback[
                "candidate_id"
            ].astype(str)
            ==
            candidate_id
        )
        &
        (
            readback[
                "question_id"
            ].astype(str)
            ==
            question_id
        )
    ]

    if len(row) != 1:

        raise RuntimeError(
            f"STOP — Missing read-back row: "
            f"{candidate_id}/{question_id}"
        )

    if str(
        row.iloc[0][
            "comparison_eligibility"
        ]
    ) != "BLOCKED_BY_HUMAN_CORRESPONDENCE":

        raise RuntimeError(
            f"STOP — Correspondence firewall lost for "
            f"{candidate_id}/{question_id}"
        )


# Verify all 47 eligible rows.

eligible_readback = (
    readback[
        "comparison_eligibility"
    ]
    ==
    "ELIGIBLE_PENDING_HUMAN_ASSESSMENT"
)

if int(
    eligible_readback.sum()
) != 47:

    raise RuntimeError(
        "STOP — Read-back eligible count is not 47."
    )


# Verify no automatic methodological judgments.

for field_name in comparison_fields:

    allowed = {
        "PENDING_HUMAN_ASSESSMENT",
        "NOT_ASSESSED_DUE_TO_CORRESPONDENCE",
    }

    if not set(
        readback[
            field_name
        ].astype(str)
    ).issubset(allowed):

        raise RuntimeError(
            f"STOP — Read-back contains an unexpected "
            f"methodological judgment in {field_name}."
        )


print("✓ 50 rows read back")
print("✓ 50 unique slots read back")
print("✓ 47 eligible slots preserved")
print("✓ 3 correspondence-blocked slots preserved")
print("✓ No automatic methodological judgments")


# ================================================================================================================
# STEP 24 — HASH STABILITY
# ================================================================================================================

print()
print("[STEP 24] HASH STABILITY")
print("-" * 112)

cell18_parquet_sha_recheck = sha256_file(
    CELL18_V001_PARQUET
)

if (
    cell18_parquet_sha_recheck
    !=
    cell18_parquet_sha
):

    raise RuntimeError(
        "STOP — Cell18 V001 Parquet SHA changed."
    )

print("✓ Cell18 V001 Parquet SHA stable")
print(
    "SHA256:",
    cell18_parquet_sha_recheck
)


# ================================================================================================================
# STEP 25 — FINAL GO / NO-GO
# ================================================================================================================

print()
print("=" * 112)
print("CELL 18 V001 — FINAL GO / NO-GO")
print("=" * 112)

print()

print(
    "Scientific Value                  : PASS"
)

print(
    "Cell17 V002 Trusted Baseline     : PASS"
)

print(
    "Legacy Cell18 Used               : NO"
)

print(
    "Candidates                        : 5 / 5"
)

print(
    "Questions                         : 10 / 10"
)

print(
    "Comparison Slots                 : 50 / 50"
)

print(
    "Eligible Human Comparison        : 47"
)

print(
    "Blocked by Human Correspondence  : 3"
)

print(
    "Automatic Novelty Judgment       : BLOCKED"
)

print(
    "Automatic Equivalence Judgment   : BLOCKED"
)

print(
    "Automatic Overlap Classification : BLOCKED"
)

print(
    "Automatic Methodological Inference: BLOCKED"
)

print(
    "Dataset Modification             : BLOCKED"
)

print(
    "199 Cohort Modification           : BLOCKED"
)

print(
    "400 Validation Reopen             : BLOCKED"
)

print(
    "Gold / Y+ / Y- / Preference      : BLOCKED"
)

print(
    "Independent Read-Back            : PASS"
)

print(
    "Output Hash Stability             : PASS"
)

print()
print("FINAL STATUS:")
print()

print(
    "CELL 18 V001 = CONTROLLED_COMPARISON_FRAME_PASS"
)

print()
print("NEXT AUTHORIZED STAGE:")
print()

print(
    "Human source-grounded methodological assessment "
    "of the 47 eligible comparison slots."
)

print()
print("=" * 112)
print("CELL 18 V001 — END")
print("=" * 112)
```

```python
# ================================================================================================================
# CELL 19 V009 — HUMAN SOURCE-GROUNDED METHODOLOGICAL ASSESSMENT WORKSPACE
# CLEAN REBUILD FROM TRUSTED CELL18 V001
# ================================================================================================================

import os
import json
import hashlib
import pandas as pd

print("=" * 100)
print("CELL 19 V009 — HUMAN SOURCE-GROUNDED METHODOLOGICAL ASSESSMENT WORKSPACE")
print("=" * 100)

# ================================================================================================================
# STEP 0 — SCIENTIFIC VALUE GATE
# ================================================================================================================

print("\n[STEP 0] SCIENTIFIC VALUE GATE")
print("-" * 100)

SCIENTIFIC_VALUE = True

print("Scientific value : PASS")
print(
    "Purpose          : Create a frozen source-grounded human-assessment workspace "
    "for the 47 eligible Cell18 methodological comparison records across D01-D10."
)

assert SCIENTIFIC_VALUE

# ================================================================================================================
# STEP 1 — TRUSTED CELL18 ARTIFACTS
# ================================================================================================================

PROJECT_ROOT = (
    "/content/drive/MyDrive/X-AutoAlign_Legal_RAG"
)

CELL18_DIR = (
    PROJECT_ROOT +
    "/04_AUDIT/NOVELTY_GAP_AUDIT/CHECKPOINTS/"
    "CELL18_CONTROLLED_METHODOLOGICAL_COMPARISON_V001"
)

CELL18_PARQUET = (
    CELL18_DIR +
    "/CELL18_CONTROLLED_METHODOLOGICAL_COMPARISON_V001.parquet"
)

CELL18_CSV = (
    CELL18_DIR +
    "/CELL18_CONTROLLED_METHODOLOGICAL_COMPARISON_V001.csv"
)

CELL18_MANIFEST = (
    CELL18_DIR +
    "/CELL18_MANIFEST_V001.json"
)

print("\n[STEP 1] TRUSTED CELL18 ARTIFACTS")
print("-" * 100)

for label, path in [
    ("Cell18 directory", CELL18_DIR),
    ("Cell18 Parquet", CELL18_PARQUET),
    ("Cell18 CSV", CELL18_CSV),
    ("Cell18 Manifest", CELL18_MANIFEST),
]:
    print(f"{label:20s}: {path}")
    assert os.path.exists(path), (
        f"STOP — REQUIRED TRUSTED ARTIFACT MISSING:\n{path}"
    )

print("✓ All trusted Cell18 artifacts physically exist")

# ================================================================================================================
# STEP 2 — CELL18 SHA256
# ================================================================================================================

def sha256_file(path, chunk_size=1024 * 1024):

    h = hashlib.sha256()

    with open(path, "rb") as f:

        while True:

            chunk = f.read(chunk_size)

            if not chunk:
                break

            h.update(chunk)

    return h.hexdigest()


EXPECTED_CELL18_SHA = (
    "16dec31f6aa158c3a87891d0347564de34f408322a595378fbce7539fdc41cf7"
)

actual_cell18_sha = sha256_file(
    CELL18_PARQUET
)

print("\n[STEP 2] CELL18 PARQUET SHA256")
print("-" * 100)

print("Expected :", EXPECTED_CELL18_SHA)
print("Actual   :", actual_cell18_sha)

assert actual_cell18_sha == EXPECTED_CELL18_SHA, (
    "STOP — Cell18 Parquet SHA mismatch.\n"
    "NO PATCHING / NO GUESSING."
)

print("✓ Cell18 Parquet SHA verified")

# ================================================================================================================
# STEP 3 — CELL18 MANIFEST TRUST CHECK
# ================================================================================================================

print("\n[STEP 3] CELL18 MANIFEST TRUST CHECK")
print("-" * 100)

with open(
    CELL18_MANIFEST,
    "r",
    encoding="utf-8"
) as f:

    cell18_manifest = json.load(f)

# Exact values established by Cell19B V001.

assert cell18_manifest["cell"] == "CELL 18"
assert cell18_manifest["version"] == "V001"
assert cell18_manifest["status"] == (
    "CONTROLLED_COMPARISON_FRAME_PASS"
)

assert cell18_manifest["scientific_value"] is True

assert cell18_manifest["candidate_count"] == 5
assert cell18_manifest["question_count"] == 10
assert cell18_manifest["slot_count"] == 50

assert cell18_manifest[
    "eligible_for_human_comparison"
] == 47

assert cell18_manifest[
    "blocked_by_human_correspondence"
] == 3

# Automatic judgment protections

assert cell18_manifest[
    "automatic_novelty_judgment"
] is False

assert cell18_manifest[
    "automatic_equivalence_judgment"
] is False

assert cell18_manifest[
    "automatic_overlap_classification"
] is False

assert cell18_manifest[
    "automatic_methodological_inference"
] is False

# Project protections

assert cell18_manifest[
    "dataset_modified"
] is False

assert cell18_manifest[
    "199_cohort_modified"
] is False

assert cell18_manifest[
    "400_validation_reopened"
] is False

assert cell18_manifest[
    "gold_created"
] is False

assert cell18_manifest[
    "y_plus_created"
] is False

assert cell18_manifest[
    "y_minus_created"
] is False

assert cell18_manifest[
    "preference_created"
] is False

print("Manifest cell    :", cell18_manifest["cell"])
print("Manifest version :", cell18_manifest["version"])
print("Manifest status  :", cell18_manifest["status"])
print("Candidates       :", cell18_manifest["candidate_count"])
print("Questions        :", cell18_manifest["question_count"])
print("Slots            :", cell18_manifest["slot_count"])
print("Eligible         :", cell18_manifest["eligible_for_human_comparison"])
print("Blocked          :", cell18_manifest["blocked_by_human_correspondence"])

print("✓ Manifest identity verified")
print("✓ Manifest safety flags verified")

# ================================================================================================================
# STEP 4 — LOAD CELL18
# ================================================================================================================

print("\n[STEP 4] CELL18 DATA LOAD")
print("-" * 100)

df = pd.read_parquet(
    CELL18_PARQUET
)

print("Shape:", df.shape)

assert df.shape == (
    50,
    43
)

# ================================================================================================================
# STEP 5 — EXACT 43-COLUMN SCHEMA
# ================================================================================================================

EXPECTED_COLUMNS = [
    "candidate_id",
    "question_id",
    "title",
    "authors",
    "year",
    "venue",
    "primary_source",
    "primary_source_filename",
    "primary_source_path",
    "primary_source_sha256",
    "question_text",
    "source_excerpt",
    "source_location",
    "source_text_status",
    "source_text_unit_start",
    "source_text_unit_end",
    "extraction_method",
    "extraction_interpretation",
    "extraction_status",
    "source_paraphrase",
    "methodological_note",
    "equivalence_judgment",
    "novelty_judgment",
    "overlap_class",
    "human_correspondence_status",
    "methodological_comparison_status",
    "novelty_status",
    "equivalence_status",
    "overlap_status",
    "comparison_eligibility",
    "post_retrieval_stage_assessment",
    "verification_input_assessment",
    "relevance_sufficiency_assessment",
    "legal_domain_support_assessment",
    "evidence_sufficiency_assessment",
    "contradiction_conflict_risk_assessment",
    "pre_generation_assessment",
    "eligibility_risk_abstention_assessment",
    "human_grounded_suitability_assessment",
    "overall_fdr_equivalence_assessment",
    "source_grounding_status",
    "assessment_method",
    "automatic_judgment",
]

print("\n[STEP 5] EXACT CELL18 SCHEMA")
print("-" * 100)

assert list(df.columns) == EXPECTED_COLUMNS, (
    "STOP — Cell18 schema mismatch.\n"
    "NO PATCHING / NO COLUMN RECONSTRUCTION."
)

print("✓ Exact 43-column schema verified")

# ================================================================================================================
# STEP 6 — IDENTITY / SLOT INTEGRITY
# ================================================================================================================

print("\n[STEP 6] IDENTITY / SLOT INTEGRITY")
print("-" * 100)

EXPECTED_CANDIDATES = [
    "ANCHOR-001",
    "LIT-001",
    "LIT-002",
    "LIT-003",
    "LIT-004",
]

EXPECTED_QUESTIONS = [
    f"Q{i:02d}"
    for i in range(1, 11)
]

assert sorted(
    df["candidate_id"].unique().tolist()
) == sorted(
    EXPECTED_CANDIDATES
)

assert sorted(
    df["question_id"].unique().tolist()
) == sorted(
    EXPECTED_QUESTIONS
)

slot_counts = (
    df.groupby(
        ["candidate_id", "question_id"]
    ).size()
)

assert len(df) == 50
assert len(slot_counts) == 50
assert slot_counts.eq(1).all()

print("Candidates  :", df["candidate_id"].nunique())
print("Questions   :", df["question_id"].nunique())
print("Rows        :", len(df))
print("Unique slots:", len(slot_counts))

print("✓ 50 unique candidate × question slots verified")

# ================================================================================================================
# STEP 7 — EXACT PROTECTED VALUE CONTRACT
# ================================================================================================================

print("\n[STEP 7] EXACT PROTECTED VALUE CONTRACT")
print("-" * 100)

EXPECTED_PROTECTED_VALUES = {

    "equivalence_judgment": {
        "NOT_ASSESSED": 50
    },

    "novelty_judgment": {
        "NOT_ASSESSED": 50
    },

    "overlap_class": {
        "PENDING_HUMAN_ASSESSMENT": 47,
        "NOT_ASSESSED_DUE_TO_CORRESPONDENCE": 3
    },

    "human_correspondence_status": {
        "NOT_REVIEWED": 47,
        "DOES_NOT_CORRESPOND": 3
    },

    "methodological_comparison_status": {
        "NOT_ASSESSED": 50
    },

    "novelty_status": {
        "NOT_ASSESSED": 50
    },

    "equivalence_status": {
        "NOT_ASSESSED": 50
    },

    "overlap_status": {
        "NOT_ASSESSED": 50
    },

    "comparison_eligibility": {
        "ELIGIBLE_PENDING_HUMAN_ASSESSMENT": 47,
        "BLOCKED_BY_HUMAN_CORRESPONDENCE": 3
    },

    "source_grounding_status": {
        "SOURCE_GROUNDED_EVIDENCE_AVAILABLE": 50
    },

    "assessment_method": {
        "HUMAN_SOURCE_GROUNDED_COMPARISON": 50
    },

    "automatic_judgment": {
        "False": 50
    },
}

for col, expected_counts in EXPECTED_PROTECTED_VALUES.items():

    actual_counts = (
        df[col]
        .astype(str)
        .value_counts()
        .to_dict()
    )

    print(f"\n{col}")

    print("Expected:", expected_counts)
    print("Actual  :", actual_counts)

    assert actual_counts == expected_counts, (
        f"STOP — Protected value contract mismatch in {col}.\n"
        "NO PATCHING / NO GUESSING."
    )

print("\n✓ Complete protected value contract verified")

# ================================================================================================================
# STEP 8 — ELIGIBILITY PARTITION
# ================================================================================================================

print("\n[STEP 8] ELIGIBILITY PARTITION")
print("-" * 100)

ELIGIBLE_STATUS = (
    "ELIGIBLE_PENDING_HUMAN_ASSESSMENT"
)

BLOCKED_STATUS = (
    "BLOCKED_BY_HUMAN_CORRESPONDENCE"
)

eligible_df = df.loc[
    df["comparison_eligibility"].eq(
        ELIGIBLE_STATUS
    )
].copy()

blocked_df = df.loc[
    df["comparison_eligibility"].eq(
        BLOCKED_STATUS
    )
].copy()

assert len(eligible_df) == 47
assert len(blocked_df) == 3

assert eligible_df[
    "overlap_class"
].eq(
    "PENDING_HUMAN_ASSESSMENT"
).all()

assert blocked_df[
    "overlap_class"
].eq(
    "NOT_ASSESSED_DUE_TO_CORRESPONDENCE"
).all()

print("Eligible records :", len(eligible_df))
print("Blocked records  :", len(blocked_df))

print("✓ Eligibility partition verified")
print("✓ Overlap-class state correctly preserved")

# ================================================================================================================
# STEP 9 — CORRESPONDENCE FIREWALL
# ================================================================================================================

print("\n[STEP 9] HUMAN CORRESPONDENCE FIREWALL")
print("-" * 100)

anchor_blocked = df.loc[
    df["candidate_id"].eq(
        "ANCHOR-001"
    )
    &
    df["question_id"].isin(
        ["Q05", "Q06", "Q08"]
    )
]

assert len(anchor_blocked) == 3

assert anchor_blocked[
    "human_correspondence_status"
].eq(
    "DOES_NOT_CORRESPOND"
).all()

assert anchor_blocked[
    "comparison_eligibility"
].eq(
    BLOCKED_STATUS
).all()

print("✓ ANCHOR-001 Q05/Q06/Q08 preserved")
print("✓ No correspondence decision modified")

# ================================================================================================================
# STEP 10 — METHODOLOGICAL JUDGMENT FIREWALL
# ================================================================================================================

print("\n[STEP 10] METHODOLOGICAL JUDGMENT FIREWALL")
print("-" * 100)

assert df[
    "equivalence_judgment"
].eq(
    "NOT_ASSESSED"
).all()

assert df[
    "novelty_judgment"
].eq(
    "NOT_ASSESSED"
).all()

assert df[
    "methodological_comparison_status"
].eq(
    "NOT_ASSESSED"
).all()

assert df[
    "novelty_status"
].eq(
    "NOT_ASSESSED"
).all()

assert df[
    "equivalence_status"
].eq(
    "NOT_ASSESSED"
).all()

assert df[
    "overlap_status"
].eq(
    "NOT_ASSESSED"
).all()

print("✓ No methodological judgment")
print("✓ No novelty judgment")
print("✓ No equivalence judgment")
print("✓ No overlap status judgment")

# ================================================================================================================
# STEP 11 — AUTOMATIC JUDGMENT FIREWALL
# ================================================================================================================

print("\n[STEP 11] AUTOMATIC JUDGMENT FIREWALL")
print("-" * 100)

assert df[
    "automatic_judgment"
].astype(str).eq(
    "False"
).all()

print("✓ automatic_judgment = False for all 50")

# ================================================================================================================
# STEP 12 — FROZEN Q → D MAPPING
# ================================================================================================================

print("\n[STEP 12] FROZEN Q → D MAPPING")
print("-" * 100)

Q_TO_D = {
    "Q01": "D01",
    "Q02": "D02",
    "Q03": "D03",
    "Q04": "D04",
    "Q05": "D05",
    "Q06": "D06",
    "Q07": "D07",
    "Q08": "D08",
    "Q09": "D09",
    "Q10": "D10",
}

for q, d in Q_TO_D.items():
    print(f"{q} → {d}")

assert set(
    Q_TO_D.keys()
) == set(
    EXPECTED_QUESTIONS
)

print("✓ Frozen Q01–Q10 → D01–D10 verified")

# ================================================================================================================
# STEP 13 — HUMAN ASSESSMENT FIELD INITIALIZATION
# ================================================================================================================

print("\n[STEP 13] HUMAN ASSESSMENT FIELD INITIALIZATION")
print("-" * 100)

ASSESSMENT_COLUMNS = [
    "post_retrieval_stage_assessment",
    "verification_input_assessment",
    "relevance_sufficiency_assessment",
    "legal_domain_support_assessment",
    "evidence_sufficiency_assessment",
    "contradiction_conflict_risk_assessment",
    "pre_generation_assessment",
    "eligibility_risk_abstention_assessment",
    "human_grounded_suitability_assessment",
    "overall_fdr_equivalence_assessment",
]

assert len(ASSESSMENT_COLUMNS) == 10

for col in ASSESSMENT_COLUMNS:

    assert col in EXPECTED_COLUMNS

    eligible_df[col] = (
        "NOT_ASSESSED"
    )

eligible_df[
    "assessment_method"
] = (
    "HUMAN_SOURCE_GROUNDED_REVIEW_PENDING"
)

eligible_df[
    "automatic_judgment"
] = False

assert all(
    eligible_df[col].eq(
        "NOT_ASSESSED"
    ).all()
    for col in ASSESSMENT_COLUMNS
)

print("Dimensions initialized :", len(ASSESSMENT_COLUMNS))
print("Eligible records       :", len(eligible_df))
print("Initial state          : NOT_ASSESSED")

print("✓ All 10 dimensions initialized")

# ================================================================================================================
# STEP 14 — REVIEW QUEUE
# ================================================================================================================

print("\n[STEP 14] HUMAN REVIEW QUEUE")
print("-" * 100)

REVIEW_COLUMNS = [
    "candidate_id",
    "question_id",
    "title",
    "authors",
    "year",
    "venue",
    "primary_source",
    "primary_source_filename",
    "primary_source_path",
    "primary_source_sha256",
    "question_text",
    "source_excerpt",
    "source_location",
    "source_text_status",
    "source_text_unit_start",
    "source_text_unit_end",
    "extraction_method",
    "extraction_interpretation",
    "extraction_status",
    "methodological_note",
    "comparison_eligibility",
    *ASSESSMENT_COLUMNS,
    "assessment_method",
    "automatic_judgment",
]

review_queue = eligible_df[
    REVIEW_COLUMNS
].copy()

assert len(review_queue) == 47

assert review_queue.groupby(
    ["candidate_id", "question_id"]
).size().eq(1).all()

print("Review queue rows    :", len(review_queue))
print("Review queue columns :", len(review_queue.columns))

print("✓ 47 unique human-review records")

# ================================================================================================================
# STEP 15 — OUTPUT PATHS
# ================================================================================================================

OUTPUT_DIR = (
    PROJECT_ROOT +
    "/04_AUDIT/NOVELTY_GAP_AUDIT/CHECKPOINTS/"
    "CELL19_HUMAN_SOURCE_GROUNDED_METHODOLOGICAL_ASSESSMENT_V009"
)

os.makedirs(
    OUTPUT_DIR,
    exist_ok=True
)

OUTPUT_PARQUET = (
    OUTPUT_DIR +
    "/CELL19_HUMAN_SOURCE_GROUNDED_METHODOLOGICAL_ASSESSMENT_V009.parquet"
)

OUTPUT_CSV = (
    OUTPUT_DIR +
    "/CELL19_HUMAN_SOURCE_GROUNDED_METHODOLOGICAL_ASSESSMENT_V009.csv"
)

OUTPUT_MANIFEST = (
    OUTPUT_DIR +
    "/CELL19_MANIFEST_V009.json"
)

# ================================================================================================================
# STEP 16 — PERSIST
# ================================================================================================================

print("\n[STEP 16] PERSIST")
print("-" * 100)

review_queue.to_parquet(
    OUTPUT_PARQUET,
    index=False
)

review_queue.to_csv(
    OUTPUT_CSV,
    index=False,
    encoding="utf-8-sig"
)

print("✓ Parquet written")
print("✓ CSV written")

# ================================================================================================================
# STEP 17 — CELL19 MANIFEST
# ================================================================================================================

cell19_manifest = {

    "cell": "CELL 19",

    "version": "V009",

    "status":
        "HUMAN_ASSESSMENT_WORKSPACE_INITIALIZED",

    "scientific_value":
        True,

    "trusted_input": {
        "cell": "CELL 18",
        "version": "V001",
        "parquet": CELL18_PARQUET,
        "parquet_sha256": actual_cell18_sha,
        "manifest": CELL18_MANIFEST,
    },

    "input_rows": 50,

    "eligible_status":
        ELIGIBLE_STATUS,

    "eligible_rows":
        47,

    "blocked_status":
        BLOCKED_STATUS,

    "blocked_rows":
        3,

    "review_queue_rows":
        47,

    "candidate_count":
        5,

    "question_count":
        10,

    "slot_count":
        50,

    "q_to_d_mapping":
        Q_TO_D,

    "assessment_dimensions":
        ASSESSMENT_COLUMNS,

    "initial_state":
        "NOT_ASSESSED",

    "automatic_judgment":
        False,

    "automatic_methodological_inference":
        False,

    "automatic_novelty_judgment":
        False,

    "automatic_equivalence_judgment":
        False,

    "automatic_overlap_classification":
        False,

    "dataset_modified":
        False,

    "199_cohort_modified":
        False,

    "400_validation_reopened":
        False,

    "gold_created":
        False,

    "y_plus_created":
        False,

    "y_minus_created":
        False,

    "preference_created":
        False,

    "firewall":
        "NO_PATCHING_NO_GUESSING",
}

with open(
    OUTPUT_MANIFEST,
    "w",
    encoding="utf-8"
) as f:

    json.dump(
        cell19_manifest,
        f,
        ensure_ascii=False,
        indent=2
    )

print("✓ Cell19 manifest written")

# ================================================================================================================
# STEP 18 — INDEPENDENT READ-BACK
# ================================================================================================================

print("\n[STEP 18] INDEPENDENT READ-BACK")
print("-" * 100)

rb = pd.read_parquet(
    OUTPUT_PARQUET
)

assert rb.shape == review_queue.shape

assert list(
    rb.columns
) == list(
    review_queue.columns
)

assert len(rb) == 47

assert rb[
    "comparison_eligibility"
].eq(
    ELIGIBLE_STATUS
).all()

for col in ASSESSMENT_COLUMNS:

    assert rb[col].eq(
        "NOT_ASSESSED"
    ).all()

assert rb[
    "automatic_judgment"
].astype(str).eq(
    "False"
).all()

assert rb.groupby(
    ["candidate_id", "question_id"]
).size().eq(1).all()

print("✓ Read-back shape PASS")
print("✓ Read-back schema PASS")
print("✓ Read-back eligibility PASS")
print("✓ Read-back 47 unique slots PASS")
print("✓ Read-back assessment firewall PASS")
print("✓ Read-back automatic judgment firewall PASS")

# ================================================================================================================
# STEP 19 — OUTPUT SHA
# ================================================================================================================

output_sha = sha256_file(
    OUTPUT_PARQUET
)

print("\n[STEP 19] OUTPUT SHA256")
print("-" * 100)

print(
    "Cell19 V009 Parquet SHA256:",
    output_sha
)

# ================================================================================================================
# FINAL GO / NO-GO
# ================================================================================================================

print("\n" + "=" * 100)
print("CELL 19 V009 — FINAL GO / NO-GO")
print("=" * 100)

print("Scientific value                       : PASS")
print("Cell18 artifacts                       : PASS")
print("Cell18 Parquet SHA                     : PASS")
print("Cell18 Manifest identity               : PASS")
print("Cell18 schema (43 columns)             : PASS")
print("Cell18 rows                             : PASS (50)")
print("Candidate identity                     : PASS (5)")
print("Question identity                      : PASS (Q01–Q10)")
print("Unique slots                            : PASS (50)")
print("Protected value contract               : PASS")
print("Eligibility partition                  : PASS (47 + 3)")
print("Human correspondence firewall           : PASS")
print("Methodological judgment firewall        : PASS")
print("Automatic judgment firewall             : PASS")
print("Q01 → D01 ... Q10 → D10                : PASS")
print("Human assessment initialization         : PASS")
print("47-record review queue                  : PASS")
print("Independent read-back                  : PASS")
print("Output persistence                     : PASS")

print("\nFINAL STATUS:")
print("✓ CELL 19 V009 = HUMAN ASSESSMENT WORKSPACE READY")
print("✓ NO methodological judgment performed.")
print("✓ NO novelty/equivalence/overlap conclusion generated.")
print("✓ NO dataset modified.")
print("✓ NO 199 cohort modified.")
print("✓ NO 400 validation reopened.")
print("✓ NO Gold / Y+ / Y− / preference generated.")

print("\nOutput directory:")
print(OUTPUT_DIR)

print("\nManifest:")
print(OUTPUT_MANIFEST)

print("=" * 100)
```

```python
# ================================================================================================================
# CELL 20A V001 — CLEAN STRUCTURAL GATE FOR CELL20
# READ ONLY — NO PATCHING / NO GUESSING / NO ARTIFACT MODIFICATION
#
# PURPOSE:
#   Establish the exact structural assessment unit before building Cell20 human-review artifacts.
#
#   Specifically:
#     1. Validate trusted Cell19 V009.
#     2. Verify the 47 eligible source-grounded records.
#     3. Inspect candidate × question completeness.
#     4. Map Q01-Q10 to D01-D10.
#     5. Determine which D01-D09 records actually exist.
#     6. Verify whether D10 can be treated as a candidate-level synthesis.
#     7. DO NOT create any assessment form.
#     8. DO NOT generate any methodological / novelty / equivalence judgment.
#
# FIREWALL:
#   NO DATASET MODIFICATION
#   NO 199-COHORT MODIFICATION
#   NO 400-VALIDATION REOPEN
#   NO GOLD / Y+ / Y- / PREFERENCE
#   NO HUMAN JUDGMENT
#   NO AUTOMATIC NOVELTY / EQUIVALENCE / OVERLAP
# ================================================================================================================

import os
import json
import hashlib
import pandas as pd

print("=" * 100)
print("CELL 20A V001 — CLEAN STRUCTURAL GATE FOR CELL20")
print("=" * 100)

# ================================================================================================================
# STEP 0 — SCIENTIFIC VALUE GATE
# ================================================================================================================

print("\n[STEP 0] SCIENTIFIC VALUE GATE")
print("-" * 100)

SCIENTIFIC_VALUE = True

SCIENTIFIC_PURPOSE = (
    "Establish the exact structural assessment unit for Cell20 from trusted Cell19 V009 "
    "before any human assessment artifact is created, with special validation of the D10 "
    "overall-assessment dependency on D01-D09."
)

print("Scientific value :", "PASS" if SCIENTIFIC_VALUE else "FAIL")
print("Purpose          :", SCIENTIFIC_PURPOSE)

assert SCIENTIFIC_VALUE

# ================================================================================================================
# STEP 1 — TRUSTED CELL19 V009
# ================================================================================================================

PROJECT_ROOT = (
    "/content/drive/MyDrive/X-AutoAlign_Legal_RAG"
)

CELL19_DIR = (
    PROJECT_ROOT +
    "/04_AUDIT/NOVELTY_GAP_AUDIT/CHECKPOINTS/"
    "CELL19_HUMAN_SOURCE_GROUNDED_METHODOLOGICAL_ASSESSMENT_V009"
)

CELL19_PARQUET = (
    CELL19_DIR +
    "/CELL19_HUMAN_SOURCE_GROUNDED_METHODOLOGICAL_ASSESSMENT_V009.parquet"
)

CELL19_MANIFEST = (
    CELL19_DIR +
    "/CELL19_MANIFEST_V009.json"
)

print("\n[STEP 1] TRUSTED CELL19 V009")
print("-" * 100)

assert os.path.isdir(CELL19_DIR), (
    f"STOP — Cell19 directory missing:\n{CELL19_DIR}"
)

assert os.path.isfile(CELL19_PARQUET), (
    f"STOP — Cell19 Parquet missing:\n{CELL19_PARQUET}"
)

assert os.path.isfile(CELL19_MANIFEST), (
    f"STOP — Cell19 Manifest missing:\n{CELL19_MANIFEST}"
)

print("Cell19 directory :", CELL19_DIR)
print("Cell19 Parquet   :", CELL19_PARQUET)
print("Cell19 Manifest  :", CELL19_MANIFEST)

print("✓ Trusted Cell19 artifacts exist")

# ================================================================================================================
# STEP 2 — CELL19 SHA
# ================================================================================================================

print("\n[STEP 2] CELL19 SHA256")
print("-" * 100)

def sha256_file(path, chunk_size=1024 * 1024):
    h = hashlib.sha256()

    with open(path, "rb") as f:
        while True:
            chunk = f.read(chunk_size)

            if not chunk:
                break

            h.update(chunk)

    return h.hexdigest()


EXPECTED_CELL19_SHA = (
    "f45efc2555929932f417c7152baa358f1d026d2e0b6cceca5f354a1daa0fd670"
)

actual_cell19_sha = sha256_file(
    CELL19_PARQUET
)

print("Expected :", EXPECTED_CELL19_SHA)
print("Actual   :", actual_cell19_sha)

assert actual_cell19_sha == EXPECTED_CELL19_SHA, (
    "STOP — Cell19 V009 SHA mismatch.\n"
    "NO PATCHING / NO GUESSING."
)

print("✓ Cell19 V009 SHA verified")

# ================================================================================================================
# STEP 3 — CELL19 MANIFEST
# ================================================================================================================

print("\n[STEP 3] CELL19 MANIFEST")
print("-" * 100)

with open(
    CELL19_MANIFEST,
    "r",
    encoding="utf-8"
) as f:
    manifest = json.load(f)

assert manifest["cell"] == "CELL 19"
assert manifest["version"] == "V009"
assert manifest["status"] == "HUMAN_ASSESSMENT_WORKSPACE_INITIALIZED"

assert manifest["input_rows"] == 50
assert manifest["eligible_rows"] == 47
assert manifest["blocked_rows"] == 3
assert manifest["review_queue_rows"] == 47

assert manifest["candidate_count"] == 5
assert manifest["question_count"] == 10
assert manifest["slot_count"] == 50

assert manifest["automatic_judgment"] is False
assert manifest["automatic_methodological_inference"] is False
assert manifest["automatic_novelty_judgment"] is False
assert manifest["automatic_equivalence_judgment"] is False
assert manifest["automatic_overlap_classification"] is False

assert manifest["dataset_modified"] is False
assert manifest["199_cohort_modified"] is False
assert manifest["400_validation_reopened"] is False
assert manifest["gold_created"] is False
assert manifest["y_plus_created"] is False
assert manifest["y_minus_created"] is False
assert manifest["preference_created"] is False

print("Manifest cell       :", manifest["cell"])
print("Manifest version    :", manifest["version"])
print("Manifest status     :", manifest["status"])
print("Input rows          :", manifest["input_rows"])
print("Eligible rows       :", manifest["eligible_rows"])
print("Blocked rows        :", manifest["blocked_rows"])
print("Review queue rows   :", manifest["review_queue_rows"])

print("✓ Cell19 manifest identity verified")
print("✓ Cell19 safety firewall verified")

# ================================================================================================================
# STEP 4 — LOAD CELL19
# ================================================================================================================

print("\n[STEP 4] LOAD CELL19")
print("-" * 100)

df = pd.read_parquet(
    CELL19_PARQUET
)

print("Shape:", df.shape)

assert df.shape == (47, 33)

print("✓ Cell19 shape verified: 47 × 33")

# ================================================================================================================
# STEP 5 — EXACT CELL19 SCHEMA
# ================================================================================================================

EXPECTED_COLUMNS = [
    "candidate_id",
    "question_id",
    "title",
    "authors",
    "year",
    "venue",
    "primary_source",
    "primary_source_filename",
    "primary_source_path",
    "primary_source_sha256",
    "question_text",
    "source_excerpt",
    "source_location",
    "source_text_status",
    "source_text_unit_start",
    "source_text_unit_end",
    "extraction_method",
    "extraction_interpretation",
    "extraction_status",
    "methodological_note",
    "comparison_eligibility",
    "post_retrieval_stage_assessment",
    "verification_input_assessment",
    "relevance_sufficiency_assessment",
    "legal_domain_support_assessment",
    "evidence_sufficiency_assessment",
    "contradiction_conflict_risk_assessment",
    "pre_generation_assessment",
    "eligibility_risk_abstention_assessment",
    "human_grounded_suitability_assessment",
    "overall_fdr_equivalence_assessment",
    "assessment_method",
    "automatic_judgment",
]

print("\n[STEP 5] EXACT CELL19 SCHEMA")
print("-" * 100)

assert list(df.columns) == EXPECTED_COLUMNS

print("✓ Exact 33-column schema verified")

# ================================================================================================================
# STEP 6 — ELIGIBILITY FIREWALL
# ================================================================================================================

print("\n[STEP 6] ELIGIBILITY FIREWALL")
print("-" * 100)

assert df["comparison_eligibility"].eq(
    "ELIGIBLE_PENDING_HUMAN_ASSESSMENT"
).all()

assert df["automatic_judgment"].astype(str).eq(
    "False"
).all()

print("✓ All 47 records are eligible pending human assessment")
print("✓ automatic_judgment = False for all records")

# ================================================================================================================
# STEP 7 — CANDIDATE / QUESTION INVENTORY
# ================================================================================================================

print("\n[STEP 7] CANDIDATE / QUESTION INVENTORY")
print("-" * 100)

candidates = sorted(
    df["candidate_id"].unique().tolist()
)

questions = sorted(
    df["question_id"].unique().tolist()
)

EXPECTED_CANDIDATES = [
    "ANCHOR-001",
    "LIT-001",
    "LIT-002",
    "LIT-003",
    "LIT-004",
]

EXPECTED_QUESTIONS = [
    f"Q{i:02d}"
    for i in range(1, 11)
]

assert candidates == EXPECTED_CANDIDATES
assert questions == EXPECTED_QUESTIONS

print("Candidates:")
for c in candidates:
    print("  •", c)

print("\nQuestions:")
for q in questions:
    print("  •", q)

print("\nCandidate count :", len(candidates))
print("Question count  :", len(questions))

print("✓ Candidate identity verified")
print("✓ Question identity verified")

# ================================================================================================================
# STEP 8 — EXACT Q → D MAPPING
# ================================================================================================================

print("\n[STEP 8] EXACT Q → D MAPPING")
print("-" * 100)

Q_TO_D = {
    "Q01": "D01",
    "Q02": "D02",
    "Q03": "D03",
    "Q04": "D04",
    "Q05": "D05",
    "Q06": "D06",
    "Q07": "D07",
    "Q08": "D08",
    "Q09": "D09",
    "Q10": "D10",
}

for q in EXPECTED_QUESTIONS:
    print(f"{q} → {Q_TO_D[q]}")

assert list(Q_TO_D.keys()) == EXPECTED_QUESTIONS

print("✓ Q01–Q10 mapping exactly covers D01–D10")

# ================================================================================================================
# STEP 9 — CANDIDATE × QUESTION COMPLETENESS MATRIX
# ================================================================================================================

print("\n[STEP 9] CANDIDATE × QUESTION COMPLETENESS MATRIX")
print("-" * 100)

presence = (
    df.assign(present=1)
      .pivot_table(
          index="candidate_id",
          columns="question_id",
          values="present",
          aggfunc="max",
          fill_value=0
      )
      .reindex(
          index=EXPECTED_CANDIDATES,
          columns=EXPECTED_QUESTIONS,
          fill_value=0
      )
)

print("\nPresence matrix (1 = record exists, 0 = absent):")
print(presence.to_string())

expected_presence = pd.DataFrame(
    1,
    index=EXPECTED_CANDIDATES,
    columns=EXPECTED_QUESTIONS
)

# Known blocked source-grounded records:
# ANCHOR-001 / Q05, Q06, Q08
expected_presence.loc[
    "ANCHOR-001",
    ["Q05", "Q06", "Q08"]
] = 0

assert presence.equals(
    expected_presence
), (
    "STOP — Candidate × question completeness does not match "
    "the trusted Cell19 eligibility structure."
)

print("\n✓ Candidate × question structure exactly matches the frozen 47-record population")

# ================================================================================================================
# STEP 10 — PER-CANDIDATE QUESTION COUNTS
# ================================================================================================================

print("\n[STEP 10] PER-CANDIDATE QUESTION COUNTS")
print("-" * 100)

candidate_question_counts = (
    df.groupby("candidate_id")["question_id"]
      .nunique()
      .reindex(EXPECTED_CANDIDATES)
)

print(candidate_question_counts.to_string())

assert candidate_question_counts.loc["ANCHOR-001"] == 7

for candidate in EXPECTED_CANDIDATES[1:]:
    assert candidate_question_counts.loc[candidate] == 10

print("\n✓ Anchor has 7 eligible question records")
print("✓ Each LIT candidate has all 10 question records")

# ================================================================================================================
# STEP 11 — DIMENSION AVAILABILITY D01-D09
# ================================================================================================================

print("\n[STEP 11] D01-D09 DIMENSION AVAILABILITY")
print("-" * 100)

D01_D09 = [
    "D01", "D02", "D03", "D04",
    "D05", "D06", "D07", "D08", "D09"
]

dimension_to_question = {
    d: q
    for q, d in Q_TO_D.items()
}

dimension_rows = []

for candidate in EXPECTED_CANDIDATES:

    available = []

    for dimension in D01_D09:

        q = dimension_to_question[dimension]

        if (
            df.loc[
                df["candidate_id"].eq(candidate),
                "question_id"
            ].eq(q).any()
        ):
            available.append(dimension)

    dimension_rows.append({
        "candidate_id": candidate,
        "available_dimensions": available,
        "available_count": len(available),
        "missing_dimensions": [
            d for d in D01_D09
            if d not in available
        ],
    })

dimension_availability = pd.DataFrame(
    dimension_rows
)

print(
    dimension_availability.to_string(
        index=False
    )
)

# Expected:
# ANCHOR = 6 D01-D09 dimensions because Q05/Q06/Q08 are absent.
# Each LIT = 9 D01-D09 dimensions.

assert (
    dimension_availability.loc[
        dimension_availability["candidate_id"].eq("ANCHOR-001"),
        "available_count"
    ].iloc[0]
    == 6
)

for candidate in EXPECTED_CANDIDATES[1:]:

    assert (
        dimension_availability.loc[
            dimension_availability["candidate_id"].eq(candidate),
            "available_count"
        ].iloc[0]
        == 9
    )

total_d01_d09_available = (
    dimension_availability["available_count"].sum()
)

assert total_d01_d09_available == 42

print("\nTotal available D01-D09 dimension records :", total_d01_d09_available)
print("✓ 42 available D01-D09 assessment inputs established")

# ================================================================================================================
# STEP 12 — D10 AVAILABILITY
# ================================================================================================================

print("\n[STEP 12] D10 AVAILABILITY")
print("-" * 100)

q10_records = df[
    df["question_id"].eq("Q10")
].copy()

print("Q10 records:", len(q10_records))

assert len(q10_records) == 5

assert sorted(
    q10_records["candidate_id"].tolist()
) == EXPECTED_CANDIDATES

print("✓ Exactly one Q10 record exists for each candidate")
print("✓ Q10 is available for all 5 candidates")

# ================================================================================================================
# STEP 13 — D10 DEPENDENCY STRUCTURE
# ================================================================================================================

print("\n[STEP 13] D10 DEPENDENCY STRUCTURE")
print("-" * 100)

print(
    "D10 is defined as an OVERALL_FDR_EQUIVALENCE assessment."
)

print(
    "Therefore the structural gate checks whether D10 can be associated "
    "with the candidate-level synthesis of available D01-D09 dimensions."
)

d10_dependency_table = []

for candidate in EXPECTED_CANDIDATES:

    available_d01_d09 = dimension_availability.loc[
        dimension_availability["candidate_id"].eq(candidate),
        "available_count"
    ].iloc[0]

    d10_dependency_table.append({
        "candidate_id": candidate,
        "available_D01_D09": int(available_d01_d09),
        "D10_record_available": True,
        "D10_dependency_ready": (
            available_d01_d09 > 0
        ),
    })

d10_dependency = pd.DataFrame(
    d10_dependency_table
)

print(
    d10_dependency.to_string(
        index=False
    )
)

assert d10_dependency[
    "D10_record_available"
].eq(True).all()

assert d10_dependency[
    "D10_dependency_ready"
].eq(True).all()

print("\n✓ D10 has one candidate-level source record per candidate")
print("✓ Each candidate has preceding D01-D09 evidence available")
print("✓ D10 can be structurally reserved as candidate-level synthesis")

# ================================================================================================================
# STEP 14 — CRITICAL STRUCTURAL RESULT
# ================================================================================================================

print("\n" + "=" * 100)
print("CRITICAL STRUCTURAL RESULT")
print("=" * 100)

print(
    "Cell19 population                : 47 candidate × question records"
)

print(
    "Available D01-D09 records        : 42"
)

print(
    "Available D10 records             : 5"
)

print(
    "Total human assessment units     : 47"
)

print(
    "D01-D09 unit                     : candidate × dimension"
)

print(
    "D10 unit                         : candidate-level overall synthesis"
)

print(
    "D10 prerequisite                 : available D01-D09 evidence for same candidate"
)

print("=" * 100)

assert total_d01_d09_available + len(q10_records) == 47

# ================================================================================================================
# STEP 15 — NO-ASSESSMENT FIREWALL
# ================================================================================================================

print("\n[STEP 15] NO-ASSESSMENT FIREWALL")
print("-" * 100)

# Verify Cell19 contains no human methodological judgments.
ASSESSMENT_COLUMNS = [
    "post_retrieval_stage_assessment",
    "verification_input_assessment",
    "relevance_sufficiency_assessment",
    "legal_domain_support_assessment",
    "evidence_sufficiency_assessment",
    "contradiction_conflict_risk_assessment",
    "pre_generation_assessment",
    "eligibility_risk_abstention_assessment",
    "human_grounded_suitability_assessment",
    "overall_fdr_equivalence_assessment",
]

for col in ASSESSMENT_COLUMNS:

    values = (
        df[col]
        .astype(str)
        .unique()
        .tolist()
    )

    assert values == ["NOT_ASSESSED"], (
        f"STOP — Unexpected pre-existing assessment in {col}: {values}"
    )

assert df["automatic_judgment"].astype(str).eq(
    "False"
).all()

print("✓ All assessment fields remain NOT_ASSESSED")
print("✓ automatic_judgment remains False")

# ================================================================================================================
# STEP 16 — STRUCTURAL FIREWALL SUMMARY
# ================================================================================================================

print("\n" + "=" * 100)
print("CELL 20A V001 — FINAL GO / NO-GO")
print("=" * 100)

print("Scientific value                     : PASS")
print("Trusted Cell19 V009                 : PASS")
print("Cell19 SHA                           : PASS")
print("Cell19 manifest                     : PASS")
print("Cell19 schema                        : PASS (47 × 33)")
print("Candidate identity                  : PASS (5)")
print("Question identity                   : PASS (Q01–Q10)")
print("Frozen Q → D mapping                : PASS")
print("Candidate × question structure      : PASS")
print("D01-D09 available records           : PASS (42)")
print("D10 records                          : PASS (5)")
print("Total structural assessment units   : PASS (47)")
print("D10 candidate-level dependency      : PASS")
print("No-assessment firewall              : PASS")
print("No artifact modification            : PASS")

print("\nFINAL STRUCTURAL DECISION:")
print("✓ CELL 20A V001 = STRUCTURAL GATE PASS")
print("✓ Cell20 may now be rebuilt on the verified assessment-unit architecture")
print("✓ D01-D09 = candidate × dimension")
print("✓ D10 = candidate-level overall synthesis")
print("✓ No human judgment has been entered")
print("✓ No novelty/equivalence/overlap conclusion generated")
print("✓ No dataset / 199 / 400 artifact modified")

print("=" * 100)
```

```python
# ================================================================================================================
# CELL 20 V002 — HUMAN SOURCE-GROUNDED METHODOLOGICAL ASSESSMENT WORKSPACE
# CLEAN REBUILD AFTER CELL20A STRUCTURAL GATE PASS
#
# TRUSTED UPSTREAM:
#   CELL19 V009
#   CELL20A V001 — STRUCTURAL GATE PASS
#
# STRUCTURAL CONTRACT:
#   D01-D09 = candidate × available dimension
#   D10      = candidate-level overall synthesis after D01-D09
#   TOTAL    = 42 + 5 = 47 assessment units
#
# IMPORTANT:
#   This cell creates the human-review workspace only.
#   It does NOT enter human judgments.
#
# FIREWALL:
#   NO PATCHING / NO GUESSING
#   NO DATASET MODIFICATION
#   NO 199-COHORT MODIFICATION
#   NO 400-VALIDATION REOPEN
#   NO GOLD / Y+ / Y- / PREFERENCE
#   NO AUTOMATIC NOVELTY / EQUIVALENCE / OVERLAP
# ================================================================================================================

import os
import json
import hashlib
import pandas as pd

print("=" * 100)
print("CELL 20 V002 — HUMAN SOURCE-GROUNDED METHODOLOGICAL ASSESSMENT WORKSPACE")
print("=" * 100)

# ================================================================================================================
# STEP 0 — SCIENTIFIC VALUE GATE
# ================================================================================================================

print("\n[STEP 0] SCIENTIFIC VALUE GATE")
print("-" * 100)

SCIENTIFIC_VALUE = True

SCIENTIFIC_PURPOSE = (
    "Create the structurally verified human source-grounded methodological assessment "
    "workspace defined by Cell20A V001, with D01-D09 assessed at candidate × dimension "
    "level and D10 reserved for candidate-level overall synthesis."
)

print("Scientific value : PASS")
print("Purpose          :", SCIENTIFIC_PURPOSE)

assert SCIENTIFIC_VALUE

# ================================================================================================================
# STEP 1 — TRUSTED CELL19 V009
# ================================================================================================================

PROJECT_ROOT = (
    "/content/drive/MyDrive/X-AutoAlign_Legal_RAG"
)

CELL19_DIR = (
    PROJECT_ROOT +
    "/04_AUDIT/NOVELTY_GAP_AUDIT/CHECKPOINTS/"
    "CELL19_HUMAN_SOURCE_GROUNDED_METHODOLOGICAL_ASSESSMENT_V009"
)

CELL19_PARQUET = (
    CELL19_DIR +
    "/CELL19_HUMAN_SOURCE_GROUNDED_METHODOLOGICAL_ASSESSMENT_V009.parquet"
)

CELL19_MANIFEST = (
    CELL19_DIR +
    "/CELL19_MANIFEST_V009.json"
)

print("\n[STEP 1] TRUSTED CELL19 V009")
print("-" * 100)

assert os.path.isdir(CELL19_DIR)
assert os.path.isfile(CELL19_PARQUET)
assert os.path.isfile(CELL19_MANIFEST)

print("✓ Cell19 directory exists")
print("✓ Cell19 Parquet exists")
print("✓ Cell19 Manifest exists")

# ================================================================================================================
# STEP 2 — CELL19 SHA256
# ================================================================================================================

def sha256_file(path, chunk_size=1024 * 1024):
    h = hashlib.sha256()

    with open(path, "rb") as f:
        while True:
            chunk = f.read(chunk_size)

            if not chunk:
                break

            h.update(chunk)

    return h.hexdigest()


print("\n[STEP 2] CELL19 SHA256")
print("-" * 100)

EXPECTED_CELL19_SHA = (
    "f45efc2555929932f417c7152baa358f1d026d2e0b6cceca5f354a1daa0fd670"
)

actual_cell19_sha = sha256_file(CELL19_PARQUET)

print("Expected :", EXPECTED_CELL19_SHA)
print("Actual   :", actual_cell19_sha)

assert actual_cell19_sha == EXPECTED_CELL19_SHA

print("✓ Cell19 V009 SHA verified")

# ================================================================================================================
# STEP 3 — CELL19 MANIFEST TRUST
# ================================================================================================================

print("\n[STEP 3] CELL19 MANIFEST TRUST")
print("-" * 100)

with open(
    CELL19_MANIFEST,
    "r",
    encoding="utf-8"
) as f:
    cell19_manifest = json.load(f)

assert cell19_manifest["cell"] == "CELL 19"
assert cell19_manifest["version"] == "V009"
assert cell19_manifest["status"] == (
    "HUMAN_ASSESSMENT_WORKSPACE_INITIALIZED"
)

assert cell19_manifest["input_rows"] == 50
assert cell19_manifest["eligible_rows"] == 47
assert cell19_manifest["blocked_rows"] == 3
assert cell19_manifest["review_queue_rows"] == 47

assert cell19_manifest["candidate_count"] == 5
assert cell19_manifest["question_count"] == 10
assert cell19_manifest["slot_count"] == 50

assert cell19_manifest["automatic_judgment"] is False
assert cell19_manifest["automatic_methodological_inference"] is False
assert cell19_manifest["automatic_novelty_judgment"] is False
assert cell19_manifest["automatic_equivalence_judgment"] is False
assert cell19_manifest["automatic_overlap_classification"] is False

assert cell19_manifest["dataset_modified"] is False
assert cell19_manifest["199_cohort_modified"] is False
assert cell19_manifest["400_validation_reopened"] is False
assert cell19_manifest["gold_created"] is False
assert cell19_manifest["y_plus_created"] is False
assert cell19_manifest["y_minus_created"] is False
assert cell19_manifest["preference_created"] is False

assert (
    cell19_manifest["trusted_input"]["parquet_sha256"]
    ==
    "16dec31f6aa158c3a87891d0347564de34f408322a595378fbce7539fdc41cf7"
)

print("✓ Cell19 manifest identity verified")
print("✓ Cell19 safety flags verified")

# ================================================================================================================
# STEP 4 — LOAD CELL19
# ================================================================================================================

print("\n[STEP 4] LOAD CELL19")
print("-" * 100)

df = pd.read_parquet(CELL19_PARQUET)

print("Shape:", df.shape)

assert df.shape == (47, 33)

print("✓ Cell19 = 47 × 33")

# ================================================================================================================
# STEP 5 — EXACT CELL19 SCHEMA
# ================================================================================================================

EXPECTED_COLUMNS = [
    "candidate_id",
    "question_id",
    "title",
    "authors",
    "year",
    "venue",
    "primary_source",
    "primary_source_filename",
    "primary_source_path",
    "primary_source_sha256",
    "question_text",
    "source_excerpt",
    "source_location",
    "source_text_status",
    "source_text_unit_start",
    "source_text_unit_end",
    "extraction_method",
    "extraction_interpretation",
    "extraction_status",
    "methodological_note",
    "comparison_eligibility",
    "post_retrieval_stage_assessment",
    "verification_input_assessment",
    "relevance_sufficiency_assessment",
    "legal_domain_support_assessment",
    "evidence_sufficiency_assessment",
    "contradiction_conflict_risk_assessment",
    "pre_generation_assessment",
    "eligibility_risk_abstention_assessment",
    "human_grounded_suitability_assessment",
    "overall_fdr_equivalence_assessment",
    "assessment_method",
    "automatic_judgment",
]

print("\n[STEP 5] EXACT CELL19 SCHEMA")
print("-" * 100)

assert list(df.columns) == EXPECTED_COLUMNS

print("✓ Exact 33-column schema verified")

# ================================================================================================================
# STEP 6 — ELIGIBILITY FIREWALL
# ================================================================================================================

print("\n[STEP 6] ELIGIBILITY FIREWALL")
print("-" * 100)

assert df["comparison_eligibility"].eq(
    "ELIGIBLE_PENDING_HUMAN_ASSESSMENT"
).all()

assert df["automatic_judgment"].astype(str).eq(
    "False"
).all()

print("✓ All 47 records are eligible")
print("✓ automatic_judgment = False")

# ================================================================================================================
# STEP 7 — CANDIDATE / QUESTION IDENTITY
# ================================================================================================================

print("\n[STEP 7] CANDIDATE / QUESTION IDENTITY")
print("-" * 100)

EXPECTED_CANDIDATES = [
    "ANCHOR-001",
    "LIT-001",
    "LIT-002",
    "LIT-003",
    "LIT-004",
]

EXPECTED_QUESTIONS = [
    f"Q{i:02d}"
    for i in range(1, 11)
]

assert sorted(
    df["candidate_id"].unique().tolist()
) == EXPECTED_CANDIDATES

assert sorted(
    df["question_id"].unique().tolist()
) == EXPECTED_QUESTIONS

print("Candidates :", EXPECTED_CANDIDATES)
print("Questions  :", EXPECTED_QUESTIONS)

# ================================================================================================================
# STEP 8 — FROZEN Q → D MAPPING
# ================================================================================================================

print("\n[STEP 8] FROZEN Q → D MAPPING")
print("-" * 100)

Q_TO_D = {
    "Q01": "D01",
    "Q02": "D02",
    "Q03": "D03",
    "Q04": "D04",
    "Q05": "D05",
    "Q06": "D06",
    "Q07": "D07",
    "Q08": "D08",
    "Q09": "D09",
    "Q10": "D10",
}

for q in EXPECTED_QUESTIONS:
    print(f"{q} → {Q_TO_D[q]}")

# ================================================================================================================
# STEP 9 — DIMENSION DEFINITIONS
# ================================================================================================================

print("\n[STEP 9] DIMENSION DEFINITIONS")
print("-" * 100)

DIMENSIONS = {

    "D01": {
        "question_id": "Q01",
        "name": "POST_RETRIEVAL_STAGE",
        "definition":
            "Does the prior method operate as a verification or control step after retrieval?"
    },

    "D02": {
        "question_id": "Q02",
        "name": "VERIFICATION_INPUT",
        "definition":
            "Does the verification or control process use the relevant question and retrieved/current evidence or context as input?"
    },

    "D03": {
        "question_id": "Q03",
        "name": "RELEVANCE_VS_SUFFICIENCY",
        "definition":
            "Does the method distinguish evidence relevance or topicality from evidence sufficiency or support?"
    },

    "D04": {
        "question_id": "Q04",
        "name": "LEGAL_OR_DOMAIN_SUPPORT",
        "definition":
            "Does the method provide legal or domain-specific evidentiary support rather than generic retrieval relevance alone?"
    },

    "D05": {
        "question_id": "Q05",
        "name": "EVIDENCE_SUFFICIENCY",
        "definition":
            "Does the method explicitly assess whether available evidence is sufficient for the intended answer or decision?"
    },

    "D06": {
        "question_id": "Q06",
        "name": "CONTRADICTION_CONFLICT_RISK",
        "definition":
            "Does the method detect or explicitly handle contradiction, conflict, uncertainty, or evidence risk?"
    },

    "D07": {
        "question_id": "Q07",
        "name": "PRE_GENERATION_OPERATION",
        "definition":
            "Does the method operate before answer generation or before the final answer is produced?"
    },

    "D08": {
        "question_id": "Q08",
        "name": "ELIGIBILITY_RISK_ABSTENTION",
        "definition":
            "Does the method provide an eligibility, risk, abstention, selective-answering, or equivalent evidence-based control?"
    },

    "D09": {
        "question_id": "Q09",
        "name": "HUMAN_GROUNDED_SUITABILITY",
        "definition":
            "Does the method provide or directly support human-grounded assessment of evidence suitability?"
    },

    "D10": {
        "question_id": "Q10",
        "name": "OVERALL_FDR_EQUIVALENCE",
        "definition":
            "After reviewing the available D01-D09 evidence for the same candidate, how closely does the prior method align with the proposed F-D+R evidence-verification role?"
    },
}

for dimension_id, spec in DIMENSIONS.items():
    print(
        f"{dimension_id} ← {spec['question_id']} | "
        f"{spec['name']}"
    )

# ================================================================================================================
# STEP 10 — ASSESSMENT VALUE CONTRACT
# ================================================================================================================

print("\n[STEP 10] ASSESSMENT VALUE CONTRACT")
print("-" * 100)

DIMENSION_VALUES = [
    "PRESENT",
    "PARTIAL",
    "ABSENT",
    "UNCLEAR",
]

EVIDENCE_STRENGTH_VALUES = [
    "DIRECT",
    "INDIRECT",
    "INSUFFICIENT",
]

D10_VALUES = [
    "DISTINCT",
    "PARTIALLY_OVERLAPPING",
    "CLOSELY_ALIGNED",
    "UNCLEAR",
]

print("D01-D09 values:")
for value in DIMENSION_VALUES:
    print("  •", value)

print("\nEvidence strength:")
for value in EVIDENCE_STRENGTH_VALUES:
    print("  •", value)

print("\nD10 overall values:")
for value in D10_VALUES:
    print("  •", value)

# ================================================================================================================
# STEP 11 — REVIEW RULES
# ================================================================================================================

print("\n[STEP 11] HUMAN REVIEW RULES")
print("-" * 100)

REVIEW_RULES = [
    "Use the primary source as the authoritative evidence.",
    "Start from source_excerpt and source_location, then inspect surrounding original context when required.",
    "Do not infer a methodological feature from a related concept alone.",
    "PRESENT requires explicit source-grounded evidence.",
    "PARTIAL means only part of the dimension is supported or the implementation is materially weaker/incomplete.",
    "ABSENT means the source does not provide the dimension.",
    "UNCLEAR means the evidence is insufficient for a safe determination.",
    "DIRECT means the source explicitly describes the relevant operation/input/control/output.",
    "INDIRECT means the source supports the interpretation only through a closely related statement.",
    "INSUFFICIENT means available source evidence is not enough for a defensible assessment.",
    "D10 is assessed only after the available D01-D09 evidence for the same candidate has been reviewed.",
    "Do not make novelty, superiority, causal, or population-level claims during this review.",
    "Do not modify dataset, 199 cohort, 400 validation, Gold, Y+, Y-, or preference artifacts.",
    "When uncertain, use UNCLEAR rather than reconstructing the authors' intent.",
]

for i, rule in enumerate(REVIEW_RULES, 1):
    print(f"{i:02d}. {rule}")

# ================================================================================================================
# STEP 12 — IDENTIFY AVAILABLE D01-D09 UNITS
# ================================================================================================================

print("\n[STEP 12] AVAILABLE D01-D09 UNITS")
print("-" * 100)

D01_D09 = [
    "D01", "D02", "D03", "D04",
    "D05", "D06", "D07", "D08", "D09"
]

dimension_to_question = {
    dimension_id: spec["question_id"]
    for dimension_id, spec in DIMENSIONS.items()
}

d01_d09_units = []

for candidate_id in EXPECTED_CANDIDATES:

    candidate_df = df[
        df["candidate_id"].eq(candidate_id)
    ].copy()

    for dimension_id in D01_D09:

        qid = dimension_to_question[dimension_id]

        match = candidate_df[
            candidate_df["question_id"].eq(qid)
        ]

        if len(match) == 0:
            continue

        assert len(match) == 1

        record = match.iloc[0]

        d01_d09_units.append({

            "assessment_unit_id":
                f"{candidate_id}__{dimension_id}",

            "assessment_level":
                "CANDIDATE_DIMENSION",

            "candidate_id":
                candidate_id,

            "dimension_id":
                dimension_id,

            "question_id":
                qid,

            "dimension_name":
                DIMENSIONS[dimension_id]["name"],

            "dimension_definition":
                DIMENSIONS[dimension_id]["definition"],

            "title":
                record["title"],

            "authors":
                record["authors"],

            "year":
                record["year"],

            "venue":
                record["venue"],

            "primary_source":
                record["primary_source"],

            "primary_source_filename":
                record["primary_source_filename"],

            "primary_source_path":
                record["primary_source_path"],

            "primary_source_sha256":
                record["primary_source_sha256"],

            "source_record_question_id":
                record["question_id"],

            "question_text":
                record["question_text"],

            "source_excerpt":
                record["source_excerpt"],

            "source_location":
                record["source_location"],

            "source_text_status":
                record["source_text_status"],

            "extraction_method":
                record["extraction_method"],

            "extraction_status":
                record["extraction_status"],

            "comparison_eligibility":
                record["comparison_eligibility"],

            "assessment":
                "NOT_ASSESSED",

            "evidence_strength":
                "NOT_ASSESSED",

            "evidence_note":
                "",

            "reviewer_note":
                "",

            "assessment_status":
                "PENDING_HUMAN_REVIEW",

            "automatic_judgment":
                False,
        })

print("D01-D09 units:", len(d01_d09_units))

assert len(d01_d09_units) == 42

print("✓ 42 D01-D09 assessment units established")

# ================================================================================================================
# STEP 13 — D10 CANDIDATE-LEVEL UNITS
# ================================================================================================================

print("\n[STEP 13] D10 CANDIDATE-LEVEL UNITS")
print("-" * 100)

d10_units = []

for candidate_id in EXPECTED_CANDIDATES:

    candidate_df = df[
        df["candidate_id"].eq(candidate_id)
    ].copy()

    q10_match = candidate_df[
        candidate_df["question_id"].eq("Q10")
    ]

    assert len(q10_match) == 1

    q10_record = q10_match.iloc[0]

    available_d01_d09 = []

    for dimension_id in D01_D09:

        qid = dimension_to_question[dimension_id]

        if (
            candidate_df["question_id"]
            .eq(qid)
            .any()
        ):
            available_d01_d09.append(
                dimension_id
            )

    assert len(available_d01_d09) > 0

    d10_units.append({

        "assessment_unit_id":
            f"{candidate_id}__D10",

        "assessment_level":
            "CANDIDATE_OVERALL",

        "candidate_id":
            candidate_id,

        "dimension_id":
            "D10",

        "question_id":
            "Q10",

        "dimension_name":
            DIMENSIONS["D10"]["name"],

        "dimension_definition":
            DIMENSIONS["D10"]["definition"],

        "title":
            q10_record["title"],

        "authors":
            q10_record["authors"],

        "year":
            q10_record["year"],

        "venue":
            q10_record["venue"],

        "primary_source":
            q10_record["primary_source"],

        "primary_source_filename":
            q10_record["primary_source_filename"],

        "primary_source_path":
            q10_record["primary_source_path"],

        "primary_source_sha256":
            q10_record["primary_source_sha256"],

        "source_record_question_id":
            "Q10",

        "question_text":
            q10_record["question_text"],

        "source_excerpt":
            q10_record["source_excerpt"],

        "source_location":
            q10_record["source_location"],

        "source_text_status":
            q10_record["source_text_status"],

        "extraction_method":
            q10_record["extraction_method"],

        "extraction_status":
            q10_record["extraction_status"],

        "comparison_eligibility":
            q10_record["comparison_eligibility"],

        # Structural context only:
        "available_preceding_dimensions":
            "|".join(available_d01_d09),

        "assessment":
            "NOT_ASSESSED",

        "evidence_strength":
            "NOT_ASSESSED",

        "evidence_note":
            "",

        "reviewer_note":
            "",

        "assessment_status":
            "PENDING_HUMAN_REVIEW",

        "automatic_judgment":
            False,
    })

print("D10 units:", len(d10_units))

assert len(d10_units) == 5

print("✓ 5 D10 candidate-level units established")

# ================================================================================================================
# STEP 14 — COMBINE ASSESSMENT UNITS
# ================================================================================================================

print("\n[STEP 14] COMBINE ASSESSMENT UNITS")
print("-" * 100)

assessment_form = pd.DataFrame(
    d01_d09_units + d10_units
)

assert len(assessment_form) == 47

assert assessment_form[
    "assessment_unit_id"
].is_unique

print("Total assessment units :", len(assessment_form))
print("D01-D09               :", len(d01_d09_units))
print("D10                   :", len(d10_units))

print("✓ 42 + 5 = 47 total assessment units")

# ================================================================================================================
# STEP 15 — STRUCTURAL FIREWALL
# ================================================================================================================

print("\n[STEP 15] STRUCTURAL FIREWALL")
print("-" * 100)

assert assessment_form[
    "assessment"
].eq("NOT_ASSESSED").all()

assert assessment_form[
    "evidence_strength"
].eq("NOT_ASSESSED").all()

assert assessment_form[
    "evidence_note"
].eq("").all()

assert assessment_form[
    "reviewer_note"
].eq("").all()

assert assessment_form[
    "assessment_status"
].eq("PENDING_HUMAN_REVIEW").all()

assert assessment_form[
    "automatic_judgment"
].eq(False).all()

assert assessment_form[
    "assessment_level"
].isin(
    ["CANDIDATE_DIMENSION", "CANDIDATE_OVERALL"]
).all()

print("✓ No human assessment entered")
print("✓ No evidence judgment entered")
print("✓ No automatic judgment")

# ================================================================================================================
# STEP 16 — D10 DEPENDENCY FIREWALL
# ================================================================================================================

print("\n[STEP 16] D10 DEPENDENCY FIREWALL")
print("-" * 100)

d10_df = assessment_form[
    assessment_form["dimension_id"].eq("D10")
].copy()

assert len(d10_df) == 5

for _, row in d10_df.iterrows():

    preceding = [
        x for x in str(
            row["available_preceding_dimensions"]
        ).split("|")
        if x
    ]

    assert len(preceding) > 0

print("✓ D10 has one candidate-level unit per candidate")
print("✓ Each D10 unit records available D01-D09 prerequisites")
print("✓ D10 remains unassessed")

# ================================================================================================================
# STEP 17 — OUTPUT PATHS
# ================================================================================================================

print("\n[STEP 17] OUTPUT PATHS")
print("-" * 100)

OUTPUT_DIR = (
    PROJECT_ROOT +
    "/04_AUDIT/NOVELTY_GAP_AUDIT/CHECKPOINTS/"
    "CELL20_HUMAN_ASSESSMENT_WORKSPACE_V002"
)

os.makedirs(
    OUTPUT_DIR,
    exist_ok=True
)

OUTPUT_PARQUET = (
    OUTPUT_DIR +
    "/CELL20_HUMAN_ASSESSMENT_WORKSPACE_V002.parquet"
)

OUTPUT_CSV = (
    OUTPUT_DIR +
    "/CELL20_HUMAN_ASSESSMENT_WORKSPACE_V002.csv"
)

OUTPUT_MANIFEST = (
    OUTPUT_DIR +
    "/CELL20_MANIFEST_V002.json"
)

# ================================================================================================================
# STEP 18 — PERSIST
# ================================================================================================================

print("\n[STEP 18] PERSIST")
print("-" * 100)

assessment_form.to_parquet(
    OUTPUT_PARQUET,
    index=False
)

assessment_form.to_csv(
    OUTPUT_CSV,
    index=False,
    encoding="utf-8-sig"
)

print("✓ Parquet written")
print("✓ CSV written")

# ================================================================================================================
# STEP 19 — MANIFEST
# ================================================================================================================

print("\n[STEP 19] MANIFEST")
print("-" * 100)

manifest = {

    "cell":
        "CELL 20",

    "version":
        "V002",

    "status":
        "HUMAN_ASSESSMENT_WORKSPACE_READY",

    "scientific_value":
        True,

    "trusted_input": {
        "cell":
            "CELL 19",
        "version":
            "V009",
        "parquet":
            CELL19_PARQUET,
        "parquet_sha256":
            actual_cell19_sha,
        "manifest":
            CELL19_MANIFEST,
    },

    "structural_gate": {
        "cell":
            "CELL 20A",
        "version":
            "V001",
        "status":
            "STRUCTURAL_GATE_PASS",
        "D01_D09_units":
            42,
        "D10_units":
            5,
        "total_units":
            47,
    },

    "candidate_count":
        5,

    "eligible_source_records":
        47,

    "assessment_units":
        47,

    "D01_D09_assessment_units":
        42,

    "D10_candidate_overall_units":
        5,

    "D10_dependency":
        "AVAILABLE_D01_D09_EVIDENCE_FOR_SAME_CANDIDATE",

    "initial_assessment_state":
        "NOT_ASSESSED",

    "assessment_status":
        "PENDING_HUMAN_REVIEW",

    "automatic_judgment":
        False,

    "automatic_methodological_inference":
        False,

    "automatic_novelty_judgment":
        False,

    "automatic_equivalence_judgment":
        False,

    "automatic_overlap_classification":
        False,

    "dataset_modified":
        False,

    "199_cohort_modified":
        False,

    "400_validation_reopened":
        False,

    "gold_created":
        False,

    "y_plus_created":
        False,

    "y_minus_created":
        False,

    "preference_created":
        False,

    "firewall":
        "NO_PATCHING_NO_GUESSING",
}

with open(
    OUTPUT_MANIFEST,
    "w",
    encoding="utf-8"
) as f:

    json.dump(
        manifest,
        f,
        ensure_ascii=False,
        indent=2
    )

print("✓ Manifest written")

# ================================================================================================================
# STEP 20 — INDEPENDENT READ-BACK
# ================================================================================================================

print("\n[STEP 20] INDEPENDENT READ-BACK")
print("-" * 100)

rb = pd.read_parquet(
    OUTPUT_PARQUET
)

assert rb.shape == assessment_form.shape

assert list(
    rb.columns
) == list(
    assessment_form.columns
)

assert len(rb) == 47

assert rb[
    "assessment_unit_id"
].is_unique

assert rb[
    "assessment"
].eq(
    "NOT_ASSESSED"
).all()

assert rb[
    "evidence_strength"
].eq(
    "NOT_ASSESSED"
).all()

assert rb[
    "assessment_status"
].eq(
    "PENDING_HUMAN_REVIEW"
).all()

assert rb[
    "automatic_judgment"
].eq(
    False
).all()

assert rb[
    "assessment_level"
].isin(
    ["CANDIDATE_DIMENSION", "CANDIDATE_OVERALL"]
).all()

assert rb[
    rb["dimension_id"].isin(
        D01_D09
    )
]["assessment_level"].eq(
    "CANDIDATE_DIMENSION"
).all()

assert rb[
    rb["dimension_id"].eq("D10")
]["assessment_level"].eq(
    "CANDIDATE_OVERALL"
).all()

print("✓ Read-back shape PASS")
print("✓ Read-back schema PASS")
print("✓ Read-back 47 units PASS")
print("✓ Read-back NOT_ASSESSED PASS")
print("✓ Read-back pending-review PASS")
print("✓ Read-back automatic firewall PASS")
print("✓ Read-back D10 candidate-level structure PASS")

# ================================================================================================================
# STEP 21 — OUTPUT SHA
# ================================================================================================================

output_sha = sha256_file(
    OUTPUT_PARQUET
)

print("\n[STEP 21] OUTPUT SHA256")
print("-" * 100)

print(
    "Cell20 V002 Parquet SHA256:",
    output_sha
)

# ================================================================================================================
# FINAL GO / NO-GO
# ================================================================================================================

print("\n" + "=" * 100)
print("CELL 20 V002 — FINAL GO / NO-GO")
print("=" * 100)

print("Scientific value                     : PASS")
print("Cell19 V009 trusted input             : PASS")
print("Cell19 SHA                            : PASS")
print("Cell19 manifest                       : PASS")
print("Cell19 schema                          : PASS (47 × 33)")
print("Candidate identity                    : PASS (5)")
print("Question identity                     : PASS (Q01–Q10)")
print("Cell20A structural gate               : PASS")
print("D01-D09 assessment units              : PASS (42)")
print("D10 candidate-level units             : PASS (5)")
print("Total assessment units                : PASS (47)")
print("D10 dependency structure              : PASS")
print("Initial assessment state              : PASS")
print("No human judgment                     : PASS")
print("Automatic judgment firewall            : PASS")
print("Dataset firewall                      : PASS")
print("199 cohort firewall                   : PASS")
print("400 validation firewall               : PASS")
print("Independent read-back                 : PASS")
print("Output persistence                    : PASS")

print("\nFINAL STATUS:")
print("✓ CELL 20 V002 = HUMAN ASSESSMENT WORKSPACE READY")
print("✓ 42 D01-D09 units + 5 D10 units = 47 total units")
print("✓ D10 is candidate-level and depends on available D01-D09 review")
print("✓ NO human methodological judgment entered")
print("✓ NO novelty conclusion generated")
print("✓ NO equivalence conclusion generated")
print("✓ NO overlap conclusion generated")
print("✓ NO dataset / 199 / 400 artifact modified")

print("\nOutput directory:")
print(OUTPUT_DIR)

print("\nManifest:")
print(OUTPUT_MANIFEST)

print("=" * 100)
```

```python
# ================================================================================================================
# CELL 21 V001 — PILOT HUMAN REVIEW
# CLEAN BUILD FROM TRUSTED CELL20 V002
#
# SCIENTIFIC PURPOSE:
#   Test the human source-grounded methodological assessment procedure on ONE complete
#   candidate before expanding the review to all remaining candidates.
#
# PILOT SELECTION RULE:
#   Select the first candidate_id, in frozen lexical order, that has all D01-D09 records available.
#
# CURRENT VERIFIED PILOT:
#   LIT-001
#
# REVIEW SCOPE:
#   D01-D09 = 9 human assessments
#   D10     = 1 candidate-level overall assessment
#   TOTAL   = 10 pilot assessment units
#
# IMPORTANT:
#   D10 is LOCKED until D01-D09 are reviewed.
#
# FIREWALL:
#   NO PATCHING / NO GUESSING
#   NO DATASET MODIFICATION
#   NO 199-COHORT MODIFICATION
#   NO 400-VALIDATION REOPEN
#   NO GOLD / Y+ / Y- / PREFERENCE
#   NO AUTOMATIC METHODOLOGICAL JUDGMENT
#   NO AUTOMATIC NOVELTY / EQUIVALENCE / OVERLAP
# ================================================================================================================

import os
import json
import hashlib
import pandas as pd

print("=" * 100)
print("CELL 21 V001 — PILOT HUMAN REVIEW")
print("=" * 100)

# ================================================================================================================
# STEP 0 — SCIENTIFIC VALUE GATE
# ================================================================================================================

print("\n[STEP 0] SCIENTIFIC VALUE GATE")
print("-" * 100)

SCIENTIFIC_VALUE = True

SCIENTIFIC_PURPOSE = (
    "Test the human source-grounded methodological assessment procedure on one complete "
    "candidate before expanding the review to the remaining candidates. The pilot evaluates "
    "reviewability, evidence sufficiency, judgment clarity, note requirements, and D10 sequencing."
)

print("Scientific value : PASS")
print("Purpose          :", SCIENTIFIC_PURPOSE)

assert SCIENTIFIC_VALUE

# ================================================================================================================
# STEP 1 — TRUSTED CELL20 V002
# ================================================================================================================

PROJECT_ROOT = (
    "/content/drive/MyDrive/X-AutoAlign_Legal_RAG"
)

CELL20_DIR = (
    PROJECT_ROOT +
    "/04_AUDIT/NOVELTY_GAP_AUDIT/CHECKPOINTS/"
    "CELL20_HUMAN_ASSESSMENT_WORKSPACE_V002"
)

CELL20_PARQUET = (
    CELL20_DIR +
    "/CELL20_HUMAN_ASSESSMENT_WORKSPACE_V002.parquet"
)

CELL20_MANIFEST = (
    CELL20_DIR +
    "/CELL20_MANIFEST_V002.json"
)

print("\n[STEP 1] TRUSTED CELL20 V002")
print("-" * 100)

assert os.path.isdir(CELL20_DIR), (
    f"STOP — Cell20 directory missing:\n{CELL20_DIR}"
)

assert os.path.isfile(CELL20_PARQUET), (
    f"STOP — Cell20 Parquet missing:\n{CELL20_PARQUET}"
)

assert os.path.isfile(CELL20_MANIFEST), (
    f"STOP — Cell20 Manifest missing:\n{CELL20_MANIFEST}"
)

print("✓ Cell20 directory exists")
print("✓ Cell20 Parquet exists")
print("✓ Cell20 Manifest exists")

# ================================================================================================================
# STEP 2 — CELL20 SHA256
# ================================================================================================================

def sha256_file(path, chunk_size=1024 * 1024):

    h = hashlib.sha256()

    with open(path, "rb") as f:

        while True:

            chunk = f.read(chunk_size)

            if not chunk:
                break

            h.update(chunk)

    return h.hexdigest()


EXPECTED_CELL20_SHA = (
    "af47badc20503dc84a313c3aaae4306d84ef684624219ada9e39dc94693d68eb"
)

actual_cell20_sha = sha256_file(
    CELL20_PARQUET
)

print("\n[STEP 2] CELL20 V002 SHA256")
print("-" * 100)

print("Expected :", EXPECTED_CELL20_SHA)
print("Actual   :", actual_cell20_sha)

assert actual_cell20_sha == EXPECTED_CELL20_SHA, (
    "STOP — Cell20 V002 SHA mismatch.\n"
    "NO PATCHING / NO GUESSING."
)

print("✓ Cell20 V002 SHA verified")

# ================================================================================================================
# STEP 3 — CELL20 MANIFEST
# ================================================================================================================

print("\n[STEP 3] CELL20 MANIFEST")
print("-" * 100)

with open(
    CELL20_MANIFEST,
    "r",
    encoding="utf-8"
) as f:
    manifest = json.load(f)

assert manifest["cell"] == "CELL 20"
assert manifest["version"] == "V002"
assert manifest["status"] == (
    "HUMAN_ASSESSMENT_WORKSPACE_READY"
)

assert manifest["assessment_units"] == 47
assert manifest["D01_D09_assessment_units"] == 42
assert manifest["D10_candidate_overall_units"] == 5

assert manifest["automatic_judgment"] is False
assert manifest["automatic_methodological_inference"] is False
assert manifest["automatic_novelty_judgment"] is False
assert manifest["automatic_equivalence_judgment"] is False
assert manifest["automatic_overlap_classification"] is False

assert manifest["dataset_modified"] is False
assert manifest["199_cohort_modified"] is False
assert manifest["400_validation_reopened"] is False
assert manifest["gold_created"] is False
assert manifest["y_plus_created"] is False
assert manifest["y_minus_created"] is False
assert manifest["preference_created"] is False

print("Manifest cell              :", manifest["cell"])
print("Manifest version           :", manifest["version"])
print("Manifest status            :", manifest["status"])
print("Total assessment units     :", manifest["assessment_units"])
print("D01-D09 units              :", manifest["D01_D09_assessment_units"])
print("D10 units                  :", manifest["D10_candidate_overall_units"])

print("✓ Cell20 manifest verified")
print("✓ Cell20 firewall verified")

# ================================================================================================================
# STEP 4 — LOAD CELL20
# ================================================================================================================

print("\n[STEP 4] LOAD CELL20")
print("-" * 100)

df = pd.read_parquet(
    CELL20_PARQUET
)

print("Shape:", df.shape)

assert len(df) == 47

print("✓ 47 trusted assessment units loaded")

# ================================================================================================================
# STEP 5 — EXACT EXPECTED STRUCTURE
# ================================================================================================================

EXPECTED_COLUMNS = [
    "assessment_unit_id",
    "assessment_level",
    "candidate_id",
    "dimension_id",
    "question_id",
    "dimension_name",
    "dimension_definition",
    "title",
    "authors",
    "year",
    "venue",
    "primary_source",
    "primary_source_filename",
    "primary_source_path",
    "primary_source_sha256",
    "source_record_question_id",
    "question_text",
    "source_excerpt",
    "source_location",
    "source_text_status",
    "extraction_method",
    "extraction_status",
    "comparison_eligibility",
    "assessment",
    "evidence_strength",
    "evidence_note",
    "reviewer_note",
    "assessment_status",
    "automatic_judgment",
    "available_preceding_dimensions",
]

print("\n[STEP 5] CELL20 SCHEMA")
print("-" * 100)

assert list(df.columns) == EXPECTED_COLUMNS, (
    "STOP — Cell20 V002 schema mismatch.\n"
    "NO PATCHING / NO GUESSING."
)

print("✓ Cell20 schema verified")

# ================================================================================================================
# STEP 6 — INITIAL-STATE FIREWALL
# ================================================================================================================

print("\n[STEP 6] INITIAL-STATE FIREWALL")
print("-" * 100)

assert df["assessment"].eq(
    "NOT_ASSESSED"
).all()

assert df["evidence_strength"].eq(
    "NOT_ASSESSED"
).all()

assert df["evidence_note"].eq("").all()

assert df["reviewer_note"].eq("").all()

assert df["automatic_judgment"].eq(
    False
).all()

print("✓ All trusted units are still NOT_ASSESSED")
print("✓ No reviewer notes exist")
print("✓ automatic_judgment = False")

# ================================================================================================================
# STEP 7 — IDENTIFY COMPLETE CANDIDATES
# ================================================================================================================

print("\n[STEP 7] COMPLETE-CANDIDATE IDENTIFICATION")
print("-" * 100)

D01_D09 = [
    "D01", "D02", "D03", "D04",
    "D05", "D06", "D07", "D08", "D09"
]

expected_candidate_order = sorted(
    df["candidate_id"].unique().tolist()
)

complete_candidates = []

for candidate_id in expected_candidate_order:

    candidate_df = df[
        df["candidate_id"].eq(candidate_id)
    ]

    available = set(
        candidate_df[
            candidate_df["dimension_id"].isin(D01_D09)
        ]["dimension_id"]
    )

    if set(D01_D09).issubset(available):
        complete_candidates.append(candidate_id)

print("Candidates with D01-D09 complete:")

for candidate_id in complete_candidates:
    print("  •", candidate_id)

assert len(complete_candidates) >= 1

# ================================================================================================================
# STEP 8 — FROZEN PILOT SELECTION RULE
# ================================================================================================================

print("\n[STEP 8] FROZEN PILOT SELECTION")
print("-" * 100)

# Deterministic rule:
# first candidate in lexical order with complete D01-D09 coverage.

PILOT_CANDIDATE = complete_candidates[0]

print("Pilot selection rule:")
print(
    "First candidate in lexical order with complete D01-D09 coverage."
)

print("\nSelected pilot candidate:", PILOT_CANDIDATE)

assert PILOT_CANDIDATE == "LIT-001"

print("✓ Pilot candidate deterministically established: LIT-001")

# ================================================================================================================
# STEP 9 — BUILD PILOT INPUT SET
# ================================================================================================================

print("\n[STEP 9] BUILD PILOT INPUT SET")
print("-" * 100)

pilot_dimension_df = df[
    df["candidate_id"].eq(PILOT_CANDIDATE)
    &
    df["dimension_id"].isin(D01_D09)
].copy()

pilot_d10_df = df[
    df["candidate_id"].eq(PILOT_CANDIDATE)
    &
    df["dimension_id"].eq("D10")
].copy()

assert len(pilot_dimension_df) == 9
assert len(pilot_d10_df) == 1

assert set(
    pilot_dimension_df["dimension_id"]
) == set(D01_D09)

assert pilot_d10_df["dimension_id"].eq(
    "D10"
).all()

print("D01-D09 pilot units:", len(pilot_dimension_df))
print("D10 pilot unit      :", len(pilot_d10_df))
print("Total pilot units   :", len(pilot_dimension_df) + len(pilot_d10_df))

# ================================================================================================================
# STEP 10 — FREEZE PILOT REVIEW ORDER
# ================================================================================================================

print("\n[STEP 10] FREEZE PILOT REVIEW ORDER")
print("-" * 100)

DIMENSION_ORDER = [
    "D01", "D02", "D03", "D04", "D05",
    "D06", "D07", "D08", "D09", "D10"
]

pilot_dimension_df["dimension_order"] = (
    pilot_dimension_df["dimension_id"]
    .map(
        {
            d: i
            for i, d in enumerate(
                DIMENSION_ORDER
            )
        }
    )
)

pilot_dimension_df = (
    pilot_dimension_df
    .sort_values("dimension_order")
    .drop(columns=["dimension_order"])
)

pilot_d10_df = pilot_d10_df.copy()

pilot_form = pd.concat(
    [
        pilot_dimension_df,
        pilot_d10_df
    ],
    ignore_index=True
)

assert list(
    pilot_form["dimension_id"]
) == DIMENSION_ORDER

print("Review order:")
for i, d in enumerate(DIMENSION_ORDER, start=1):
    print(f"{i:02d}. {d}")

print("✓ Pilot order frozen")

# ================================================================================================================
# STEP 11 — D10 SEQUENCING FIREWALL
# ================================================================================================================

print("\n[STEP 11] D10 SEQUENCING FIREWALL")
print("-" * 100)

# D10 cannot be assessed yet.
pilot_form.loc[
    pilot_form["dimension_id"].eq("D10"),
    "assessment_status"
] = "LOCKED_PENDING_D01_D09"

pilot_form.loc[
    pilot_form["dimension_id"].eq("D10"),
    "assessment"
] = "NOT_ASSESSED"

pilot_form.loc[
    pilot_form["dimension_id"].eq("D10"),
    "evidence_strength"
] = "NOT_ASSESSED"

pilot_form.loc[
    pilot_form["dimension_id"].eq("D10"),
    "evidence_note"
] = ""

pilot_form.loc[
    pilot_form["dimension_id"].eq("D10"),
    "reviewer_note"
] = ""

pilot_form.loc[
    pilot_form["dimension_id"].eq("D10"),
    "automatic_judgment"
] = False

# D01-D09 remain pending human review.
pilot_form.loc[
    pilot_form["dimension_id"].isin(D01_D09),
    "assessment_status"
] = "PENDING_HUMAN_REVIEW"

assert pilot_form.loc[
    pilot_form["dimension_id"].isin(D01_D09),
    "assessment_status"
].eq(
    "PENDING_HUMAN_REVIEW"
).all()

assert pilot_form.loc[
    pilot_form["dimension_id"].eq("D10"),
    "assessment_status"
].eq(
    "LOCKED_PENDING_D01_D09"
).all()

print("✓ D01-D09 = PENDING_HUMAN_REVIEW")
print("✓ D10 = LOCKED_PENDING_D01_D09")

# ================================================================================================================
# STEP 12 — HUMAN ASSESSMENT VALUE CONTRACT
# ================================================================================================================

print("\n[STEP 12] HUMAN ASSESSMENT VALUE CONTRACT")
print("-" * 100)

DIMENSION_VALUES = [
    "PRESENT",
    "PARTIAL",
    "ABSENT",
    "UNCLEAR",
]

EVIDENCE_STRENGTH_VALUES = [
    "DIRECT",
    "INDIRECT",
    "INSUFFICIENT",
]

D10_VALUES = [
    "DISTINCT",
    "PARTIALLY_OVERLAPPING",
    "CLOSELY_ALIGNED",
    "UNCLEAR",
]

print("D01-D09 allowed assessment:")
print("  ", DIMENSION_VALUES)

print("\nEvidence strength:")
print("  ", EVIDENCE_STRENGTH_VALUES)

print("\nD10 allowed assessment:")
print("  ", D10_VALUES)

# ================================================================================================================
# STEP 13 — HUMAN REVIEW INSTRUCTIONS
# ================================================================================================================

print("\n[STEP 13] PILOT REVIEW INSTRUCTIONS")
print("-" * 100)

REVIEW_INSTRUCTIONS = [
    "Review the primary source material for the current dimension.",
    "Use source_excerpt and source_location as the initial anchor.",
    "Inspect surrounding original context when the excerpt alone is insufficient.",
    "Do not infer an unstated methodological capability.",
    "PRESENT requires explicit source-grounded evidence.",
    "PARTIAL means only part of the dimension is supported or the implementation is materially incomplete.",
    "ABSENT means the method does not provide the dimension.",
    "UNCLEAR means the source evidence does not safely resolve the dimension.",
    "DIRECT evidence explicitly states the relevant operation, input, control, or output.",
    "INDIRECT evidence supports the interpretation through a closely related statement.",
    "INSUFFICIENT means there is not enough source evidence for a defensible assessment.",
    "Record the exact methodological basis in evidence_note.",
    "Use reviewer_note only for reviewer reasoning, ambiguity, or procedural observations.",
    "Do not enter novelty, superiority, causal, or population-level claims.",
    "Do not modify dataset, 199 cohort, 400 validation, Gold, Y+, Y-, or preference artifacts.",
    "Do not assess D10 until D01-D09 for the pilot candidate are completed.",
]

for i, instruction in enumerate(
    REVIEW_INSTRUCTIONS,
    start=1
):
    print(f"{i:02d}. {instruction}")

# ================================================================================================================
# STEP 14 — PILOT REVIEW PACKET
# ================================================================================================================

print("\n[STEP 14] PILOT REVIEW PACKET")
print("-" * 100)

for _, row in pilot_form.iterrows():

    print("\n" + "=" * 100)
    print(
        f"{row['dimension_id']} | "
        f"{row['dimension_name']} | "
        f"{PILOT_CANDIDATE}"
    )
    print("=" * 100)

    print("Assessment unit :", row["assessment_unit_id"])
    print("Question ID     :", row["question_id"])
    print("Definition      :", row["dimension_definition"])
    print("Paper           :", row["title"])
    print("Source location :", row["source_location"])
    print("Source status   :", row["source_text_status"])
    print("Extraction      :", row["extraction_method"])
    print("Extraction stat :", row["extraction_status"])

    print("\nQuestion:")
    print(row["question_text"])

    print("\nSource excerpt:")
    print(row["source_excerpt"])

    print("\nCurrent state:")
    print("Assessment       :", row["assessment"])
    print("Evidence strength:", row["evidence_strength"])
    print("Assessment status:", row["assessment_status"])

    if row["dimension_id"] == "D10":
        print("\nD10 CONTROL:")
        print("⚠ D10 remains LOCKED until D01-D09 are reviewed.")

# ================================================================================================================
# STEP 15 — OUTPUT PATH
# ================================================================================================================

print("\n" + "=" * 100)
print("STEP 15 — OUTPUT")
print("=" * 100)

OUTPUT_DIR = (
    PROJECT_ROOT +
    "/04_AUDIT/NOVELTY_GAP_AUDIT/CHECKPOINTS/"
    "CELL21_PILOT_HUMAN_REVIEW_V001"
)

os.makedirs(
    OUTPUT_DIR,
    exist_ok=True
)

OUTPUT_PARQUET = (
    OUTPUT_DIR +
    "/CELL21_PILOT_HUMAN_REVIEW_V001.parquet"
)

OUTPUT_CSV = (
    OUTPUT_DIR +
    "/CELL21_PILOT_HUMAN_REVIEW_V001.csv"
)

OUTPUT_MANIFEST = (
    OUTPUT_DIR +
    "/CELL21_MANIFEST_V001.json"
)

# ================================================================================================================
# STEP 16 — PERSIST
# ================================================================================================================

pilot_form.to_parquet(
    OUTPUT_PARQUET,
    index=False
)

pilot_form.to_csv(
    OUTPUT_CSV,
    index=False,
    encoding="utf-8-sig"
)

print("✓ Pilot Parquet written")
print("✓ Pilot CSV written")

# ================================================================================================================
# STEP 17 — MANIFEST
# ================================================================================================================

pilot_manifest = {

    "cell":
        "CELL 21",

    "version":
        "V001",

    "status":
        "PILOT_HUMAN_REVIEW_READY",

    "scientific_value":
        True,

    "purpose":
        SCIENTIFIC_PURPOSE,

    "trusted_input": {
        "cell":
            "CELL 20",

        "version":
            "V002",

        "parquet":
            CELL20_PARQUET,

        "parquet_sha256":
            actual_cell20_sha,

        "manifest":
            CELL20_MANIFEST,
    },

    "pilot_selection_rule":
        "FIRST_LEXICAL_CANDIDATE_WITH_COMPLETE_D01_D09_COVERAGE",

    "pilot_candidate":
        PILOT_CANDIDATE,

    "pilot_units":
        10,

    "D01_D09_units":
        9,

    "D10_units":
        1,

    "D10_status":
        "LOCKED_PENDING_D01_D09",

    "review_order":
        DIMENSION_ORDER,

    "initial_assessment":
        "NOT_ASSESSED",

    "automatic_judgment":
        False,

    "automatic_methodological_inference":
        False,

    "automatic_novelty_judgment":
        False,

    "automatic_equivalence_judgment":
        False,

    "automatic_overlap_classification":
        False,

    "dataset_modified":
        False,

    "199_cohort_modified":
        False,

    "400_validation_reopened":
        False,

    "gold_created":
        False,

    "y_plus_created":
        False,

    "y_minus_created":
        False,

    "preference_created":
        False,

    "firewall":
        "NO_PATCHING_NO_GUESSING",
}

with open(
    OUTPUT_MANIFEST,
    "w",
    encoding="utf-8"
) as f:

    json.dump(
        pilot_manifest,
        f,
        ensure_ascii=False,
        indent=2
    )

print("✓ Pilot manifest written")

# ================================================================================================================
# STEP 18 — INDEPENDENT READ-BACK
# ================================================================================================================

print("\n[STEP 18] INDEPENDENT READ-BACK")
print("-" * 100)

rb = pd.read_parquet(
    OUTPUT_PARQUET
)

assert rb.shape == pilot_form.shape

assert list(
    rb.columns
) == list(
    pilot_form.columns
)

assert len(rb) == 10

assert rb[
    "assessment_unit_id"
].is_unique

assert list(
    rb["dimension_id"]
) == DIMENSION_ORDER

assert rb.loc[
    rb["dimension_id"].isin(D01_D09),
    "assessment_status"
].eq(
    "PENDING_HUMAN_REVIEW"
).all()

assert rb.loc[
    rb["dimension_id"].eq("D10"),
    "assessment_status"
].eq(
    "LOCKED_PENDING_D01_D09"
).all()

assert rb[
    "assessment"
].eq(
    "NOT_ASSESSED"
).all()

assert rb[
    "evidence_strength"
].eq(
    "NOT_ASSESSED"
).all()

assert rb[
    "automatic_judgment"
].eq(
    False
).all()

print("✓ Read-back shape PASS")
print("✓ Read-back 10 pilot units PASS")
print("✓ Read-back review order PASS")
print("✓ Read-back D01-D09 pending state PASS")
print("✓ Read-back D10 lock PASS")
print("✓ Read-back NOT_ASSESSED firewall PASS")
print("✓ Read-back automatic judgment firewall PASS")

# ================================================================================================================
# STEP 19 — OUTPUT SHA
# ================================================================================================================

output_sha = sha256_file(
    OUTPUT_PARQUET
)

print("\n[STEP 19] OUTPUT SHA256")
print("-" * 100)

print(
    "Cell21 Pilot Parquet SHA256:",
    output_sha
)

# ================================================================================================================
# FINAL GO / NO-GO
# ================================================================================================================

print("\n" + "=" * 100)
print("CELL 21 V001 — FINAL GO / NO-GO")
print("=" * 100)

print("Scientific value                    : PASS")
print("Cell20 V002 trusted input            : PASS")
print("Cell20 SHA                            : PASS")
print("Cell20 manifest                       : PASS")
print("Cell20 schema                         : PASS")
print("Complete pilot candidate             : PASS (LIT-001)")
print("D01-D09 pilot units                   : PASS (9)")
print("D10 candidate-level unit              : PASS (1)")
print("Total pilot units                     : PASS (10)")
print("Pilot order                           : PASS")
print("D01-D09 review state                  : PASS")
print("D10 sequencing lock                   : PASS")
print("Human judgment entered                : NO")
print("Automatic methodological judgment     : NO")
print("Automatic novelty/equivalence/overlap : NO")
print("Dataset modification                  : NO")
print("199 cohort modification               : NO")
print("400 validation reopen                 : NO")
print("Independent read-back                 : PASS")
print("Output persistence                    : PASS")

print("\nFINAL STATUS:")
print("✓ CELL 21 V001 = PILOT HUMAN REVIEW READY")
print("✓ Pilot candidate = LIT-001")
print("✓ 9 D01-D09 human assessments are ready")
print("✓ D10 is explicitly locked until D01-D09 are completed")
print("✓ NO judgment has been entered automatically")
print("✓ NO novelty conclusion generated")
print("✓ NO dataset / 199 / 400 artifact modified")

print("\nPilot Parquet:")
print(OUTPUT_PARQUET)

print("\nPilot CSV:")
print(OUTPUT_CSV)

print("\nManifest:")
print(OUTPUT_MANIFEST)

print("=" * 100)
```

```python
# =============================================================================
# CELL 22C V001 — FROZEN ARTIFACT ROUND-TRIP DIAGNOSTIC
# =============================================================================
# SCIENTIFIC PURPOSE
# -----------------------------------------------------------------------------
# Identify the exact reason why the trusted Cell22 V003 artifact has a stable
# file SHA but pandas DataFrame equality fails after Parquet read-back.
#
# READ-ONLY ONLY
# NO PATCHING
# NO REWRITE
# NO JUDGMENT
# NO ARTIFACT MODIFICATION
# =============================================================================

from pathlib import Path
import hashlib
import pandas as pd
import numpy as np

print("=" * 100)
print("CELL 22C V001 — FROZEN ARTIFACT ROUND-TRIP DIAGNOSTIC")
print("=" * 100)

# -----------------------------------------------------------------------------
# STEP 1 — TRUSTED WRITTEN ARTIFACT
# -----------------------------------------------------------------------------

PARQUET_PATH = Path(
    "/content/drive/MyDrive/X-AutoAlign_Legal_RAG/04_AUDIT/"
    "NOVELTY_GAP_AUDIT/CHECKPOINTS/"
    "CELL22_FROZEN_HUMAN_METHODOLOGICAL_ASSESSMENT_V003/"
    "FROZEN_HUMAN_METHODOLOGICAL_ASSESSMENT_V003.parquet"
)

EXPECTED_SHA = (
    "f786aaede9ab4f9cc705262062b7879605f61d545467a0acb919933cdd03c0ad"
)

print("\n[STEP 1] TRUSTED ARTIFACT")
print("-" * 100)

if not PARQUET_PATH.exists():
    raise RuntimeError(
        "STOP: Cell22 V003 Parquet not found. "
        "Do not guess another artifact."
    )

actual_sha = hashlib.sha256(
    PARQUET_PATH.read_bytes()
).hexdigest()

print(f"path exists : True")
print(f"expected SHA: {EXPECTED_SHA}")
print(f"actual SHA  : {actual_sha}")
print(f"SHA PASS    : {actual_sha == EXPECTED_SHA}")

if actual_sha != EXPECTED_SHA:
    raise RuntimeError(
        "STOP: Cell22 V003 SHA mismatch."
    )


# -----------------------------------------------------------------------------
# STEP 2 — READ ARTIFACT
# -----------------------------------------------------------------------------

df_disk = pd.read_parquet(PARQUET_PATH)

print("\n[STEP 2] DISK ARTIFACT")
print("-" * 100)
print(f"shape : {df_disk.shape}")

print("\nDtypes:")
print(df_disk.dtypes.to_string())


# -----------------------------------------------------------------------------
# STEP 3 — RECONSTRUCT THE EXPECTED IN-MEMORY FRAME FROM THE ARTIFACT ITSELF
# -----------------------------------------------------------------------------
# This does NOT rewrite anything.
# The purpose is only to check whether the parquet round-trip preserves the
# observable values, nullability, types and column ordering.

print("\n[STEP 3] BASIC STRUCTURAL COMPARISON")
print("-" * 100)

print("columns equal:",
      list(df_disk.columns) == list(df_disk.columns))

print("shape equal:",
      df_disk.shape == (50, 30))


# -----------------------------------------------------------------------------
# STEP 4 — NULL / NaN INVENTORY
# -----------------------------------------------------------------------------

print("\n[STEP 4] NULL / NaN INVENTORY")
print("-" * 100)

for col in df_disk.columns:
    null_count = int(df_disk[col].isna().sum())

    if null_count > 0:
        print(f"{col:35s}: {null_count} null/NaN")


# -----------------------------------------------------------------------------
# STEP 5 — INTERNAL ROUND-TRIP COPY
# -----------------------------------------------------------------------------

print("\n[STEP 5] INTERNAL ROUND-TRIP TEST")
print("-" * 100)

tmp_roundtrip = PARQUET_PATH.parent / "_CELL22C_TEMP_ROUNDTRIP.parquet"

df_disk.to_parquet(tmp_roundtrip, index=False)

roundtrip_sha = hashlib.sha256(
    tmp_roundtrip.read_bytes()
).hexdigest()

df_roundtrip = pd.read_parquet(tmp_roundtrip)

print(f"temporary SHA : {roundtrip_sha}")
print(f"second shape  : {df_roundtrip.shape}")
print(f"equals        : {df_disk.equals(df_roundtrip)}")


# -----------------------------------------------------------------------------
# STEP 6 — COLUMN-BY-COLUMN TYPE COMPARISON
# -----------------------------------------------------------------------------

print("\n[STEP 6] COLUMN TYPE COMPARISON")
print("-" * 100)

for col in df_disk.columns:
    print(
        f"{col:35s} | "
        f"{str(df_disk[col].dtype):20s}"
    )


# -----------------------------------------------------------------------------
# STEP 7 — VALUE-LEVEL COMPARISON AGAINST A SELF-COPY
# -----------------------------------------------------------------------------
# Detect distinctions such as NaN/None/object/string representation.

print("\n[STEP 7] VALUE-LEVEL NORMALIZED DIAGNOSTIC")
print("-" * 100)

differences = []

a = df_disk.reset_index(drop=True)
b = df_roundtrip.reset_index(drop=True)

for col in a.columns:

    s1 = a[col]
    s2 = b[col]

    # Compare missingness first.
    missing_diff = s1.isna().ne(s2.isna())

    if missing_diff.any():
        differences.append(
            (col, "MISSINGNESS_DIFFERENCE", int(missing_diff.sum()))
        )
        continue

    # Compare stringified observable values to expose object representation.
    v1 = s1.astype(object).map(
        lambda x: "<NA>" if pd.isna(x) else repr(x)
    )
    v2 = s2.astype(object).map(
        lambda x: "<NA>" if pd.isna(x) else repr(x)
    )

    diff = v1.ne(v2)

    if diff.any():
        differences.append(
            (col, "VALUE_REPRESENTATION_DIFFERENCE", int(diff.sum()))
        )

if differences:
    print("Differences detected:")
    for item in differences:
        print(item)
else:
    print("No observable column/value differences detected.")


# -----------------------------------------------------------------------------
# STEP 8 — CLEAN TEMPORARY FILE
# -----------------------------------------------------------------------------

if tmp_roundtrip.exists():
    tmp_roundtrip.unlink()

print("\n[STEP 8] TEMPORARY ARTIFACT CLEANUP")
print("-" * 100)
print("Temporary diagnostic artifact removed.")


# -----------------------------------------------------------------------------
# FINAL
# -----------------------------------------------------------------------------

print("\n" + "=" * 100)
print("CELL 22C V001 — READ-ONLY ROUND-TRIP DIAGNOSTIC COMPLETE")
print("NO PROJECT ARTIFACT MODIFIED")
print("NO HUMAN ASSESSMENT MODIFIED")
print("NO JUDGMENT MODIFIED")
print("=" * 100)
```

```python
# =============================================================================
# CELL 22 V004 — FINAL FROZEN HUMAN METHODOLOGICAL ASSESSMENT
# =============================================================================
# SCIENTIFIC PURPOSE
# -----------------------------------------------------------------------------
# Officially freeze the completed human methodological assessment for the five
# literature sources across D01-D10 as an independently verifiable artifact.
#
# THIS CELL IS A REBUILD FROM ZERO.
#
# TRUSTED UPSTREAMS
# -----------------------------------------------------------------------------
# Cell20 V002 = structural baseline
# Cell18 V001 = question-level source provenance when Cell20 has no row
#
# FIREWALL
# -----------------------------------------------------------------------------
# NO PATCHING
# NO GUESSING
# NO NEW HUMAN JUDGMENT
# NO AUTOMATIC JUDGMENT
# NO NOVELTY INFERENCE
# NO DATASET MODIFICATION
# NO 199-COHORT MODIFICATION
# NO 400-VALIDATION MODIFICATION
# NO Q82 MODIFICATION
# NO MODIFICATION OF CELL20
# NO MODIFICATION OF CELL18
#
# KEY INTEGRITY CHANGE FROM V003
# -----------------------------------------------------------------------------
# Do NOT use:
#     reload_df.equals(original_in_memory_df)
#
# Instead use:
#     SHA
#     exact schema
#     exact slot count
#     key uniqueness
#     normalized content hash
#     independent reload verification
# =============================================================================

from pathlib import Path
import hashlib
import json
from datetime import datetime, timezone

import pandas as pd


print("=" * 100)
print("CELL 22 V004 — FINAL FROZEN HUMAN METHODOLOGICAL ASSESSMENT")
print("=" * 100)


# =============================================================================
# STEP 0 — SCIENTIFIC VALUE GATE
# =============================================================================

SCIENTIFIC_VALUE = True

print("\n[STEP 0] SCIENTIFIC VALUE")
print("-" * 100)
print(f"scientific_value : {SCIENTIFIC_VALUE}")
print(
    "reason           : Officially freeze the completed human methodological "
    "assessment as an auditable artifact before Gap/Novelty Synthesis."
)

if not SCIENTIFIC_VALUE:
    raise RuntimeError("STOP: Scientific-value gate failed.")


# =============================================================================
# STEP 1 — TRUSTED CELL20 V002
# =============================================================================

CELL20_PARQUET = Path(
    "/content/drive/MyDrive/X-AutoAlign_Legal_RAG/04_AUDIT/"
    "NOVELTY_GAP_AUDIT/CHECKPOINTS/"
    "CELL20_HUMAN_ASSESSMENT_WORKSPACE_V002/"
    "CELL20_HUMAN_ASSESSMENT_WORKSPACE_V002.parquet"
)

CELL20_EXPECTED_SHA = (
    "af47badc20503dc84a313c3aaae4306d84ef684624219ada9e39dc94693d68eb"
)

print("\n[STEP 1] TRUSTED CELL20 V002")
print("-" * 100)

if not CELL20_PARQUET.exists():
    raise RuntimeError(
        "STOP: Trusted Cell20 V002 artifact not found."
    )

cell20_sha = hashlib.sha256(
    CELL20_PARQUET.read_bytes()
).hexdigest()

print(f"path exists : True")
print(f"expected SHA: {CELL20_EXPECTED_SHA}")
print(f"actual SHA  : {cell20_sha}")
print(f"SHA PASS    : {cell20_sha == CELL20_EXPECTED_SHA}")

if cell20_sha != CELL20_EXPECTED_SHA:
    raise RuntimeError(
        "STOP: Cell20 V002 SHA mismatch."
    )

cell20 = pd.read_parquet(CELL20_PARQUET)

print(f"shape       : {cell20.shape}")


# =============================================================================
# STEP 2 — TRUSTED CELL18 V001
# =============================================================================

CELL18_PARQUET = Path(
    "/content/drive/MyDrive/X-AutoAlign_Legal_RAG/04_AUDIT/"
    "NOVELTY_GAP_AUDIT/CHECKPOINTS/"
    "CELL18_CONTROLLED_METHODOLOGICAL_COMPARISON_V001/"
    "CELL18_CONTROLLED_METHODOLOGICAL_COMPARISON_V001.parquet"
)

CELL18_EXPECTED_SHA = (
    "16dec31f6aa158c3a87891d0347564de34f408322a595378fbce7539fdc41cf7"
)

print("\n[STEP 2] TRUSTED CELL18 V001")
print("-" * 100)

if not CELL18_PARQUET.exists():
    raise RuntimeError(
        "STOP: Trusted Cell18 V001 artifact not found."
    )

cell18_sha = hashlib.sha256(
    CELL18_PARQUET.read_bytes()
).hexdigest()

print(f"path exists : True")
print(f"expected SHA: {CELL18_EXPECTED_SHA}")
print(f"actual SHA  : {cell18_sha}")
print(f"SHA PASS    : {cell18_sha == CELL18_EXPECTED_SHA}")

if cell18_sha != CELL18_EXPECTED_SHA:
    raise RuntimeError(
        "STOP: Cell18 V001 SHA mismatch."
    )

cell18 = pd.read_parquet(CELL18_PARQUET)

print(f"shape       : {cell18.shape}")


# =============================================================================
# STEP 3 — EXACT CELL20 CONTRACT
# =============================================================================

EXPECTED_COLUMNS = [
    "assessment_unit_id",
    "assessment_level",
    "candidate_id",
    "dimension_id",
    "question_id",
    "dimension_name",
    "dimension_definition",
    "title",
    "authors",
    "year",
    "venue",
    "primary_source",
    "primary_source_filename",
    "primary_source_path",
    "primary_source_sha256",
    "source_record_question_id",
    "question_text",
    "source_excerpt",
    "source_location",
    "source_text_status",
    "extraction_method",
    "extraction_status",
    "comparison_eligibility",
    "assessment",
    "evidence_strength",
    "evidence_note",
    "reviewer_note",
    "assessment_status",
    "automatic_judgment",
    "available_preceding_dimensions",
]

print("\n[STEP 3] CELL20 EXACT SCHEMA")
print("-" * 100)

if list(cell20.columns) != EXPECTED_COLUMNS:
    raise RuntimeError(
        "STOP: Cell20 schema mismatch."
    )

print("schema PASS : 30 columns exactly")


# =============================================================================
# STEP 4 — FROZEN DIMENSION / QUESTION MAP
# =============================================================================

D_TO_Q = {
    "D01": "Q01",
    "D02": "Q02",
    "D03": "Q03",
    "D04": "Q04",
    "D05": "Q05",
    "D06": "Q06",
    "D07": "Q07",
    "D08": "Q08",
    "D09": "Q09",
    "D10": "Q10",
}

D_TO_NAME = {
    "D01": "POST_RETRIEVAL_STAGE",
    "D02": "VERIFICATION_INPUT",
    "D03": "RELEVANCE_SUFFICIENCY",
    "D04": "LEGAL_DOMAIN_SUPPORT",
    "D05": "EVIDENCE_SUFFICIENCY",
    "D06": "CONTRADICTION_CONFLICT_RISK",
    "D07": "PRE_GENERATION",
    "D08": "ELIGIBILITY_RISK_ABSTENTION",
    "D09": "HUMAN_GROUNDED_SUITABILITY",
    "D10": "OVERALL_FDR_EQUIVALENCE",
}


# =============================================================================
# STEP 5 — FINAL HUMAN ASSESSMENTS
# =============================================================================

HUMAN_ASSESSMENTS = {
    "ANCHOR-001": {
        "D01": ("ABSENT", "DIRECT"),
        "D02": ("ABSENT", "DIRECT"),
        "D03": ("ABSENT", "DIRECT"),
        "D04": ("PRESENT", "DIRECT"),
        "D05": ("ABSENT", "DIRECT"),
        "D06": ("ABSENT", "DIRECT"),
        "D07": ("ABSENT", "DIRECT"),
        "D08": ("ABSENT", "DIRECT"),
        "D09": ("ABSENT", "DIRECT"),
        "D10": ("DISTINCT", "DIRECT"),
    },

    "LIT-001": {
        "D01": ("PRESENT", "DIRECT"),
        "D02": ("PRESENT", "DIRECT"),
        "D03": ("PRESENT", "DIRECT"),
        "D04": ("ABSENT", "DIRECT"),
        "D05": ("PRESENT", "DIRECT"),
        "D06": ("PRESENT", "DIRECT"),
        "D07": ("PRESENT", "DIRECT"),
        "D08": ("PRESENT", "DIRECT"),
        "D09": ("PARTIAL", "DIRECT"),
        "D10": ("PARTIALLY_OVERLAPPING", "DIRECT"),
    },

    "LIT-002": {
        "D01": ("ABSENT", "DIRECT"),
        "D02": ("ABSENT", "DIRECT"),
        "D03": ("PARTIAL", "DIRECT"),
        "D04": ("ABSENT", "DIRECT"),
        "D05": ("PRESENT", "DIRECT"),
        "D06": ("PRESENT", "DIRECT"),
        "D07": ("PARTIAL", "DIRECT"),
        "D08": ("PRESENT", "DIRECT"),
        "D09": ("PARTIAL", "DIRECT"),
        "D10": ("PARTIALLY_OVERLAPPING", "DIRECT"),
    },

    "LIT-003": {
        "D01": ("PRESENT", "DIRECT"),
        "D02": ("PRESENT", "DIRECT"),
        "D03": ("PRESENT", "DIRECT"),
        "D04": ("PRESENT", "DIRECT"),
        "D05": ("PRESENT", "DIRECT"),
        "D06": ("PRESENT", "DIRECT"),
        "D07": ("PRESENT", "DIRECT"),
        "D08": ("PRESENT", "DIRECT"),
        "D09": ("PRESENT", "DIRECT"),
        "D10": ("CLOSELY_ALIGNED", "DIRECT"),
    },

    "LIT-004": {
        "D01": ("PRESENT", "DIRECT"),
        "D02": ("PRESENT", "DIRECT"),
        "D03": ("PRESENT", "DIRECT"),
        "D04": ("ABSENT", "DIRECT"),
        "D05": ("PRESENT", "DIRECT"),
        "D06": ("PRESENT", "DIRECT"),
        "D07": ("PRESENT", "DIRECT"),
        "D08": ("PRESENT", "DIRECT"),
        "D09": ("PRESENT", "DIRECT"),
        "D10": ("CLOSELY_ALIGNED", "DIRECT"),
    },
}

EXPECTED_CANDIDATES = set(HUMAN_ASSESSMENTS)
EXPECTED_DIMENSIONS = set(D_TO_Q)


# =============================================================================
# STEP 6 — HUMAN ASSESSMENT CONTRACT
# =============================================================================

print("\n[STEP 6] HUMAN ASSESSMENT CONTRACT")
print("-" * 100)

for candidate_id, dims in HUMAN_ASSESSMENTS.items():

    if set(dims.keys()) != EXPECTED_DIMENSIONS:
        raise RuntimeError(
            f"STOP: {candidate_id} does not contain exactly D01-D10."
        )

print(f"candidates              : {len(EXPECTED_CANDIDATES)}")
print("dimensions/candidate    : 10")
print("expected frozen records : 50")
print("actual frozen records   : 50")


# =============================================================================
# STEP 7 — PROVENANCE RESOLUTION
# =============================================================================

print("\n[STEP 7] PROVENANCE RESOLUTION")
print("-" * 100)

rows = []

special_anchor_questions = {"Q05", "Q06", "Q08"}

for candidate_id in sorted(EXPECTED_CANDIDATES):

    for dim_id in [f"D{i:02d}" for i in range(1, 11)]:

        q_id = D_TO_Q[dim_id]

        # -------------------------------------------------------------
        # First attempt: trusted Cell20
        # -------------------------------------------------------------
        match20 = cell20[
            (cell20["candidate_id"] == candidate_id) &
            (cell20["question_id"] == q_id)
        ].copy()

        if len(match20) == 1:

            row = match20.iloc[0].copy()
            provenance_source = "CELL20_V002"

        # -------------------------------------------------------------
        # Controlled fallback ONLY for Anchor Q05/Q06/Q08:
        # Cell18 is indexed by candidate_id + question_id.
        # -------------------------------------------------------------
        elif (
            candidate_id == "ANCHOR-001"
            and q_id in special_anchor_questions
        ):

            match18 = cell18[
                (cell18["candidate_id"] == candidate_id) &
                (cell18["question_id"] == q_id)
            ].copy()

            if len(match18) != 1:
                raise RuntimeError(
                    f"STOP: Expected exactly one Cell18 source row for "
                    f"{candidate_id}/{q_id}; found {len(match18)}."
                )

            src = match18.iloc[0]

            row = pd.Series(
                index=EXPECTED_COLUMNS,
                dtype="object"
            )

            # Structural identity
            row["assessment_unit_id"] = (
                f"{candidate_id}__{dim_id}"
            )
            row["assessment_level"] = (
                "CANDIDATE_LEVEL"
                if dim_id == "D10"
                else "CANDIDATE_DIMENSION"
            )

            row["candidate_id"] = candidate_id
            row["dimension_id"] = dim_id
            row["question_id"] = q_id
            row["dimension_name"] = D_TO_NAME[dim_id]
            row["dimension_definition"] = ""

            # Exact source provenance from Cell18
            for col in [
                "title",
                "authors",
                "year",
                "venue",
                "primary_source",
                "primary_source_filename",
                "primary_source_path",
                "primary_source_sha256",
                "question_text",
                "source_excerpt",
                "source_location",
                "source_text_status",
                "extraction_method",
                "extraction_status",
                "comparison_eligibility",
                "automatic_judgment",
            ]:
                if col in src.index:
                    row[col] = src[col]

            row["source_record_question_id"] = q_id
            provenance_source = "CELL18_V001_QUESTION_LEVEL"

        else:

            raise RuntimeError(
                f"STOP: Missing trusted provenance for "
                f"{candidate_id}/{dim_id}/{q_id}."
            )

        # -------------------------------------------------------------
        # Frozen human assessment — NO inference.
        # -------------------------------------------------------------

        assessment, evidence_strength = (
            HUMAN_ASSESSMENTS[candidate_id][dim_id]
        )

        row["assessment_unit_id"] = (
            f"{candidate_id}__{dim_id}"
        )

        row["assessment_level"] = (
            "CANDIDATE_LEVEL"
            if dim_id == "D10"
            else "CANDIDATE_DIMENSION"
        )

        row["candidate_id"] = candidate_id
        row["dimension_id"] = dim_id
        row["question_id"] = q_id
        row["dimension_name"] = D_TO_NAME[dim_id]

        row["assessment"] = assessment
        row["evidence_strength"] = evidence_strength

        row["evidence_note"] = (
            "FINAL HUMAN METHODOLOGICAL ASSESSMENT — "
            "SOURCE-GROUNDED REVIEW"
        )

        row["reviewer_note"] = (
            "Frozen human assessment. No automatic inference."
        )

        row["assessment_status"] = (
            "HUMAN_ASSESSMENT_FROZEN"
        )

        row["automatic_judgment"] = False

        # Internal provenance marker is NOT added as a new column.
        # It is kept separately in a temporary list for QC.
        rows.append(
            (row, provenance_source)
        )


# =============================================================================
# STEP 8 — MATERIALIZE FINAL DATAFRAME
# =============================================================================

frozen = pd.DataFrame(
    [item[0] for item in rows],
    columns=EXPECTED_COLUMNS
)

provenance_sources = [
    item[1]
    for item in rows
]


# =============================================================================
# STEP 9 — STRUCTURAL VALIDATION
# =============================================================================

print("\n[STEP 9] STRUCTURAL VALIDATION")
print("-" * 100)

print(f"shape : {frozen.shape}")

if frozen.shape != (50, 30):
    raise RuntimeError(
        f"STOP: Expected (50,30), got {frozen.shape}."
    )

slot_count = frozen[
    ["candidate_id", "dimension_id"]
].drop_duplicates().shape[0]

print(f"unique candidate×dimension slots : {slot_count}")

if slot_count != 50:
    raise RuntimeError(
        "STOP: Candidate×dimension uniqueness failed."
    )

print("50 unique slots : PASS")


# =============================================================================
# STEP 10 — VALUE VALIDATION
# =============================================================================

print("\n[STEP 10] HUMAN VALUE VALIDATION")
print("-" * 100)

D01_D09_ALLOWED = {
    "PRESENT",
    "PARTIAL",
    "ABSENT",
    "UNCLEAR",
}

D10_ALLOWED = {
    "DISTINCT",
    "PARTIALLY_OVERLAPPING",
    "CLOSELY_ALIGNED",
    "UNCLEAR",
}

for _, r in frozen.iterrows():

    if r["dimension_id"] == "D10":
        if r["assessment"] not in D10_ALLOWED:
            raise RuntimeError(
                f"STOP: Invalid D10 assessment."
            )
    else:
        if r["assessment"] not in D01_D09_ALLOWED:
            raise RuntimeError(
                f"STOP: Invalid D01-D09 assessment."
            )

    if r["evidence_strength"] != "DIRECT":
        raise RuntimeError(
            "STOP: Evidence strength must be DIRECT."
        )

    if r["assessment_status"] != "HUMAN_ASSESSMENT_FROZEN":
        raise RuntimeError(
            "STOP: Assessment status is not frozen."
        )

    if bool(r["automatic_judgment"]):
        raise RuntimeError(
            "STOP: Automatic judgment detected."
        )

if frozen["assessment"].eq("NOT_ASSESSED").any():
    raise RuntimeError(
        "STOP: NOT_ASSESSED remains."
    )

print("No NOT_ASSESSED      : PASS")
print("All evidence DIRECT  : PASS")
print("All assessments HUMAN: PASS")
print("Automatic judgment   : FALSE")


# =============================================================================
# STEP 11 — CORRESPONDENCE FIREWALL PRESERVATION
# =============================================================================

print("\n[STEP 11] CORRESPONDENCE FIREWALL PRESERVATION")
print("-" * 100)

anchor_special = frozen[
    (frozen["candidate_id"] == "ANCHOR-001") &
    (frozen["dimension_id"].isin(["D05", "D06", "D08"]))
][[
    "candidate_id",
    "dimension_id",
    "question_id",
    "comparison_eligibility",
    "assessment",
]]

print(anchor_special.to_string(index=False))

if not anchor_special[
    "comparison_eligibility"
].eq("BLOCKED_BY_HUMAN_CORRESPONDENCE").all():
    raise RuntimeError(
        "STOP: Original Anchor correspondence status changed."
    )

print("Original correspondence status : PRESERVED")


# =============================================================================
# STEP 12 — PROVENANCE INVENTORY
# =============================================================================

print("\n[STEP 12] PROVENANCE INVENTORY")
print("-" * 100)

print(
    "Cell20 provenance rows :",
    provenance_sources.count("CELL20_V002")
)

print(
    "Cell18 provenance rows :",
    provenance_sources.count("CELL18_V001_QUESTION_LEVEL")
)

if provenance_sources.count(
    "CELL18_V001_QUESTION_LEVEL"
) != 3:
    raise RuntimeError(
        "STOP: Expected exactly 3 controlled Cell18 provenance rows."
    )

print("Controlled provenance split : PASS")


# =============================================================================
# STEP 13 — OUTPUT DIRECTORY
# =============================================================================

OUTPUT_DIR = Path(
    "/content/drive/MyDrive/X-AutoAlign_Legal_RAG/04_AUDIT/"
    "NOVELTY_GAP_AUDIT/CHECKPOINTS/"
    "CELL22_FROZEN_HUMAN_METHODOLOGICAL_ASSESSMENT_V004"
)

print("\n[STEP 13] OUTPUT DIRECTORY")
print("-" * 100)
print(OUTPUT_DIR)

if OUTPUT_DIR.exists():
    raise RuntimeError(
        "STOP: V004 directory already exists. "
        "No overwrite is permitted."
    )

OUTPUT_DIR.mkdir(
    parents=True,
    exist_ok=False
)

PARQUET_PATH = OUTPUT_DIR / (
    "FROZEN_HUMAN_METHODOLOGICAL_ASSESSMENT_V004.parquet"
)

CSV_PATH = OUTPUT_DIR / (
    "FROZEN_HUMAN_METHODOLOGICAL_ASSESSMENT_V004.csv"
)

MANIFEST_PATH = OUTPUT_DIR / (
    "CELL22_MANIFEST_V004.json"
)


# =============================================================================
# STEP 14 — NORMALIZED CONTENT HASH
# =============================================================================
# This hash is independent of pandas object identity and robust to
# serialization-level representation differences.

def normalized_dataframe_hash(df: pd.DataFrame) -> str:

    work = df.copy()

    # Stable column order
    work = work[EXPECTED_COLUMNS]

    # Stable row order
    work = work.sort_values(
        ["candidate_id", "dimension_id"],
        kind="stable"
    ).reset_index(drop=True)

    # Normalize nulls and scalar representation
    def normalize_value(x):

        if pd.isna(x):
            return "<NA>"

        if isinstance(x, bool):
            return "BOOL:" + str(x)

        return "VAL:" + str(x)

    for col in work.columns:
        work[col] = work[col].map(normalize_value)

    payload = work.to_csv(
        index=False,
        lineterminator="\n"
    ).encode("utf-8")

    return hashlib.sha256(payload).hexdigest()


normalized_prewrite_hash = normalized_dataframe_hash(
    frozen
)

print("\n[STEP 14] NORMALIZED CONTENT HASH")
print("-" * 100)
print(f"normalized pre-write hash : {normalized_prewrite_hash}")


# =============================================================================
# STEP 15 — WRITE ARTIFACTS
# =============================================================================

print("\n[STEP 15] WRITE FROZEN ARTIFACTS")
print("-" * 100)

frozen.to_parquet(
    PARQUET_PATH,
    index=False
)

frozen.to_csv(
    CSV_PATH,
    index=False,
    encoding="utf-8-sig"
)

parquet_sha = hashlib.sha256(
    PARQUET_PATH.read_bytes()
).hexdigest()

csv_sha = hashlib.sha256(
    CSV_PATH.read_bytes()
).hexdigest()

print(f"Parquet path : {PARQUET_PATH}")
print(f"CSV path     : {CSV_PATH}")
print(f"Parquet SHA  : {parquet_sha}")
print(f"CSV SHA      : {csv_sha}")


# =============================================================================
# STEP 16 — INDEPENDENT READ-BACK
# =============================================================================

print("\n[STEP 16] INDEPENDENT READ-BACK")
print("-" * 100)

reloaded = pd.read_parquet(
    PARQUET_PATH
)

reload_sha = hashlib.sha256(
    PARQUET_PATH.read_bytes()
).hexdigest()

normalized_reload_hash = normalized_dataframe_hash(
    reloaded
)

print(f"reload shape             : {reloaded.shape}")
print(f"write SHA                : {parquet_sha}")
print(f"reload SHA               : {reload_sha}")
print(f"SHA stable               : {reload_sha == parquet_sha}")
print(
    f"normalized content hash  : "
    f"{normalized_reload_hash}"
)
print(
    "normalized hash stable   :",
    normalized_reload_hash == normalized_prewrite_hash
)

if reload_sha != parquet_sha:
    raise RuntimeError(
        "STOP: File SHA changed after read-back."
    )

if normalized_reload_hash != normalized_prewrite_hash:
    raise RuntimeError(
        "STOP: Normalized content changed after read-back."
    )

# Exact observable-value check after normalizing both frames.
expected_norm = frozen.copy()
reloaded_norm = reloaded.copy()

for col in EXPECTED_COLUMNS:

    expected_norm[col] = expected_norm[col].map(
        lambda x: "<NA>" if pd.isna(x) else "VAL:" + str(x)
    )

    reloaded_norm[col] = reloaded_norm[col].map(
        lambda x: "<NA>" if pd.isna(x) else "VAL:" + str(x)
    )

expected_norm = expected_norm.sort_values(
    ["candidate_id", "dimension_id"],
    kind="stable"
).reset_index(drop=True)

reloaded_norm = reloaded_norm.sort_values(
    ["candidate_id", "dimension_id"],
    kind="stable"
).reset_index(drop=True)

if expected_norm.to_csv(
    index=False,
    lineterminator="\n"
) != reloaded_norm.to_csv(
    index=False,
    lineterminator="\n"
):
    raise RuntimeError(
        "STOP: Normalized observable content mismatch."
    )

print("Normalized observable content : PASS")


# =============================================================================
# STEP 17 — RELOAD STRUCTURAL GATES
# =============================================================================

print("\n[STEP 17] RELOAD STRUCTURAL GATES")
print("-" * 100)

reload_checks = {
    "shape_50x30":
        reloaded.shape == (50, 30),

    "schema_exact":
        list(reloaded.columns) == EXPECTED_COLUMNS,

    "five_candidates":
        reloaded["candidate_id"].nunique() == 5,

    "ten_dimensions":
        reloaded["dimension_id"].nunique() == 10,

    "fifty_unique_slots":
        reloaded[
            ["candidate_id", "dimension_id"]
        ].drop_duplicates().shape[0] == 50,

    "no_not_assessed":
        not reloaded["assessment"].eq("NOT_ASSESSED").any(),

    "all_direct":
        reloaded["evidence_strength"].eq("DIRECT").all(),

    "all_frozen":
        reloaded["assessment_status"]
        .eq("HUMAN_ASSESSMENT_FROZEN")
        .all(),

    "all_manual":
        (~reloaded["automatic_judgment"].astype(bool)).all(),

    "anchor_correspondence_preserved":
        reloaded[
            (reloaded["candidate_id"] == "ANCHOR-001") &
            (reloaded["dimension_id"].isin(
                ["D05", "D06", "D08"]
            ))
        ]["comparison_eligibility"]
        .eq("BLOCKED_BY_HUMAN_CORRESPONDENCE")
        .all(),
}

for name, result in reload_checks.items():
    print(
        f"{name:35s}: {'PASS' if result else 'FAIL'}"
    )

if not all(reload_checks.values()):
    raise RuntimeError(
        "STOP: One or more reload freeze gates failed."
    )


# =============================================================================
# STEP 18 — PAPER-LEVEL D10 SUMMARY
# =============================================================================

print("\n[STEP 18] FROZEN PAPER-LEVEL D10 SUMMARY")
print("-" * 100)

summary = (
    reloaded[
        reloaded["dimension_id"] == "D10"
    ][
        [
            "candidate_id",
            "assessment",
            "evidence_strength",
        ]
    ]
    .sort_values("candidate_id")
)

print(summary.to_string(index=False))


# =============================================================================
# STEP 19 — MANIFEST
# =============================================================================

manifest = {
    "artifact": "FROZEN_HUMAN_METHODOLOGICAL_ASSESSMENT",
    "cell": "CELL 22",
    "version": "V004",
    "status": "FROZEN",

    "scientific_purpose": (
        "Official freeze of the completed human methodological assessment "
        "for the five reviewed literature sources before Gap/Novelty "
        "Synthesis."
    ),

    "scope": {
        "candidate_count": 5,
        "dimension_count": 10,
        "record_count": 50,
        "candidates": sorted(EXPECTED_CANDIDATES),
        "dimensions": [f"D{i:02d}" for i in range(1, 11)],
    },

    "dimension_question_mapping": D_TO_Q,

    "trusted_upstreams": {
        "cell20_v002": {
            "path": str(CELL20_PARQUET),
            "sha256": CELL20_EXPECTED_SHA,
        },
        "cell18_v001": {
            "path": str(CELL18_PARQUET),
            "sha256": CELL18_EXPECTED_SHA,
        },
    },

    "human_assessment_policy": {
        "assessment_source": "FINALIZED_HUMAN_SOURCE_GROUNDED_REVIEW",
        "evidence_strength": "DIRECT",
        "automatic_judgment": False,
        "new_human_judgment_created": False,
        "novelty_inference": False,
        "superiority_inference": False,
        "causal_claim": False,
        "population_claim": False,
    },

    "correspondence_separation": {
        "anchor_blocked_questions": [
            "Q05",
            "Q06",
            "Q08",
        ],
        "anchor_special_dimensions": [
            "D05",
            "D06",
            "D08",
        ],
        "original_status_preserved": True,
    },

    "integrity": {
        "normalized_prewrite_hash":
            normalized_prewrite_hash,
        "normalized_readback_hash":
            normalized_reload_hash,
        "normalized_hash_stable":
            normalized_reload_hash == normalized_prewrite_hash,
        "parquet_sha256":
            parquet_sha,
        "csv_sha256":
            csv_sha,
    },

    "artifact_paths": {
        "parquet": str(PARQUET_PATH),
        "csv": str(CSV_PATH),
        "manifest": str(MANIFEST_PATH),
    },

    "generated_utc":
        datetime.now(timezone.utc).isoformat(),
}

with open(
    MANIFEST_PATH,
    "w",
    encoding="utf-8"
) as f:
    json.dump(
        manifest,
        f,
        ensure_ascii=False,
        indent=2
    )

print("\n[STEP 19] MANIFEST WRITTEN")
print(MANIFEST_PATH)


# =============================================================================
# STEP 20 — FINAL FREEZE DECISION
# =============================================================================

print("\n[STEP 20] FINAL FREEZE DECISION")
print("-" * 100)

FINAL_CHECKS = {
    "scientific_value":
        SCIENTIFIC_VALUE,

    "cell20_sha":
        cell20_sha == CELL20_EXPECTED_SHA,

    "cell18_sha":
        cell18_sha == CELL18_EXPECTED_SHA,

    "shape_50x30":
        reloaded.shape == (50, 30),

    "schema":
        list(reloaded.columns) == EXPECTED_COLUMNS,

    "unique_slots":
        reloaded[
            ["candidate_id", "dimension_id"]
        ].drop_duplicates().shape[0] == 50,

    "human_complete":
        not reloaded["assessment"].eq("NOT_ASSESSED").any(),

    "direct_evidence":
        reloaded["evidence_strength"].eq("DIRECT").all(),

    "frozen_status":
        reloaded["assessment_status"]
        .eq("HUMAN_ASSESSMENT_FROZEN")
        .all(),

    "no_automatic_judgment":
        (~reloaded["automatic_judgment"].astype(bool)).all(),

    "normalized_hash_stable":
        normalized_reload_hash == normalized_prewrite_hash,

    "parquet_sha_stable":
        reload_sha == parquet_sha,

    "anchor_firewall_preserved":
        reloaded[
            (reloaded["candidate_id"] == "ANCHOR-001") &
            (reloaded["dimension_id"].isin(
                ["D05", "D06", "D08"]
            ))
        ]["comparison_eligibility"]
        .eq("BLOCKED_BY_HUMAN_CORRESPONDENCE")
        .all(),
}

for name, result in FINAL_CHECKS.items():
    print(
        f"{name:30s}: {'PASS' if result else 'FAIL'}"
    )

if not all(FINAL_CHECKS.values()):
    raise RuntimeError(
        "STOP: HUMAN METHODOLOGICAL ASSESSMENT FREEZE FAILED."
    )


# =============================================================================
# STEP 21 — FINAL STATUS
# =============================================================================

print("\n" + "=" * 100)
print("CELL 22 V004 — FINAL STATUS")
print("=" * 100)

print("Human Methodological Assessment : FROZEN")
print("Sources                         : 5")
print("Dimensions                      : D01-D10")
print("Frozen records                  : 50")
print("Evidence strength               : DIRECT")
print("Automatic judgment              : FALSE")
print("Correspondence firewall         : PRESERVED")
print("Novelty synthesis               : NOT PERFORMED")
print("Dataset modification            : NONE")
print("199 cohort modification         : NONE")
print("400 validation modification     : NONE")
print("Q82 modification                : NONE")

print("\nFINAL RESULT:")
print(
    "CELL 22 V004 = FROZEN HUMAN METHODOLOGICAL ASSESSMENT PASS"
)

print(
    "GO → Gap / Novelty Synthesis may now begin from this frozen artifact."
)

print("=" * 100)
```

```python
# =============================================================================
# CELL 23A V001 — GAP DECISION LOGIC AUDIT
# =============================================================================
# EXPERIMENT : FDR-NOVELTY-GAP-001
#
# SCIENTIFIC PURPOSE
# -----------------------------------------------------------------------------
# Audit the decision logic used by CELL 23 V001 to determine whether its
# GAP_NOT_DEFENSIBLE conclusion is actually licensed by the frozen
# methodological assessment and the frozen novelty-audit protocol.
#
# THIS CELL IS READ-ONLY.
#
# It MUST NOT:
#   - re-assess any paper
#   - change any human assessment
#   - introduce new literature
#   - modify Cell22
#   - modify Cell23 V001
#   - modify the dataset
#   - make a new novelty decision
#   - implement architecture
#
# NO PATCHING
# NO GUESSING
# =============================================================================

from pathlib import Path
import hashlib
import json
import pandas as pd


print("=" * 108)
print("CELL 23A V001 — GAP DECISION LOGIC AUDIT")
print("=" * 108)


# =============================================================================
# STEP 0 — SCIENTIFIC VALUE GATE
# =============================================================================

SCIENTIFIC_VALUE = True

print("\n[STEP 0] SCIENTIFIC VALUE")
print("-" * 108)
print(f"scientific_value : {SCIENTIFIC_VALUE}")
print(
    "reason           : Verify that the GAP decision logic used by Cell23 "
    "is licensed by the frozen protocol and human assessment, without "
    "changing either."
)

if not SCIENTIFIC_VALUE:
    raise RuntimeError("STOP: Scientific-value gate failed.")


# =============================================================================
# STEP 1 — TRUSTED CELL22 V004
# =============================================================================

CELL22_PATH = Path(
    "/content/drive/MyDrive/X-AutoAlign_Legal_RAG/04_AUDIT/"
    "NOVELTY_GAP_AUDIT/CHECKPOINTS/"
    "CELL22_FROZEN_HUMAN_METHODOLOGICAL_ASSESSMENT_V004/"
    "FROZEN_HUMAN_METHODOLOGICAL_ASSESSMENT_V004.parquet"
)

CELL22_EXPECTED_SHA = (
    "4e2d9bd060761862a02b4e21ce087d0a0ebe0dd000d7877df05d1b2fb10a6ff9"
)

print("\n[STEP 1] TRUSTED CELL22 V004")
print("-" * 108)

if not CELL22_PATH.exists():
    raise RuntimeError(
        "STOP: Cell22 V004 not found."
    )

cell22_sha = hashlib.sha256(
    CELL22_PATH.read_bytes()
).hexdigest()

print(f"expected SHA : {CELL22_EXPECTED_SHA}")
print(f"actual SHA   : {cell22_sha}")
print(f"SHA PASS     : {cell22_sha == CELL22_EXPECTED_SHA}")

if cell22_sha != CELL22_EXPECTED_SHA:
    raise RuntimeError(
        "STOP: Cell22 SHA mismatch."
    )

cell22 = pd.read_parquet(CELL22_PATH)

print(f"shape        : {cell22.shape}")


# =============================================================================
# STEP 2 — TRUSTED CELL23 V001 REPORT
# =============================================================================

CELL23_REPORT = Path(
    "/content/drive/MyDrive/X-AutoAlign_Legal_RAG/04_AUDIT/"
    "NOVELTY_GAP_AUDIT/CHECKPOINTS/"
    "CELL23_GAP_NOVELTY_SYNTHESIS_V001/"
    "CELL23_GAP_NOVELTY_REPORT_V001.json"
)

print("\n[STEP 2] CELL23 V001 REPORT")
print("-" * 108)

if not CELL23_REPORT.exists():
    raise RuntimeError(
        "STOP: Cell23 V001 report not found."
    )

with open(CELL23_REPORT, "r", encoding="utf-8") as f:
    cell23_report = json.load(f)

print(f"report exists : {CELL23_REPORT.exists()}")


# =============================================================================
# STEP 3 — RECONSTRUCT FROZEN D10 HUMAN ASSESSMENTS
# =============================================================================

print("\n[STEP 3] FROZEN D10 ASSESSMENTS")
print("-" * 108)

d10 = (
    cell22[
        cell22["dimension_id"] == "D10"
    ][
        [
            "candidate_id",
            "assessment",
            "evidence_strength",
        ]
    ]
    .sort_values("candidate_id")
    .reset_index(drop=True)
)

print(d10.to_string(index=False))


# =============================================================================
# STEP 4 — RECONSTRUCT D01-D09 HUMAN ASSESSMENTS
# =============================================================================

print("\n[STEP 4] D01-D09 HUMAN ASSESSMENTS")
print("-" * 108)

d01_d09 = (
    cell22[
        cell22["dimension_id"] != "D10"
    ][
        [
            "candidate_id",
            "dimension_id",
            "assessment",
        ]
    ]
    .sort_values(
        ["candidate_id", "dimension_id"]
    )
    .reset_index(drop=True)
)

print(d01_d09.to_string(index=False))


# =============================================================================
# STEP 5 — RECONSTRUCT CELL23 V001'S DECISION LOGIC
# =============================================================================
#
# This is the exact logic concept used by V001:
#
#   If UNCLEAR -> GAP_INCONCLUSIVE
#   Else if CLOSELY_ALIGNED exists:
#       If a CLOSELY_ALIGNED candidate has all D01-D09 PRESENT:
#           GAP_NOT_DEFENSIBLE
#       Else:
#           GAP_PARTIALLY_OVERLAPPING
#   Else:
#       GAP_DEFENSIBLE
#
# We reproduce this logic only to audit it.
# =============================================================================

print("\n[STEP 5] CELL23 V001 DECISION LOGIC RECONSTRUCTION")
print("-" * 108)

closest_candidates = (
    d10[
        d10["assessment"] == "CLOSELY_ALIGNED"
    ]["candidate_id"]
    .tolist()
)

print(
    "CLOSELY_ALIGNED candidates :",
    closest_candidates
)

full_d01_d09 = {}

for candidate_id in closest_candidates:

    sub = d01_d09[
        d01_d09["candidate_id"] == candidate_id
    ]

    full_present = (
        len(sub) == 9 and
        sub["assessment"].eq("PRESENT").all()
    )

    full_d01_d09[candidate_id] = bool(full_present)

    print(
        f"{candidate_id}: "
        f"D01-D09 all PRESENT = {full_present}"
    )


# =============================================================================
# STEP 6 — AUDIT THE LOGICAL IMPLICATION
# =============================================================================

print("\n[STEP 6] LOGICAL IMPLICATION AUDIT")
print("-" * 108)

audit_findings = []

# Finding A:
# D10 is a human overall methodological assessment, but V001 does not use
# D10 as the decisive human judgment. Instead it derives a stronger conclusion
# from D01-D09 counts.

if closest_candidates:

    audit_findings.append({
        "finding_id": "F01",
        "status": "ISSUE",
        "finding": (
            "Cell23 V001 converts D01-D09 all-PRESENT into "
            "GAP_NOT_DEFENSIBLE without an independently frozen rule "
            "authorizing that implication."
        ),
    })


# Finding B:
# D10 itself is CLOSELY_ALIGNED, not an explicit equivalence or complete-cover
# category.

for candidate_id in closest_candidates:

    d10_value = d10[
        d10["candidate_id"] == candidate_id
    ]["assessment"].iloc[0]

    if d10_value == "CLOSELY_ALIGNED":
        audit_findings.append({
            "finding_id": "F02",
            "status": "ISSUE",
            "finding": (
                f"{candidate_id} is human-assessed as CLOSELY_ALIGNED "
                "at D10. CLOSELY_ALIGNED is not itself the same label "
                "as GAP_NOT_DEFENSIBLE."
            ),
        })


# Finding C:
# The frozen protocol says GAP_NOT_DEFENSIBLE means prior work substantially
# covers the proposed contribution; therefore the synthesis needs an explicit
# boundary argument, not just dimension counting.

audit_findings.append({
    "finding_id": "F03",
    "status": "ISSUE",
    "finding": (
        "The frozen decision definition for GAP_NOT_DEFENSIBLE requires "
        "substantial coverage of the proposed contribution; a raw "
        "PRESENT-count rule does not by itself demonstrate substantial "
        "equivalence of task formulation and control boundary."
    ),
})


# Finding D:
# Human assessment vs synthesis must remain separate.

audit_findings.append({
    "finding_id": "F04",
    "status": "CHECK_REQUIRED",
    "finding": (
        "The synthesis must preserve the distinction between human "
        "methodological assessment and the later scientific gap decision."
    ),
})


audit_df = pd.DataFrame(audit_findings)

print(audit_df.to_string(index=False))


# =============================================================================
# STEP 7 — CHECK FOR OVERREACH IN CELL23 V001 GAP STATEMENT
# =============================================================================

print("\n[STEP 7] GAP STATEMENT OVERREACH AUDIT")
print("-" * 108)

original_decision = cell23_report.get(
    "gap_decision"
)

original_statement = cell23_report.get(
    "gap_statement"
)

print("Cell23 V001 decision:")
print(original_decision)

print("\nCell23 V001 statement:")
print(original_statement)

overreach_flags = {
    "decision_uses_count_only":
        bool(closest_candidates) and any(full_d01_d09.values()),

    "d10_is_only_closely_aligned":
        all(
            d10[
                d10["candidate_id"] == cid
            ]["assessment"].iloc[0] == "CLOSELY_ALIGNED"
            for cid in closest_candidates
        ) if closest_candidates else False,

    "substantial_coverage_argument_missing":
        True,
}

print("\nFlags:")
for name, value in overreach_flags.items():
    print(f"{name:40s}: {value}")


# =============================================================================
# STEP 8 — COMPARE WITH FROZEN PROTOCOL DECISION STATES
# =============================================================================

print("\n[STEP 8] FROZEN DECISION-STATE CONTRACT")
print("-" * 108)

DECISION_STATES = {
    "GAP_DEFENSIBLE":
        "No prior work identified with substantially equivalent task formulation and control boundary.",

    "GAP_PARTIALLY_OVERLAPPING":
        "Related components exist, but the complete F-D+R formulation remains distinguishable.",

    "GAP_NOT_DEFENSIBLE":
        "A prior work or tightly connected body of work substantially covers the proposed contribution.",

    "GAP_INCONCLUSIVE":
        "Available evidence is insufficient to make a defensible novelty decision.",
}

for state, definition in DECISION_STATES.items():
    print(f"{state:28s} -> {definition}")


# =============================================================================
# STEP 9 — SAFE LOGICAL CONCLUSION OF THIS AUDIT
# =============================================================================
#
# This is NOT a new gap decision.
# It only determines whether Cell23 V001's decision logic is licensed.
# =============================================================================

print("\n[STEP 9] AUDIT CONCLUSION")
print("-" * 108)

logic_is_licensed = False

print("Cell23 V001 logic licensed : FALSE")

print(
    "\nReason:"
)

print(
    "The available evidence shows that LIT-003 is human-assessed "
    "CLOSELY_ALIGNED and D01-D09 are all PRESENT, but the frozen protocol "
    "does not authorize replacing the required substantial-coverage "
    "boundary argument with an automatic PRESENT-count rule."
)

print(
    "\nTherefore:"
)

print(
    "CELL23 V001 GAP_NOT_DEFENSIBLE = NOT VALIDATED BY THIS AUDIT"
)

print(
    "This does NOT mean GAP_DEFENSIBLE."
)

print(
    "This does NOT mean GAP_PARTIALLY_OVERLAPPING."
)

print(
    "This means ONLY that the V001 decision rule is insufficiently justified."
)


# =============================================================================
# STEP 10 — REQUIRED REBUILD CONDITION
# =============================================================================

print("\n[STEP 10] REBUILD CONDITION")
print("-" * 108)

print(
    "CELL23 V001 should NOT be patched."
)

print(
    "A future synthesis cell must rebuild the decision logic explicitly "
    "around:"
)

print(
    "  1. Frozen D01-D09 assessments"
)

print(
    "  2. Frozen D10 human overall assessment"
)

print(
    "  3. Explicit task-formulation comparison"
)

print(
    "  4. Explicit control-boundary comparison"
)

print(
    "  5. Explicit statement of what remains distinguishable "
    "or what is substantially covered"
)

print(
    "  6. No numerical PRESENT-count used as a substitute for "
    "methodological equivalence reasoning"
)


# =============================================================================
# STEP 11 — FIREWALL
# =============================================================================

print("\n[STEP 11] FIREWALL")
print("-" * 108)

print("Cell22 modified              : NO")
print("Cell23 V001 modified        : NO")
print("Human assessments modified  : NO")
print("New literature introduced   : NO")
print("Dataset modified            : NO")
print("199 cohort modified         : NO")
print("400 validation modified     : NO")
print("Q82 modified                : NO")
print("New novelty decision made   : NO")
print("Architecture implemented    : NO")


# =============================================================================
# STEP 12 — FINAL STATUS
# =============================================================================

print("\n" + "=" * 108)
print("CELL 23A V001 — FINAL STATUS")
print("=" * 108)

print("Purpose                         : LOGIC AUDIT ONLY")
print("Cell23 V001 artifact            : PRESERVED")
print("Human assessment                : PRESERVED")
print("Decision logic licensed        : NO")
print("New gap decision                : NOT MADE")
print("Scientific action               : REBUILD DECISION LOGIC")
print("=" * 108)
```

```python
# =============================================================================
# CELL 23 V002 — REBUILT GAP / NOVELTY SYNTHESIS
# =============================================================================
# EXPERIMENT : FDR-NOVELTY-GAP-001
#
# SCIENTIFIC PURPOSE
# -----------------------------------------------------------------------------
# Rebuild the Gap / Novelty Synthesis using the FROZEN human methodological
# assessment from Cell22 V004 without using numerical PRESENT-count rules.
#
# IMPORTANT:
#   - Cell23 V001 is historical/rejected and is NOT used as a decision source.
#   - Cell23A V001 established that the V001 decision rule was not licensed.
#   - This cell therefore rebuilds the synthesis logic from zero.
#
# TRUSTED INPUT:
#   CELL22 V004 ONLY
#
# FIREWALL:
#   NO HUMAN REASSESSMENT
#   NO NEW LITERATURE
#   NO DATASET MODIFICATION
#   NO 199-COHORT MODIFICATION
#   NO 400-VALIDATION MODIFICATION
#   NO Q82 MODIFICATION
#   NO MODEL
#   NO TRAINING
#   NO INFERENCE
#   NO ARCHITECTURE IMPLEMENTATION
#   NO PROMPT REVISION
#   NO GOLD / Y+ / Y- / PREFERENCE
#
# DECISION PRINCIPLE:
#   D01-D09 = dimensional human evidence
#   D10     = human overall methodological assessment
#
# NO:
#   "9 PRESENT => GAP_NOT_DEFENSIBLE"
#   "8 PRESENT => GAP_PARTIALLY_OVERLAPPING"
#   or any other count-based novelty rule.
#
# The synthesis must distinguish:
#   (1) overlap,
#   (2) complete methodological equivalence,
#   (3) the remaining F-D+R boundary.
# =============================================================================

from pathlib import Path
import hashlib
import json
from datetime import datetime, timezone

import pandas as pd


print("=" * 110)
print("CELL 23 V002 — REBUILT GAP / NOVELTY SYNTHESIS")
print("=" * 110)


# =============================================================================
# STEP 0 — SCIENTIFIC VALUE GATE
# =============================================================================

SCIENTIFIC_VALUE = True

print("\n[STEP 0] SCIENTIFIC VALUE")
print("-" * 110)
print(f"scientific_value : {SCIENTIFIC_VALUE}")
print(
    "reason           : Rebuild the Gap/Novelty Synthesis from the frozen "
    "human assessment without relying on the invalidated count-based rule."
)

if not SCIENTIFIC_VALUE:
    raise RuntimeError("STOP: Scientific-value gate failed.")


# =============================================================================
# STEP 1 — TRUSTED CELL22 V004
# =============================================================================

CELL22_PATH = Path(
    "/content/drive/MyDrive/X-AutoAlign_Legal_RAG/04_AUDIT/"
    "NOVELTY_GAP_AUDIT/CHECKPOINTS/"
    "CELL22_FROZEN_HUMAN_METHODOLOGICAL_ASSESSMENT_V004/"
    "FROZEN_HUMAN_METHODOLOGICAL_ASSESSMENT_V004.parquet"
)

CELL22_EXPECTED_SHA = (
    "4e2d9bd060761862a02b4e21ce087d0a0ebe0dd000d7877df05d1b2fb10a6ff9"
)

print("\n[STEP 1] TRUSTED CELL22 V004")
print("-" * 110)

if not CELL22_PATH.exists():
    raise RuntimeError(
        "STOP: Cell22 V004 trusted artifact not found. "
        "Do not guess another path."
    )

actual_sha = hashlib.sha256(
    CELL22_PATH.read_bytes()
).hexdigest()

print(f"expected SHA : {CELL22_EXPECTED_SHA}")
print(f"actual SHA   : {actual_sha}")
print(f"SHA PASS     : {actual_sha == CELL22_EXPECTED_SHA}")

if actual_sha != CELL22_EXPECTED_SHA:
    raise RuntimeError(
        "STOP: Cell22 V004 SHA mismatch."
    )

df = pd.read_parquet(CELL22_PATH)

print(f"shape        : {df.shape}")

if df.shape != (50, 30):
    raise RuntimeError(
        f"STOP: Expected Cell22 shape (50,30), got {df.shape}."
    )


# =============================================================================
# STEP 2 — EXACT INPUT CONTRACT
# =============================================================================

EXPECTED_COLUMNS = [
    "assessment_unit_id",
    "assessment_level",
    "candidate_id",
    "dimension_id",
    "question_id",
    "dimension_name",
    "dimension_definition",
    "title",
    "authors",
    "year",
    "venue",
    "primary_source",
    "primary_source_filename",
    "primary_source_path",
    "primary_source_sha256",
    "source_record_question_id",
    "question_text",
    "source_excerpt",
    "source_location",
    "source_text_status",
    "extraction_method",
    "extraction_status",
    "comparison_eligibility",
    "assessment",
    "evidence_strength",
    "evidence_note",
    "reviewer_note",
    "assessment_status",
    "automatic_judgment",
    "available_preceding_dimensions",
]

print("\n[STEP 2] INPUT SCHEMA")
print("-" * 110)

if list(df.columns) != EXPECTED_COLUMNS:
    raise RuntimeError(
        "STOP: Cell22 V004 schema differs from trusted contract."
    )

print("schema PASS : 30 columns exactly")


# =============================================================================
# STEP 3 — HUMAN-ASSESSMENT FIREWALL
# =============================================================================

print("\n[STEP 3] HUMAN-ASSESSMENT FIREWALL")
print("-" * 110)

checks = {
    "all_frozen":
        df["assessment_status"].eq(
            "HUMAN_ASSESSMENT_FROZEN"
        ).all(),

    "no_not_assessed":
        (~df["assessment"].eq("NOT_ASSESSED")).all(),

    "all_direct":
        df["evidence_strength"].eq("DIRECT").all(),

    "no_automatic_judgment":
        (~df["automatic_judgment"].astype(bool)).all(),
}

for name, result in checks.items():
    print(
        f"{name:30s}: {'PASS' if result else 'FAIL'}"
    )

if not all(checks.values()):
    raise RuntimeError(
        "STOP: Cell22 V004 human-assessment firewall failed."
    )


# =============================================================================
# STEP 4 — F-D+R BOUNDARY
# =============================================================================

print("\n[STEP 4] F-D+R BOUNDARY")
print("-" * 110)

FDR_BOUNDARY = {
    "stage": "POST_RETRIEVAL",
    "object": "CANDIDATE_EVIDENCE",
    "target": "EVIDENCE_SUITABILITY",
    "input_boundary": "QUESTION + CANDIDATE_EVIDENCE",
    "control": "PRE_GENERATION_EVIDENCE_ELIGIBILITY_AND_RISK_GATE",
}

for key, value in FDR_BOUNDARY.items():
    print(f"{key:18s}: {value}")


# =============================================================================
# STEP 5 — FROZEN ASSESSMENT MATRIX
# =============================================================================

EXPECTED_CANDIDATES = [
    "ANCHOR-001",
    "LIT-001",
    "LIT-002",
    "LIT-003",
    "LIT-004",
]

EXPECTED_DIMS = [
    "D01",
    "D02",
    "D03",
    "D04",
    "D05",
    "D06",
    "D07",
    "D08",
    "D09",
    "D10",
]

assessment_matrix = (
    df.pivot(
        index="candidate_id",
        columns="dimension_id",
        values="assessment"
    )
    .reindex(
        index=EXPECTED_CANDIDATES,
        columns=EXPECTED_DIMS
    )
)

print("\n[STEP 5] FROZEN HUMAN ASSESSMENT MATRIX")
print("-" * 110)

print(assessment_matrix.to_string())


# =============================================================================
# STEP 6 — HUMAN D10 IS THE OVERALL ASSESSMENT
# =============================================================================
#
# IMPORTANT:
# D10 is NOT recomputed here.
# Its existing human value is taken exactly as frozen.
# =============================================================================

print("\n[STEP 6] HUMAN D10 OVERALL ASSESSMENT")
print("-" * 110)

d10_summary = (
    df[df["dimension_id"] == "D10"]
    [["candidate_id", "assessment", "evidence_strength"]]
    .sort_values("candidate_id")
    .reset_index(drop=True)
)

print(d10_summary.to_string(index=False))


# =============================================================================
# STEP 7 — CLOSEST COMPETITOR IDENTIFICATION
# =============================================================================
#
# This uses only frozen D10 values.
# It is not a new judgment.
# =============================================================================

closest_candidates = d10_summary[
    d10_summary["assessment"] == "CLOSELY_ALIGNED"
]["candidate_id"].tolist()

partial_candidates = d10_summary[
    d10_summary["assessment"] == "PARTIALLY_OVERLAPPING"
]["candidate_id"].tolist()

distinct_candidates = d10_summary[
    d10_summary["assessment"] == "DISTINCT"
]["candidate_id"].tolist()

print("\n[STEP 7] CLOSEST COMPETITORS")
print("-" * 110)

print(
    "CLOSELY_ALIGNED         :",
    closest_candidates
)

print(
    "PARTIALLY_OVERLAPPING   :",
    partial_candidates
)

print(
    "DISTINCT                :",
    distinct_candidates
)


# =============================================================================
# STEP 8 — CORE F-D+R DIMENSION GROUPS
# =============================================================================
#
# These groups organize the frozen human dimensions for synthesis.
# No new assessment is introduced.
# =============================================================================

CORE_VERIFICATION = [
    "D01",
    "D02",
    "D03",
]

CORE_EVIDENCE_CONTROL = [
    "D05",
    "D06",
    "D07",
    "D08",
]

HUMAN_GROUNDING = [
    "D09",
]

DOMAIN = [
    "D04",
]

print("\n[STEP 8] CORE SYNTHESIS GROUPS")
print("-" * 110)

print("Verification          :", CORE_VERIFICATION)
print("Evidence control      :", CORE_EVIDENCE_CONTROL)
print("Human grounding       :", HUMAN_GROUNDING)
print("Domain-specific       :", DOMAIN)


# =============================================================================
# STEP 9 — EXPLICIT CLOSEST-COMPETITOR SYNTHESIS
# =============================================================================
#
# This is the critical correction over V001.
#
# We do NOT say:
#    "all PRESENT -> not novel"
#
# Instead:
#   A) identify the human D10 status
#   B) identify whether the frozen D01-D09 evidence establishes exact
#      equivalence or merely strong component overlap
#   C) preserve unresolved distinctions instead of inventing equivalence
# =============================================================================

competitor_rows = []

for candidate_id in closest_candidates:

    values = assessment_matrix.loc[candidate_id]

    non_present_dims = [
        dim
        for dim in EXPECTED_DIMS[:-1]
        if values[dim] != "PRESENT"
    ]

    # D10 remains the human overall judgment.
    d10_value = values["D10"]

    if len(non_present_dims) == 0:

        dimensional_pattern = (
            "All assessed D01-D09 dimensions are PRESENT."
        )

        equivalence_state = (
            "NOT_ESTABLISHED_BY_DIMENSIONAL_EVIDENCE"
        )

        synthesis_basis = (
            "The prior work covers all assessed D01-D09 components, "
            "but the human overall assessment remains CLOSELY_ALIGNED, "
            "not an explicit equivalence judgment. Therefore dimensional "
            "coverage alone cannot be promoted to substantial equivalence."
        )

    else:

        dimensional_pattern = (
            "Non-PRESENT assessed dimensions: "
            + ", ".join(non_present_dims)
        )

        equivalence_state = (
            "NOT_ESTABLISHED"
        )

        synthesis_basis = (
            "The prior work is close at D10, but at least one assessed "
            "F-D+R dimension is not fully PRESENT."
        )

    competitor_rows.append({
        "candidate_id": candidate_id,
        "d10_human_assessment": d10_value,
        "dimensional_pattern": dimensional_pattern,
        "equivalence_state": equivalence_state,
        "synthesis_basis": synthesis_basis,
    })

competitor_df = pd.DataFrame(competitor_rows)

print("\n[STEP 9] CLOSEST-COMPETITOR SYNTHESIS")
print("-" * 110)

if competitor_df.empty:
    print("No CLOSELY_ALIGNED candidate exists.")
else:
    print(competitor_df.to_string(index=False))


# =============================================================================
# STEP 10 — EXPLICIT TASK / CONTROL BOUNDARY TEST
# =============================================================================
#
# This is a qualitative logic gate.
#
# A GAP_NOT_DEFENSIBLE decision requires evidence of BOTH:
#   1) substantially equivalent task formulation
#   2) substantially equivalent control boundary
#
# Neither is allowed to be inferred merely from D01-D09 coverage.
#
# Since D10 for the closest candidate(s) is CLOSELY_ALIGNED rather than
# an explicit equivalence state, this cell must NOT manufacture complete
# equivalence.
# =============================================================================

print("\n[STEP 10] TASK / CONTROL BOUNDARY TEST")
print("-" * 110)

task_equivalence_established = False
control_boundary_equivalence_established = False

print(
    "Task formulation equivalence established      : "
    f"{task_equivalence_established}"
)

print(
    "Control-boundary equivalence established      : "
    f"{control_boundary_equivalence_established}"
)

print(
    "\nRule:"
)

print(
    "GAP_NOT_DEFENSIBLE cannot be issued unless substantial "
    "equivalence of task formulation AND control boundary is "
    "explicitly established by the available frozen evidence."
)


# =============================================================================
# STEP 11 — OVERLAP STATUS
# =============================================================================
#
# IMPORTANT:
# This is synthesis, not a new D10 judgment.
#
# If a CLOSELY_ALIGNED human assessment exists, strong methodological overlap
# is established.
#
# If exact substantial equivalence is NOT established, the conservative
# synthesis state is GAP_PARTIALLY_OVERLAPPING.
# =============================================================================

print("\n[STEP 11] SYNTHESIS OVERLAP STATE")
print("-" * 110)

if closest_candidates:

    GAP_DECISION = "GAP_PARTIALLY_OVERLAPPING"

    GAP_DECISION_BASIS = (
        "The frozen human assessment identifies closely aligned prior work "
        "(AGR and AdaKG-RAG). The evidence demonstrates substantial component "
        "overlap, but the frozen assessment does not establish an explicit "
        "substantially equivalent task formulation and control boundary. "
        "Therefore the defensible bounded synthesis state is "
        "GAP_PARTIALLY_OVERLAPPING rather than GAP_NOT_DEFENSIBLE."
    )

elif partial_candidates:

    GAP_DECISION = "GAP_PARTIALLY_OVERLAPPING"

    GAP_DECISION_BASIS = (
        "The frozen human assessment identifies partially overlapping prior "
        "work, while no closely aligned competitor is present."
    )

else:

    GAP_DECISION = "GAP_DEFENSIBLE"

    GAP_DECISION_BASIS = (
        "No closely aligned or partially overlapping prior work exists "
        "within the frozen assessed set."
    )

print(f"GAP DECISION : {GAP_DECISION}")
print(f"BASIS        : {GAP_DECISION_BASIS}")


# =============================================================================
# STEP 12 — WHAT THE RESULT DOES AND DOES NOT MEAN
# =============================================================================

print("\n[STEP 12] SCIENTIFIC INTERPRETATION BOUNDARY")
print("-" * 110)

INTERPRETATION = {
    "supported": [
        "There is substantial methodological overlap with prior work.",
        "AGR and AdaKG-RAG are the closest assessed works.",
        "The current evidence does not establish complete methodological equivalence.",
        "The current bounded state is GAP_PARTIALLY_OVERLAPPING.",
    ],

    "not_supported": [
        "F-D+R is globally novel.",
        "F-D+R is the first evidence-verification mechanism.",
        "F-D+R is superior to prior work.",
        "AGR is identical to F-D+R.",
        "The literature audit is exhaustive.",
        "A causal advantage is established.",
    ],
}

print("SUPPORTED:")
for item in INTERPRETATION["supported"]:
    print(f"  ✓ {item}")

print("\nNOT SUPPORTED:")
for item in INTERPRETATION["not_supported"]:
    print(f"  ✗ {item}")


# =============================================================================
# STEP 13 — BOUNDED GAP STATEMENT
# =============================================================================

print("\n[STEP 13] BOUNDED GAP STATEMENT")
print("-" * 110)

GAP_STATEMENT = (
    "The bounded human-grounded literature comparison identifies substantial "
    "methodological overlap between the proposed F-D+R formulation and prior "
    "evidence-verification approaches, with AGR and AdaKG-RAG assessed as "
    "the closest related works. However, the frozen human assessment does not "
    "establish complete methodological equivalence between those approaches "
    "and the full F-D+R task and control boundary. Accordingly, the current "
    "scientific conclusion is GAP_PARTIALLY_OVERLAPPING: the contribution "
    "cannot be defended as wholly distinct at the component level, and any "
    "remaining research contribution must be articulated around the specific "
    "distinguishing formulation and boundary rather than around evidence "
    "verification or sufficiency alone."
)

print(GAP_STATEMENT)


# =============================================================================
# STEP 14 — NEXT-ACTION GATE
# =============================================================================

print("\n[STEP 14] NEXT-ACTION GATE")
print("-" * 110)

if GAP_DECISION == "GAP_NOT_DEFENSIBLE":

    NEXT_ACTION = (
        "STOP architecture experimentation. Reconsider the research "
        "contribution."
    )

elif GAP_DECISION == "GAP_INCONCLUSIVE":

    NEXT_ACTION = (
        "STOP architecture experimentation. Perform only the minimum "
        "additional bounded literature audit necessary for resolution."
    )

elif GAP_DECISION == "GAP_PARTIALLY_OVERLAPPING":

    NEXT_ACTION = (
        "Do NOT claim broad novelty. First articulate the residual "
        "distinguishing F-D+R boundary and verify that the architecture "
        "protocol is explicitly aligned to that boundary."
    )

else:

    NEXT_ACTION = (
        "Gap remains distinguishable under the bounded assessment; "
        "architecture-protocol review may proceed."
    )

print(NEXT_ACTION)


# =============================================================================
# STEP 15 — SYNTHESIS TABLE
# =============================================================================

print("\n[STEP 15] FINAL SYNTHESIS TABLE")
print("-" * 110)

synthesis_table = (
    d10_summary.rename(
        columns={
            "assessment": "D10_human_overall",
            "evidence_strength": "D10_evidence_strength",
        }
    )
)

print(
    synthesis_table[
        [
            "candidate_id",
            "D10_human_overall",
            "D10_evidence_strength",
        ]
    ].to_string(index=False)
)


# =============================================================================
# STEP 16 — OUTPUT DIRECTORY
# =============================================================================

OUTPUT_DIR = Path(
    "/content/drive/MyDrive/X-AutoAlign_Legal_RAG/04_AUDIT/"
    "NOVELTY_GAP_AUDIT/CHECKPOINTS/"
    "CELL23_GAP_NOVELTY_SYNTHESIS_V002"
)

print("\n[STEP 16] OUTPUT DIRECTORY")
print("-" * 110)

print(OUTPUT_DIR)

if OUTPUT_DIR.exists():
    raise RuntimeError(
        "STOP: Cell23 V002 output directory already exists. "
        "No overwrite permitted."
    )

OUTPUT_DIR.mkdir(
    parents=True,
    exist_ok=False
)


# =============================================================================
# STEP 17 — OUTPUT ARTIFACTS
# =============================================================================

SYNTHESIS_CSV = OUTPUT_DIR / (
    "GAP_NOVELTY_SYNTHESIS_V002.csv"
)

D10_CSV = OUTPUT_DIR / (
    "FROZEN_D10_SUMMARY_V002.csv"
)

COMPETITOR_CSV = OUTPUT_DIR / (
    "CLOSEST_COMPETITOR_SYNTHESIS_V002.csv"
)

REPORT_JSON = OUTPUT_DIR / (
    "CELL23_GAP_NOVELTY_REPORT_V002.json"
)

synthesis_table.to_csv(
    SYNTHESIS_CSV,
    index=False,
    encoding="utf-8-sig"
)

d10_summary.to_csv(
    D10_CSV,
    index=False,
    encoding="utf-8-sig"
)

competitor_df.to_csv(
    COMPETITOR_CSV,
    index=False,
    encoding="utf-8-sig"
)


# =============================================================================
# STEP 18 — HASH OUTPUTS
# =============================================================================

def sha256_file(path):
    return hashlib.sha256(
        path.read_bytes()
    ).hexdigest()


artifact_hashes = {
    "synthesis_csv": sha256_file(SYNTHESIS_CSV),
    "d10_csv": sha256_file(D10_CSV),
    "competitor_csv": sha256_file(COMPETITOR_CSV),
}

print("\n[STEP 18] OUTPUT SHA256")
print("-" * 110)

for name, sha in artifact_hashes.items():
    print(f"{name:24s}: {sha}")


# =============================================================================
# STEP 19 — MACHINE-READABLE REPORT
# =============================================================================

report = {
    "experiment_id": "FDR-NOVELTY-GAP-001",
    "cell": "CELL 23",
    "version": "V002",

    "status": "SYNTHESIS_COMPLETE",

    "trusted_input": {
        "path": str(CELL22_PATH),
        "sha256": CELL22_EXPECTED_SHA,
    },

    "scope": {
        "candidates": EXPECTED_CANDIDATES,
        "dimensions": EXPECTED_DIMS,
        "human_records": 50,
    },

    "decision_logic": {
        "count_based_rule": False,
        "d10_human_assessment_preserved": True,
        "task_equivalence_established":
            task_equivalence_established,
        "control_boundary_equivalence_established":
            control_boundary_equivalence_established,
    },

    "closest_candidates": closest_candidates,

    "partial_candidates": partial_candidates,

    "distinct_candidates": distinct_candidates,

    "gap_decision": GAP_DECISION,

    "gap_decision_basis": GAP_DECISION_BASIS,

    "gap_statement": GAP_STATEMENT,

    "next_action": NEXT_ACTION,

    "firewall": {
        "human_reassessment": False,
        "new_literature": False,
        "dataset_modification": False,
        "model_inference": False,
        "training": False,
        "architecture_implementation": False,
        "prompt_revision": False,
        "gold_generation": False,
        "y_plus_generation": False,
        "y_minus_generation": False,
        "preference_generation": False,
        "population_claim": False,
        "causal_claim": False,
    },

    "artifact_hashes": artifact_hashes,

    "generated_utc":
        datetime.now(timezone.utc).isoformat(),
}

with open(
    REPORT_JSON,
    "w",
    encoding="utf-8"
) as f:
    json.dump(
        report,
        f,
        ensure_ascii=False,
        indent=2
    )

print("\n[STEP 19] REPORT")
print("-" * 110)
print(REPORT_JSON)


# =============================================================================
# STEP 20 — INDEPENDENT READ-BACK
# =============================================================================

print("\n[STEP 20] INDEPENDENT READ-BACK")
print("-" * 110)

reloaded_synthesis = pd.read_csv(
    SYNTHESIS_CSV
)

reloaded_d10 = pd.read_csv(
    D10_CSV
)

reloaded_competitor = pd.read_csv(
    COMPETITOR_CSV
)

readback_checks = {
    "synthesis_exists":
        SYNTHESIS_CSV.exists(),

    "d10_exists":
        D10_CSV.exists(),

    "competitor_exists":
        COMPETITOR_CSV.exists(),

    "report_exists":
        REPORT_JSON.exists(),

    "synthesis_rows":
        len(reloaded_synthesis) == 5,

    "d10_rows":
        len(reloaded_d10) == 5,

    "competitor_rows":
        len(reloaded_competitor) == len(
            closest_candidates
        ),

    "decision_present":
        GAP_DECISION in {
            "GAP_DEFENSIBLE",
            "GAP_PARTIALLY_OVERLAPPING",
            "GAP_NOT_DEFENSIBLE",
            "GAP_INCONCLUSIVE",
        },

    "decision_not_count_based":
        True,
}

for name, result in readback_checks.items():
    print(
        f"{name:35s}: {'PASS' if result else 'FAIL'}"
    )

if not all(readback_checks.values()):
    raise RuntimeError(
        "STOP: Cell23 V002 read-back validation failed."
    )


# =============================================================================
# STEP 21 — FINAL SYNTHESIS GATES
# =============================================================================

print("\n[STEP 21] FINAL SYNTHESIS GATES")
print("-" * 110)

FINAL_CHECKS = {
    "trusted_cell22_sha":
        actual_sha == CELL22_EXPECTED_SHA,

    "human_assessments_unchanged":
        len(df) == 50,

    "five_candidates":
        df["candidate_id"].nunique() == 5,

    "ten_dimensions":
        df["dimension_id"].nunique() == 10,

    "d10_preserved":
        len(d10_summary) == 5,

    "closest_candidates_identified":
        closest_candidates == [
            "LIT-003",
            "LIT-004",
        ],

    "no_new_human_judgment":
        True,

    "no_new_literature":
        True,

    "no_dataset_modification":
        True,

    "decision_valid":
        GAP_DECISION in {
            "GAP_DEFENSIBLE",
            "GAP_PARTIALLY_OVERLAPPING",
            "GAP_NOT_DEFENSIBLE",
            "GAP_INCONCLUSIVE",
        },

    "decision_not_based_on_present_count":
        True,

    "outputs_persisted":
        all(
            path.exists()
            for path in [
                SYNTHESIS_CSV,
                D10_CSV,
                COMPETITOR_CSV,
                REPORT_JSON,
            ]
        ),
}

for name, result in FINAL_CHECKS.items():
    print(
        f"{name:40s}: {'PASS' if result else 'FAIL'}"
    )

if not all(FINAL_CHECKS.values()):
    raise RuntimeError(
        "STOP: One or more Cell23 V002 synthesis gates failed."
    )


# =============================================================================
# STEP 22 — FINAL STATUS
# =============================================================================

print("\n" + "=" * 110)
print("CELL 23 V002 — FINAL STATUS")
print("=" * 110)

print("Trusted input                   : CELL22 V004")
print("Human reassessment              : NO")
print("New literature                 : NO")
print("Dataset modification            : NONE")
print("199 cohort modification         : NONE")
print("400 validation modification     : NONE")
print("Q82 modification                : NONE")
print("Count-based decision rule       : NOT USED")

print("\nClosest assessed works:")
for candidate in closest_candidates:
    print(f"  - {candidate}")

print(f"\nGap decision                    : {GAP_DECISION}")

print("\nGap statement:")
print(GAP_STATEMENT)

print("\nFINAL RESULT:")
print("CELL 23 V002 = REBUILT GAP / NOVELTY SYNTHESIS PASS")

print("\nNEXT:")
print(NEXT_ACTION)

print("=" * 110)
```

```python
# ================================================================================================================
# X-AutoAlign Legal RAG
# F-D+R
#
# CELL 25 V003 — SEPARATE CONTROLLED ARCHITECTURE STUDY DESIGN GATE
# REBUILT FROM ZERO
#
# V001 = HISTORICAL / REJECTED
# V002 = HISTORICAL / REJECTED
#
# V003 PRINCIPLE:
#   - Never assume upstream filename semantics.
#   - Never assume upstream JSON key names.
#   - Discover actual JSON artifacts.
#   - Validate them by frozen scientific content.
#   - STOP if content cannot be uniquely identified.
# ================================================================================================================

import os
import json
import hashlib
from pathlib import Path
from datetime import datetime, timezone


print("=" * 120)
print("CELL 25 V003 — SEPARATE CONTROLLED ARCHITECTURE STUDY DESIGN GATE")
print("REBUILT FROM ZERO")
print("=" * 120)


# ================================================================================================================
# STEP 0 — SCIENTIFIC VALUE GATE
# ================================================================================================================

print("\n[STEP 0] SCIENTIFIC VALUE GATE")
print("-" * 120)

SCIENTIFIC_VALUE = {
    "status": "PASS",
    "question": "ما فائدتها للرسالة العلمية أو لبناء الـdataset؟",
    "answer": (
        "Freeze the separately controlled architecture-study protocol before "
        "implementation or inference, making the study reproducible, auditable, "
        "and protected from post-hoc tuning."
    ),
    "design_only": True,
    "model_inference": False,
    "training": False,
    "dataset_modification": False,
}

assert SCIENTIFIC_VALUE["status"] == "PASS"
assert SCIENTIFIC_VALUE["design_only"] is True
assert SCIENTIFIC_VALUE["model_inference"] is False
assert SCIENTIFIC_VALUE["training"] is False
assert SCIENTIFIC_VALUE["dataset_modification"] is False

print("Mandatory question:")
print("  ما فائدتها للرسالة العلمية أو لبناء الـdataset؟")
print()
print("Answer:")
print(f"  {SCIENTIFIC_VALUE['answer']}")
print()
print("✓ DIRECT SCIENTIFIC VALUE")
print("✓ DESIGN ONLY")
print("✓ NO MODEL INFERENCE")
print("✓ NO TRAINING")
print("✓ NO DATASET MODIFICATION")


# ================================================================================================================
# STEP 1 — EXPERIMENT IDENTITY
# ================================================================================================================

print("\n[STEP 1] EXPERIMENT IDENTITY")
print("-" * 120)

EXPERIMENT_ID = "FDR-ARCH-CMP-001"
CELL_ID = "CELL_25"
CELL_VERSION = "V003"
EXPERIMENT_TYPE = "SEPARATE_CONTROLLED_ARCHITECTURE_STUDY_DESIGN"
STATUS = "DESIGN_GATE_REBUILT"

print(f"Experiment ID   : {EXPERIMENT_ID}")
print(f"Cell            : {CELL_ID}")
print(f"Version         : {CELL_VERSION}")
print(f"Experiment Type : {EXPERIMENT_TYPE}")
print(f"Status          : {STATUS}")


# ================================================================================================================
# STEP 2 — CANONICAL DIRECTORIES
# ================================================================================================================

print("\n[STEP 2] CANONICAL DIRECTORIES")
print("-" * 120)

BASE_DIR = Path(
    "/content/drive/MyDrive/F-D+R/DATASET_CONSTRUCTION/DESIGN"
)

CELL24_DIR = (
    BASE_DIR /
    "CELL_24_RESIDUAL_FDR_BOUNDARY_V002"
)

CELL90_DIR = (
    BASE_DIR /
    "CELL_90_ARCHITECTURE_RECONSIDERATION_GATE"
)

CELL87_DIR = (
    BASE_DIR /
    "CELL_87_SCIENTIFIC_DECISION_GATE"
)

CELL25_DIR = (
    BASE_DIR /
    "CELL_25_SEPARATE_CONTROLLED_ARCHITECTURE_STUDY_DESIGN_V003"
)

for label, path in [
    ("BASE_DIR", BASE_DIR),
    ("CELL24_DIR", CELL24_DIR),
    ("CELL90_DIR", CELL90_DIR),
    ("CELL87_DIR", CELL87_DIR),
    ("CELL25_DIR", CELL25_DIR),
]:
    print(f"{label:12s}: {path}")

assert BASE_DIR.exists(), (
    f"❌ Missing DESIGN directory:\n{BASE_DIR}"
)

assert CELL24_DIR.exists(), (
    f"❌ Missing Cell 24 directory:\n{CELL24_DIR}"
)

assert CELL90_DIR.exists(), (
    f"❌ Missing Cell 90 directory:\n{CELL90_DIR}"
)

assert CELL87_DIR.exists(), (
    f"❌ Missing Cell 87.13 directory:\n{CELL87_DIR}"
)

CELL25_DIR.mkdir(
    parents=True,
    exist_ok=True
)

print("\n✓ All required directories exist")
print("✓ Cell 25 V003 output directory ready")


# ================================================================================================================
# STEP 3 — GENERIC JSON UTILITIES
# ================================================================================================================

print("\n[STEP 3] JSON CONTENT INSPECTION UTILITIES")
print("-" * 120)

def load_json(path: Path):
    with open(path, "r", encoding="utf-8") as f:
        return json.load(f)


def sha256_file(path: Path) -> str:
    h = hashlib.sha256()
    with open(path, "rb") as f:
        for block in iter(
            lambda: f.read(1024 * 1024),
            b""
        ):
            h.update(block)
    return h.hexdigest()


def flatten_json_values(obj):
    """
    Recursively collect scalar values from arbitrary JSON structures.

    This intentionally does NOT assume key names.
    """
    values = []

    if isinstance(obj, dict):
        for value in obj.values():
            values.extend(flatten_json_values(value))

    elif isinstance(obj, list):
        for value in obj:
            values.extend(flatten_json_values(value))

    else:
        values.append(obj)

    return values


def json_contains_value(obj, target):
    """
    Exact scalar/value presence check anywhere in JSON.
    """
    return any(
        value == target
        for value in flatten_json_values(obj)
    )


def json_contains_all(obj, targets):
    return all(
        json_contains_value(obj, target)
        for target in targets
    )


print("✓ Content-based inspection utilities ready")
print("✓ No upstream key-name assumptions")


# ================================================================================================================
# STEP 4 — ACTUAL CELL 24 JSON DISCOVERY
# ================================================================================================================

print("\n[STEP 4] ACTUAL CELL 24 JSON DISCOVERY")
print("-" * 120)

cell24_json_files = sorted(
    CELL24_DIR.glob("*.json")
)

print("JSON files actually present:")

for path in cell24_json_files:
    print(f"  - {path.name}")

assert len(cell24_json_files) > 0, (
    "❌ No JSON artifacts found in Cell 24 directory."
)

print(
    f"\n✓ {len(cell24_json_files)} JSON artifact(s) discovered"
)


# ================================================================================================================
# STEP 5 — CELL 24 SCIENTIFIC CONTENT FINGERPRINT
# ================================================================================================================

print("\n[STEP 5] CELL 24 SCIENTIFIC CONTENT FINGERPRINT")
print("-" * 120)

CELL24_SCIENTIFIC_FINGERPRINT = [
    "GAP_PARTIALLY_OVERLAPPING",
    "BOUNDARY_ALIGNED_PENDING_EMPIRICAL_VALIDATION",
    "POST_RETRIEVAL",
    "CANDIDATE_EVIDENCE",
    "EVIDENCE_SUITABILITY",
    "QUESTION + CANDIDATE_EVIDENCE",
    "PRE_GENERATION_EVIDENCE_ELIGIBILITY_AND_RISK_GATE",
    "LEGAL_RAG",
]

print("Required frozen scientific values:")

for value in CELL24_SCIENTIFIC_FINGERPRINT:
    print(f"  ✓ {value}")


# ================================================================================================================
# STEP 6 — CONTENT-BASED CELL 24 IDENTIFICATION
# ================================================================================================================

print("\n[STEP 6] CONTENT-BASED CELL 24 IDENTIFICATION")
print("-" * 120)

cell24_loaded = []

for path in cell24_json_files:

    try:
        payload = load_json(path)
    except Exception as exc:
        print(
            f"  ⚠ Could not parse {path.name}: {exc}"
        )
        continue

    matched_values = [
        value
        for value in CELL24_SCIENTIFIC_FINGERPRINT
        if json_contains_value(payload, value)
    ]

    print(
        f"\n{path.name}"
    )
    print(
        f"  Scientific fingerprint matches: "
        f"{len(matched_values)}/"
        f"{len(CELL24_SCIENTIFIC_FINGERPRINT)}"
    )

    for value in matched_values:
        print(f"    ✓ {value}")

    cell24_loaded.append(
        {
            "path": path,
            "payload": payload,
            "matched_values": matched_values,
        }
    )


matching_cell24 = [
    item
    for item in cell24_loaded
    if len(item["matched_values"])
    == len(CELL24_SCIENTIFIC_FINGERPRINT)
]

print(
    f"\nUnique full scientific matches: "
    f"{len(matching_cell24)}"
)

assert len(matching_cell24) == 1, (
    "❌ Cell 24 scientific artifact could not be uniquely "
    "identified by its frozen content. STOP — no guessing."
)

CELL24_ARTIFACT_PATH = matching_cell24[0]["path"]
cell24 = matching_cell24[0]["payload"]

print(
    "\n✓ Cell 24 artifact uniquely identified by "
    "scientific content:"
)

print(
    f"  {CELL24_ARTIFACT_PATH.name}"
)


# ================================================================================================================
# STEP 7 — CELL 24 ARTIFACT HASH
# ================================================================================================================

print("\n[STEP 7] CELL 24 ARTIFACT HASH LOCK")
print("-" * 120)

CELL24_SHA256 = sha256_file(
    CELL24_ARTIFACT_PATH
)

print(
    f"Cell 24 SHA256 : {CELL24_SHA256}"
)

assert len(CELL24_SHA256) == 64

print("✓ Cell 24 hash recorded")


# ================================================================================================================
# STEP 8 — CELL 24 SCIENTIFIC VALUES LOCK
# ================================================================================================================

print("\n[STEP 8] CELL 24 SCIENTIFIC VALUES LOCK")
print("-" * 120)

assert json_contains_value(
    cell24,
    "GAP_PARTIALLY_OVERLAPPING"
)

assert json_contains_value(
    cell24,
    "BOUNDARY_ALIGNED_PENDING_EMPIRICAL_VALIDATION"
)

assert json_contains_value(
    cell24,
    "POST_RETRIEVAL"
)

assert json_contains_value(
    cell24,
    "CANDIDATE_EVIDENCE"
)

assert json_contains_value(
    cell24,
    "EVIDENCE_SUITABILITY"
)

assert json_contains_value(
    cell24,
    "LEGAL_RAG"
)

print("✓ GAP_PARTIALLY_OVERLAPPING")
print("✓ BOUNDARY_ALIGNED_PENDING_EMPIRICAL_VALIDATION")
print("✓ POST_RETRIEVAL")
print("✓ CANDIDATE_EVIDENCE")
print("✓ EVIDENCE_SUITABILITY")
print("✓ LEGAL_RAG")


# ================================================================================================================
# STEP 9 — CELL 90 V003 EXACT CANONICAL ARTIFACT
# ================================================================================================================

print("\n[STEP 9] CELL 90 V003 CANONICAL ARTIFACT")
print("-" * 120)

CELL90_SUMMARY_PATH = (
    CELL90_DIR /
    "FDR_ARCHITECTURE_RECONSIDERATION_GATE_90_V003_SUMMARY.json"
)

assert CELL90_SUMMARY_PATH.exists(), (
    f"❌ Missing Cell 90 V003 summary:\n"
    f"{CELL90_SUMMARY_PATH}"
)

cell90 = load_json(
    CELL90_SUMMARY_PATH
)

CELL90_SHA256 = sha256_file(
    CELL90_SUMMARY_PATH
)

print("✓ Cell 90 summary exists")
print(f"✓ Cell 90 SHA256 : {CELL90_SHA256}")


# ================================================================================================================
# STEP 10 — CELL 90 IDENTITY / AUTHORIZATION LOCK
# ================================================================================================================

print("\n[STEP 10] CELL 90 IDENTITY / AUTHORIZATION LOCK")
print("-" * 120)

cell90_id = cell90.get(
    "experiment_id"
)

cell90_status = cell90.get(
    "status"
)

cell90_scope = cell90.get(
    "scope"
)

cell90_flag = cell90.get(
    "architecture_flag"
)

print(f"Experiment ID     : {cell90_id}")
print(f"Status            : {cell90_status}")
print(f"Scope             : {cell90_scope}")
print(f"Architecture flag : {cell90_flag}")

assert cell90_id == (
    "CELL90_ARCHITECTURE_RECONSIDERATION_GATE_V003"
)

assert cell90_status == (
    "COMPLETED_READ_ONLY"
)

assert cell90_scope == (
    "FROZEN_Q82_01_Q82_10"
)

assert cell90_flag == (
    "ARCHITECTURE_RECONSIDERATION_WARRANTED"
)

print("✓ Cell 90 identity locked")
print("✓ Separate-study authorization confirmed")
print("✓ Frozen Q82 scope confirmed")


# ================================================================================================================
# STEP 11 — CELL 87.13 BASELINE LOCK
# ================================================================================================================

print("\n[STEP 11] CELL 87.13 BASELINE LOCK")
print("-" * 120)

CELL87_SUMMARY_PATH = (
    CELL87_DIR /
    "FDR_SCIENTIFIC_DECISION_GATE_87_13_SUMMARY_V002.json"
)

assert CELL87_SUMMARY_PATH.exists(), (
    f"❌ Missing Cell 87.13 summary:\n"
    f"{CELL87_SUMMARY_PATH}"
)

cell87 = load_json(
    CELL87_SUMMARY_PATH
)

CELL87_SHA256 = sha256_file(
    CELL87_SUMMARY_PATH
)

cell87_decision = cell87.get(
    "scientific_decision"
)

cell87_flag = cell87.get(
    "architecture_flag"
)

cell87_scope = cell87.get(
    "scope"
)

print(f"Scientific decision : {cell87_decision}")
print(f"Architecture flag   : {cell87_flag}")
print(f"Scope               : {cell87_scope}")

assert cell87_decision == (
    "STOP_PROMPT_REVISION_LOOP"
)

assert cell87_flag == (
    "ARCHITECTURE_RECONSIDERATION_WARRANTED"
)

assert cell87_scope == (
    "FROZEN_Q82_01_Q82_10"
)

print("✓ Cell 87.13 baseline decision locked")
print("✓ Prompt Revision Loop remains closed")


# ================================================================================================================
# STEP 12 — BOUNDED ARCHITECTURE RESEARCH QUESTION
# ================================================================================================================

print("\n[STEP 12] BOUNDED ARCHITECTURE RESEARCH QUESTION")
print("-" * 120)

RESEARCH_QUESTION = (
    "Under a separately controlled future study, can alternative proposition "
    "representation or architecture designs reduce the structural drift and "
    "mechanical instability observed in the frozen Q82_01 → Q82_10 diagnostic "
    "without introducing unacceptable regression or structural distortion?"
)

print(RESEARCH_QUESTION)

assert "Q82_01" in RESEARCH_QUESTION
assert "Q82_10" in RESEARCH_QUESTION
assert "structural drift" in RESEARCH_QUESTION
assert "mechanical instability" in RESEARCH_QUESTION
assert "unacceptable regression" in RESEARCH_QUESTION

print("✓ Research question locked")


# ================================================================================================================
# STEP 13 — EXPERIMENTAL SCOPE
# ================================================================================================================

print("\n[STEP 13] EXPERIMENTAL SCOPE")
print("-" * 120)

SCOPE = {
    "scope_id": "FROZEN_Q82_01_Q82_10",
    "case_count": 10,
    "expansion_allowed": False,
    "population_claim_allowed": False,
    "causal_claim_allowed": False,
}

for key, value in SCOPE.items():
    print(f"{key:32s}: {value}")

assert SCOPE["scope_id"] == (
    "FROZEN_Q82_01_Q82_10"
)

assert SCOPE["case_count"] == 10
assert SCOPE["expansion_allowed"] is False
assert SCOPE["population_claim_allowed"] is False
assert SCOPE["causal_claim_allowed"] is False

print("\n✓ Scope frozen")


# ================================================================================================================
# STEP 14 — PHASE 1 ARCHITECTURE CONDITIONS
# ================================================================================================================

print("\n[STEP 14] PHASE 1 ARCHITECTURE CONDITIONS")
print("-" * 120)

CONDITIONS = {
    "BASELINE": {
        "name": "Current Cell 87.13 architecture",
        "role": "FROZEN_BASELINE",
        "target": "REFERENCE",
        "phase": "PHASE_1",
    },

    "CANDIDATE_A": {
        "name": "Schema-constrained decoding",
        "role": "ALTERNATIVE",
        "target": "GENERATION_FORMAT_AND_SCHEMA_FAILURE",
        "phase": "PHASE_1",
    },

    "CANDIDATE_B": {
        "name": (
            "Decoupled proposition extraction → "
            "independent proposition judgment → "
            "deterministic aggregation"
        ),
        "role": "ALTERNATIVE",
        "target": "ARCHITECTURAL_CONFLATION",
        "phase": "PHASE_1",
    },

    "CANDIDATE_C": {
        "name": "Trained verifier",
        "role": "DEFERRED",
        "target": "VERIFIER_LEARNING",
        "phase": "LATER_STUDY",
    },
}

for condition_id, specification in CONDITIONS.items():
    print(
        f"{condition_id:14s} | "
        f"{specification['role']:20s} | "
        f"{specification['phase']:16s} | "
        f"{specification['name']}"
    )

assert CONDITIONS["BASELINE"]["phase"] == "PHASE_1"
assert CONDITIONS["CANDIDATE_A"]["phase"] == "PHASE_1"
assert CONDITIONS["CANDIDATE_B"]["phase"] == "PHASE_1"
assert CONDITIONS["CANDIDATE_C"]["phase"] == "LATER_STUDY"

print("\n✓ Baseline + Candidate A + Candidate B locked")
print("✓ Candidate C deferred")


# ================================================================================================================
# STEP 15 — FIXED INPUT / OUTPUT BOUNDARY
# ================================================================================================================

print("\n[STEP 15] FIXED INPUT / OUTPUT BOUNDARY")
print("-" * 120)

INPUT_OUTPUT_BOUNDARY = {
    "input_scope": "FROZEN_Q82_01_Q82_10",
    "case_order": "IDENTICAL_ACROSS_CONDITIONS",
    "output_schema": "COMMON_FIXED_SCHEMA",

    "reference_answer": False,
    "human_label": False,
    "gold": False,
    "y_plus": False,
    "y_minus": False,
    "preference": False,

    "retrieval_rank_as_factuality_evidence": False,
    "dataset_modification": False,
    "post_hoc_tuning": False,
}

for key, value in INPUT_OUTPUT_BOUNDARY.items():
    print(f"{key:48s}: {value}")

for key in [
    "reference_answer",
    "human_label",
    "gold",
    "y_plus",
    "y_minus",
    "preference",
    "retrieval_rank_as_factuality_evidence",
    "dataset_modification",
    "post_hoc_tuning",
]:
    assert INPUT_OUTPUT_BOUNDARY[key] is False

print("\n✓ Input/output boundary frozen")
print("✓ No reference-answer leakage")
print("✓ No human-label leakage")
print("✓ No dataset modification")
print("✓ No post-hoc tuning")


# ================================================================================================================
# STEP 16 — EXECUTION CONTROL
# ================================================================================================================

print("\n[STEP 16] EXECUTION CONTROL")
print("-" * 120)

EXECUTION_CONTROL = {
    "fixed_configuration_per_condition": True,
    "all_settings_logged": True,
    "deterministic_decoding_for_generative_candidates": True,
    "post_hoc_tuning": False,
    "result_driven_prompt_tuning": False,
    "repeated_runs": 3,
    "repeated_runs_purpose": "REPRODUCIBILITY_ONLY",
}

for key, value in EXECUTION_CONTROL.items():
    print(f"{key:52s}: {value}")

assert EXECUTION_CONTROL[
    "fixed_configuration_per_condition"
] is True

assert EXECUTION_CONTROL[
    "all_settings_logged"
] is True

assert EXECUTION_CONTROL[
    "post_hoc_tuning"
] is False

assert EXECUTION_CONTROL[
    "result_driven_prompt_tuning"
] is False

assert EXECUTION_CONTROL[
    "repeated_runs"
] == 3

assert EXECUTION_CONTROL[
    "repeated_runs_purpose"
] == "REPRODUCIBILITY_ONLY"

print("\n✓ Execution controls frozen")


# ================================================================================================================
# STEP 17 — PREDEFINED METRICS
# ================================================================================================================

print("\n[STEP 17] PREDEFINED METRICS")
print("-" * 120)

METRICS = {
    "semantic_agreement":
        "Agreement with frozen researcher adjudication.",

    "target_decomposition_type_relation_correctness":
        "Correct target decomposition/type/relation where applicable.",

    "evidence_focus_correctness":
        "Correct evidence-focused proposition target.",

    "structure_schema_validity":
        "Structural / JSON schema validity.",

    "language_preservation":
        "Preservation of intended language.",

    "answer_neutrality":
        "Neutral evidence-oriented proposition representation.",

    "regression_rate":
        "Regression relative to the frozen baseline.",

    "mechanical_reproducibility":
        "Run-to-run reproducibility.",

    "structural_drift_reduction":
        "Reduction in the predefined structural-drift failure mode.",
}

for metric, definition in METRICS.items():
    print(
        f"{metric:50s} | {definition}"
    )

assert len(METRICS) == 9

print("\n✓ Metric framework frozen")


# ================================================================================================================
# STEP 18 — ACCEPTANCE CRITERIA
# ================================================================================================================

print("\n[STEP 18] PREDEFINED ACCEPTANCE CRITERIA")
print("-" * 120)

ACCEPTANCE_CRITERIA = {
    "no_unacceptable_regression": True,

    "no_new_schema_failure": True,

    "mechanical_sd_max": 0.10,

    "targeted_failure_mode_improvement_required":
        True,

    "baseline_noninferiority_alone_is_not_pass":
        True,

    "population_claim_allowed":
        False,

    "causal_claim_allowed":
        False,
}

for key, value in ACCEPTANCE_CRITERIA.items():
    print(f"{key:54s}: {value}")

assert ACCEPTANCE_CRITERIA[
    "no_unacceptable_regression"
] is True

assert ACCEPTANCE_CRITERIA[
    "no_new_schema_failure"
] is True

assert ACCEPTANCE_CRITERIA[
    "mechanical_sd_max"
] == 0.10

assert ACCEPTANCE_CRITERIA[
    "targeted_failure_mode_improvement_required"
] is True

assert ACCEPTANCE_CRITERIA[
    "baseline_noninferiority_alone_is_not_pass"
] is True

print("\n✓ Acceptance criteria frozen")
print("✓ Targeted improvement required")
print("✓ Non-inferiority alone is NOT PASS")


# ================================================================================================================
# STEP 19 — CANDIDATE-SPECIFIC TARGETS
# ================================================================================================================

print("\n[STEP 19] CANDIDATE-SPECIFIC TARGETS")
print("-" * 120)

TARGETS = {
    "CANDIDATE_A": {
        "failure_mode":
            "GENERATION_FORMAT_AND_SCHEMA_FAILURE",
        "primary_metric":
            "structure_schema_validity",
    },

    "CANDIDATE_B": {
        "failure_mode":
            "ARCHITECTURAL_CONFLATION",
        "primary_metric":
            "structural_drift_reduction",
    },
}

for candidate, specification in TARGETS.items():
    print(
        f"{candidate:14s} | "
        f"failure={specification['failure_mode']} | "
        f"primary_metric={specification['primary_metric']}"
    )

assert TARGETS["CANDIDATE_A"][
    "primary_metric"
] == "structure_schema_validity"

assert TARGETS["CANDIDATE_B"][
    "primary_metric"
] == "structural_drift_reduction"

print("\n✓ Candidate-specific improvement targets frozen")


# ================================================================================================================
# STEP 20 — MECHANICAL REPRODUCIBILITY POLICY
# ================================================================================================================

print("\n[STEP 20] MECHANICAL REPRODUCIBILITY POLICY")
print("-" * 120)

REPRODUCIBILITY = {
    "runs": 3,
    "measurement_only": True,
    "tuning_loop": False,
    "reported_separately": True,
    "run_selection_after_observation": False,
}

for key, value in REPRODUCIBILITY.items():
    print(f"{key:48s}: {value}")

assert REPRODUCIBILITY["runs"] == 3
assert REPRODUCIBILITY["measurement_only"] is True
assert REPRODUCIBILITY["tuning_loop"] is False
assert REPRODUCIBILITY["reported_separately"] is True
assert (
    REPRODUCIBILITY[
        "run_selection_after_observation"
    ]
    is False
)

print("\n✓ Reproducibility is separated from architectural robustness")


# ================================================================================================================
# STEP 21 — STOP RULE
# ================================================================================================================

print("\n[STEP 21] STOP RULE")
print("-" * 120)

STOP_RULE = {
    "ZERO_PASS":
        "Reject both candidates and formulate one new bounded architectural hypothesis.",

    "ONE_PASS":
        "Freeze the surviving representation as the Phase-2 baseline.",

    "BOTH_PASS":
        "Require an explicit new decision record before any expanded comparison.",

    "INCONCLUSIVE":
        "Allow one bounded follow-up with a new decision record; no open-ended tuning.",
}

for state, action in STOP_RULE.items():
    print(f"{state:12s} → {action}")

assert len(STOP_RULE) == 4

print("\n✓ Stop rules frozen")


# ================================================================================================================
# STEP 22 — FUTURE EXECUTION OUTPUT CONTRACT
# ================================================================================================================

print("\n[STEP 22] FUTURE EXECUTION OUTPUT CONTRACT")
print("-" * 120)

OUTPUT_CONTRACT = [
    "Configuration manifest",
    "Baseline outputs",
    "Candidate A outputs",
    "Candidate B outputs",
    "Reproducibility logs",
    "Metric table",
    "Paired regression table",
    "Failure-mode comparison",
    "Final PASS / FAIL / INCONCLUSIVE decision record",
]

for i, item in enumerate(
    OUTPUT_CONTRACT,
    start=1
):
    print(f"{i:02d}. {item}")

assert len(OUTPUT_CONTRACT) == 9

print("\n✓ Output contract frozen")


# ================================================================================================================
# STEP 23 — HARD SCIENTIFIC FIREWALL
# ================================================================================================================

print("\n[STEP 23] HARD SCIENTIFIC FIREWALL")
print("-" * 120)

FIREWALL = {
    "read_only_upstream": True,

    "dataset_modification": False,
    "new_adjudication": False,
    "automatic_gold_generation": False,
    "y_plus_generation": False,
    "y_minus_generation": False,
    "preference_generation": False,

    "prompt_revision": False,
    "architecture_implementation": False,
    "model_training": False,
    "model_inference": False,

    "reopen_199_cohort": False,
    "reopen_400_validation": False,
    "expand_q82": False,

    "population_claim": False,
    "causal_claim": False,
    "broad_novelty_claim": False,

    "reference_answer_leakage": False,
    "human_label_leakage": False,
}

for key, value in FIREWALL.items():
    print(
        f"{key:42s}: {value}"
    )

assert FIREWALL[
    "read_only_upstream"
] is True

for key, value in FIREWALL.items():
    if key != "read_only_upstream":
        assert value is False, (
            f"❌ Firewall violation: {key}"
        )

print("\n✓ HARD FIREWALL PASS")


# ================================================================================================================
# STEP 24 — BUILD PROTOCOL
# ================================================================================================================

print("\n[STEP 24] BUILD PROTOCOL")
print("-" * 120)

PROTOCOL = {
    "experiment_id": EXPERIMENT_ID,
    "cell": CELL_ID,
    "version": CELL_VERSION,
    "experiment_type": EXPERIMENT_TYPE,

    "status": "READY_FOR_FREEZE",

    "research_question": RESEARCH_QUESTION,

    "scope": SCOPE,

    "conditions": CONDITIONS,

    "input_output_boundary":
        INPUT_OUTPUT_BOUNDARY,

    "execution_control":
        EXECUTION_CONTROL,

    "metrics":
        METRICS,

    "acceptance_criteria":
        ACCEPTANCE_CRITERIA,

    "candidate_targets":
        TARGETS,

    "reproducibility":
        REPRODUCIBILITY,

    "stop_rule":
        STOP_RULE,

    "output_contract":
        OUTPUT_CONTRACT,

    "firewall":
        FIREWALL,

    "upstream_locks": {
        "cell24_artifact":
            str(CELL24_ARTIFACT_PATH),

        "cell24_sha256":
            CELL24_SHA256,

        "cell24_gap_state":
            "GAP_PARTIALLY_OVERLAPPING",

        "cell24_architecture_alignment":
            "BOUNDARY_ALIGNED_PENDING_EMPIRICAL_VALIDATION",

        "cell90_summary":
            str(CELL90_SUMMARY_PATH),

        "cell90_sha256":
            CELL90_SHA256,

        "cell90_status":
            cell90_status,

        "cell90_scope":
            cell90_scope,

        "cell87_summary":
            str(CELL87_SUMMARY_PATH),

        "cell87_sha256":
            CELL87_SHA256,

        "cell87_decision":
            cell87_decision,

        "cell87_scope":
            cell87_scope,
    },
}

print("✓ Protocol object built")


# ================================================================================================================
# STEP 25 — PROTOCOL HASH
# ================================================================================================================

print("\n[STEP 25] PROTOCOL CONTENT HASH")
print("-" * 120)

protocol_serialized = json.dumps(
    PROTOCOL,
    ensure_ascii=False,
    sort_keys=True,
    separators=(",", ":"),
)

PROTOCOL_SHA256 = hashlib.sha256(
    protocol_serialized.encode("utf-8")
).hexdigest()

print(
    f"Protocol SHA256 : {PROTOCOL_SHA256}"
)

assert len(PROTOCOL_SHA256) == 64


# ================================================================================================================
# STEP 26 — WRITE ARTIFACTS
# ================================================================================================================

print("\n[STEP 26] WRITE ARTIFACTS")
print("-" * 120)

PROTOCOL_PATH = (
    CELL25_DIR /
    "FDR_ARCH_CMP_001_PROTOCOL_V003.json"
)

SUMMARY_PATH = (
    CELL25_DIR /
    "FDR_ARCH_CMP_001_CELL25_V003_SUMMARY.json"
)

MANIFEST_PATH = (
    CELL25_DIR /
    "FDR_ARCH_CMP_001_CELL25_V003_MANIFEST.json"
)

SUMMARY = {
    "experiment_id": EXPERIMENT_ID,
    "cell_id": CELL_ID,
    "cell_version": CELL_VERSION,
    "experiment_type": EXPERIMENT_TYPE,

    "status":
        "READY_FOR_FREEZE",

    "scientific_value":
        SCIENTIFIC_VALUE,

    "research_question":
        RESEARCH_QUESTION,

    "scope":
        SCOPE,

    "phase_1_conditions": [
        "BASELINE",
        "CANDIDATE_A",
        "CANDIDATE_B",
    ],

    "candidate_c_status":
        "DEFERRED",

    "cell24_gap_state":
        "GAP_PARTIALLY_OVERLAPPING",

    "cell24_architecture_alignment":
        "BOUNDARY_ALIGNED_PENDING_EMPIRICAL_VALIDATION",

    "cell90_status":
        cell90_status,

    "cell90_scope":
        cell90_scope,

    "cell87_decision":
        cell87_decision,

    "protocol_sha256":
        PROTOCOL_SHA256,

    "design_only":
        True,

    "inference_in_this_cell":
        False,

    "training_in_this_cell":
        False,

    "dataset_modification":
        False,

    "next_gate":
        "GO_TO_CONTROLLED_ARCHITECTURE_COMPARISON_EXECUTION",

    "created_utc":
        datetime.now(
            timezone.utc
        ).isoformat(),
}

MANIFEST = {
    "experiment_id":
        EXPERIMENT_ID,

    "cell_id":
        CELL_ID,

    "cell_version":
        CELL_VERSION,

    "upstream": {
        "cell24": {
            "path":
                str(CELL24_ARTIFACT_PATH),

            "sha256":
                CELL24_SHA256,
        },

        "cell90": {
            "path":
                str(CELL90_SUMMARY_PATH),

            "sha256":
                CELL90_SHA256,
        },

        "cell87_13": {
            "path":
                str(CELL87_SUMMARY_PATH),

            "sha256":
                CELL87_SHA256,
        },
    },

    "outputs": {
        "protocol":
            str(PROTOCOL_PATH),

        "summary":
            str(SUMMARY_PATH),

        "manifest":
            str(MANIFEST_PATH),
    },

    "protocol_sha256":
        PROTOCOL_SHA256,

    "design_only":
        True,

    "created_utc":
        SUMMARY["created_utc"],
}


with open(
    PROTOCOL_PATH,
    "w",
    encoding="utf-8"
) as f:
    json.dump(
        PROTOCOL,
        f,
        ensure_ascii=False,
        indent=2
    )


with open(
    SUMMARY_PATH,
    "w",
    encoding="utf-8"
) as f:
    json.dump(
        SUMMARY,
        f,
        ensure_ascii=False,
        indent=2
    )


with open(
    MANIFEST_PATH,
    "w",
    encoding="utf-8"
) as f:
    json.dump(
        MANIFEST,
        f,
        ensure_ascii=False,
        indent=2
    )


print(f"✓ Protocol : {PROTOCOL_PATH}")
print(f"✓ Summary  : {SUMMARY_PATH}")
print(f"✓ Manifest : {MANIFEST_PATH}")


# ================================================================================================================
# STEP 27 — INDEPENDENT READ-BACK
# ================================================================================================================

print("\n[STEP 27] INDEPENDENT READ-BACK")
print("-" * 120)

assert PROTOCOL_PATH.exists()
assert SUMMARY_PATH.exists()
assert MANIFEST_PATH.exists()

with open(
    PROTOCOL_PATH,
    "r",
    encoding="utf-8"
) as f:
    protocol_reload = json.load(f)

with open(
    SUMMARY_PATH,
    "r",
    encoding="utf-8"
) as f:
    summary_reload = json.load(f)

with open(
    MANIFEST_PATH,
    "r",
    encoding="utf-8"
) as f:
    manifest_reload = json.load(f)


protocol_reload_serialized = json.dumps(
    protocol_reload,
    ensure_ascii=False,
    sort_keys=True,
    separators=(",", ":"),
)

PROTOCOL_RELOAD_SHA256 = hashlib.sha256(
    protocol_reload_serialized.encode("utf-8")
).hexdigest()

print(
    f"Original protocol SHA256  : "
    f"{PROTOCOL_SHA256}"
)

print(
    f"Read-back protocol SHA256 : "
    f"{PROTOCOL_RELOAD_SHA256}"
)

assert (
    PROTOCOL_RELOAD_SHA256
    == PROTOCOL_SHA256
)

assert (
    summary_reload["experiment_id"]
    == EXPERIMENT_ID
)

assert (
    summary_reload["cell_id"]
    == CELL_ID
)

assert (
    summary_reload["cell_version"]
    == CELL_VERSION
)

assert (
    summary_reload["design_only"]
    is True
)

assert (
    manifest_reload["protocol_sha256"]
    == PROTOCOL_SHA256
)

assert (
    manifest_reload["design_only"]
    is True
)

print("✓ Protocol round-trip stable")
print("✓ Summary identity stable")
print("✓ Manifest identity stable")


# ================================================================================================================
# STEP 28 — FINAL SCIENTIFIC GATES
# ================================================================================================================

print("\n[STEP 28] FINAL SCIENTIFIC GATES")
print("-" * 120)

FINAL_GATES = {
    "scientific_value":
        SCIENTIFIC_VALUE["status"] == "PASS",

    "cell24_unique_content_identification":
        len(matching_cell24) == 1,

    "cell24_gap_state":
        "GAP_PARTIALLY_OVERLAPPING"
        in flatten_json_values(cell24),

    "cell24_alignment":
        "BOUNDARY_ALIGNED_PENDING_EMPIRICAL_VALIDATION"
        in flatten_json_values(cell24),

    "cell90_verified":
        (
            cell90_id
            == "CELL90_ARCHITECTURE_RECONSIDERATION_GATE_V003"
            and
            cell90_status
            == "COMPLETED_READ_ONLY"
        ),

    "cell90_scope":
        cell90_scope
        == "FROZEN_Q82_01_Q82_10",

    "cell87_verified":
        cell87_decision
        == "STOP_PROMPT_REVISION_LOOP",

    "baseline_defined":
        "BASELINE" in CONDITIONS,

    "candidate_a_defined":
        "CANDIDATE_A" in CONDITIONS,

    "candidate_b_defined":
        "CANDIDATE_B" in CONDITIONS,

    "candidate_c_deferred":
        CONDITIONS["CANDIDATE_C"]["phase"]
        == "LATER_STUDY",

    "scope_frozen":
        SCOPE["expansion_allowed"] is False,

    "targeted_improvement_required":
        ACCEPTANCE_CRITERIA[
            "targeted_failure_mode_improvement_required"
        ] is True,

    "noninferiority_not_enough":
        ACCEPTANCE_CRITERIA[
            "baseline_noninferiority_alone_is_not_pass"
        ] is True,

    "mechanical_sd_threshold":
        ACCEPTANCE_CRITERIA[
            "mechanical_sd_max"
        ] == 0.10,

    "reproducibility_separate":
        REPRODUCIBILITY[
            "reported_separately"
        ] is True,

    "stop_rule_defined":
        len(STOP_RULE) == 4,

    "firewall_clean":
        (
            FIREWALL["read_only_upstream"] is True
            and
            all(
                value is False
                for key, value in FIREWALL.items()
                if key != "read_only_upstream"
            )
        ),

    "artifacts_exist":
        (
            PROTOCOL_PATH.exists()
            and SUMMARY_PATH.exists()
            and MANIFEST_PATH.exists()
        ),

    "protocol_roundtrip":
        PROTOCOL_RELOAD_SHA256
        == PROTOCOL_SHA256,
}


for key, value in FINAL_GATES.items():
    print(
        f"{key:48s}: "
        f"{'PASS' if value else 'FAIL'}"
    )


assert all(
    FINAL_GATES.values()
)


# ================================================================================================================
# STEP 29 — FINAL STATUS
# ================================================================================================================

print("\n" + "=" * 120)
print("CELL 25 V003 — FINAL STATUS")
print("=" * 120)

print("""
SCIENTIFIC VALUE:
    PASS

REBUILD STATUS:
    REBUILT FROM ZERO
    No V001/V002 artifact used as a decision source.

UPSTREAM:
    Cell 24 V002:
        GAP_PARTIALLY_OVERLAPPING
        BOUNDARY_ALIGNED_PENDING_EMPIRICAL_VALIDATION

    Cell 90 V003:
        ARCHITECTURE_RECONSIDERATION_WARRANTED
        FROZEN_Q82_01_Q82_10

    Cell 87.13:
        STOP_PROMPT_REVISION_LOOP
        CURRENT ARCHITECTURE = FROZEN BASELINE

PHASE 1:
    BASELINE
    CANDIDATE A = Schema-Constrained Decoding
    CANDIDATE B = Decoupled Extraction → Independent Judgment
                   → Deterministic Aggregation

    CANDIDATE C = DEFERRED

SCOPE:
    Q82_01 → Q82_10 ONLY

PASS PRINCIPLE:
    Targeted improvement is REQUIRED.
    No unacceptable regression.
    No new schema failure.
    Mechanical SD <= 0.10.
    Non-inferiority alone is NOT sufficient.

FIREWALL:
    Dataset unchanged
    199 cohort untouched
    400 validation closed
    Q82 unchanged
    Prompt Revision Loop closed
    No adjudication
    No Gold / Y+ / Y- / Preference
    No training
    No inference
    No architecture implementation
    No population claim
    No causal claim
    Broad novelty claim blocked

DESIGN STATUS:
    READY FOR FREEZE

NEXT GATE:
    GO_TO_CONTROLLED_ARCHITECTURE_COMPARISON_EXECUTION

FINAL RESULT:
    CELL 25 V003 = PASS
    — SEPARATE CONTROLLED ARCHITECTURE STUDY DESIGN FROZEN
""")

print(
    f"Protocol SHA256 : {PROTOCOL_SHA256}"
)

print(
    f"Output directory: {CELL25_DIR}"
)

print("=" * 120)
```
